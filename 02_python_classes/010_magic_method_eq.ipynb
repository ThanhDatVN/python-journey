## The __eq__() Method — Phương thức __eq__()

`__eq__()` định nghĩa cách so sánh **bằng nhau** giữa hai object khi dùng toán tử `==`.

**Mặc định**, `==` giữa hai object chỉ trả về `True` khi chúng là **cùng một object** — dù dữ liệu giống hệt nhau:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Emil", 36)
p2 = Person("Emil", 36)

print(p1 == p2)
```

Cài đặt `__eq__()` để so sánh theo **dữ liệu**:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __eq__(self, other):
        return self.name == other.name and self.age == other.age

p1 = Person("Emil", 36)
p2 = Person("Emil", 36)
p3 = Person("Tobias", 25)

print(p1 == p2)
print(p1 == p3)
print(p1 != p3)   # != tự động là phủ định của __eq__
```

## Comparing with Other Types — So sánh với kiểu khác

Nên kiểm tra kiểu của `other`; nếu không so sánh được, trả về `NotImplemented` để Python xử lý tiếp:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __eq__(self, other):
        if not isinstance(other, Person):
            return NotImplemented
        return self.name == other.name and self.age == other.age

p1 = Person("Emil", 36)
print(p1 == "Emil")
print(p1 == Person("Emil", 36))
```

---

## 🧠 Ghi nhớ nhanh

> - `__eq__(self, other)` → định nghĩa `==` theo **dữ liệu**; `!=` tự suy ra.
> - Mặc định `==` giữa object chỉ đúng khi là **cùng một object** (như `is`).
> - Kiểu khác → `return NotImplemented`.
