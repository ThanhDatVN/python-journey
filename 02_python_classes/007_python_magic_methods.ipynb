## What are Magic Methods? — Magic method là gì?

**Magic methods** (còn gọi là *dunder methods* — viết tắt của "**d**ouble **under**score") là các phương thức đặc biệt có tên bắt đầu và kết thúc bằng **hai dấu gạch dưới**, ví dụ `__init__()`, `__str__()`.

Bạn **không gọi trực tiếp** chúng: Python **tự động gọi** khi một hành động tương ứng xảy ra — tạo object, `print()`, so sánh `==`, cộng `+`, `len()`…

```python
class Person:
    def __init__(self, name, age):      # gọi khi tạo object
        self.name = name
        self.age = age

    def __str__(self):                  # gọi khi print(object)
        return f"{self.name} ({self.age})"

p1 = Person("Emil", 36)
print(p1)
```

## List of Magic Methods — Danh sách magic method

| Phương thức | Được gọi khi | Mô tả | Bài |
|---|---|---|---|
| `__init__()` | Tạo object | Khởi tạo thuộc tính | [__init__](003_python_init_method.ipynb) |
| `__str__()` | `print(obj)`, `str(obj)` | Chuỗi dễ đọc cho người dùng | [__str__](008_magic_method_str.ipynb) |
| `__repr__()` | `repr(obj)`, hiển thị trong console | Chuỗi "chính thức" cho lập trình viên | [__repr__](009_magic_method_repr.ipynb) |
| `__eq__()` | `obj1 == obj2` | So sánh bằng | [__eq__](010_magic_method_eq.ipynb) |
| `__add__()` | `obj1 + obj2` | Phép cộng | [__add__](011_magic_method_add.ipynb) |
| `__len__()` | `len(obj)` | Độ dài | [__len__](012_magic_method_len.ipynb) |
| `__lt__()` | `obj1 < obj2` | So sánh nhỏ hơn | [__lt__](013_magic_method_lt.ipynb) |
| `__contains__()` | `x in obj` | Kiểm tra thành viên | [__contains__](014_magic_method_contains.ipynb) |
| `__call__()` | `obj()` | Gọi object như hàm | [__call__](015_magic_method_call.ipynb) |

## ➕ Mở rộng: Python dùng magic method ở khắp nơi

Ngay cả kiểu có sẵn cũng hoạt động nhờ magic method:

```python
print((5).__add__(3))          # 5 + 3
print("abc".__len__())         # len("abc")
print([1, 2, 3].__contains__(2))   # 2 in [1, 2, 3]
print(sorted(m for m in dir(list) if m.startswith("__"))[:12])
```

## 🧠 Ghi nhớ nhanh

> - Magic method = phương thức dạng `__ten__()`, Python **tự gọi** khi có hành động tương ứng.
> - `__init__` tạo · `__str__`/`__repr__` hiển thị · `__eq__`/`__lt__` so sánh · `__add__` cộng · `__len__` độ dài · `__contains__` `in` · `__call__` gọi như hàm.
