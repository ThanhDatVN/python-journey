## Python Encapsulation — Đóng gói

**Đóng gói (encapsulation)** là việc **bảo vệ dữ liệu** bên trong class, và chỉ cho phép truy cập/thay đổi qua các phương thức được kiểm soát.

Bằng cách đó, bạn tránh được việc dữ liệu bị sửa sai từ bên ngoài.

## Private Properties — Thuộc tính private

Thêm **hai dấu gạch dưới** `__` vào trước tên thuộc tính để biến nó thành **private** — không truy cập trực tiếp từ bên ngoài được:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age    # thuộc tính private

p1 = Person("Emil", 25)
print(p1.name)
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age

p1 = Person("Emil", 25)
print(p1.__age)
```

## Get Private Property Value — Lấy giá trị thuộc tính private

Tạo phương thức **getter** để đọc giá trị:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age

    def get_age(self):
        return self.__age

p1 = Person("Tobias", 25)
print(p1.get_age())
```

## Set Private Property Value — Gán giá trị thuộc tính private

Tạo phương thức **setter** để sửa giá trị — kèm **kiểm tra hợp lệ**:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age

    def get_age(self):
        return self.__age

    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age must be positive")

p1 = Person("Tobias", 25)
print(p1.get_age())

p1.set_age(26)
print(p1.get_age())

p1.set_age(-5)
print(p1.get_age())
```

## Why Use Encapsulation? — Vì sao cần đóng gói?

- **Bảo vệ dữ liệu**: tránh bị sửa tuỳ tiện.
- **Kiểm tra hợp lệ**: chỉ chấp nhận giá trị đúng.
- **Linh hoạt**: thay đổi cách lưu bên trong mà không ảnh hưởng code bên ngoài.
- **Kiểm soát**: quyết định thuộc tính nào chỉ đọc, thuộc tính nào sửa được.

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.__grade = 0

    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade
        else:
            print("Grade must be between 0 and 100")

    def get_grade(self):
        return self.__grade

    def get_status(self):
        return "Passed" if self.__grade >= 60 else "Failed"

student = Student("Emil")
student.set_grade(85)
print(student.get_grade())
print(student.get_status())
student.set_grade(150)
```

## Protected Properties — Thuộc tính protected

**Một dấu gạch dưới** `_` là **quy ước** cho thuộc tính *protected* — "chỉ nên dùng nội bộ". Python **không chặn** truy cập; đây chỉ là lời nhắc cho lập trình viên:

```python
class Person:
    def __init__(self, name, salary):
        self.name = name
        self._salary = salary   # protected

p1 = Person("Linus", 50000)
print(p1._salary)   # vẫn truy cập được, nhưng không nên
```

## Private Methods — Phương thức private

Phương thức cũng có thể private bằng `__`:

```python
class Calculator:
    def __init__(self):
        self.result = 0

    def __validate(self, num):
        if not isinstance(num, (int, float)):
            return False
        return True

    def add(self, num):
        if self.__validate(num):
            self.result += num
        else:
            print("Invalid number")

calc = Calculator()
calc.add(10)
calc.add(5)
calc.add("x")
print(calc.result)
```

## Name Mangling — Đổi tên ngầm

Thực ra Python **không** ẩn hoàn toàn thuộc tính private. Nó **tự đổi tên** `__age` thành `_TenClass__age` (gọi là *name mangling*):

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age

p1 = Person("Emil", 30)
print(p1._Person__age)   # truy cập được, nhưng KHÔNG nên làm
```

## ➕ Mở rộng: @property — getter/setter kiểu Pythonic

```python
class Person:
    def __init__(self, age):
        self.age = age          # gọi setter bên dưới

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age must be positive")
        self.__age = value

p = Person(20)
p.age = 21          # trông như gán thuộc tính, nhưng có kiểm tra
print(p.age)
```

> 🧪 **Try it Yourself** — Viết class BankAccount với __balance private, deposit() chỉ nhận số dương, get_balance().

```python
class BankAccount:
    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Số tiền không hợp lệ")

    def get_balance(self):
        return self.__balance

acc = BankAccount()
acc.deposit(100)
acc.deposit(-50)
print(acc.get_balance())
```

## 🧠 Ghi nhớ nhanh

> - `__ten` → **private** (không truy cập trực tiếp từ ngoài) · `_ten` → **protected** (chỉ là quy ước).
> - Đọc/sửa qua **getter/setter** để kiểm tra hợp lệ.
> - Name mangling: `__age` thành `_Person__age`.
> - Cách Pythonic: `@property` + `@x.setter`.
