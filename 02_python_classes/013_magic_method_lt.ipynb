## The __lt__() Method — Phương thức __lt__()

`__lt__()` (*less than*) định nghĩa phép so sánh **nhỏ hơn** `<` giữa hai object:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __lt__(self, other):
        return self.age < other.age

p1 = Person("Emil", 36)
p2 = Person("Tobias", 25)

print(p1 < p2)
print(p2 < p1)
```

## Sorting Objects — Sắp xếp object

Khi có `__lt__()`, các hàm như `sorted()`, `min()`, `max()` hoạt động được với object:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __lt__(self, other):
        return self.age < other.age

    def __repr__(self):
        return f"{self.name}({self.age})"

people = [Person("Emil", 36), Person("Tobias", 25), Person("Linus", 30)]
print(sorted(people))
print(min(people), max(people))
```

## ➕ Mở rộng: Các phép so sánh khác và total_ordering

| Toán tử | Magic method |
|---|---|
| `<` | `__lt__()` |
| `<=` | `__le__()` |
| `>` | `__gt__()` |
| `>=` | `__ge__()` |
| `==` | `__eq__()` |
| `!=` | `__ne__()` |

Decorator `functools.total_ordering` tự sinh các phép còn lại khi bạn có `__eq__` và `__lt__`:

```python
from functools import total_ordering

@total_ordering
class Score:
    def __init__(self, value):
        self.value = value

    def __eq__(self, other):
        return self.value == other.value

    def __lt__(self, other):
        return self.value < other.value

print(Score(8) >= Score(7), Score(5) > Score(9))
```

---

## 🧠 Ghi nhớ nhanh

> - `__lt__(self, other)` → định nghĩa `<`.
> - Có `__lt__` → dùng được `sorted()`, `min()`, `max()` với object.
> - `@functools.total_ordering` + `__eq__` + `__lt__` → có đủ mọi phép so sánh.
