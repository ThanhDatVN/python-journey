Từ *"polymorphism"* (đa hình) nghĩa là **"nhiều hình dạng"**. Trong lập trình, nó chỉ các phương thức/hàm/toán tử **cùng tên** nhưng có thể thực thi trên **nhiều object hoặc class khác nhau**.

## Function Polymorphism — Đa hình với hàm

Ví dụ điển hình là hàm `len()` — dùng được với nhiều kiểu object khác nhau.

### String

Với chuỗi, `len()` trả về số ký tự:

```python
x = "Hello World!"
print(len(x))
```

### Tuple

Với tuple, `len()` trả về số phần tử:

```python
mytuple = ("apple", "banana", "cherry")
print(len(mytuple))
```

### Dictionary

Với dictionary, `len()` trả về số cặp key/value:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(len(thisdict))
```

## Class Polymorphism — Đa hình với class

Đa hình thường thấy trong các class có **phương thức cùng tên**.

Ví dụ ba class `Car`, `Boat`, `Plane` đều có phương thức `move()`:

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Drive!")

class Boat:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Sail!")

class Plane:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Fly!")

car1 = Car("Ford", "Mustang")
boat1 = Boat("Ibiza", "Touring 20")
plane1 = Plane("Boeing", "747")

for x in (car1, boat1, plane1):
    x.move()
```

Nhờ đa hình, ta có thể gọi **cùng một phương thức** cho cả ba class trong một vòng lặp.

## Inheritance Class Polymorphism — Đa hình với kế thừa

Nếu tạo class cha `Vehicle` và cho `Car`, `Boat`, `Plane` kế thừa, các class con sẽ có sẵn thuộc tính và phương thức của cha, đồng thời có thể **ghi đè** chúng:

```python
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def move(self):
        print("Move!")

class Car(Vehicle):
    pass

class Boat(Vehicle):
    def move(self):
        print("Sail!")

class Plane(Vehicle):
    def move(self):
        print("Fly!")

car1 = Car("Ford", "Mustang")
boat1 = Boat("Ibiza", "Touring 20")
plane1 = Plane("Boeing", "747")

for x in (car1, boat1, plane1):
    print(x.brand)
    print(x.model)
    x.move()
```

- `Car` để trống nên **kế thừa** `move()` của `Vehicle` → in "Move!".
- `Boat` và `Plane` **ghi đè** `move()`.

## ➕ Mở rộng: Duck typing

Python không cần class chung để dùng đa hình: *"Nếu nó đi như vịt và kêu như vịt, thì nó là vịt"*:

```python
class Duck:
    def speak(self):
        return "Quack!"

class Robot:
    def speak(self):
        return "Beep!"

for thing in (Duck(), Robot()):
    print(thing.speak())
```

> 🧪 **Try it Yourself** — Thêm class Bicycle(Vehicle) với move() in "Pedal!" rồi đưa vào vòng lặp.

```python
class Bicycle(Vehicle):
    def move(self):
        print("Pedal!")

for v in (Car("Ford", "Focus"), Bicycle("Giant", "ATX")):
    v.move()
```

## 🧠 Ghi nhớ nhanh

> - Đa hình = **cùng tên** phương thức/hàm, **nhiều cách thực hiện**.
> - Hàm `len()` là ví dụ: dùng cho str, tuple, dict…
> - Nhiều class có cùng phương thức → gọi chung trong một vòng lặp.
> - Kế thừa + ghi đè (override) → class con tuỳ biến hành vi của class cha.
