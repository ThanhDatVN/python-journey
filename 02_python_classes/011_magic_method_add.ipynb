## The __add__() Method — Phương thức __add__()

`__add__()` định nghĩa điều gì xảy ra khi dùng toán tử `+` giữa hai object. Kỹ thuật này gọi là **nạp chồng toán tử (operator overloading)**.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

p1 = Point(1, 2)
p2 = Point(3, 4)
p3 = p1 + p2

print(p3)
```

Không có `__add__()`, phép cộng sẽ báo lỗi:

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

print(Point(1, 2) + Point(3, 4))
```

## Adding Different Types — Cộng với kiểu khác

`__add__()` có thể xử lý cả trường hợp cộng với số:

```python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __add__(self, other):
        if isinstance(other, Money):
            return Money(self.amount + other.amount)
        return Money(self.amount + other)

    def __str__(self):
        return f"{self.amount:,} đ"

wallet = Money(50_000) + Money(20_000) + 5_000
print(wallet)
```

## ➕ Mở rộng: Các toán tử khác

| Toán tử | Magic method |
|---|---|
| `-` | `__sub__()` |
| `*` | `__mul__()` |
| `/` | `__truediv__()` |
| `//` | `__floordiv__()` |
| `%` | `__mod__()` |
| `**` | `__pow__()` |

```python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    def __mul__(self, k):
        return Vector(self.x * k, self.y * k)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

print(Vector(5, 5) - Vector(1, 2))
print(Vector(1, 2) * 3)
```

---

## 🧠 Ghi nhớ nhanh

> - `__add__(self, other)` → định nghĩa phép `+` (operator overloading).
> - Nên **trả về object mới**, không sửa object gốc.
> - Toán tử khác: `__sub__` `-` · `__mul__` `*` · `__truediv__` `/`.
