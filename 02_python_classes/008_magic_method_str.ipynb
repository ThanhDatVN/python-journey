# Python __str__() Method

> 🎯 **Trong bài này:** The __str__() Method · __str__() Must Return a String

## The __str__() Method — Phương thức __str__()

`__str__()` quyết định object được **hiển thị** thế nào khi bạn `print()` nó hoặc chuyển nó thành chuỗi bằng `str()`.

**Không có** `__str__()`, Python in ra tên class và địa chỉ bộ nhớ:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Emil", 36)
print(p1)
```

**Có** `__str__()`, kết quả dễ đọc hơn nhiều:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} ({self.age})"

p1 = Person("Tobias", 25)
print(p1)
print(str(p1))
print(f"Thông tin: {p1}")
```

## __str__() Must Return a String — Phải trả về chuỗi

`__str__()` **bắt buộc trả về chuỗi**. Trả về kiểu khác sẽ gây `TypeError`:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return self.age   # sai: trả về số

p1 = Person("Linus", 30)
print(p1)
```

Sửa bằng cách chuyển sang chuỗi:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return str(self.age)

p1 = Person("Linus", 30)
print(p1)
```

> 🧪 **Try it Yourself** — Viết __str__ cho class Product để in ra dạng: "Laptop - 15,000,000 đ".

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name} - {self.price:,} đ"

print(Product("Laptop", 15_000_000))
```

## 🧠 Ghi nhớ nhanh

> - `__str__()` được gọi bởi `print(obj)`, `str(obj)`, f-string.
> - Mục đích: chuỗi **dễ đọc cho người dùng**.
> - ⚠️ Bắt buộc **trả về str**, nếu không → `TypeError`.
