## The __repr__() Method — Phương thức __repr__()

`__repr__()` trả về **chuỗi biểu diễn "chính thức"** của object — hướng tới **lập trình viên**, dùng khi debug. Lý tưởng nhất, chuỗi này trông giống **code Python tạo lại được object**.

`__repr__()` được gọi khi dùng hàm `repr()`, hoặc khi object được hiển thị trong console/notebook mà không dùng `print()`:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

p1 = Person("Emil", 36)
print(repr(p1))
p1
```

## __str__() vs __repr__()

- `__str__()` → chuỗi **thân thiện với người dùng**, dùng bởi `print()`.
- `__repr__()` → chuỗi **chi tiết, rõ ràng cho lập trình viên**.

Khi có cả hai, `print()` dùng `__str__()`, còn `repr()` dùng `__repr__()`:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} ({self.age})"

    def __repr__(self):
        return f"Person('{self.name}', {self.age})"

p1 = Person("Tobias", 25)
print(p1)
print(repr(p1))
```

## __repr__() as Fallback — __repr__ làm phương án dự phòng

Nếu class **không có** `__str__()`, Python dùng `__repr__()` cho `print()`:

```python
class Person:
    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f"Person('{self.name}')"

print(Person("Linus"))
```

> 💡 `__repr__()` cũng được dùng khi object nằm **trong list/dict**:

```python
class Point:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

points = [Point(1, 2), Point(3, 4)]
print(points)
```

## 🧠 Ghi nhớ nhanh

> - `__repr__()` → biểu diễn **chính thức, cho lập trình viên** (nên trông như code tạo object).
> - Gọi bởi `repr(obj)`, hiển thị trong console, và khi object nằm **trong list/dict**.
> - Không có `__str__()` → `print()` dùng `__repr__()`.
> - Mẹo: luôn viết `__repr__`; thêm `__str__` khi cần bản hiển thị thân thiện.
