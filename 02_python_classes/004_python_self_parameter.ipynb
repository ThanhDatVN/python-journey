## The self Parameter — Tham số self

Tham số `self` là một **tham chiếu tới object hiện tại** của class. Nó dùng để truy cập các thuộc tính và phương thức thuộc về object đó:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print("Hello, my name is " + self.name)

p1 = Person("Emil", 25)
p1.greet()
```

> 📌 `self` phải là **tham số đầu tiên** của mọi phương thức trong class, kể cả `__init__()`.

## Why Use self? — Vì sao cần self?

Nếu không có `self`, Python không biết bạn đang nói tới thuộc tính của **object nào**. `self` gắn phương thức với **object cụ thể** đang gọi nó:

```python
class Person:
    def __init__(self, name):
        self.name = name

    def printname(self):
        print(self.name)

p1 = Person("Tobias")
p2 = Person("Linus")

p1.printname()
p2.printname()
```

Khi gọi `p1.printname()`, Python thực chất gọi `Person.printname(p1)` — `self` chính là `p1`:

```python
class Person:
    def __init__(self, name):
        self.name = name

    def printname(self):
        print(self.name)

p1 = Person("Tobias")
Person.printname(p1)
```

## self Does Not Have to Be Named "self" — Không bắt buộc tên self

Tham số đầu tiên **không bắt buộc** phải tên là `self` — đặt tên gì cũng được, miễn là nó là **tham số đầu tiên**:

```python
class Person:
    def __init__(myobject, name, age):
        myobject.name = name
        myobject.age = age

    def greet(abc):
        print("Hello, my name is " + abc.name)

p1 = Person("Emil", 36)
p1.greet()
```

> 💡 Dù được phép, **hãy luôn dùng tên `self`** — đó là quy ước mà mọi lập trình viên Python đều hiểu.

## Accessing Properties with self — Truy cập thuộc tính bằng self

```python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def display_info(self):
        print(f"{self.year} {self.brand} {self.model}")

car1 = Car("Toyota", "Corolla", 2020)
car1.display_info()
```

## Calling Methods with self — Gọi phương thức bằng self

Một phương thức có thể gọi phương thức khác của cùng object qua `self`:

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        return "Hello, " + self.name

    def welcome(self):
        message = self.greet()
        print(message + "! Welcome to our website.")

p1 = Person("Tobias")
p1.welcome()
```

```python
class Person:
    def greet():          # quên self
        print("Hello")

p = Person()
p.greet()
```

> 🧪 **Try it Yourself** — Thêm phương thức have_birthday() tăng tuổi lên 1, rồi gọi introduce() để in lại.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Tôi là {self.name}, {self.age} tuổi")

    def have_birthday(self):
        self.age += 1

p = Person("An", 20)
p.have_birthday()
p.introduce()
```

## 🧠 Ghi nhớ nhanh

> - `self` = **object hiện tại**; là **tham số đầu tiên** của mọi phương thức.
> - `obj.method()` ⇔ `Class.method(obj)` → Python tự truyền `obj` vào `self`.
> - Dùng `self.thuoc_tinh` và `self.phuong_thuc()` bên trong class.
> - ⚠️ Quên `self` trong định nghĩa phương thức → `TypeError` khi gọi.
