## Class Methods — Phương thức

**Phương thức (method)** là hàm thuộc về một class, định nghĩa **hành vi** của object.

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello, my name is " + self.name)

p1 = Person("Emil")
p1.greet()
```

## Methods with Parameters — Phương thức có tham số

Ngoài `self`, phương thức có thể nhận thêm tham số:

```python
class Calculator:
    def add(self, a, b):
        return a + b

    def multiply(self, a, b):
        return a * b

calc = Calculator()
print(calc.add(5, 3))
print(calc.multiply(4, 7))
```

## Methods Accessing Properties — Phương thức đọc thuộc tính

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_info(self):
        return f"{self.name} is {self.age} years old"

p1 = Person("Tobias", 28)
print(p1.get_info())
```

## Methods Modifying Properties — Phương thức thay đổi thuộc tính

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def celebrate_birthday(self):
        self.age += 1
        print(f"Happy birthday! You are now {self.age}")

p1 = Person("Linus", 25)
p1.celebrate_birthday()
p1.celebrate_birthday()
```

## The __str__() Method — Phương thức __str__()

`__str__()` quyết định object được hiển thị thế nào khi in ra (hoặc chuyển thành chuỗi).

**Không có** `__str__()`, in object ra một chuỗi khó hiểu:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Emil", 36)
print(p1)
```

**Có** `__str__()`:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} ({self.age})"

p1 = Person("Tobias", 36)
print(p1)
```

## Multiple Methods — Nhiều phương thức

Một class có thể có nhiều phương thức phối hợp với nhau:

```python
class Playlist:
    def __init__(self, name):
        self.name = name
        self.songs = []

    def add_song(self, song):
        self.songs.append(song)
        print(f"Added: {song}")

    def remove_song(self, song):
        if song in self.songs:
            self.songs.remove(song)
            print(f"Removed: {song}")

    def show_songs(self):
        print(f"Playlist '{self.name}':")
        for song in self.songs:
            print(f"- {song}")

my_playlist = Playlist("Favorites")
my_playlist.add_song("Bohemian Rhapsody")
my_playlist.add_song("Stairway to Heaven")
my_playlist.show_songs()
```

## Delete Methods — Xoá phương thức

Có thể xoá phương thức khỏi class bằng `del`:

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello!")

p1 = Person("Emil")

del Person.greet

p1.greet()  # lỗi: phương thức greet không còn tồn tại
```

## ➕ Mở rộng: @classmethod và @staticmethod

Ngoài phương thức thường (nhận `self`), Python còn có:

- `@classmethod` — nhận `cls` (chính class), hay dùng làm "hàm khởi tạo thay thế".
- `@staticmethod` — không nhận `self` hay `cls`, chỉ là hàm tiện ích đặt trong class.

```python
class Date:
    def __init__(self, day, month, year):
        self.day, self.month, self.year = day, month, year

    @classmethod
    def from_string(cls, text):
        d, m, y = map(int, text.split("/"))
        return cls(d, m, y)

    @staticmethod
    def is_leap(year):
        return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)

d = Date.from_string("03/10/2026")
print(d.day, d.month, d.year)
print(Date.is_leap(2028))
```

> 🧪 **Try it Yourself** — Viết class BankAccount có deposit(amount), withdraw(amount) (không cho rút quá số dư) và __str__.

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Số dư không đủ!")
        else:
            self.balance -= amount

    def __str__(self):
        return f"{self.owner}: {self.balance:,} đ"

acc = BankAccount("An", 100_000)
acc.deposit(50_000)
acc.withdraw(500_000)
print(acc)
```

## 🧠 Ghi nhớ nhanh

> - Phương thức = hàm trong class, tham số đầu là `self`.
> - Phương thức có thể **đọc** (`self.x`) và **sửa** (`self.x = ...`) thuộc tính.
> - `__str__()` → quyết định cách `print(obj)` hiển thị.
> - `@classmethod` nhận `cls` · `@staticmethod` không nhận gì đặc biệt.
