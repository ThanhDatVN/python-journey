## Class Properties — Thuộc tính

**Thuộc tính (property)** là các biến thuộc về một class. Chúng lưu **dữ liệu** của mỗi object.

Tạo thuộc tính trong `__init__()`:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Emil", 36)

print(p1.name)
print(p1.age)
```

## Access Properties — Truy cập thuộc tính

Dùng **dấu chấm** để truy cập thuộc tính của object:

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

car1 = Car("Toyota", "Corolla")

print(car1.brand)
print(car1.model)
```

## Modify Properties — Sửa thuộc tính

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Tobias", 25)
print(p1.age)

p1.age = 26
print(p1.age)
```

## Delete Properties — Xoá thuộc tính

Xoá thuộc tính bằng từ khoá `del`:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Linus", 30)

del p1.age

print(p1.name)  # vẫn hoạt động
print(p1.age)   # lỗi: thuộc tính age không còn tồn tại
```

## Class Properties vs Object Properties — Thuộc tính class và thuộc tính object

- Thuộc tính khai báo **bên trong class, ngoài các phương thức** là **thuộc tính class** — **dùng chung** cho mọi object.
- Thuộc tính tạo bằng `self.` trong `__init__()` là **thuộc tính object** — **riêng** cho từng object.

```python
class Person:
    species = "Human"   # thuộc tính class

    def __init__(self, name):
        self.name = name  # thuộc tính object

p1 = Person("Emil")
p2 = Person("Tobias")

print(p1.name, p1.species)
print(p2.name, p2.species)
```

## Modifying Class Properties — Sửa thuộc tính class

Sửa thuộc tính class (qua **tên class**) sẽ ảnh hưởng tới **mọi object**:

```python
class Person:
    lastname = ""

    def __init__(self, name):
        self.name = name

p1 = Person("Linus")
p2 = Person("Emil")

Person.lastname = "Refsnes"

print(p1.lastname)
print(p2.lastname)
```

## Add New Properties — Thêm thuộc tính mới

Có thể thêm thuộc tính cho **một object cụ thể** sau khi tạo — các object khác không có thuộc tính đó:

```python
class Person:
    def __init__(self, name):
        self.name = name

p1 = Person("Tobias")
p2 = Person("Linus")

p1.age = 25
p1.city = "Oslo"

print(p1.name, p1.age, p1.city)
print(hasattr(p2, "age"))
```

## ➕ Mở rộng: Bẫy khi gán thuộc tính class qua object

Gán qua **object** (`p1.species = ...`) sẽ tạo **thuộc tính object mới** che mất thuộc tính class, chứ không sửa thuộc tính class:

```python
class Person:
    species = "Human"

p1 = Person()
p2 = Person()
p1.species = "Robot"

print(p1.species, p2.species, Person.species)
```

> 🧪 **Try it Yourself** — Tạo class Student có thuộc tính class school dùng chung; đổi tên trường qua class và kiểm tra.

```python
class Student:
    school = "THPT ABC"

    def __init__(self, name):
        self.name = name

s1 = Student("An")
s2 = Student("Bình")
Student.school = "THPT XYZ"
print(s1.name, s1.school)
print(s2.name, s2.school)
```

## 🧠 Ghi nhớ nhanh

> - Truy cập/sửa: `obj.thuoc_tinh` · xoá: `del obj.thuoc_tinh`.
> - **Thuộc tính class** (khai báo trong class) → dùng chung · **thuộc tính object** (`self.x`) → riêng từng object.
> - Sửa thuộc tính class qua **tên class** → mọi object thay đổi theo.
> - Có thể thêm thuộc tính mới cho một object bất kỳ lúc nào.
