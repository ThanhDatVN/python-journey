## What is OOP? — OOP là gì?

**OOP** là viết tắt của *Object-Oriented Programming* — **Lập trình hướng đối tượng**.

Python là ngôn ngữ hướng đối tượng: nó cho phép bạn tổ chức code bằng **class** (lớp) và **object** (đối tượng).

## Advantages of OOP — Ưu điểm của OOP

- Cấu trúc chương trình **rõ ràng**.
- Code **dễ bảo trì**, dễ tái sử dụng và dễ gỡ lỗi.
- Giúp code tuân theo nguyên tắc **DRY** (*Don't Repeat Yourself* — đừng lặp lại chính mình).
- Có thể xây dựng các ứng dụng tái sử dụng được với ít code hơn.

> 💡 Nguyên tắc **DRY**: gom những đoạn code dùng chung vào một chỗ rồi tái sử dụng, thay vì viết đi viết lại.

## What are Classes and Objects? — Class và Object là gì?

**Class** và **object** là hai khái niệm cốt lõi của OOP.

- **Class** là **bản thiết kế (khuôn mẫu)** định nghĩa object sẽ có dữ liệu gì và làm được gì.
- **Object** là một **thực thể cụ thể** được tạo ra từ class.

| Class | Object |
|---|---|
| Fruit (Trái cây) | Apple, Banana, Mango |
| Car (Xe hơi) | Volvo, Audi, Toyota |

Khi tạo object, nó **thừa hưởng** tất cả biến (thuộc tính) và hàm (phương thức) được định nghĩa trong class.

Xem trước một class đơn giản (bạn sẽ học chi tiết ở các bài tiếp theo):

```python
class Car:
    def __init__(self, brand):
        self.brand = brand

    def honk(self):
        print(self.brand, "says: Beep beep!")

volvo = Car("Volvo")
audi = Car("Audi")

volvo.honk()
audi.honk()
```
