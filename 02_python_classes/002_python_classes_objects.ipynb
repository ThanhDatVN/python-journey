## Python Classes/Objects

Python là ngôn ngữ hướng đối tượng. **Hầu như mọi thứ trong Python đều là object**, với thuộc tính (properties) và phương thức (methods) riêng.

Một class giống như một **"hàm khởi tạo" object**, hay một **bản thiết kế** để tạo object.

## Create a Class — Tạo class

Dùng từ khoá `class` để tạo class.

Tạo class tên `MyClass` có một thuộc tính tên `x`:

````python
class MyClass:
    x = 5

print(MyClass)
````

## Create Object — Tạo object

Giờ có thể dùng class `MyClass` để tạo object.

Tạo object tên `p1` và in giá trị của `x`:

````python
class MyClass:
    x = 5

p1 = MyClass()
print(p1.x)
````

## Delete Objects — Xoá object

Xoá object bằng từ khoá `del`:

````python
class MyClass:
    x = 5

p1 = MyClass()
del p1
print(p1)
````

## Multiple Objects — Nhiều object

Có thể tạo **nhiều object** từ cùng một class. Mỗi object **độc lập** với nhau:

````python
class MyClass:
    x = 5

p1 = MyClass()
p2 = MyClass()
p3 = MyClass()

print(p1.x)
print(p2.x)
print(p3.x)
````

## The pass Statement — Câu lệnh pass

Định nghĩa class không được để trống. Nếu chưa có nội dung, đặt `pass` để tránh lỗi:

````python
class Person:
    pass

p = Person()
print(type(p))
````

## ➕ Mở rộng: Mọi thứ đều là object

````python
print(type(5))
print(type("hello"))
print(type([1, 2]))
print(isinstance(5, int))
print("hello".upper())   # chuỗi là object nên có phương thức
````

> 🧪 **Try it Yourself** — Tạo class Dog có thuộc tính sound = "Woof", tạo 2 object và in sound của mỗi con.

````python
class Dog:
    sound = "Woof"

d1 = Dog()
d2 = Dog()
print(d1.sound, d2.sound)
````

## 🧠 Ghi nhớ nhanh

> - Tạo class: `class TenClass:` (tên viết **PascalCase**).
> - Tạo object: `obj = TenClass()` · truy cập thuộc tính: `obj.x`.
> - Một class tạo được **nhiều object độc lập**.
> - `del obj` xoá object · class trống thì dùng `pass`.
