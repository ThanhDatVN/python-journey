## The __call__() Method — Phương thức __call__()

`__call__()` cho phép **gọi object như một hàm**: `obj()`.

```python
class Greeter:
    def __init__(self, greeting):
        self.greeting = greeting

    def __call__(self, name):
        return f"{self.greeting}, {name}!"

hello = Greeter("Hello")
print(hello("Emil"))
print(hello("Tobias"))
```

## Objects That Remember State — Object có trạng thái

Khác hàm thường, object "gọi được" có thể **ghi nhớ trạng thái** giữa các lần gọi:

```python
class Counter:
    def __init__(self):
        self.count = 0

    def __call__(self):
        self.count += 1
        return self.count

counter = Counter()
print(counter())
print(counter())
print(counter())
```

## callable() — Kiểm tra gọi được

Hàm có sẵn `callable()` cho biết một object có gọi được hay không:

```python
class Greeter:
    def __call__(self):
        return "Hi"

class Plain:
    pass

print(callable(Greeter()))
print(callable(Plain()))
print(callable(print))
```

```python
class Plain:
    pass

p = Plain()
p()
```

---

## 🧠 Ghi nhớ nhanh

> - `__call__()` → cho phép gọi `obj(...)` như hàm.
> - Object gọi được có thể **ghi nhớ trạng thái** (bộ đếm, cấu hình…).
> - `callable(x)` kiểm tra `x` có gọi được không.
