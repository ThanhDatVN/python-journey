# Python __len__() Method

> 🎯 **Trong bài này:** The __len__() Method · __len__() Must Return an Integer · __len__() and Truth Value

## The __len__() Method — Phương thức __len__()

`__len__()` định nghĩa giá trị trả về khi gọi hàm `len()` trên object:

```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)

my_list = Playlist(["Song A", "Song B", "Song C"])
print(len(my_list))
```

Không có `__len__()`, `len()` báo lỗi:

```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs

print(len(Playlist(["Song A"])))
```

## __len__() Must Return an Integer — Phải trả về số nguyên không âm

```python
class Weird:
    def __len__(self):
        return -1

len(Weird())
```

## __len__() and Truth Value — __len__ và giá trị logic

Nếu class không có `__bool__()`, Python dùng `__len__()` để quyết định object là `True` hay `False`: độ dài `0` → `False`:

```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs

    def __len__(self):
        return len(self.songs)

empty = Playlist([])
full = Playlist(["Song A"])

print(bool(empty), bool(full))
if not empty:
    print("Playlist is empty")
```

---

## 🧠 Ghi nhớ nhanh

> - `__len__()` → giá trị trả về của `len(obj)`.
> - ⚠️ Phải trả về **số nguyên ≥ 0**.
> - Độ dài `0` → object được coi là `False` (nếu không có `__bool__`).
