## The __init__() Method — Phương thức __init__()

Các ví dụ ở bài trước là class và object ở dạng đơn giản nhất, chưa thật hữu ích trong thực tế.

Để hiểu ý nghĩa của class, cần hiểu phương thức có sẵn `__init__()`.

Mọi class đều có phương thức `__init__()`, được **tự động gọi khi tạo object**. Dùng nó để **gán giá trị cho thuộc tính** của object, hoặc làm các thao tác cần thiết khi object được tạo:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Emil", 36)

print(p1.name)
print(p1.age)
```

> 📌 `__init__()` được gọi **tự động** mỗi khi class được dùng để tạo object mới.

## Why Use __init__()? — Vì sao dùng __init__()?

**Không** có `__init__()`, bạn phải gán thuộc tính thủ công cho từng object:

```python
class Person:
    pass

p1 = Person()
p1.name = "Tobias"
p1.age = 25

print(p1.name)
print(p1.age)
```

**Có** `__init__()`, các thuộc tính được gán ngay khi tạo object — gọn hơn và không quên thuộc tính nào:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person("Tobias", 25)

print(p1.name)
print(p1.age)
```

## Default Values in __init__() — Giá trị mặc định

Tham số của `__init__()` có thể có giá trị mặc định:

```python
class Person:
    def __init__(self, name, age=18):
        self.name = name
        self.age = age

p1 = Person("Emil")
p2 = Person("Tobias", 25)

print(p1.name, p1.age)
print(p2.name, p2.age)
```

## Multiple Parameters — Nhiều tham số

`__init__()` nhận bao nhiêu tham số cũng được:

```python
class Person:
    def __init__(self, name, age, city, country):
        self.name = name
        self.age = age
        self.city = city
        self.country = country

p1 = Person("Linus", 30, "Oslo", "Norway")

print(p1.name)
print(p1.age)
print(p1.city)
print(p1.country)
```

> 🧪 **Try it Yourself** — Tạo class Book với __init__ nhận title, author, pages; tạo 2 cuốn sách và in thông tin.

```python
class Book:
    def __init__(self, title, author, pages=100):
        self.title = title
        self.author = author
        self.pages = pages

b1 = Book("Dế Mèn phiêu lưu ký", "Tô Hoài", 150)
b2 = Book("Python cơ bản", "W3", )
print(b1.title, "-", b1.author, "-", b1.pages)
print(b2.title, "-", b2.author, "-", b2.pages)
```

---

## 🧠 Ghi nhớ nhanh

> - `__init__(self, ...)` chạy **tự động** khi tạo object (hàm khởi tạo).
> - Gán thuộc tính: `self.name = name`.
> - Tham số có thể có **giá trị mặc định**: `def __init__(self, name, age=18)`.
> - Tạo object: `Person("Emil", 36)` — không truyền `self`.
