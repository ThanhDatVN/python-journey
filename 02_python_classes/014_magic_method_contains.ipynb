## The __contains__() Method — Phương thức __contains__()

`__contains__()` định nghĩa hành vi của toán tử `in` (và `not in`) với object:

```python
class Team:
    def __init__(self, members):
        self.members = members

    def __contains__(self, name):
        return name in self.members

team = Team(["Emil", "Tobias", "Linus"])

print("Emil" in team)
print("John" in team)
print("John" not in team)
```

## Custom Membership Logic — Logic kiểm tra tuỳ chỉnh

`__contains__()` có thể chứa logic riêng, ví dụ không phân biệt hoa thường:

```python
class Team:
    def __init__(self, members):
        self.members = [m.lower() for m in members]

    def __contains__(self, name):
        return name.lower() in self.members

team = Team(["Emil", "Tobias"])
print("EMIL" in team)
```

Hoặc kiểm tra một số có nằm trong khoảng:

```python
class Range:
    def __init__(self, low, high):
        self.low, self.high = low, high

    def __contains__(self, value):
        return self.low <= value <= self.high

working_hours = Range(8, 17)
print(10 in working_hours, 20 in working_hours)
```
