## Merge Sort — Sắp xếp trộn

**Merge sort** là thuật toán **chia để trị (divide and conquer)**:

1. **Chia** mảng làm đôi, rồi tiếp tục chia đôi các mảng con cho tới khi mỗi mảng chỉ còn **một** phần tử.
2. **Trộn (merge)** các mảng con lại với nhau theo đúng thứ tự, cho tới khi được mảng hoàn chỉnh đã sắp xếp.

```text
[12, 8, 9, 3, 11, 5, 4]
[12, 8, 9] [3, 11, 5, 4]
[12] [8, 9] [3, 11] [5, 4]
[12] [8] [9] [3] [11] [5] [4]   ← chia xong
[12] [8, 9] [3, 11] [4, 5]      ← bắt đầu trộn
[8, 9, 12] [3, 4, 5, 11]
[3, 4, 5, 8, 9, 11, 12]
```

## Implement Merge Sort in Python — Cài đặt đệ quy

````python
def mergeSort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    leftHalf = arr[:mid]
    rightHalf = arr[mid:]

    sortedLeft = mergeSort(leftHalf)
    sortedRight = mergeSort(rightHalf)

    return merge(sortedLeft, sortedRight)

def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result

mylist = [3, 7, 6, -10, 15, 23.5, 55, -13]
mysortedlist = mergeSort(mylist)
print("Sorted array:", mysortedlist)
````

## Merge Sort without Recursion — Không dùng đệ quy

Merge sort cũng cài được bằng **vòng lặp**: trộn từng cặp mảng con có độ dài 1, rồi 2, rồi 4…

````python
def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def mergeSort(arr):
    step = 1
    length = len(arr)
    while step < length:
        for i in range(0, length, 2 * step):
            left = arr[i:i + step]
            right = arr[i + step:i + 2 * step]
            merged = merge(left, right)
            for j, val in enumerate(merged):
                arr[i + j] = val
        step *= 2
    return arr

mylist = [3, 7, 6, -10, 15, 23.5, 55, -13]
mergeSort(mylist)
print(mylist)
````

## Merge Sort Time Complexity — Độ phức tạp

Merge sort **luôn** là **O(n log n)** — trong mọi trường hợp (tốt nhất, trung bình, xấu nhất). Đổi lại, nó cần thêm **O(n)** bộ nhớ phụ.

## ➕ Mở rộng: So sánh các thuật toán sắp xếp

````python
import random
import time

def bubble(arr):
    a = arr[:]
    n = len(a)
    for i in range(n - 1):
        for j in range(n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
    return a

data = random.sample(range(100_000), 2000)
for name, fn in [("Bubble O(n²)", bubble), ("Merge O(n log n)", lambda a: mergeSort(a[:])),
                 ("sorted() Timsort", sorted)]:
    t = time.perf_counter()
    fn(data)
    print(f"{name:<20}{time.perf_counter() - t:.4f} giây")
````

| Thuật toán | Tốt nhất | Trung bình | Xấu nhất | Bộ nhớ phụ |
|---|---|---|---|---|
| Bubble | O(n) | O(n²) | O(n²) | O(1) |
| Selection | O(n²) | O(n²) | O(n²) | O(1) |
| Insertion | O(n) | O(n²) | O(n²) | O(1) |
| Quick | O(n log n) | O(n log n) | O(n²) | O(log n) |
| Counting | O(n + k) | O(n + k) | O(n + k) | O(k) |
| Radix | O(n·k) | O(n·k) | O(n·k) | O(n + k) |
| Merge | O(n log n) | O(n log n) | O(n log n) | O(n) |

## 🧠 Ghi nhớ nhanh

> - Merge sort: **chia đôi** tới khi còn 1 phần tử, rồi **trộn** các mảng đã sắp xếp.
> - Luôn **O(n log n)**, ổn định, nhưng cần **O(n)** bộ nhớ phụ.
> - Trong thực tế dùng `sorted()` / `list.sort()` (Timsort = merge sort + insertion sort).
