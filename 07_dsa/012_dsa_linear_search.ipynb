## Linear Search — Tìm kiếm tuyến tính

**Tìm kiếm tuyến tính** duyệt qua mảng và trả về **chỉ số** của giá trị cần tìm.

```text
[ 3, 7, 2, 9, 5 ]   tìm 9
  ↑  ↑  ↑  ↑  ✔
```

## How it Works — Cách hoạt động

1. Đi qua từng giá trị trong mảng, bắt đầu từ đầu.
2. So sánh giá trị hiện tại với giá trị cần tìm.
3. Nếu bằng nhau, trả về chỉ số hiện tại.
4. Nếu đi hết mảng mà không thấy, trả về `-1`.

## Implementing Linear Search in Python — Cài đặt

````python
def linearSearch(arr, targetVal):
    for i in range(len(arr)):
        if arr[i] == targetVal:
            return i
    return -1

mylist = [3, 7, 2, 9, 5, 1, 8, 4, 6]
x = 4

result = linearSearch(mylist, x)

if result != -1:
    print("Found at index", result)
else:
    print("Not found")
````

## Linear Search Time Complexity — Độ phức tạp

Trường hợp **xấu nhất**: giá trị nằm cuối mảng hoặc không có → phải so sánh `n` lần → **O(n)**.

````python
import matplotlib.pyplot as plt

def steps_linear(n):
    return n          # số lần so sánh tệ nhất

ns = range(1, 101)
plt.plot(ns, [steps_linear(n) for n in ns])
plt.xlabel("n (số phần tử)")
plt.ylabel("Số lần so sánh")
plt.title("Linear Search: O(n)")
plt.show()
````

> 💡 Tìm kiếm tuyến tính dùng được với **mọi** mảng, kể cả mảng **chưa sắp xếp**. Python có sẵn: `mylist.index(x)` và toán tử `in`.

## 🧠 Ghi nhớ nhanh

> - Linear search: duyệt **từng phần tử** từ đầu tới cuối.
> - Không cần mảng sắp xếp; trả về chỉ số hoặc `-1`.
> - Độ phức tạp **O(n)**.
