## Bubble Sort — Sắp xếp nổi bọt

**Bubble sort** sắp xếp mảng từ nhỏ đến lớn bằng cách **so sánh từng cặp kề nhau** và **đổi chỗ** nếu sai thứ tự. Sau mỗi lượt, giá trị **lớn nhất** "nổi" lên cuối mảng — giống bọt khí nổi lên mặt nước.

## How it Works — Cách hoạt động

1. Duyệt mảng, mỗi lần xét một cặp phần tử kề nhau.
2. Nếu phần tử trước **lớn hơn** phần tử sau → **đổi chỗ**.
3. Lặp lại cho tới khi mảng được sắp xếp (n-1 lượt).

## Implement Bubble Sort in Python — Cài đặt

```python
mylist = [64, 34, 25, 12, 22, 11, 90, 5]

n = len(mylist)
for i in range(n - 1):
    for j in range(n - i - 1):
        if mylist[j] > mylist[j + 1]:
            mylist[j], mylist[j + 1] = mylist[j + 1], mylist[j]

print(mylist)
```

> 📌 Vòng trong chạy `n - i - 1` lần vì sau mỗi lượt, phần tử lớn nhất đã nằm đúng chỗ ở cuối — không cần xét lại.

## Bubble Sort Improvement — Cải tiến

Nếu một lượt duyệt **không có lần đổi chỗ nào**, mảng **đã được sắp xếp** — có thể dừng sớm:

```python
mylist = [7, 3, 9, 12, 11]

n = len(mylist)
for i in range(n - 1):
    swapped = False
    for j in range(n - i - 1):
        if mylist[j] > mylist[j + 1]:
            mylist[j], mylist[j + 1] = mylist[j + 1], mylist[j]
            swapped = True
    if not swapped:
        print("Dừng sớm sau lượt", i + 1)
        break

print("Sorted list:", mylist)
```

## Bubble Sort Time Complexity — Độ phức tạp

Bubble sort cần khoảng `n²/2` lần so sánh → **O(n²)**. Trường hợp **tốt nhất** (mảng đã sắp xếp, có cải tiến) là **O(n)**.

```python
import random
import matplotlib.pyplot as plt

def bubble_steps(arr):
    arr = arr[:]
    steps = 0
    n = len(arr)
    for i in range(n - 1):
        for j in range(n - i - 1):
            steps += 1
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
    return steps

ns = list(range(10, 201, 10))
plt.plot(ns, [bubble_steps(random.sample(range(1000), n)) for n in ns], marker="o")
plt.xlabel("n")
plt.ylabel("Số lần so sánh")
plt.title("Bubble Sort: O(n²)")
plt.show()
```

> - Cải tiến: lượt nào **không đổi chỗ** → dừng sớm.
