## Lists and Arrays — List và mảng

Trong Python, **list** là cấu trúc dữ liệu có sẵn dùng để lưu nhiều phần tử trong một biến. List đóng vai trò như **mảng (array)** ở các ngôn ngữ khác.

```python
x = [9, 12, 7, 4, 11]

# thêm phần tử:
x.append(8)

print(x)
```

## Algorithm: Find The Lowest Value in a List — Tìm giá trị nhỏ nhất

Tìm giá trị nhỏ nhất bằng một **giải thuật** — các bước:

1. Tạo biến `minVal` và gán bằng phần tử **đầu tiên** của list.
2. Duyệt qua từng phần tử của list.
3. Nếu phần tử hiện tại **nhỏ hơn** `minVal`, cập nhật `minVal` bằng phần tử đó.
4. Sau khi duyệt hết, `minVal` là giá trị nhỏ nhất.

```python
my_array = [7, 12, 9, 4, 11, 8]
minVal = my_array[0]

for i in my_array:
    if i < minVal:
        minVal = i

print('Lowest value:', minVal)
```

## Algorithm Time Complexity — Độ phức tạp thời gian

Để tìm giá trị nhỏ nhất, giải thuật phải **so sánh mỗi phần tử một lần**. Với `n` phần tử, cần `n` lần so sánh.

→ Độ phức tạp thời gian là **O(n)**: thời gian chạy tăng **tuyến tính** theo số phần tử.

> 💡 Python có sẵn hàm `min()` — cũng là O(n), nhưng viết bằng C nên nhanh hơn:

```python
my_array = [7, 12, 9, 4, 11, 8]
print(min(my_array))
```

## ➕ Mở rộng: Độ phức tạp các thao tác trên list

| Thao tác | Ví dụ | Big O |
|---|---|---|
| Truy cập theo chỉ số | `a[i]` | O(1) |
| Thêm vào cuối | `a.append(x)` | O(1) |
| Chèn/xoá ở đầu | `a.insert(0, x)`, `a.pop(0)` | O(n) |
| Tìm kiếm | `x in a` | O(n) |
| Sắp xếp | `a.sort()` | O(n log n) |

---

## 🧠 Ghi nhớ nhanh

> - List Python đóng vai trò **mảng**: truy cập theo chỉ số O(1).
> - Tìm min/max bằng duyệt một lượt → **O(n)**.
> - ⚠️ Chèn/xoá ở **đầu** list là O(n) (phải dịch chuyển phần tử).
