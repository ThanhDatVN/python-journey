## Hash Table — Bảng băm

**Hash table (bảng băm)** là cấu trúc dữ liệu được thiết kế để **tìm kiếm, thêm, xoá cực nhanh** — trung bình **O(1)**.

Hash table lưu dữ liệu trong các **bucket (ngăn)**. Vị trí bucket của mỗi phần tử được tính bằng **hàm băm (hash function)**.

## Build a Hash Table from Scratch — Tự xây dựng bảng băm

Ví dụ: lưu tên người, dùng list 10 phần tử làm các bucket:

```python
my_list = [None, None, None, None, None, None, None, None, None, None]
```

### The Hash Function — Hàm băm

Hàm băm đơn giản: **cộng mã Unicode** của các ký tự trong tên, rồi **chia lấy dư** cho 10 để ra chỉ số bucket (0-9):

```python
def hash_function(value):
    sum_of_chars = 0
    for char in value:
        sum_of_chars += ord(char)
    return sum_of_chars % 10

print("'Bob' has hash code:", hash_function('Bob'))
```

### Storing Elements — Lưu phần tử

Lưu "Bob" vào bucket mà hàm băm chỉ ra:

```python
def add(name):
    index = hash_function(name)
    my_list[index] = name

add('Bob')
print(my_list)
```

Thêm vài tên nữa:

```python
add('Pete')
add('Jones')
add('Lisa')
add('Siri')
print(my_list)
```

### Looking Up a Name — Tra cứu

Tra cứu chỉ cần tính lại hàm băm và **nhìn đúng một bucket** — không phải duyệt cả list:

```python
def contains(name):
    index = hash_function(name)
    return my_list[index] == name

print("'Pete' is in the Hash Table:", contains('Pete'))
```

## Handling Collisions — Xử lý va chạm

Nếu hai tên có **cùng mã băm**, chúng tranh cùng một bucket — gọi là **va chạm (collision)**.

Cách xử lý phổ biến là **chaining**: mỗi bucket là **một list** chứa nhiều phần tử:

```python
my_list = [[] for _ in range(10)]

def hash_function(value):
    return sum(ord(char) for char in value) % 10

def add(name):
    index = hash_function(name)
    if name not in my_list[index]:
        my_list[index].append(name)

def contains(name):
    index = hash_function(name)
    return name in my_list[index]

for name in ['Bob', 'Pete', 'Jones', 'Lisa', 'Siri', 'Stuart']:
    add(name)

for i, bucket in enumerate(my_list):
    print(i, bucket)

print("'Stuart' is in the Hash Table:", contains('Stuart'))
```

## Python Dictionaries and Sets — Hash table có sẵn

Trong Python, **dictionary** và **set** chính là hash table — đó là lý do kiểm tra `in` với chúng rất nhanh:

```python
phone_book = {"Bob": "0901", "Pete": "0902"}
print(phone_book["Pete"])

print(hash("Bob"))   # hàm băm có sẵn của Python
```

## 🧠 Ghi nhớ nhanh

> - Hash table: **hàm băm** tính vị trí bucket → tìm/thêm/xoá trung bình **O(1)**.
> - **Va chạm** (cùng mã băm) → xử lý bằng **chaining** (mỗi bucket là một list).
> - `dict` và `set` của Python chính là hash table.
