## Queues — Hàng đợi

**Queue (hàng đợi)** là cấu trúc dữ liệu hoạt động theo nguyên tắc **FIFO** — *First In, First Out* (vào trước, ra trước).

Hãy tưởng tượng **xếp hàng ở siêu thị**: người đến trước được phục vụ trước.

| Thao tác | Ý nghĩa |
|---|---|
| **Enqueue** | Thêm phần tử vào **cuối** hàng |
| **Dequeue** | Lấy và xoá phần tử ở **đầu** hàng |
| **Peek** | Xem phần tử đầu hàng |
| **isEmpty** | Kiểm tra hàng rỗng |
| **Size** | Số phần tử |

Queue dùng cho: lập lịch tác vụ, hàng đợi in, xử lý yêu cầu theo thứ tự, duyệt theo chiều rộng (BFS) trên đồ thị…

## Queue Implementation using Python Lists — Dùng list

```python
queue = []

# Enqueue
queue.append('A')
queue.append('B')
queue.append('C')
print("Queue: ", queue)

# Peek
frontElement = queue[0]
print("Peek: ", frontElement)

# Dequeue
poppedElement = queue.pop(0)
print("Dequeue: ", poppedElement)

print("Queue after Dequeue: ", queue)

# isEmpty
isEmpty = not bool(queue)
print("isEmpty: ", isEmpty)

# Size
print("Size: ", len(queue))
```

> ⚠️ `pop(0)` trên list là **O(n)** vì mọi phần tử phải dịch lên. Với hàng đợi lớn, dùng `collections.deque` (xem phần mở rộng).

Đóng gói thành class:

```python
class Queue:
    def __init__(self):
        self.queue = []

    def enqueue(self, element):
        self.queue.append(element)

    def dequeue(self):
        if self.isEmpty():
            return "Queue is empty"
        return self.queue.pop(0)

    def peek(self):
        if self.isEmpty():
            return "Queue is empty"
        return self.queue[0]

    def isEmpty(self):
        return len(self.queue) == 0

    def size(self):
        return len(self.queue)

myQueue = Queue()
myQueue.enqueue('A')
myQueue.enqueue('B')
myQueue.enqueue('C')

print("Queue: ", myQueue.queue)
print("Dequeue: ", myQueue.dequeue())
print("Peek: ", myQueue.peek())
print("isEmpty: ", myQueue.isEmpty())
print("Size: ", myQueue.size())
```

## Queue Implementation using Linked Lists — Dùng danh sách liên kết

Linked list giữ con trỏ tới **đầu (front)** và **cuối (rear)** → enqueue và dequeue đều **O(1)**:

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class Queue:
    def __init__(self):
        self.front = None
        self.rear = None
        self.length = 0

    def enqueue(self, element):
        new_node = Node(element)
        if self.rear is None:
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node
        self.length += 1

    def dequeue(self):
        if self.isEmpty():
            return "Queue is empty"
        temp = self.front
        self.front = temp.next
        self.length -= 1
        if self.front is None:
            self.rear = None
        return temp.data

    def peek(self):
        if self.isEmpty():
            return "Queue is empty"
        return self.front.data

    def isEmpty(self):
        return self.length == 0

    def size(self):
        return self.length

    def printQueue(self):
        temp = self.front
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print()

myQueue = Queue()
myQueue.enqueue('A')
myQueue.enqueue('B')
myQueue.enqueue('C')
myQueue.printQueue()

print("Dequeue: ", myQueue.dequeue())
print("Peek: ", myQueue.peek())
print("Size: ", myQueue.size())
```

## ➕ Mở rộng: collections.deque

```python
from collections import deque

q = deque()
q.append("A")       # enqueue
q.append("B")
q.append("C")
print(q.popleft())  # dequeue - O(1)
print(q)
```

> - Ứng dụng: lập lịch, hàng đợi in, BFS.
