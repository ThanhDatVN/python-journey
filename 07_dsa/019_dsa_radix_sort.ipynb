## Radix Sort — Sắp xếp cơ số

**Radix sort** sắp xếp số nguyên theo **từng chữ số**, bắt đầu từ chữ số **hàng đơn vị** (LSD — *Least Significant Digit*), rồi hàng chục, hàng trăm…

Từ "radix" nghĩa là **cơ số** — với hệ thập phân, cơ số là 10 (các chữ số 0-9).

## How it Works — Cách hoạt động

1. Bắt đầu từ chữ số hàng đơn vị.
2. Đưa mỗi số vào một trong **10 bucket (0-9)** theo chữ số đang xét.
3. Lấy các số ra khỏi bucket **theo thứ tự**, ghép lại thành mảng.
4. Chuyển sang chữ số tiếp theo (hàng chục…) và lặp lại, cho tới chữ số cao nhất của số lớn nhất.

> 📌 Việc lấy ra theo thứ tự **giữ nguyên thứ tự tương đối** của các số cùng bucket (sắp xếp **ổn định — stable**) — điều kiện để radix sort cho kết quả đúng.

## Implement Radix Sort in Python — Cài đặt

```python
mylist = [170, 45, 75, 90, 802, 24, 2, 66]
print("Original array:", mylist)

radixArray = [[] for _ in range(10)]
maxVal = max(mylist)
exp = 1

while maxVal // exp > 0:
    while len(mylist) > 0:
        val = mylist.pop()
        radixIndex = (val // exp) % 10
        radixArray[radixIndex].append(val)

    for bucket in radixArray:
        while len(bucket) > 0:
            val = bucket.pop()
            mylist.append(val)

    print(f"Sau khi xét chữ số exp={exp}:", mylist)
    exp *= 10

print("Sorted array:", mylist)
```

## Radix Sort Using Other Sorting Algorithms — Kết hợp thuật toán khác

Radix sort có thể dùng **bất kỳ thuật toán sắp xếp ổn định nào** để sắp theo từng chữ số — ví dụ bubble sort (vì bubble sort ổn định):

```python
def bubbleSort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

def radixSortWithBubbleSort(arr):
    max_val = max(arr)
    exp = 1

    while max_val // exp > 0:
        radixArray = [[], [], [], [], [], [], [], [], [], []]

        for num in arr:
            radixIndex = (num // exp) % 10
            radixArray[radixIndex].append(num)

        for bucket in radixArray:
            bubbleSort(bucket)

        i = 0
        for bucket in radixArray:
            for num in bucket:
                arr[i] = num
                i += 1

        exp *= 10

mylist = [170, 45, 75, 90, 802, 24, 2, 66]
radixSortWithBubbleSort(mylist)
print(mylist)
```

## Radix Sort Time Complexity — Độ phức tạp

**O(n · k)**, với `n` là số phần tử, `k` là **số chữ số** của giá trị lớn nhất.

## 🧠 Ghi nhớ nhanh

> - Radix sort: sắp theo **từng chữ số**, từ hàng đơn vị lên (LSD), dùng 10 bucket.
> - Yêu cầu sắp xếp **ổn định** ở mỗi bước.
> - Độ phức tạp **O(n · k)** (k = số chữ số); chỉ dùng cho số nguyên không âm.
