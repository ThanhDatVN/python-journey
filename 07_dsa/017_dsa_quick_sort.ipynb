## Quicksort — Sắp xếp nhanh

**Quicksort** là một trong những thuật toán sắp xếp **nhanh nhất**. Nó chọn một phần tử làm **pivot (chốt)**, rồi **phân hoạch** mảng: các giá trị **nhỏ hơn** pivot sang trái, **lớn hơn** sang phải. Sau đó áp dụng **đệ quy** cho hai nửa.

## How it Works — Cách hoạt động

1. Chọn pivot (ở đây là phần tử **cuối** mảng con).
2. Phân hoạch: đưa phần tử nhỏ hơn pivot về bên trái, lớn hơn về bên phải; đặt pivot vào giữa.
3. Gọi đệ quy quicksort cho mảng con bên trái và bên phải pivot.
4. Dừng khi mảng con có ít hơn 2 phần tử.

## Implement Quicksort in Python — Cài đặt

```python
def partition(array, low, high):
    pivot = array[high]
    i = low - 1

    for j in range(low, high):
        if array[j] <= pivot:
            i += 1
            array[i], array[j] = array[j], array[i]

    array[i + 1], array[high] = array[high], array[i + 1]
    return i + 1

def quicksort(array, low=0, high=None):
    if high is None:
        high = len(array) - 1

    if low < high:
        pivot_index = partition(array, low, high)
        quicksort(array, low, pivot_index - 1)
        quicksort(array, pivot_index + 1, high)

mylist = [64, 34, 25, 5, 22, 11, 90, 12]
quicksort(mylist)
print(mylist)
```

## Quicksort Time Complexity — Độ phức tạp

- Trung bình: **O(n log n)** — mỗi lần phân hoạch chia mảng khoảng đôi.
- Xấu nhất: **O(n²)** — khi pivot luôn là phần tử lớn/nhỏ nhất (ví dụ mảng đã sắp xếp và chọn pivot cuối).

## ➕ Mở rộng: Phiên bản ngắn gọn kiểu Python

Dễ đọc hơn nhưng tốn thêm bộ nhớ (tạo list mới):

```python
def quicksort_simple(arr):
    if len(arr) < 2:
        return arr
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quicksort_simple(left) + middle + quicksort_simple(right)

print(quicksort_simple([64, 34, 25, 5, 22, 11, 90, 12]))
```

## 🧠 Ghi nhớ nhanh

> - Quicksort: chọn **pivot**, **phân hoạch** (nhỏ trái, lớn phải), **đệ quy** hai nửa.
> - Trung bình **O(n log n)**, xấu nhất **O(n²)** (pivot tồi).
> - Sắp xếp **tại chỗ**, không cần mảng phụ (phiên bản partition).
