## Selection Sort — Sắp xếp chọn

**Selection sort** tìm giá trị **nhỏ nhất** trong phần chưa sắp xếp, rồi **đưa nó lên đầu** phần đó. Lặp lại cho tới khi mảng được sắp xếp.

## How it Works — Cách hoạt động

1. Duyệt mảng để tìm giá trị **nhỏ nhất**.
2. Đưa giá trị nhỏ nhất lên **đầu** phần chưa sắp xếp.
3. Lặp lại với phần còn lại — `n - 1` lượt.

## Implement Selection Sort in Python — Cài đặt

```python
mylist = [64, 34, 25, 5, 22, 11, 90, 12]

n = len(mylist)
for i in range(n - 1):
    min_index = i
    for j in range(i + 1, n):
        if mylist[j] < mylist[min_index]:
            min_index = j
    min_value = mylist.pop(min_index)
    mylist.insert(i, min_value)

print(mylist)
```

## Selection Sort Shifting Problem — Vấn đề dịch chuyển

Cách trên dùng `pop()` rồi `insert()` — mỗi lần chèn, các phần tử phía sau phải **dịch chuyển**, tốn thêm thời gian.

## Solution: Swap Values! — Giải pháp: đổi chỗ

Thay vì pop/insert, chỉ cần **đổi chỗ** giá trị nhỏ nhất với phần tử ở vị trí `i`:

```python
mylist = [64, 34, 25, 12, 22, 11, 90, 5]

n = len(mylist)
for i in range(n):
    min_index = i
    for j in range(i + 1, n):
        if mylist[j] < mylist[min_index]:
            min_index = j
    mylist[i], mylist[min_index] = mylist[min_index], mylist[i]

print(mylist)
```

## Selection Sort Time Complexity — Độ phức tạp

Luôn cần khoảng `n²/2` lần so sánh, kể cả khi mảng đã sắp xếp → **O(n²)**.

> 💡 Ưu điểm: số lần **đổi chỗ** ít (tối đa `n - 1`) — hữu ích khi việc ghi dữ liệu tốn kém.

> - Selection sort: mỗi lượt **chọn giá trị nhỏ nhất** của phần chưa sắp xếp, đưa lên đầu.
> - Dùng **đổi chỗ** thay vì pop/insert để tránh dịch chuyển phần tử.
> - Độ phức tạp **O(n²)** (cả khi mảng đã sắp xếp).
