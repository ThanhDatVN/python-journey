## Binary Search Trees — Cây nhị phân tìm kiếm

**Binary Search Tree (BST)** là cây nhị phân thoả mãn, với **mọi** node `X`:

- Mọi giá trị trong **cây con trái** của X **nhỏ hơn** giá trị của X.
- Mọi giá trị trong **cây con phải** của X **lớn hơn** giá trị của X.
- Cả cây con trái và phải cũng là BST.

```text
            13
          /    \
         7      15
        / \    /  \
       3   8  14   19
                   /
                  18
```

Nhờ thứ tự này, **tìm kiếm** chỉ cần đi một nhánh mỗi bước → trung bình **O(log n)** (nếu cây cân bằng).

## Create a BST — Tạo BST

````python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def insert(node, data):
    if node is None:
        return TreeNode(data)
    if data < node.data:
        node.left = insert(node.left, data)
    elif data > node.data:
        node.right = insert(node.right, data)
    return node

root = None
for value in [13, 7, 15, 3, 8, 14, 19, 18]:
    root = insert(root, value)
````

## Traversal of a BST — Duyệt BST

Duyệt **in-order** một BST cho ra các giá trị **đã sắp xếp**:

````python
def inOrderTraversal(node):
    if node is None:
        return
    inOrderTraversal(node.left)
    print(node.data, end=", ")
    inOrderTraversal(node.right)

inOrderTraversal(root)
````

## Search for a Value in a BST — Tìm kiếm

Bắt đầu từ gốc: nhỏ hơn → đi trái, lớn hơn → đi phải:

````python
def search(node, target):
    if node is None:
        return None
    elif node.data == target:
        return node
    elif target < node.data:
        return search(node.left, target)
    else:
        return search(node.right, target)

result = search(root, 14)
print("Found the node with value:", result.data if result else "Not found")
print(search(root, 99))
````

## Insert a Node in a BST — Chèn node

Chèn luôn diễn ra ở vị trí **lá** phù hợp — hàm `insert` ở trên đã làm việc đó:

````python
insert(root, 10)
inOrderTraversal(root)
````

## Find The Lowest Value in a BST Subtree — Tìm giá trị nhỏ nhất

Giá trị nhỏ nhất nằm ở node **trái nhất**:

````python
def minValueNode(node):
    current = node
    while current.left is not None:
        current = current.left
    return current

print("Lowest value:", minValueNode(root).data)
````

## Delete a Node in a BST — Xoá node

Ba trường hợp:

1. Node là **lá** → xoá luôn.
2. Node có **một** con → nối con lên thay node.
3. Node có **hai** con → thay giá trị bằng **node nhỏ nhất của cây con phải** (in-order successor), rồi xoá node đó.

````python
def delete(node, data):
    if not node:
        return None

    if data < node.data:
        node.left = delete(node.left, data)
    elif data > node.data:
        node.right = delete(node.right, data)
    else:
        # node có 0 hoặc 1 con
        if not node.left:
            return node.right
        elif not node.right:
            return node.left
        # node có 2 con
        node.data = minValueNode(node.right).data
        node.right = delete(node.right, node.data)

    return node

root = delete(root, 15)
inOrderTraversal(root)
````

---

## 🧠 Ghi nhớ nhanh

> - BST: con **trái < node < con phải** với mọi node.
> - In-order traversal → giá trị **tăng dần**.
> - Tìm/chèn/xoá: đi một nhánh mỗi bước → O(h), cân bằng thì **O(log n)**.
> - Xoá node 2 con: thay bằng **node nhỏ nhất của cây con phải**.
