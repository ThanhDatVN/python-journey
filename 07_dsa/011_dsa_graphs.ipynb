## Graphs — Đồ thị

**Đồ thị (graph)** là cấu trúc dữ liệu **phi tuyến** gồm các **đỉnh (vertex)** và các **cạnh (edge)** nối các đỉnh.

```text
    A — B
    | \ |
    C — D — E
```

Đồ thị biểu diễn rất nhiều thứ ngoài đời: **mạng xã hội** (người là đỉnh, quan hệ bạn bè là cạnh), **bản đồ** (thành phố là đỉnh, đường đi là cạnh), **Internet**…

## Graph Terminology — Thuật ngữ

| Thuật ngữ | Ý nghĩa |
|---|---|
| **Vertex** | Đỉnh |
| **Edge** | Cạnh nối hai đỉnh |
| **Weighted** | Cạnh có **trọng số** (khoảng cách, chi phí…) |
| **Directed** | Cạnh có **hướng** (A → B khác B → A) |
| **Connected** | Mọi đỉnh đều đi tới được từ mọi đỉnh khác |
| **Cyclic** | Có **chu trình** — đường đi quay về điểm xuất phát |

## Graph Representations — Biểu diễn đồ thị

Cách phổ biến là **ma trận kề (adjacency matrix)**: ô `[i][j]` = 1 nếu có cạnh giữa đỉnh i và j.

````python
vertexData = ['A', 'B', 'C', 'D']

adjacency_matrix = [
    [0, 1, 1, 1],   # A
    [1, 0, 1, 0],   # B
    [1, 1, 0, 0],   # C
    [1, 0, 0, 0]    # D
]

def print_adjacency_list(matrix):
    for i, row in enumerate(matrix):
        neighbors = [vertexData[j] for j, v in enumerate(row) if v]
        print(vertexData[i] + ":", neighbors)

print_adjacency_list(adjacency_matrix)
````

## Graph Implementation — Cài đặt bằng class

````python
class Graph:
    def __init__(self, size):
        self.adj_matrix = [[0] * size for _ in range(size)]
        self.size = size
        self.vertex_data = [''] * size

    def add_edge(self, u, v):
        if 0 <= u < self.size and 0 <= v < self.size:
            self.adj_matrix[u][v] = 1
            self.adj_matrix[v][u] = 1   # vô hướng

    def add_vertex_data(self, vertex, data):
        if 0 <= vertex < self.size:
            self.vertex_data[vertex] = data

    def print_graph(self):
        print("Adjacency Matrix:")
        for row in self.adj_matrix:
            print(' '.join(map(str, row)))
        print("\nVertex Data:")
        for vertex, data in enumerate(self.vertex_data):
            print(f"Vertex {vertex}: {data}")

g = Graph(4)
for i, name in enumerate("ABCD"):
    g.add_vertex_data(i, name)
g.add_edge(0, 1)   # A - B
g.add_edge(0, 2)   # A - C
g.add_edge(0, 3)   # A - D
g.add_edge(1, 2)   # B - C

g.print_graph()
````

## Graph Traversal — Duyệt đồ thị

### Depth First Search (DFS) — Duyệt theo chiều sâu

Đi **sâu nhất có thể** theo một nhánh, rồi mới quay lui. Thường cài bằng **đệ quy** (hoặc stack):

````python
class Graph:
    def __init__(self, size):
        self.adj_matrix = [[0] * size for _ in range(size)]
        self.size = size
        self.vertex_data = [''] * size

    def add_edge(self, u, v):
        self.adj_matrix[u][v] = 1
        self.adj_matrix[v][u] = 1

    def add_vertex_data(self, vertex, data):
        self.vertex_data[vertex] = data

    def dfs_util(self, v, visited):
        visited[v] = True
        print(self.vertex_data[v], end=' ')
        for i in range(self.size):
            if self.adj_matrix[v][i] == 1 and not visited[i]:
                self.dfs_util(i, visited)

    def dfs(self, start_vertex_data):
        visited = [False] * self.size
        start_vertex = self.vertex_data.index(start_vertex_data)
        self.dfs_util(start_vertex, visited)

    def bfs(self, start_vertex_data):
        queue = [self.vertex_data.index(start_vertex_data)]
        visited = [False] * self.size
        visited[queue[0]] = True
        while queue:
            current_vertex = queue.pop(0)
            print(self.vertex_data[current_vertex], end=' ')
            for i in range(self.size):
                if self.adj_matrix[current_vertex][i] == 1 and not visited[i]:
                    queue.append(i)
                    visited[i] = True

g = Graph(7)
for i, name in enumerate("ABCDEFG"):
    g.add_vertex_data(i, name)

for u, v in [(3, 0), (0, 2), (0, 3), (0, 4), (4, 2), (2, 5), (2, 1), (2, 6), (1, 5)]:
    g.add_edge(u, v)

print("Depth First Search starting from vertex D:")
g.dfs('D')
````

### Breadth First Search (BFS) — Duyệt theo chiều rộng

Thăm **tất cả đỉnh kề** trước, rồi mới đi tiếp ra xa. Cài bằng **queue**:

````python
print("Breadth First Search starting from vertex D:")
g.bfs('D')
````
