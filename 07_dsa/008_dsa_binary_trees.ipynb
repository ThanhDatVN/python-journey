## Binary Trees — Cây nhị phân

**Cây nhị phân** là cây mà mỗi node có **tối đa hai** con: **con trái** và **con phải**.

```text
          R
        /   \
       A     B
      / \   / \
     C   D E   F
    /
   G
```

### Types of Binary Trees — Các dạng cây nhị phân

- **Balanced (cân bằng)**: chiều cao cây con trái và phải của mỗi node chênh nhau **tối đa 1**.
- **Complete (đầy đủ theo mức)**: mọi mức đều đầy, trừ mức cuối được điền **từ trái sang**.
- **Full (đủ)**: mỗi node có **0 hoặc 2** con.
- **Perfect (hoàn hảo)**: mọi node trong có 2 con và mọi lá ở **cùng một mức**.

## Binary Tree Implementation — Cài đặt

````python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

root = TreeNode('R')
nodeA = TreeNode('A')
nodeB = TreeNode('B')
nodeC = TreeNode('C')
nodeD = TreeNode('D')
nodeE = TreeNode('E')
nodeF = TreeNode('F')
nodeG = TreeNode('G')

root.left = nodeA
root.right = nodeB

nodeA.left = nodeC
nodeA.right = nodeD

nodeB.left = nodeE
nodeB.right = nodeF

nodeF.left = nodeG

# Thử truy cập
print("root.right.left.data:", root.right.left.data)
````

## Binary Tree Traversal — Duyệt cây nhị phân

**Duyệt** là đi qua từng node một lần. Có hai nhóm:

- **Theo chiều rộng (BFS)**: duyệt từng mức.
- **Theo chiều sâu (DFS)**: gồm 3 kiểu — **Pre-order**, **In-order**, **Post-order**.

### Pre-order Traversal — Duyệt tiền thứ tự

Thăm **node hiện tại trước**, rồi cây con trái, rồi cây con phải. Dùng để **sao chép** cây:

````python
def preOrderTraversal(node):
    if node is None:
        return
    print(node.data, end=", ")
    preOrderTraversal(node.left)
    preOrderTraversal(node.right)

preOrderTraversal(root)
````

### In-order Traversal — Duyệt trung thứ tự

Cây con trái → **node hiện tại** → cây con phải. Với cây nhị phân tìm kiếm, cho ra các giá trị **theo thứ tự tăng dần**:

````python
def inOrderTraversal(node):
    if node is None:
        return
    inOrderTraversal(node.left)
    print(node.data, end=", ")
    inOrderTraversal(node.right)

inOrderTraversal(root)
````

### Post-order Traversal — Duyệt hậu thứ tự

Cây con trái → cây con phải → **node hiện tại sau cùng**. Dùng để **xoá** cây:

````python
def postOrderTraversal(node):
    if node is None:
        return
    postOrderTraversal(node.left)
    postOrderTraversal(node.right)
    print(node.data, end=", ")

postOrderTraversal(root)
````

## Array Implementation — Lưu cây nhị phân bằng mảng

Cây nhị phân cũng lưu được trong **mảng**: với node ở chỉ số `i`, con trái ở `2*i + 1`, con phải ở `2*i + 2`:

````python
binary_tree_array = ['R', 'A', 'B', 'C', 'D', 'E', 'F', None, None, None, None, None, None, 'G']

def left_child_index(index):
    return 2 * index + 1

def right_child_index(index):
    return 2 * index + 2

def get_data(index):
    if 0 <= index < len(binary_tree_array):
        return binary_tree_array[index]
    return None

right_child = right_child_index(0)
left_child_of_right_child = left_child_index(right_child)
print("root.right.left.data:", get_data(left_child_of_right_child))
````

## 🧠 Ghi nhớ nhanh

> - Cây nhị phân: mỗi node có tối đa **2 con** (`left`, `right`).
> - DFS: **Pre-order** (gốc-trái-phải) · **In-order** (trái-gốc-phải) · **Post-order** (trái-phải-gốc).
> - Lưu bằng mảng: con trái `2i+1`, con phải `2i+2`.
