## Trees — Cây

**Cây (tree)** là cấu trúc dữ liệu **phân cấp** gồm các **node** nối với nhau bằng các **cạnh (edge)**.

Mỗi node chứa dữ liệu, và có thể có các **node con**.

```text
            R          ← root (gốc)
          /   \
         A     B       ← A, B là con của R
        / \     \
       C   D     E     ← C, D, E là lá (leaf)
```

Cây được dùng ở nhiều nơi: hệ thống thư mục, cây DOM của HTML, cơ sở dữ liệu (chỉ mục), cây quyết định trong học máy…

## Tree Terminology — Thuật ngữ

| Thuật ngữ | Ý nghĩa |
|---|---|
| **Root** | Node gốc — node đầu tiên, không có cha |
| **Edge** | Cạnh nối hai node |
| **Parent node** | Node cha — node có node con |
| **Child node** | Node con |
| **Leaf node** | Node lá — không có con |
| **Height** | Chiều cao — số cạnh nhiều nhất từ gốc tới một lá |
| **Depth** | Độ sâu của một node — số cạnh từ gốc tới node đó |
| **Subtree** | Cây con — một node cùng mọi hậu duệ của nó |

## Types of Trees — Các loại cây

- **Binary Tree** (cây nhị phân): mỗi node có **tối đa 2** con.
- **Binary Search Tree** (cây nhị phân tìm kiếm): cây nhị phân có **thứ tự** — con trái nhỏ hơn, con phải lớn hơn.
- **AVL Tree**: cây nhị phân tìm kiếm **tự cân bằng**.

Các bài tiếp theo sẽ đi sâu vào từng loại.

## ➕ Mở rộng: Cây tổng quát bằng class

````python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.children = []

    def add(self, child):
        self.children.append(child)
        return child

def print_tree(node, level=0):
    print("  " * level + "- " + node.data)
    for child in node.children:
        print_tree(child, level + 1)

root = TreeNode("Documents")
python = root.add(TreeNode("Python"))
python.add(TreeNode("lesson1.ipynb"))
python.add(TreeNode("lesson2.ipynb"))
root.add(TreeNode("Photos")).add(TreeNode("trip.jpg"))

print_tree(root)
````

---

## 🧠 Ghi nhớ nhanh

> - Cây = cấu trúc **phân cấp**: root → parent → child → leaf.
> - Height (cao nhất từ gốc) · Depth (độ sâu một node) · Subtree (cây con).
> - Các loại: Binary Tree · Binary Search Tree · AVL Tree.
