## Counting Sort — Sắp xếp đếm

**Counting sort** sắp xếp bằng cách **đếm số lần xuất hiện** của mỗi giá trị — **không cần so sánh** các phần tử với nhau.

Counting sort chỉ dùng được khi:

- Giá trị là **số nguyên không âm**.
- **Khoảng giá trị** (`k`) không quá lớn so với số phần tử (`n`).

## How it Works — Cách hoạt động

1. Tạo mảng đếm `count` có độ dài bằng giá trị lớn nhất + 1, toàn số 0.
2. Duyệt mảng gốc: với mỗi giá trị `v`, tăng `count[v]` thêm 1.
3. Duyệt mảng đếm: với mỗi chỉ số `i`, ghi `i` vào kết quả `count[i]` lần.

## Implement Counting Sort in Python — Cài đặt

```python
def countingSort(arr):
    max_val = max(arr)
    count = [0] * (max_val + 1)

    while len(arr) > 0:
        num = arr.pop(0)
        count[num] += 1

    for i in range(len(count)):
        while count[i] > 0:
            arr.append(i)
            count[i] -= 1

    return arr

mylist = [4, 2, 2, 6, 3, 3, 1, 6, 5, 2, 3]
mysortedlist = countingSort(mylist)
print(mysortedlist)
```

## Counting Sort Time Complexity — Độ phức tạp

**O(n + k)**, với `n` là số phần tử, `k` là khoảng giá trị.

- `k` nhỏ (ví dụ điểm thi 0-10, tuổi 0-120) → cực nhanh.
- `k` rất lớn (ví dụ số từ 0 tới 1 tỷ) → tốn bộ nhớ và chậm.

## ➕ Mở rộng: Dùng Counter

```python
from collections import Counter

scores = [8, 7, 9, 7, 10, 8, 8, 6]
counts = Counter(scores)
print(counts)
print([v for v in range(11) for _ in range(counts[v])])
```

## 🧠 Ghi nhớ nhanh

> - Counting sort: **đếm** số lần xuất hiện, không so sánh phần tử.
> - Chỉ dùng với **số nguyên không âm**, khoảng giá trị `k` nhỏ.
> - Độ phức tạp **O(n + k)**.
