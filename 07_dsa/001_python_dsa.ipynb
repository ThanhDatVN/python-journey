## Data Structures and Algorithms (DSA) — Cấu trúc dữ liệu và giải thuật

**Cấu trúc dữ liệu (data structure)** là cách **tổ chức và lưu trữ** dữ liệu để dùng hiệu quả.

**Giải thuật (algorithm)** là tập các **bước** để giải một bài toán.

Hiểu DSA giúp bạn chọn đúng công cụ cho từng bài toán, và viết chương trình **nhanh hơn**, **tốn ít bộ nhớ hơn**.

## Data Structures — Cấu trúc dữ liệu

Python có sẵn nhiều cấu trúc dữ liệu: **list**, **tuple**, **set**, **dictionary**.

Chương này còn giới thiệu các cấu trúc dữ liệu kinh điển khác: **Stack**, **Queue**, **Linked List**, **Hash Table**, **Tree**, **Graph** — nhiều cấu trúc trong số đó có thể **tự xây dựng** bằng Python.

## Algorithms — Giải thuật

Giải thuật là các bước giải quyết vấn đề — giống như một **công thức nấu ăn**: làm theo đúng thứ tự sẽ ra kết quả.

Ví dụ giải thuật tìm **giá trị nhỏ nhất** trong một list:

```python
my_array = [7, 12, 9, 4, 11]
minVal = my_array[0]

for i in my_array:
    if i < minVal:
        minVal = i

print('Lowest value:', minVal)
```

Chương này giới thiệu các giải thuật **tìm kiếm** (Linear Search, Binary Search) và **sắp xếp** (Bubble, Selection, Insertion, Quick, Counting, Radix, Merge Sort).

## Theory and Terminology — Lý thuyết và thuật ngữ

| Thuật ngữ | Ý nghĩa |
|---|---|
| **Algorithm** | Tập các bước giải một bài toán |
| **Data Structure** | Cách tổ chức dữ liệu để dùng hiệu quả |
| **Time Complexity** | Độ phức tạp thời gian — thời gian chạy tăng thế nào khi dữ liệu tăng |
| **Space Complexity** | Độ phức tạp bộ nhớ — bộ nhớ cần thêm khi dữ liệu tăng |
| **Big O Notation** | Ký hiệu mô tả độ phức tạp trong trường hợp xấu nhất, ví dụ `O(n)`, `O(n²)` |
| **Recursion** | Đệ quy — hàm gọi lại chính nó |
| **Divide and Conquer** | Chia để trị — chia bài toán lớn thành bài toán nhỏ hơn |
| **Brute Force** | Vét cạn — thử mọi khả năng |

## ➕ Mở rộng: Big O trực quan

```python
import math

print(f"{'n':>8}{'O(log n)':>10}{'O(n)':>10}{'O(n log n)':>12}{'O(n²)':>14}")
for n in (10, 100, 1000, 10000):
    print(f"{n:>8}{math.log2(n):>10.1f}{n:>10}{n * math.log2(n):>12.0f}{n * n:>14}")
```

---

## 🧠 Ghi nhớ nhanh

> - **Cấu trúc dữ liệu** = cách tổ chức dữ liệu · **Giải thuật** = các bước giải bài toán.
> - Đánh giá giải thuật bằng **độ phức tạp thời gian/bộ nhớ**, ký hiệu **Big O**.
> - Thứ tự từ nhanh → chậm: `O(1)` < `O(log n)` < `O(n)` < `O(n log n)` < `O(n²)`.
