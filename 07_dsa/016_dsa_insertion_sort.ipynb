## Insertion Sort — Sắp xếp chèn

**Insertion sort** chia mảng làm hai phần: **đã sắp xếp** và **chưa sắp xếp**. Lần lượt lấy từng giá trị ở phần chưa sắp xếp và **chèn** vào đúng vị trí trong phần đã sắp xếp — giống cách xếp bài trên tay.

## How it Works — Cách hoạt động

1. Phần tử đầu tiên coi như đã sắp xếp.
2. Lấy phần tử tiếp theo của phần chưa sắp xếp.
3. Dịch các phần tử lớn hơn nó trong phần đã sắp xếp sang phải, rồi **chèn** nó vào chỗ trống.
4. Lặp lại cho tới hết mảng.

## Implement Insertion Sort in Python — Cài đặt

```python
mylist = [64, 34, 25, 12, 22, 11, 90, 5]

n = len(mylist)
for i in range(1, n):
    insert_index = i
    current_value = mylist.pop(i)
    for j in range(i - 1, -1, -1):
        if mylist[j] > current_value:
            insert_index = j
    mylist.insert(insert_index, current_value)

print(mylist)
```

## Insertion Sort Improvement — Cải tiến

Cách trên dùng `pop()`/`insert()` gây dịch chuyển thừa. Cách tốt hơn: **dịch dần** các phần tử lớn hơn sang phải, rồi đặt giá trị vào chỗ trống, và **dừng** ngay khi gặp phần tử nhỏ hơn:

```python
mylist = [64, 34, 25, 12, 22, 11, 90, 5]

n = len(mylist)
for i in range(1, n):
    current_value = mylist[i]
    j = i - 1
    while j >= 0 and mylist[j] > current_value:
        mylist[j + 1] = mylist[j]
        j -= 1
    mylist[j + 1] = current_value

print(mylist)
```

## Insertion Sort Time Complexity — Độ phức tạp

- Trường hợp xấu nhất (mảng sắp ngược): **O(n²)**.
- Trường hợp tốt nhất (mảng **gần như đã sắp xếp**): **O(n)** — rất nhanh!

> 💡 Vì nhanh với dữ liệu gần sắp xếp, insertion sort được dùng **bên trong** thuật toán sắp xếp có sẵn của Python (Timsort).

## 🧠 Ghi nhớ nhanh

> - Insertion sort: lấy từng phần tử, **chèn** vào đúng chỗ trong phần đã sắp xếp (như xếp bài).
> - Cải tiến: **dịch** phần tử lớn hơn sang phải, dừng sớm khi gặp phần tử nhỏ hơn.
> - Xấu nhất **O(n²)**, tốt nhất (gần sắp xếp) **O(n)**.
