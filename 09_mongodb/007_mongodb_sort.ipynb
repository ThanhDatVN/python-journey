## Sort the Result — Sắp xếp kết quả

Dùng phương thức `sort()` để sắp xếp kết quả **tăng dần** hoặc **giảm dần**.

`sort()` nhận một tham số là **tên trường**, và một tham số tuỳ chọn là **chiều sắp xếp** (1 tăng dần là mặc định):

````python
import pymongo

myclient = pymongo.MongoClient("mongodb://localhost:27017/")
mydb = myclient["mydatabase"]
mycol = mydb["customers"]

mydoc = mycol.find().sort("name")

for x in mydoc:
    print(x)
````

## Sort Descending — Sắp giảm dần

Dùng giá trị `-1` làm tham số thứ hai:

```text
sort("name", 1)   # tăng dần
sort("name", -1)  # giảm dần
```

````python
mydoc = mycol.find().sort("name", -1)

for x in mydoc:
    print(x)
````

## ➕ Thực hành ngay với mongomock

````python
import mongomock

mycol = mongomock.MongoClient()["mydatabase"]["customers"]
mycol.insert_many([{"name": n} for n in ["Chi", "An", "Em", "Bình", "Dũng"]])

print([x["name"] for x in mycol.find().sort("name")])
print([x["name"] for x in mycol.find().sort("name", -1)])
````

---

## 🧠 Ghi nhớ nhanh

> - `find().sort("field")` → tăng dần (1) · `sort("field", -1)` → giảm dần.
