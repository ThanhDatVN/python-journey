## Creating a Database — Tạo cơ sở dữ liệu

Để tạo cơ sở dữ liệu trong MongoDB, tạo một đối tượng `MongoClient`, chỉ định **chuỗi kết nối** gồm địa chỉ IP và tên cơ sở dữ liệu muốn tạo.

MongoDB sẽ tạo cơ sở dữ liệu nếu chưa tồn tại, và kết nối tới nó:

```python
import pymongo

myclient = pymongo.MongoClient("mongodb://localhost:27017/")

mydb = myclient["mydatabase"]
```

> ⚠️ **Quan trọng:** Trong MongoDB, cơ sở dữ liệu **chưa thực sự được tạo** cho tới khi có **nội dung**!
>
> MongoDB đợi tới khi bạn tạo collection (bảng) chứa ít nhất một document (bản ghi) mới thực sự tạo cơ sở dữ liệu.

## Check if Database Exists — Kiểm tra cơ sở dữ liệu

Liệt kê mọi cơ sở dữ liệu trong hệ thống:

```python
print(myclient.list_database_names())
```

Hoặc kiểm tra một cơ sở dữ liệu cụ thể theo tên:

```python
dblist = myclient.list_database_names()
if "mydatabase" in dblist:
    print("The database exists.")
```

## ➕ Thực hành ngay với mongomock

```python
import mongomock

myclient = mongomock.MongoClient()
mydb = myclient["mydatabase"]

print("Trước khi có dữ liệu:", myclient.list_database_names())

mydb["customers"].insert_one({"name": "An", "address": "Lê Lợi 12"})

print("Sau khi có dữ liệu:", myclient.list_database_names())
if "mydatabase" in myclient.list_database_names():
    print("The database exists.")
```
