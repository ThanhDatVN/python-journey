> 🎯 **Trong bài này:** Insert Into Collection · Return the _id Field · Insert Multiple Documents · Insert Multiple Documents, with Specified IDs · ➕ Thực hành ngay với mongomock

**Document** trong MongoDB giống **bản ghi (record)** trong SQL.

## Insert Into Collection — Chèn document

Để chèn một document vào collection, dùng `insert_one()`.

Tham số đầu tiên là một **dictionary** chứa tên và giá trị các trường:

````python
import pymongo

myclient = pymongo.MongoClient("mongodb://localhost:27017/")
mydb = myclient["mydatabase"]
mycol = mydb["customers"]

mydict = {"name": "An", "address": "Lê Lợi 12"}

x = mycol.insert_one(mydict)
````

## Return the _id Field — Trường _id

`insert_one()` trả về đối tượng `InsertOneResult`, có thuộc tính `inserted_id` chứa **id** của document vừa chèn:

````python
mydict = {"name": "Bình", "address": "Trần Hưng Đạo 4"}

x = mycol.insert_one(mydict)

print(x.inserted_id)
````

Kết quả mẫu:

```text
5b1910482ddb101b7042fcd7
```

Nếu **không** chỉ định trường `_id`, MongoDB tự thêm và gán một **id duy nhất** cho mỗi document.

## Insert Multiple Documents — Chèn nhiều document

Dùng `insert_many()`, tham số là **list các dictionary**:

````python
mylist = [
    {"name": "Chi", "address": "Nguyễn Huệ 65"},
    {"name": "Dũng", "address": "Hai Bà Trưng 21"},
    {"name": "Em", "address": "Lý Thường Kiệt 34"},
]

x = mycol.insert_many(mylist)

# in danh sách _id của các document vừa chèn:
print(x.inserted_ids)
````

## Insert Multiple Documents, with Specified IDs — Tự chỉ định _id

Nếu không muốn MongoDB tự gán id, hãy tự chỉ định `_id` — giá trị phải **duy nhất**:

````python
mylist = [
    {"_id": 1, "name": "Giang", "address": "Nguyễn Trãi 2"},
    {"_id": 2, "name": "Hà", "address": "Lê Duẩn 1"},
    {"_id": 3, "name": "Khoa", "address": "Pasteur 33"},
]

x = mycol.insert_many(mylist)

print(x.inserted_ids)
````

## ➕ Thực hành ngay với mongomock

````python
import mongomock

mycol = mongomock.MongoClient()["mydatabase"]["customers"]

x = mycol.insert_one({"name": "An", "address": "Lê Lợi 12"})
print(x.inserted_id)

x = mycol.insert_many([
    {"name": "Chi", "address": "Nguyễn Huệ 65"},
    {"name": "Dũng", "address": "Hai Bà Trưng 21"},
])
print(len(x.inserted_ids), "documents inserted")

x = mycol.insert_many([
    {"_id": 1, "name": "Giang", "address": "Nguyễn Trãi 2"},
    {"_id": 2, "name": "Hà", "address": "Lê Duẩn 1"},
])
print(x.inserted_ids)
````

> 📌 `mongomock` tạo id theo định dạng riêng; với MongoDB thật, `_id` là một `ObjectId` 24 ký tự hex.

Chèn trùng `_id` sẽ báo lỗi:

````python
import mongomock

mycol = mongomock.MongoClient()["db"]["customers"]
mycol.insert_one({"_id": 1, "name": "A"})
mycol.insert_one({"_id": 1, "name": "B"})
````

---

## 🧠 Ghi nhớ nhanh

> - `insert_one(dict)` → `.inserted_id` · `insert_many(list_dict)` → `.inserted_ids`.
> - Không chỉ định `_id` → MongoDB tự tạo `ObjectId` duy nhất.
> - Tự chỉ định `_id` → phải **duy nhất**, trùng sẽ lỗi.
