# Python MongoDB Delete Document

>
> 🎯 **Trong bài này:** Delete Document · Delete Many Documents · Delete All Documents in a Collection · ➕ Thực hành ngay với mongomock

## Delete Document — Xoá một document

Xoá một document bằng `delete_one()`. Tham số đầu tiên là **đối tượng truy vấn** xác định document cần xoá.

> 📌 Nếu truy vấn khớp **nhiều** document, chỉ document **đầu tiên** bị xoá.

```python
import pymongo

myclient = pymongo.MongoClient("mongodb://localhost:27017/")
mydb = myclient["mydatabase"]
mycol = mydb["customers"]

myquery = {"address": "Hai Bà Trưng 21"}

mycol.delete_one(myquery)
```

## Delete Many Documents — Xoá nhiều document

Dùng `delete_many()`. Xoá mọi document có địa chỉ bắt đầu bằng chữ "L":

```python
myquery = {"address": {"$regex": "^L"}}

x = mycol.delete_many(myquery)

print(x.deleted_count, " documents deleted.")
```

## Delete All Documents in a Collection — Xoá toàn bộ

Truyền đối tượng truy vấn **rỗng** cho `delete_many()`:

```python
x = mycol.delete_many({})

print(x.deleted_count, " documents deleted.")
```

## ➕ Thực hành ngay với mongomock

```python
import mongomock

customers = [("An", "Lê Lợi 12"), ("Dũng", "Hai Bà Trưng 21"), ("Em", "Lý Thường Kiệt 34"),
             ("Hà", "Lê Duẩn 1"), ("Khoa", "Pasteur 33")]
mycol = mongomock.MongoClient()["mydatabase"]["customers"]
mycol.insert_many([{"name": n, "address": a} for n, a in customers])

mycol.delete_one({"address": "Hai Bà Trưng 21"})
print(mycol.count_documents({}), "documents left")

x = mycol.delete_many({"address": {"$regex": "^L"}})
print(x.deleted_count, "documents deleted.")

x = mycol.delete_many({})
print(x.deleted_count, "documents deleted.")
```

---

## 🧠 Ghi nhớ nhanh

> - `delete_one(query)` → xoá document **đầu tiên** khớp.
> - `delete_many(query)` → xoá **mọi** document khớp; `.deleted_count` là số đã xoá.
> - ⚠️ `delete_many({})` → xoá **toàn bộ** collection.
