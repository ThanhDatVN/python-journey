## Update Collection — Cập nhật document

Cập nhật một document bằng `update_one()`.

- Tham số **thứ nhất**: đối tượng truy vấn xác định document cần cập nhật.
- Tham số **thứ hai**: đối tượng mô tả **giá trị mới** (dùng toán tử `$set`).

> 📌 Nếu truy vấn khớp nhiều document, chỉ document **đầu tiên** được cập nhật.

Đổi địa chỉ "Trần Hưng Đạo 4" thành "Pasteur 123":

```python
import pymongo

myclient = pymongo.MongoClient("mongodb://localhost:27017/")
mydb = myclient["mydatabase"]
mycol = mydb["customers"]

myquery = {"address": "Trần Hưng Đạo 4"}
newvalues = {"$set": {"address": "Pasteur 123"}}

mycol.update_one(myquery, newvalues)

# in collection sau khi cập nhật:
for x in mycol.find():
    print(x)
```

## Update Many — Cập nhật nhiều document

Dùng `update_many()`. Cập nhật mọi document có địa chỉ bắt đầu bằng chữ "L":

```python
myquery = {"address": {"$regex": "^L"}}
newvalues = {"$set": {"name": "Minnie"}}

x = mycol.update_many(myquery, newvalues)

print(x.modified_count, "documents updated.")
```

## ➕ Thực hành ngay với mongomock

```python
import mongomock

customers = [("An", "Lê Lợi 12"), ("Bình", "Trần Hưng Đạo 4"), ("Hà", "Lê Duẩn 1"),
             ("Khoa", "Pasteur 33")]
mycol = mongomock.MongoClient()["mydatabase"]["customers"]
mycol.insert_many([{"name": n, "address": a} for n, a in customers])

mycol.update_one({"address": "Trần Hưng Đạo 4"}, {"$set": {"address": "Pasteur 123"}})

x = mycol.update_many({"address": {"$regex": "^L"}}, {"$set": {"name": "Minnie"}})
print(x.modified_count, "documents updated.")

for doc in mycol.find({}, {"_id": 0}):
    print(doc)
```

> ⚠️ Quên `$set` (truyền thẳng `{"address": "..."}`) sẽ **thay thế cả document** bằng nội dung mới — PyMongo mới chặn việc này trong `update_one()` và báo lỗi.

---

## 🧠 Ghi nhớ nhanh

> - `update_one(query, {"$set": {...}})` → cập nhật document **đầu tiên** khớp.
> - `update_many(query, {"$set": {...}})` → cập nhật **mọi** document khớp; `.modified_count`.
> - Luôn dùng toán tử `$set` để chỉ thay đổi các trường chỉ định.
