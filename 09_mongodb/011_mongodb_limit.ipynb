## Limit the Result — Giới hạn kết quả

Giới hạn số kết quả trả về bằng phương thức `limit()`.

`limit()` nhận một tham số là **số document** muốn trả về.

Giả sử collection `customers` có 14 document. Chỉ lấy **5** document đầu tiên:

```python
import pymongo

myclient = pymongo.MongoClient("mongodb://localhost:27017/")
mydb = myclient["mydatabase"]
mycol = mydb["customers"]

myresult = mycol.find().limit(5)

# in kết quả:
for x in myresult:
    print(x)
```

## ➕ Thực hành ngay với mongomock

```python
import mongomock

names = ["An", "Bình", "Chi", "Dũng", "Em", "Giang", "Hà", "Khoa", "Lan", "Minh"]
mycol = mongomock.MongoClient()["mydatabase"]["customers"]
mycol.insert_many([{"_id": i + 1, "name": n} for i, n in enumerate(names)])

for x in mycol.find().limit(5):
    print(x)

# Phân trang: bỏ qua 5, lấy 3 (giống LIMIT 3 OFFSET 5 của SQL)
print([x["name"] for x in mycol.find().skip(5).limit(3)])
```

---

## 🧠 Ghi nhớ nhanh

> - `find().limit(n)` → tối đa `n` document.
> - Kết hợp `skip(k).limit(n)` để **phân trang** (≈ `LIMIT n OFFSET k`).
> - Có thể nối chuỗi: `find(query).sort("name").limit(5)`.
