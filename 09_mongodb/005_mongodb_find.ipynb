# Python MongoDB Find

> 📖 **Học song song trên W3Schools:** [python_mongodb_find.asp](https://www.w3schools.com/python/python_mongodb_find.asp)
>
> 🎯 **Trong bài này:** Find One · Find All · Return Only Some Fields · ➕ Thực hành ngay với mongomock

Trong MongoDB, ta dùng các phương thức `find()` và `find_one()` để tìm dữ liệu — giống câu lệnh `SELECT` trong SQL.

## Find One — Tìm một document

`find_one()` trả về **document đầu tiên** trong collection:

```python
import pymongo

myclient = pymongo.MongoClient("mongodb://localhost:27017/")
mydb = myclient["mydatabase"]
mycol = mydb["customers"]

x = mycol.find_one()

print(x)
```

## Find All — Tìm tất cả

`find()` trả về **tất cả** document thoả điều kiện. Tham số đầu tiên là **đối tượng truy vấn** — để trống `{}` (hoặc không truyền) nghĩa là chọn tất cả, giống `SELECT *` trong SQL:

```python
for x in mycol.find():
    print(x)
```

## Return Only Some Fields — Chỉ trả về một số trường

Tham số thứ hai của `find()` là đối tượng mô tả **trường nào được trả về** (projection). Giá trị `1` = có, `0` = không:

```python
for x in mycol.find({}, {"_id": 0, "name": 1, "address": 1}):
    print(x)
```

> ⚠️ Không được trộn `0` và `1` trong cùng một projection — **trừ** trường `_id` (được phép đặt `0` cùng các trường `1`).

Loại bỏ trường `address`:

```python
for x in mycol.find({}, {"address": 0}):
    print(x)
```

## ➕ Thực hành ngay với mongomock

```python
import mongomock

mycol = mongomock.MongoClient()["mydatabase"]["customers"]
mycol.insert_many([
    {"_id": 1, "name": "An", "address": "Lê Lợi 12"},
    {"_id": 2, "name": "Bình", "address": "Trần Hưng Đạo 4"},
    {"_id": 3, "name": "Chi", "address": "Nguyễn Huệ 65"},
])

print(mycol.find_one())

for x in mycol.find():
    print(x)
```

```python
for x in mycol.find({}, {"_id": 0, "name": 1, "address": 1}):
    print(x)

for x in mycol.find({}, {"address": 0}):
    print(x)
```

---

## 🧠 Ghi nhớ nhanh

> - `find_one()` → document đầu tiên · `find()` → tất cả (≈ `SELECT *`).
> - Projection (tham số 2): `{"_id": 0, "name": 1}` — 1 lấy, 0 bỏ.
> - ⚠️ Không trộn 0 và 1 (trừ `_id`).
