## Creating a Collection — Tạo collection

**Collection** trong MongoDB giống **bảng** trong SQL.

Để tạo collection, dùng đối tượng database và chỉ định tên collection. MongoDB sẽ tạo collection nếu chưa tồn tại:

```python
import pymongo

myclient = pymongo.MongoClient("mongodb://localhost:27017/")
mydb = myclient["mydatabase"]

mycol = mydb["customers"]
```

> ⚠️ Giống database: collection **chưa được tạo** cho tới khi có nội dung (ít nhất một document).

## Check if Collection Exists — Kiểm tra collection

Liệt kê mọi collection trong cơ sở dữ liệu:

```python
print(mydb.list_collection_names())
```

Kiểm tra một collection cụ thể:

```python
collist = mydb.list_collection_names()
if "customers" in collist:
    print("The collection exists.")
```

## ➕ Thực hành ngay với mongomock

```python
import mongomock

myclient = mongomock.MongoClient()
mydb = myclient["mydatabase"]
mycol = mydb["customers"]

print(mydb.list_collection_names())
mycol.insert_one({"name": "An"})
print(mydb.list_collection_names())

if "customers" in mydb.list_collection_names():
    print("The collection exists.")
```

---

## 🧠 Ghi nhớ nhanh

> - Collection ≈ bảng: `mycol = mydb["customers"]`.
> - Chỉ thực sự được tạo khi có **document**.
> - Liệt kê: `mydb.list_collection_names()`.
