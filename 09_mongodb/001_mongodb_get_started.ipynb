Python có thể dùng trong các ứng dụng cơ sở dữ liệu.

Một trong những cơ sở dữ liệu **NoSQL** phổ biến nhất là **MongoDB**.

## MongoDB

MongoDB lưu dữ liệu dưới dạng **document** giống JSON — nhờ vậy cơ sở dữ liệu rất **linh hoạt** và **mở rộng tốt**.

| SQL (MySQL) | MongoDB |
|---|---|
| Database | Database |
| Table (bảng) | **Collection** |
| Row (dòng) | **Document** |
| Column (cột) | **Field** |

Để thực hành, bạn cần truy cập một cơ sở dữ liệu MongoDB:

- Cài **MongoDB Community** trên máy: [mongodb.com/try/download/community](https://www.mongodb.com/try/download/community)
- Hoặc dùng **MongoDB Atlas** miễn phí trên cloud: [mongodb.com/atlas](https://www.mongodb.com/atlas)

> 💡 **Không muốn cài MongoDB?** Mỗi bài có mục **➕ Thực hành ngay với mongomock** — một thư viện Python giả lập MongoDB **trong bộ nhớ**, API giống hệt `pymongo`. Cài bằng `pip install mongomock` (đã có trong `requirements.txt`).

## ➕ Thực hành ngay với mongomock

````python
import mongomock

myclient = mongomock.MongoClient()
print(myclient)
````

Với MongoDB thật, chỉ cần đổi **một dòng**:

````python
import pymongo

myclient = pymongo.MongoClient("mongodb://localhost:27017/")
````
