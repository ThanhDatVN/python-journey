## Filter the Result — Lọc kết quả

Khi tìm document, có thể lọc bằng **đối tượng truy vấn (query object)** — tham số đầu tiên của `find()`.

Tìm document có địa chỉ "Lê Lai 38":

```python
import pymongo

myclient = pymongo.MongoClient("mongodb://localhost:27017/")
mydb = myclient["mydatabase"]
mycol = mydb["customers"]

myquery = {"address": "Lê Lai 38"}

mydoc = mycol.find(myquery)

for x in mydoc:
    print(x)
```

## Advanced Query — Truy vấn nâng cao

Dùng **toán tử (modifier)** làm giá trị trong đối tượng truy vấn.

Ví dụ tìm document có địa chỉ bắt đầu bằng chữ "S" **trở về sau** (theo thứ tự bảng chữ cái), dùng toán tử "lớn hơn" `{"$gt": "S"}`:

```python
myquery = {"address": {"$gt": "S"}}

mydoc = mycol.find(myquery)

for x in mydoc:
    print(x)
```

| Toán tử | Ý nghĩa |
|---|---|
| `$gt` / `$gte` | Lớn hơn / lớn hơn hoặc bằng |
| `$lt` / `$lte` | Nhỏ hơn / nhỏ hơn hoặc bằng |
| `$ne` | Khác |
| `$in` | Nằm trong danh sách |

## Filter With Regular Expressions — Lọc bằng biểu thức chính quy

Dùng **regex** làm điều kiện — chỉ dùng được cho truy vấn **chuỗi**.

Tìm document có địa chỉ bắt đầu bằng chữ "L", dùng `{"$regex": "^L"}`:

```python
myquery = {"address": {"$regex": "^L"}}

mydoc = mycol.find(myquery)

for x in mydoc:
    print(x)
```

## ➕ Thực hành ngay với mongomock

```python
import mongomock

customers = [
    ("An", "Lê Lợi 12"), ("Bình", "Trần Hưng Đạo 4"), ("Chi", "Nguyễn Huệ 65"),
    ("Dũng", "Hai Bà Trưng 21"), ("Em", "Lý Thường Kiệt 34"), ("Giang", "Nguyễn Trãi 2"),
    ("Hà", "Lê Duẩn 1"), ("Khoa", "Pasteur 33"), ("Minh", "Võ Văn Tần 2"), ("Nam", "Lê Lai 38"),
]
mycol = mongomock.MongoClient()["mydatabase"]["customers"]
mycol.insert_many([{"name": n, "address": a} for n, a in customers])

print(list(mycol.find({"address": "Lê Lai 38"}, {"_id": 0})))
print(list(mycol.find({"address": {"$gt": "S"}}, {"_id": 0})))
print(list(mycol.find({"address": {"$regex": "^L"}}, {"_id": 0})))
```

---

## 🧠 Ghi nhớ nhanh

> - Lọc: `find({"field": "giá trị"})`.
> - Toán tử: `$gt $gte $lt $lte $ne $in` → `{"field": {"$gt": x}}`.
> - Regex: `{"field": {"$regex": "^L"}}` (chỉ cho chuỗi).
