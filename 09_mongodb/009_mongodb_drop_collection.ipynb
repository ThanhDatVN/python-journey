# Python MongoDB Drop Collection

> 📖 **Học song song trên W3Schools:** [python_mongodb_drop_collection.asp](https://www.w3schools.com/python/python_mongodb_drop_collection.asp)
>
> 🎯 **Trong bài này:** Delete Collection · ➕ Thực hành ngay với mongomock

## Delete Collection — Xoá collection

Xoá một collection (bảng) bằng phương thức `drop()`:

```python
import pymongo

myclient = pymongo.MongoClient("mongodb://localhost:27017/")
mydb = myclient["mydatabase"]
mycol = mydb["customers"]

mycol.drop()
```

> ⚠️ **Lưu ý cập nhật:** W3Schools ghi `drop()` trả về `True`/`False`, nhưng với PyMongo, `drop()` trả về `None` (kể cả khi collection không tồn tại — không báo lỗi). Muốn biết collection còn hay không, kiểm tra bằng `list_collection_names()`.

## ➕ Thực hành ngay với mongomock

```python
import mongomock

mydb = mongomock.MongoClient()["mydatabase"]
mydb["customers"].insert_one({"name": "An"})
print(mydb.list_collection_names())

mydb["customers"].drop()
print(mydb.list_collection_names())
```

---

## 🧠 Ghi nhớ nhanh

> - `mycol.drop()` → xoá **cả collection** (≈ `DROP TABLE`).
> - So sánh: `delete_many({})` xoá document nhưng collection vẫn còn.
