## Delete a File — Xoá file

Để xoá file, cần import module **`os`** và dùng hàm `os.remove()`.

Để không xoá nhầm `demofile.txt` của bài học, ta tạo một file tạm trước:

```python
with open("demofile_tmp.txt", "w", encoding="utf-8") as f:
    f.write("Tôi sắp bị xoá!")
```

Xoá file `demofile_tmp.txt`:

```python
import os

os.remove("demofile_tmp.txt")
print("Đã xoá file")
```

## Check if File exist — Kiểm tra file có tồn tại

Xoá file không tồn tại sẽ báo lỗi. Để tránh, hãy **kiểm tra trước**:

```python
import os

if os.path.exists("demofile_tmp.txt"):
    os.remove("demofile_tmp.txt")
else:
    print("The file does not exist")
```

## Delete Folder — Xoá thư mục

Để xoá **cả thư mục**, dùng `os.rmdir()`:

```python
import os

os.mkdir("myfolder")     # tạo thư mục để thực hành
print(os.path.exists("myfolder"))

os.rmdir("myfolder")
print(os.path.exists("myfolder"))
```

> 📌 `os.rmdir()` chỉ xoá được **thư mục rỗng**.

## ➕ Mở rộng: pathlib — cách hiện đại

Module `pathlib` cho cú pháp gọn và dễ đọc hơn:

```python
from pathlib import Path

p = Path("tam.txt")
p.write_text("xin chào", encoding="utf-8")
print(p.exists(), p.read_text(encoding="utf-8"))

p.unlink()                         # xoá file
print(p.exists())
Path("tam.txt").unlink(missing_ok=True)   # không lỗi nếu file không tồn tại
```

Xoá thư mục **có nội dung** dùng `shutil.rmtree()` — ⚠️ cẩn thận, không thể khôi phục:

```python
import os
import shutil

os.makedirs("thu_muc/con", exist_ok=True)
Path("thu_muc/con/a.txt").write_text("a", encoding="utf-8")

shutil.rmtree("thu_muc")
print(os.path.exists("thu_muc"))
```

> 🧪 **Try it Yourself** — Dọn dẹp các file đã tạo ở bài trước (demofile2.txt, myfile.txt, nhatky.txt) nếu chúng tồn tại.

```python
import os

for name in ["demofile2.txt", "myfile.txt", "nhatky.txt"]:
    if os.path.exists(name):
        os.remove(name)
        print("Đã xoá", name)
    else:
        print(name, "không tồn tại")
```

## 🧠 Ghi nhớ nhanh

> - Xoá file: `import os` → `os.remove("file")`.
> - Kiểm tra trước: `os.path.exists("file")` để tránh lỗi.
> - Xoá thư mục **rỗng**: `os.rmdir()`; thư mục có nội dung: `shutil.rmtree()` (⚠️ cẩn thận).
> - Hiện đại: `Path("file").unlink(missing_ok=True)`.
