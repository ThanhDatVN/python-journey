## Write to an Existing File — Ghi vào file có sẵn

Để ghi vào file có sẵn, thêm tham số chế độ cho `open()`:

- `"a"` — **Append**: ghi thêm vào **cuối** file
- `"w"` — **Write**: **ghi đè** toàn bộ nội dung cũ

Để không làm thay đổi `demofile.txt` dùng cho bài đọc, ta tạo bản sao `demofile2.txt` để thực hành:

```python
import shutil
shutil.copy("demofile.txt", "demofile2.txt")
```

Mở `demofile2.txt` và **thêm** nội dung vào cuối:

```python
with open("demofile2.txt", "a", encoding="utf-8") as f:
    f.write("Now the file has more content!")

# mở và đọc file sau khi ghi thêm:
with open("demofile2.txt", encoding="utf-8") as f:
    print(f.read())
```

## Overwrite Existing Content — Ghi đè nội dung

Dùng chế độ `"w"` để ghi đè:

```python
with open("demofile2.txt", "w", encoding="utf-8") as f:
    f.write("Woops! I have deleted the content!")

# mở và đọc file sau khi ghi đè:
with open("demofile2.txt", encoding="utf-8") as f:
    print(f.read())
```

> ⚠️ Chế độ `"w"` **xoá sạch** nội dung cũ của file!

## Create a New File — Tạo file mới

Để tạo file mới, dùng `open()` với một trong các chế độ:

- `"x"` — **Create**: tạo file, **báo lỗi** nếu file đã tồn tại
- `"a"` — **Append**: tạo file nếu chưa có
- `"w"` — **Write**: tạo file nếu chưa có

Tạo file mới tên `myfile.txt`:

```python
import os

if os.path.exists("myfile.txt"):
    os.remove("myfile.txt")   # dọn dẹp để ví dụ chạy lại được

f = open("myfile.txt", "x", encoding="utf-8")
f.close()
print(os.path.exists("myfile.txt"))
```

Tạo lần nữa với `"x"` sẽ báo lỗi vì file đã có:

```python
f = open("myfile.txt", "x")
```

Tạo file mới nếu chưa tồn tại bằng `"w"`:

```python
f = open("myfile.txt", "w", encoding="utf-8")
f.close()
```

## ➕ Mở rộng: Ghi nhiều dòng

`write()` **không tự thêm** ký tự xuống dòng — bạn phải tự thêm `\n`. Có thể dùng `writelines()` để ghi một list:

```python
lines = ["Dòng 1\n", "Dòng 2\n", "Dòng 3\n"]

with open("myfile.txt", "w", encoding="utf-8") as f:
    f.write("Tiêu đề\n")
    f.writelines(lines)

with open("myfile.txt", encoding="utf-8") as f:
    print(f.read())
```

Hoặc dùng `print(..., file=f)`:

```python
with open("myfile.txt", "w", encoding="utf-8") as f:
    for i in range(1, 4):
        print("Số", i, file=f)

with open("myfile.txt", encoding="utf-8") as f:
    print(f.read())
```

> 🧪 **Try it Yourself** — Ghi nhật ký học tập: mỗi lần chạy cell sẽ thêm một dòng vào file nhatky.txt.

```python
import datetime

with open("nhatky.txt", "a", encoding="utf-8") as f:
    f.write(f"{datetime.date.today()}: Học bài Write Files\n")

with open("nhatky.txt", encoding="utf-8") as f:
    print(f.read())
```

---

## 🧠 Ghi nhớ nhanh

> - `"a"` ghi **thêm** vào cuối · `"w"` ghi **đè** (⚠️ mất nội dung cũ) · `"x"` **tạo mới** (lỗi nếu đã có).
> - `"a"` và `"w"` tự tạo file nếu chưa tồn tại.
> - `write()` **không** tự xuống dòng → thêm `\n`; ghi list bằng `writelines()`.
