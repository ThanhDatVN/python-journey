## Open a File on the Server — Mở và đọc file

Giả sử có file `demofile.txt` nằm cùng thư mục với notebook này:

```text
Hello! Welcome to demofile.txt
This file is for testing purposes.
Xin chào! Dòng này có tiếng Việt để thử encoding.
Good Luck!
```

Để mở file, dùng hàm `open()`. Nó trả về một **file object**, có phương thức `read()` để đọc nội dung:

````python
f = open("demofile.txt", encoding="utf-8")
print(f.read())
f.close()
````

Nếu file nằm ở thư mục khác, ghi rõ đường dẫn:

````python
f = open("D:\\myfiles\\welcome.txt", encoding="utf-8")
print(f.read())
````

## Using the with statement — Dùng câu lệnh with

Có thể dùng câu lệnh `with` khi mở file. Khi đó **file tự động được đóng** sau khi ra khỏi khối `with`:

````python
with open("demofile.txt", encoding="utf-8") as f:
    print(f.read())
````

> 💡 Đây là cách được **khuyến khích**, vì bạn không bao giờ quên đóng file.

## Close Files — Đóng file

Nếu không dùng `with`, hãy luôn **đóng file** sau khi dùng xong:

````python
f = open("demofile.txt", encoding="utf-8")
print(f.readline())
f.close()
````

> 📌 Nên luôn đóng file. Trong một số trường hợp, do bộ đệm (buffering), thay đổi trên file có thể chưa được ghi cho tới khi file được đóng.

## Read Only Parts of the File — Đọc một phần file

Mặc định `read()` đọc **toàn bộ** nội dung, nhưng có thể chỉ định **số ký tự** muốn đọc:

````python
with open("demofile.txt", encoding="utf-8") as f:
    print(f.read(5))
````

## Read Lines — Đọc từng dòng

Đọc một dòng bằng `readline()`:

````python
with open("demofile.txt", encoding="utf-8") as f:
    print(f.readline())
````

Gọi `readline()` hai lần để đọc hai dòng đầu:

````python
with open("demofile.txt", encoding="utf-8") as f:
    print(f.readline())
    print(f.readline())
````

Duyệt qua file bằng vòng lặp để đọc **từng dòng**:

````python
with open("demofile.txt", encoding="utf-8") as f:
    for x in f:
        print(x)
````

> 💡 Mỗi dòng đã có sẵn ký tự xuống dòng `\n` ở cuối, nên `print()` tạo ra dòng trống xen kẽ. Dùng `x.strip()` hoặc `print(x, end="")` để bỏ.

## ➕ Mở rộng: readlines() và đếm dòng

````python
with open("demofile.txt", encoding="utf-8") as f:
    lines = f.readlines()

print(lines)
print("Số dòng:", len(lines))
for i, line in enumerate(lines, start=1):
    print(i, line.strip())
````

Thử đọc file tiếng Việt **không** chỉ định encoding (trên Windows có thể ra ký tự lạ hoặc lỗi):

````python
try:
    with open("demofile.txt") as f:
        print(f.read().splitlines()[2])
except UnicodeDecodeError as e:
    print("Lỗi encoding:", e)
````

> 🧪 **Try it Yourself** — Đếm số từ trong file demofile.txt.

````python
with open("demofile.txt", encoding="utf-8") as f:
    content = f.read()

print("Số từ:", len(content.split()))
````

---

## 🧠 Ghi nhớ nhanh

> - `f.read()` toàn bộ · `f.read(5)` 5 ký tự · `f.readline()` một dòng · `for line in f` từng dòng · `f.readlines()` list các dòng.
> - Dùng `with open(...) as f:` → **tự động đóng** file.
> - Không dùng `with` → nhớ `f.close()`.
> - Mỗi dòng có `\n` ở cuối → `line.strip()`.
