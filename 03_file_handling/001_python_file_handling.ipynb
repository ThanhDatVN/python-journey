Xử lý file là phần quan trọng của mọi ứng dụng web và phần mềm.

Python có nhiều hàm để **tạo, đọc, cập nhật và xoá** file.

## File Handling — Làm việc với file

Hàm chính để làm việc với file là **`open()`**.

`open()` nhận hai tham số chính: **tên file** (`filename`) và **chế độ** (`mode`).

Có bốn chế độ mở file:

| Chế độ | Tên | Ý nghĩa |
|---|---|---|
| `"r"` | Read | **Mặc định.** Mở để đọc; lỗi nếu file không tồn tại |
| `"a"` | Append | Mở để **ghi thêm** vào cuối; tạo file nếu chưa có |
| `"w"` | Write | Mở để **ghi đè**; tạo file nếu chưa có |
| `"x"` | Create | **Tạo** file mới; lỗi nếu file đã tồn tại |

Ngoài ra, có thể chỉ định file được xử lý ở dạng **văn bản** hay **nhị phân**:

| Chế độ | Tên | Ý nghĩa |
|---|---|---|
| `"t"` | Text | **Mặc định.** Chế độ văn bản |
| `"b"` | Binary | Chế độ nhị phân (ví dụ: ảnh) |

## Syntax — Cú pháp

Để mở một file để đọc, chỉ cần ghi tên file:

```python
f = open("demofile.txt", encoding="utf-8")
f.close()
```

Câu lệnh trên tương đương với:

```python
f = open("demofile.txt", "rt", encoding="utf-8")
f.close()
```

Vì `"r"` (đọc) và `"t"` (văn bản) là giá trị mặc định, bạn không cần ghi ra.

> 📌 Phải chắc chắn file tồn tại, nếu không sẽ báo lỗi:

```python
f = open("khong_ton_tai.txt")
```

## ➕ Mở rộng: Luôn ghi rõ `encoding="utf-8"`

Trên **Windows**, Python mặc định đọc/ghi file theo bảng mã của hệ điều hành (thường là `cp1252`), **không phải UTF-8**. Với file có tiếng Việt, điều này gây lỗi hoặc ra ký tự lạ.

```python
import locale
print("Bảng mã mặc định của máy này:", locale.getpreferredencoding())
```

👉 **Quy tắc vàng:** khi mở file văn bản, luôn thêm `encoding="utf-8"`. (Từ Python 3.15, UTF-8 mới trở thành mặc định trên mọi hệ điều hành.)

---

## 🧠 Ghi nhớ nhanh

> - `open(filename, mode)` — mode: `"r"` đọc (mặc định) · `"a"` ghi thêm · `"w"` ghi đè · `"x"` tạo mới.
> - Thêm `"t"` (văn bản, mặc định) hoặc `"b"` (nhị phân).
> - `"r"` với file không tồn tại → `FileNotFoundError`.
> - 🇻🇳 Luôn thêm `encoding="utf-8"` khi làm việc với file văn bản (nhất là trên Windows).
