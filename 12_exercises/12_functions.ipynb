````python
def shout(func):
    # ✍️ Viết code của bạn ở đây
    pass

@shout
def greet(name):
    return f"hello {name}"
````

````python
assert greet("an") == "HELLO AN!", f"Kết quả của bạn: {greet('an')!r}"
print("✅ Chính xác!")
````

<details>
<summary>💡 Xem lời giải</summary>

```python
import functools

def shout(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs).upper() + "!"
    return wrapper

@shout
def greet(name):
    return f"hello {name}"
```

</details>
