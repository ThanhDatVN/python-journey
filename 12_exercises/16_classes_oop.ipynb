### Challenge 1 — Hình chữ nhật

Viết class `Rectangle(width, height)` có phương thức `area()`, `perimeter()` và `__str__` trả về `"Rectangle(3x4)"`.

````python
class Rectangle:
    # ✍️ Viết code của bạn ở đây
    pass
````

````python
r = Rectangle(3, 4)
assert r.area() == 12
assert r.perimeter() == 14
assert str(r) == "Rectangle(3x4)"
print("✅ Chính xác!")
````

<details>
<summary>💡 Xem lời giải</summary>

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

    def __str__(self):
        return f"Rectangle({self.width}x{self.height})"
```

### Challenge 2 — Tài khoản ngân hàng

Viết class `BankAccount` với số dư **private** `__balance` (ban đầu 0), phương thức `deposit(amount)`, `withdraw(amount)` (rút quá số dư → `raise ValueError`) và `get_balance()`.

````python
class BankAccount:
    # ✍️ Viết code của bạn ở đây
    pass
````

````python
acc = BankAccount()
acc.deposit(100)
acc.withdraw(30)
assert acc.get_balance() == 70
try:
    acc.withdraw(1000)
    raise AssertionError("Rút quá số dư phải báo ValueError")
except ValueError:
    pass
assert not hasattr(acc, "balance") and not hasattr(acc, "__balance"), "Số dư phải là private (__balance)"
print("✅ Chính xác!")
````

<details>
<summary>💡 Xem lời giải</summary>

```python
class BankAccount:
    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount):
        if amount > self.__balance:
            raise ValueError("Số dư không đủ")
        self.__balance -= amount

    def get_balance(self):
        return self.__balance
```

</details>

### Challenge 3 — Đa hình với động vật

Viết class cha `Animal` có phương thức `speak()` trả về `"..."`, và hai class con `Dog`, `Cat` ghi đè `speak()` trả về `"Gâu"` và `"Meo"`.

````python
class Animal:
    pass

# ✍️ Viết Dog và Cat
````

````python
sounds = [a.speak() for a in (Dog(), Cat(), Animal())]
assert sounds == ["Gâu", "Meo", "..."], sounds
assert isinstance(Dog(), Animal) and isinstance(Cat(), Animal)
print("✅ Chính xác!")
````

<details>

```python
class Animal:
    def speak(self):
```

### Challenge 4 — Vector cộng được

Viết class `Vector(x, y)` hỗ trợ `v1 + v2`, `v1 == v2` và `repr` dạng `Vector(1, 2)`.

````python
class Vector:
    # ✍️ Viết code của bạn ở đây
    pass
````

````python
assert Vector(1, 2) + Vector(3, 4) == Vector(4, 6)
assert repr(Vector(1, 2)) == "Vector(1, 2)"
print("✅ Chính xác!")
````

<details>
<summary>💡 Xem lời giải</summary>

```python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __eq__(self, other):
        return (self.x, self.y) == (other.x, other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
```

</details>
