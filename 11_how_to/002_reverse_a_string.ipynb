# How to Reverse a String in Python

> 📖 **Học song song trên W3Schools:** [python_howto_reverse_string.asp](https://www.w3schools.com/python/python_howto_reverse_string.asp)
>
> 🎯 **Trong bài này:** Example · Example Explained · Create a Function · ➕ Mở rộng: Các cách khác và ứng dụng

Học cách đảo ngược một chuỗi trong Python.

Python không có hàm có sẵn để đảo ngược chuỗi. Cách nhanh nhất (và dễ nhất) là dùng **slice bước lùi** `-1`.

## Example — Ví dụ

Đảo ngược chuỗi "Hello World":

```python
txt = "Hello World"[::-1]
print(txt)
```

## Example Explained — Giải thích

Ta có chuỗi "Hello World" cần đảo ngược:

```python
txt = "Hello World"
```

Dùng slice bắt đầu từ cuối chuỗi và lùi về đầu. Cú pháp `[::-1]` nghĩa là: *bắt đầu từ cuối, kết thúc ở vị trí 0, bước nhảy -1* (lùi một bước mỗi lần):

```python
txt = "Hello World"[::-1]
print(txt)
```

## Create a Function — Viết thành hàm

```python
def my_function(x):
    return x[::-1]

mytxt = my_function("I wonder how this text looks like backwards")
print(mytxt)
```

## ➕ Mở rộng: Các cách khác và ứng dụng

```python
s = "Python"
print("".join(reversed(s)))     # dùng reversed() + join()

word = "level"
print(word == word[::-1])       # kiểm tra chuỗi đối xứng (palindrome)
```

> 🧪 **Try it Yourself** — Kiểm tra một câu có đối xứng không (bỏ khoảng trắng, không phân biệt hoa thường).

```python
sentence = "Never odd or even"
clean = sentence.replace(" ", "").lower()
print(clean == clean[::-1])
```

---

## 🧠 Ghi nhớ nhanh

> - Đảo ngược chuỗi: `s[::-1]` (slice với bước `-1`).
> - Cách khác: `"".join(reversed(s))`.
> - Kiểm tra palindrome: `s == s[::-1]`.
