# How to Add Two Numbers in Python

> 📖 **Học song song trên W3Schools:** [python_howto_add_two_numbers.asp](https://www.w3schools.com/python/python_howto_add_two_numbers.asp)
>
> 🎯 **Trong bài này:** Example · Add Two Numbers with User Input · ➕ Mở rộng: Cộng số thực và kiểm tra dữ liệu

Học cách cộng hai số trong Python.

## Example — Ví dụ

```python
x = 5
y = 10
print(x + y)
```

## Add Two Numbers with User Input — Cộng hai số người dùng nhập

Ví dụ này yêu cầu người dùng nhập hai số, sau đó in tổng:

```python
x = input("Type a number: ")
y = input("Type another number: ")

sum = int(x) + int(y)

print("The sum is: ", sum)
```

> 📌 `input()` luôn trả về **chuỗi**, nên phải chuyển sang số bằng `int()` trước khi cộng. Nếu không, Python sẽ **nối chuỗi**: `"7" + "15"` → `"715"`.

## ➕ Mở rộng: Cộng số thực và kiểm tra dữ liệu

```python
def read_number(prompt):
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Vui lòng nhập một số!")

a = read_number("Số thứ nhất: ")
b = read_number("Số thứ hai: ")
print(f"{a} + {b} = {a + b}")
```

> ⚠️ Ví dụ gốc dùng tên biến `sum` — trùng tên hàm có sẵn `sum()`. Trong code thật, nên đặt tên khác như `total`.

---

## 🧠 Ghi nhớ nhanh

> - Cộng hai số: `x + y`.
> - Dữ liệu từ `input()` là **chuỗi** → `int()`/`float()` trước khi cộng.
> - ⚠️ Tránh đặt tên biến trùng hàm có sẵn (`sum`, `list`, `str`…).
