# How to Remove Duplicates From a Python List

> 📖 **Học song song trên W3Schools:** [python_howto_remove_duplicates.asp](https://www.w3schools.com/python/python_howto_remove_duplicates.asp)
>
> 🎯 **Trong bài này:** Example · Example Explained · Create a Function · ➕ Mở rộng: Vì sao không dùng set()?

Học cách loại bỏ phần tử trùng lặp khỏi một list trong Python.

## Example — Ví dụ

```python
mylist = ["a", "b", "a", "c", "c"]
mylist = list(dict.fromkeys(mylist))
print(mylist)
```

## Example Explained — Giải thích

Đầu tiên ta có một list chứa phần tử trùng lặp:

```python
mylist = ["a", "b", "a", "c", "c"]
```

Tạo một **dictionary** dùng các phần tử của list làm **key**. Vì dictionary **không cho phép key trùng**, các phần tử lặp lại tự động bị loại bỏ:

```python
mylist = ["a", "b", "a", "c", "c"]
print(dict.fromkeys(mylist))
```

Sau đó chuyển dictionary ngược lại thành list:

```python
mylist = ["a", "b", "a", "c", "c"]
mylist = list(dict.fromkeys(mylist))
print(mylist)
```

## Create a Function — Viết thành hàm

Muốn dùng lại nhiều lần với các list khác nhau, hãy viết thành hàm:

```python
def my_function(x):
    return list(dict.fromkeys(x))

mylist = my_function(["a", "b", "a", "c", "c"])
print(mylist)
```

## ➕ Mở rộng: Vì sao không dùng set()?

`set()` cũng loại trùng được, nhưng **không giữ thứ tự** ban đầu. `dict.fromkeys()` giữ đúng thứ tự xuất hiện:

```python
data = ["banana", "apple", "banana", "cherry", "apple"]
print(list(set(data)))             # thứ tự có thể thay đổi
print(list(dict.fromkeys(data)))   # giữ thứ tự xuất hiện đầu tiên
```

> 🧪 **Try it Yourself** — Loại bỏ email trùng lặp nhưng giữ đúng thứ tự đăng ký.

```python
emails = ["an@x.com", "binh@x.com", "an@x.com", "chi@x.com", "binh@x.com"]
print(list(dict.fromkeys(emails)))
```

## 🧠 Ghi nhớ nhanh

> - Loại trùng **giữ thứ tự**: `list(dict.fromkeys(my_list))`.
> - `list(set(my_list))` cũng loại trùng nhưng **không giữ thứ tự**.
