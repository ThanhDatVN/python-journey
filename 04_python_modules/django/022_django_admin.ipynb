## Django Admin — Trang quản trị

**Django admin** là công cụ rất mạnh: một **giao diện web** để **tạo, đọc, sửa, xoá (CRUD)** dữ liệu trong các model — mà gần như không cần viết code.

Trang admin đã được cấu hình sẵn khi tạo dự án. Trong `urls.py` gốc có dòng:

```python
from django_helper import show_file

show_file("my_tennis_club/my_tennis_club/urls.py")
```

Dòng `path('admin/', admin.site.urls)` chính là trang admin, truy cập tại `127.0.0.1:8000/admin/`.

## 👀 Xem kết quả

Truy cập `/admin/` khi **chưa đăng nhập** sẽ bị **chuyển hướng** tới trang đăng nhập:

```python
from django_helper import preview

preview("/admin/")
```

```python
preview("/admin/login/?next=/admin/", max_chars=600)
```

Để đăng nhập, cần tạo một **tài khoản quản trị** — bài tiếp theo.

---

## 🧠 Ghi nhớ nhanh

> - Django admin: giao diện web **CRUD** có sẵn tại `/admin/`.
> - Khai báo bởi `path('admin/', admin.site.urls)` trong `urls.py` gốc.
> - Chưa đăng nhập → chuyển hướng tới `/admin/login/`.
