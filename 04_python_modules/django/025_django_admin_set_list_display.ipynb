# Django Admin - Set Fields to Display

## Set Fields to Display — Chọn cột hiển thị

Có thể điều khiển các **cột** được hiển thị trong danh sách admin, bằng cách tạo class **`MemberAdmin()`** và khai báo `list_display`:

```python
%%writefile my_tennis_club/members/admin.py
from django.contrib import admin
from .models import Member

# Register your models here.

class MemberAdmin(admin.ModelAdmin):
    list_display = ("firstname", "lastname", "joined_date",)

admin.site.register(Member, MemberAdmin)
```

> 📌 Nhớ truyền `MemberAdmin` làm đối số thứ hai của `admin.site.register()`.

Giờ danh sách thành viên trong admin hiển thị 3 cột: **Firstname**, **Lastname**, **Joined date**.

## ➕ Mở rộng: Tìm kiếm, lọc và sắp xếp

`ModelAdmin` còn nhiều tuỳ chọn hữu ích:

```python
class MemberAdmin(admin.ModelAdmin):
    list_display = ("firstname", "lastname", "joined_date",)
    search_fields = ("firstname", "lastname")   # ô tìm kiếm
    list_filter = ("joined_date",)              # bộ lọc bên phải
    ordering = ("lastname",)                    # sắp xếp mặc định
```

## 👀 Kiểm tra

```python
from django_helper import setup

setup()
from django.contrib import admin
from members.models import Member

print(admin.site._registry[Member].list_display)
```

---

## 🧠 Ghi nhớ nhanh

> - Tuỳ biến admin: `class MemberAdmin(admin.ModelAdmin)` + `list_display = (...)`.
> - Đăng ký kèm class: `admin.site.register(Member, MemberAdmin)`.
> - Thêm: `search_fields`, `list_filter`, `ordering`.
