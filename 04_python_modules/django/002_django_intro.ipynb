## What is Django? — Django là gì?

Django là một **framework web** Python giúp tạo website **an toàn, dễ bảo trì** một cách **nhanh chóng**.

Django tuân theo nguyên tắc **DRY** (Don't Repeat Yourself) và đi kèm nhiều thành phần dựng sẵn ("batteries included").

## How does Django Work? — Django hoạt động thế nào?

Django theo mô hình **MVT** — **Model, View, Template**:

| Thành phần | Vai trò | File |
|---|---|---|
| **Model** | Dữ liệu — làm việc với cơ sở dữ liệu (thường là bảng) | `models.py` |
| **View** | Hàm nhận **HTTP request** và trả về **HTTP response** | `views.py` |
| **Template** | File HTML mô tả **giao diện** hiển thị | thư mục `templates/` |

Ngoài ra còn **URLs**: Django dùng file `urls.py` để **ánh xạ URL tới view**.

### Quy trình xử lý một request

```text
Trình duyệt ──request──▶ urls.py ──▶ view ──▶ model (lấy dữ liệu)
                                       │
                                       ▼
Trình duyệt ◀──response──  HTML ◀── template (hiển thị dữ liệu)
```

1. Django nhận URL, tra trong `urls.py` để tìm **view** phù hợp.
2. View xử lý — thường lấy dữ liệu từ **model**.
3. View gửi dữ liệu cho **template**.
4. Template sinh ra **HTML** trả về cho trình duyệt.

### Model

Model cung cấp dữ liệu từ cơ sở dữ liệu. Django dùng **ORM (Object-Relational Mapping)** — bạn làm việc với dữ liệu bằng **class Python**, không cần viết SQL.

### View

View là một hàm (hoặc class) nhận request và trả về response — có thể là HTML, chuyển hướng, lỗi 404…

### Template

Template là file HTML có thêm **thẻ Django** như `{{ biến }}` và `{% for %}` để hiển thị dữ liệu động.

## Django History — Lịch sử Django

- Django được phát triển từ năm **2003** bởi **Adrian Holovaty** và **Simon Willison** tại một toà soạn báo.
- Phát hành công khai năm **2005**.
- Do **Django Software Foundation (DSF)** duy trì.
- Nhiều website lớn đã/đang dùng Django: Instagram, Pinterest, Mozilla, Disqus…

## 🧠 Ghi nhớ nhanh

> - Django dùng mô hình **MVT**: **Model** (dữ liệu) · **View** (xử lý request) · **Template** (giao diện HTML).
> - `urls.py` ánh xạ **URL → view**.
> - ORM: làm việc với cơ sở dữ liệu bằng class Python thay vì SQL.
