## Install Django — Cài Django

Khi môi trường ảo đã được kích hoạt, cài Django bằng PIP:

```bash
python -m pip install Django
```

Trong notebook:

````python
%pip install Django
````

Kết quả mẫu:

```text
Collecting Django
  Downloading Django-5.x-py3-none-any.whl
...
Successfully installed Django-5.x asgiref-3.x sqlparse-0.5.x tzdata-2024.x
```

> 📌 Từ Django **6.0**, cần **Python 3.12** trở lên. Với Python 3.10/3.11, PIP tự chọn Django 5.2 (bản LTS).

## Check Django Version — Kiểm tra phiên bản

```bash
django-admin --version
```

````python
!python -m django --version
````

## Windows, Mac, or Unix?

Django chạy trên mọi hệ điều hành. Khác biệt duy nhất là cách gọi lệnh: `py` / `python` trên Windows, `python3` trên macOS/Linux.

> - Cài: `python -m pip install Django` (trong môi trường ảo).
> - Kiểm tra: `django-admin --version` hoặc `python -m django --version`.
