## Include — Chèn template

Thẻ `{% include %}` cho phép **chèn một template** vào bên trong template hiện tại.

Hữu ích khi có những phần dùng chung cho nhiều trang: **header, menu, footer**…

Tạo template `footer.html`:

````python
%%writefile my_tennis_club/members/templates/footer.html
<p>You have reached the bottom of this page, thank you for your time.</p>
````

Chèn footer vào `template.html`:

```text
<h1>Hello</h1>

<p>This page contains a footer in a template.</p>

{% include 'footer.html' %}
```

## Variables in Include — Truyền biến khi include

Có thể truyền **biến** vào template được chèn bằng từ khoá `with`:

````python
%%writefile my_tennis_club/members/templates/mymenu.html
<div>HOME | {{ me }} | ABOUT | FORUM | {{ sponsor }}</div>
````

````python
%%writefile my_tennis_club/members/templates/template.html
<!DOCTYPE html>
<html>
<body>

{% include "mymenu.html" with me="AN" sponsor="PYTHON VIỆT" %}

<h1>Welcome</h1>

<p>This is my webpage</p>

{% include 'footer.html' %}

</body>
</html>
````

> 💡 Khác biệt: `{% extends %}` là **kế thừa bố cục** (template con điền vào template cha), còn `{% include %}` là **chèn một mảnh** template vào vị trí bất kỳ.

---

## 🧠 Ghi nhớ nhanh

> - `{% include "file.html" %}` → chèn template con (header, menu, footer…).
> - Truyền biến: `{% include "mymenu.html" with me="AN" sponsor="..." %}`.
> - `extends` = kế thừa bố cục · `include` = chèn mảnh.
