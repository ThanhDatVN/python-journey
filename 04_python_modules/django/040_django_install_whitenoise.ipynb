## WhiteNoise

Django **không** có giải pháp dựng sẵn để phục vụ file tĩnh khi `DEBUG = False` (môi trường thật). Một cách đơn giản và phổ biến là dùng thư viện **WhiteNoise**.

## Install WhiteNoise — Cài WhiteNoise

```bash
pip install whitenoise
```

````python
%pip install whitenoise
````

> 📌 WhiteNoise đã có trong `requirements.txt` của repo.

## Modify Settings — Sửa cấu hình

Để Django dùng WhiteNoise, thêm `WhiteNoiseMiddleware` vào danh sách `MIDDLEWARE` trong `settings.py`, **ngay sau** `SecurityMiddleware`:

```text
MIDDLEWARE = [
    'django.middleware.security.SecurityMiddleware',
    'whitenoise.middleware.WhiteNoiseMiddleware',
    ...
]
```

````python
from django_helper import edit_file, show_file

edit_file("my_tennis_club/my_tennis_club/settings.py",
          "    'django.middleware.security.SecurityMiddleware',\n",
          "    'django.middleware.security.SecurityMiddleware',\n"
          "    'whitenoise.middleware.WhiteNoiseMiddleware',\n")

show_file("my_tennis_club/my_tennis_club/settings.py", "MIDDLEWARE", "]")
````

## Handle Static Files — Xử lý file tĩnh

File tĩnh trong dự án (như stylesheet) **không** được phục vụ tự động khi `DEBUG = False` — chúng phải được **thu gom** về một thư mục trước. Bài tiếp theo sẽ hướng dẫn cách làm.

---

## 🧠 Ghi nhớ nhanh

> - WhiteNoise phục vụ file tĩnh khi `DEBUG = False`: `pip install whitenoise`.
> - Thêm `'whitenoise.middleware.WhiteNoiseMiddleware'` **ngay sau** `SecurityMiddleware`.
