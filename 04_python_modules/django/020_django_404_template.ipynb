# Django 404 (page not found)

> 📖 **Học song song trên W3Schools:** [django_404.php](https://www.w3schools.com/django/django_404.php)
>
> 🎯 **Trong bài này:** Page Not Found · Change Settings · Add a 404 Template · 👀 Xem kết quả

## Page Not Found — Không tìm thấy trang

Khi truy cập một trang **không tồn tại**, Django trả về lỗi **404**.

Khi `DEBUG = True` (đang phát triển), Django hiện **trang lỗi chi tiết** "Page not found at /masfdfg/" kèm danh sách mọi URL của dự án — hữu ích cho lập trình viên, nhưng **không** nên để người dùng thấy.

## Change Settings — Sửa cấu hình

Để hiện trang 404 riêng, đặt `DEBUG = False` trong `settings.py`. Khi đó **bắt buộc** khai báo `ALLOWED_HOSTS` — danh sách tên miền được phép truy cập:

```text
DEBUG = False

ALLOWED_HOSTS = ['*']
```

````python
from django_helper import edit_file

settings = "my_tennis_club/my_tennis_club/settings.py"
edit_file(settings, "DEBUG = True", "DEBUG = False")
edit_file(settings, "ALLOWED_HOSTS = []", "ALLOWED_HOSTS = ['*']")
````

> ⚠️ `ALLOWED_HOSTS = ['*']` cho phép **mọi** tên miền — chỉ dùng khi học/thử nghiệm. Khi triển khai thật, hãy ghi rõ tên miền, ví dụ `['mytennisclub.com']`.

## Add a 404 Template — Thêm template 404

Tạo file `404.html` trong thư mục `templates` — Django tự dùng nó khi gặp lỗi 404 (với `DEBUG = False`):

````python
%%writefile my_tennis_club/members/templates/404.html
<!DOCTYPE html>
<html>
<title>Wrong address</title>
<body>

<h1>Ooops!</h1>

<h2>I cannot find the file you requested!</h2>

</body>
</html>
````

## 👀 Xem kết quả

````python
from django_helper import preview

preview("/masfdfg/")
````

> 📌 Django đọc `settings.py` **một lần** khi khởi động. Nếu trước đó bạn đã chạy `preview()` trong cùng kernel, hãy **Restart Kernel** rồi chạy lại cell này để thấy trang 404 mới. (Với `runserver`, server tự khởi động lại khi file thay đổi.)

## 🧠 Ghi nhớ nhanh

> - `DEBUG = True` → trang lỗi chi tiết (chỉ dùng khi phát triển).
> - `DEBUG = False` → bắt buộc có `ALLOWED_HOSTS`; Django dùng `templates/404.html` cho lỗi 404.
> - ⚠️ Không bao giờ triển khai thật với `DEBUG = True`.
