# Django QuerySet - Get Data

>
> 🎯 **Trong bài này:** Get Data · The values() Method · Return Specific Columns · Return Specific Rows · ➕ Mở rộng: get(), first(), count()

## Get Data — Lấy dữ liệu

Có nhiều cách lấy dữ liệu từ model vào QuerySet.

## The values() Method — Phương thức values()

`values()` trả về mỗi object dưới dạng **dictionary** Python, với tên trường và giá trị là cặp key/value:

```python
from django_helper import setup

setup()
from members.models import Member

mydata = Member.objects.all().values()
for x in mydata:
    print(x)
```

## Return Specific Columns — Chỉ lấy một số cột

`values_list()` trả về **chỉ các cột** chỉ định:

```python
mydata = Member.objects.values_list('firstname')
print(mydata)
```

## Return Specific Rows — Chỉ lấy một số hàng

Lọc kết quả bằng `filter()` (bài tiếp theo):

```python
mydata = Member.objects.filter(firstname='Chi').values()
print(mydata)
```

## ➕ Mở rộng: get(), first(), count()

```python
print(Member.objects.get(id=1))               # đúng một object (lỗi nếu 0 hoặc >1 kết quả)
print(Member.objects.first(), "|", Member.objects.last())
print(Member.objects.count(), "thành viên")
print(Member.objects.values_list('firstname', flat=True))
```

---

## 🧠 Ghi nhớ nhanh

> - `.values()` → list các **dict** · `.values_list('cột')` → chỉ các cột chỉ định (`flat=True` → list giá trị).
> - `.filter(...)` → chỉ các hàng thoả điều kiện · `.get(...)` → đúng **một** object.
