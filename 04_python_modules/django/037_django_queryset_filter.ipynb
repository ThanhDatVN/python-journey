# Django QuerySet - Filter

> 📖 **Học song song trên W3Schools:** [django_queryset_filter.php](https://www.w3schools.com/django/django_queryset_filter.php)
>
> 🎯 **Trong bài này:** QuerySet Filter · AND · OR · Field Lookups

## QuerySet Filter — Lọc QuerySet

`filter()` dùng để lọc kết quả — giống mệnh đề `WHERE` trong SQL:

````python
from django_helper import setup

setup()
from members.models import Member

mydata = Member.objects.filter(firstname='An').values()
print(mydata)
````

Câu SQL tương ứng:

```text
SELECT * FROM members_member WHERE firstname = 'An';
```

## AND

`filter()` nhận nhiều đối số ngăn cách bằng dấu phẩy → điều kiện **AND**:

````python
mydata = Member.objects.filter(lastname='Nguyễn', id=1).values()
print(mydata)
````

## OR

Lọc với điều kiện **OR** khó hơn một chút. Cách thứ nhất: dùng **nhiều** `filter()` nối bằng `|`:

````python
mydata = Member.objects.filter(firstname='An').values() | Member.objects.filter(firstname='Chi').values()
print(mydata)
````

Cách thứ hai — phổ biến hơn — import và dùng **`Q`**:

````python
from django.db.models import Q

mydata = Member.objects.filter(Q(firstname='An') | Q(firstname='Chi')).values()
print(mydata)
````

## Field Lookups — Tra cứu theo trường

Django có cú pháp đặc biệt cho điều kiện phức tạp: **tên trường** + **hai dấu gạch dưới** + **từ khoá**.

Ví dụ lấy các thành viên có `firstname` bắt đầu bằng "D":

````python
mydata = Member.objects.filter(firstname__startswith='D').values()
print(mydata)
````

```text
SELECT * FROM members_member WHERE firstname LIKE 'D%';
```

### Field Lookups Reference — Một số từ khoá

| Từ khoá | Mô tả |
|---|---|
| `contains` / `icontains` | Chứa chuỗi (có/không phân biệt hoa thường) |
| `exact` / `iexact` | Bằng chính xác |
| `startswith` / `endswith` | Bắt đầu / kết thúc bằng |
| `gt` / `gte` / `lt` / `lte` | Lớn hơn / ≥ / nhỏ hơn / ≤ |
| `in` | Nằm trong danh sách |
| `range` | Nằm trong khoảng |
| `isnull` | Là NULL |
| `date`, `year`, `month`, `day`… | So sánh theo phần của ngày |
| `regex` / `iregex` | Biểu thức chính quy |

````python
print(Member.objects.filter(id__in=[1, 3, 5]).values_list("firstname", flat=True))
print(Member.objects.filter(joined_date__year__gte=2023).values_list("firstname", flat=True))
print(Member.objects.filter(lastname__icontains="n").values_list("firstname", "lastname"))
````

Xem đầy đủ: [Django Field Lookups Reference](https://www.w3schools.com/django/django_ref_field_lookups.php).

## 🧠 Ghi nhớ nhanh

> - `filter(a=1, b=2)` → **AND** · `Q(a=1) | Q(b=2)` → **OR**.
> - Field lookup: `field__keyword=value` — `startswith`, `contains`, `gt`, `in`, `range`, `year`…
> - `exclude(...)` → loại bỏ các bản ghi thoả điều kiện.
