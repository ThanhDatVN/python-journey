## URLs — Đường dẫn

Tạo file `urls.py` trong thư mục `members` với nội dung:

````python
%%writefile my_tennis_club/members/urls.py
from django.urls import path
from . import views

urlpatterns = [
    path('members/', views.members, name='members'),
]
````

File `urls.py` vừa tạo là **riêng** của app `members`. Cần thêm một bước nữa: **khai báo** nó trong file `urls.py` **gốc** của dự án (`my_tennis_club/my_tennis_club/urls.py`).

File gốc hiện tại:

````python
from django_helper import show_file

show_file("my_tennis_club/my_tennis_club/urls.py", "from django.contrib", "]")
````

Thêm `include` và đường dẫn tới `members.urls`:

````python
%%writefile my_tennis_club/my_tennis_club/urls.py
from django.contrib import admin
from django.urls import include, path

urlpatterns = [
    path('', include('members.urls')),
    path('admin/', admin.site.urls),
]
````

Chạy server và mở `127.0.0.1:8000/members/` trên trình duyệt:

```bash
python manage.py runserver
```

## 🧠 Ghi nhớ nhanh

> - App có `urls.py` riêng: `path('members/', views.members, name='members')`.
> - Khai báo trong `urls.py` gốc: `path('', include('members.urls'))`.
> - URL `/members/` → view `members` → "Hello world!".
