## Delete Records — Xoá bản ghi

Để xoá một bản ghi, trước hết lấy bản ghi đó.

Thêm một bản ghi "thử nghiệm" để xoá:

````python
from django_helper import setup

setup()
from members.models import Member

Member(firstname="Test", lastname="Delete").save()
Member.objects.all().values()
````

Lấy bản ghi cuối cùng (vừa thêm):

````python
x = Member.objects.last()
x.firstname
````

Xoá bản ghi:

````python
x.delete()
````

Kết quả cho biết số bản ghi đã bị xoá và loại model:

```text
(1, {'members.Member': 1})
```

Kiểm tra bảng `Member`:

````python
Member.objects.all().values()
````

## ➕ Mở rộng: Xoá nhiều bản ghi

````python
Member(firstname="Tạm 1", lastname="X").save()
Member(firstname="Tạm 2", lastname="X").save()

print(Member.objects.filter(lastname="X").delete())
print(Member.objects.count(), "thành viên")
````

> ⚠️ `Member.objects.all().delete()` xoá **toàn bộ** bảng — hãy cẩn thận!

---

## 🧠 Ghi nhớ nhanh

> - Xoá một bản ghi: lấy bản ghi → `.delete()`.
> - Xoá nhiều: `Member.objects.filter(...).delete()`.
> - `.delete()` trả về `(số_bản_ghi_xoá, {model: số})`.
