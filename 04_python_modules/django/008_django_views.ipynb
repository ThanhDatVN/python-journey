## Views — View

**View** trong Django là **hàm Python** nhận **HTTP request** và trả về **HTTP response** — ví dụ một trang HTML.

Một trang web dùng Django thường có nhiều view với các nhiệm vụ khác nhau.

View thường được đặt trong file `views.py` của app.

Trong thư mục `members`, file `views.py` hiện trông như sau:

```python
from django_helper import show_file

show_file("my_tennis_club/members/views.py")
```

Thay nội dung bằng view đầu tiên — trả về dòng chữ "Hello world!":

```python
%%writefile my_tennis_club/members/views.py
from django.http import HttpResponse

def members(request):
    return HttpResponse("Hello world!")
```

Đây là một ví dụ đơn giản về cách gửi phản hồi về trình duyệt.

**Nhưng làm sao để chạy view?** Phải gọi view qua một **URL** — bạn sẽ học ở bài tiếp theo.

---

## 🧠 Ghi nhớ nhanh

> - View = hàm nhận `request`, trả về `HttpResponse`.
> - Đặt trong `members/views.py`; muốn chạy view phải **gắn với URL**.
