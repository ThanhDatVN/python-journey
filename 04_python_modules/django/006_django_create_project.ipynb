## My First Project — Dự án đầu tiên

Đặt tên dự án là **`my_tennis_club`**.

Trong terminal (môi trường ảo đã kích hoạt), dùng lệnh:

```bash
django-admin startproject my_tennis_club
```

Trong notebook, dùng lệnh tương đương:

````python
!python -m django startproject my_tennis_club
````

> ⚠️ Nếu chạy lại sẽ báo lỗi "already exists" — dự án chỉ cần tạo **một lần**.

Django tạo thư mục `my_tennis_club` với cấu trúc:

````python
from pathlib import Path

for p in sorted(Path("my_tennis_club").rglob("*")):
    print("    " * (len(p.parts) - 2) + p.name + ("/" if p.is_dir() else ""))
````

```text
my_tennis_club
    manage.py
    my_tennis_club/
        __init__.py
        asgi.py
        settings.py
        urls.py
        wsgi.py
```

Đây là cấu trúc của **project**. `settings.py` chứa cấu hình, `urls.py` chứa các đường dẫn, `manage.py` là công cụ dòng lệnh để quản lý dự án.

## Run the Django Project — Chạy dự án

Trong terminal, di chuyển vào thư mục dự án và chạy server phát triển:

```bash
cd my_tennis_club
python manage.py runserver
```

````python
!cd my_tennis_club && python manage.py runserver
````

> ⚠️ `runserver` chạy **mãi** cho tới khi bạn dừng nó (`Ctrl + C`, hoặc **Kernel → Interrupt** trong Jupyter). Nên chạy nó trong **terminal** riêng.

Kết quả mẫu:

```text
Watching for file changes with StatReloader
Performing system checks...

System check identified no issues (0 silenced).
Django version 5.2, using settings 'my_tennis_club.settings'
Starting development server at http://127.0.0.1:8000/
Quit the server with CTRL-BREAK.
```

Mở trình duyệt, gõ `127.0.0.1:8000` — bạn sẽ thấy trang chào mừng của Django 🚀.
