## Delete Members — Xoá thành viên

Xoá thành viên qua trang admin:

1. Đăng nhập, vào **Members**.
2. Bấm vào thành viên cần xoá.
3. Bấm nút **DELETE** (góc trái phía dưới).
4. Xác nhận **Yes, I'm sure**.

> 💡 Muốn xoá **nhiều** thành viên cùng lúc: tích chọn ô vuông ở đầu mỗi dòng → chọn **Action: Delete selected members** → **Go**.

## Thử nghiệm bằng code

Trang xác nhận xoá nằm ở `/admin/members/member/<id>/delete/`; gửi POST với `post=yes` để xác nhận:

```python
from django_helper import setup

setup()
from django.conf import settings
from django.contrib.auth.models import User
from django.test import Client
from members.models import Member

settings.ALLOWED_HOSTS = ["*"]
client = Client()
client.force_login(User.objects.get(username="admin"))

temp = Member.objects.create(firstname="Xoá", lastname="Thử")
print("Trước:", Member.objects.count(), "thành viên")

response = client.post(f"/admin/members/member/{temp.id}/delete/", {"post": "yes"})
print("Status:", response.status_code)
print("Sau:", Member.objects.count(), "thành viên")
```

---

## 🧠 Ghi nhớ nhanh

> - Xoá một: mở thành viên → **DELETE** → xác nhận.
> - Xoá nhiều: tích chọn → Action **Delete selected** → **Go**.
