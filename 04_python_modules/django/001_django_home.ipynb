## Learn Django — Học Django

**Django** là một **web framework phía server (back-end)** viết bằng Python.

Django giúp xây dựng website nhanh chóng: nó lo sẵn những phần phức tạp như cơ sở dữ liệu, xác thực người dùng, trang quản trị, bảo mật… để bạn tập trung vào ứng dụng của mình.

## Learning by Doing — Học qua thực hành

Trong chương này, bạn sẽ được hướng dẫn **từng bước** cài đặt Django và xây dựng một dự án hoàn chỉnh: website quản lý thành viên của một câu lạc bộ tennis — **`my_tennis_club`**.

```text
my_tennis_club/            ← dự án (project)
├── manage.py
├── my_tennis_club/        ← cấu hình: settings.py, urls.py...
└── members/               ← ứng dụng (app) quản lý thành viên
    ├── models.py · views.py · urls.py · admin.py
    ├── templates/         ← các trang HTML
    └── static/            ← CSS, ảnh...
```

## Cách học chương Django trong notebook

Trên W3Schools, các lệnh được gõ trong **terminal**. Trong khóa học này, bạn có thể làm **theo cả hai cách**:

| Việc cần làm | Trong terminal (như W3Schools) | Trong notebook (khóa học) |
|---|---|---|
| Chạy lệnh | `python manage.py migrate` | `!cd my_tennis_club && python manage.py migrate` |
| Tạo/sửa file | Dùng trình soạn thảo (VS Code…) | Cell `%%writefile đường/dẫn/file.py` |
| Xem trang web | `python manage.py runserver` → mở trình duyệt | `preview("/members/")` — in HTML ngay dưới cell |
