# Django QuerySet - Order By

> 📖 **Học song song trên W3Schools:** [django_queryset_orderby.php](https://www.w3schools.com/django/django_queryset_orderby.php)
>
> 🎯 **Trong bài này:** Order By · Descending Order · Multiple Order Bys · ➕ Mở rộng: Kết hợp filter + order_by + giới hạn

## Order By — Sắp xếp

Dùng phương thức `order_by()` để sắp xếp QuerySet.

Sắp xếp theo `firstname` (theo thứ tự chữ cái):

````python
from django_helper import setup

setup()
from members.models import Member

mydata = Member.objects.all().order_by('firstname').values()
print([x['firstname'] for x in mydata])
````

```text
SELECT * FROM members_member ORDER BY firstname;
```

## Descending Order — Giảm dần

Mặc định là tăng dần. Thêm dấu **trừ** `-` trước tên trường để sắp **giảm dần**:

````python
mydata = Member.objects.all().order_by('-firstname').values()
print([x['firstname'] for x in mydata])
````

```text
SELECT * FROM members_member ORDER BY firstname DESC;
```

## Multiple Order Bys — Sắp xếp theo nhiều trường

Sắp xếp theo `lastname` tăng dần, rồi theo `id` giảm dần:

````python
mydata = Member.objects.all().order_by('lastname', '-id').values()
print([(x['lastname'], x['id']) for x in mydata])
````

```text
SELECT * FROM members_member ORDER BY lastname ASC, id DESC;
```

## ➕ Mở rộng: Kết hợp filter + order_by + giới hạn

````python
newest = Member.objects.filter(joined_date__isnull=False).order_by('-joined_date')[:3]
for m in newest:
    print(m, m.joined_date)
````

---

## 🧠 Ghi nhớ nhanh

> - `order_by('field')` tăng dần · `order_by('-field')` giảm dần.
> - Nhiều trường: `order_by('lastname', '-id')`.
> - Kết hợp: `filter(...).order_by(...)[:n]` (cắt = `LIMIT`).
