## Bootstrap 5

**Bootstrap** là framework CSS phổ biến giúp làm giao diện **đẹp và responsive** (hiển thị tốt trên điện thoại) nhanh chóng.

Có hai cách thêm Bootstrap 5 vào Django.

## Cách 1 (W3Schools): Package django-bootstrap-v5

```bash
pip install django-bootstrap-v5
```

Thêm vào `INSTALLED_APPS`:

```text
INSTALLED_APPS = [
    ...
    'members',
    'bootstrap5',
]
```

Trong template:

```text
{% load bootstrap5 %}
{% bootstrap_css %}
{% bootstrap_javascript %}
```

## Cách 2: Dùng CDN — không cần cài gì

Chỉ cần thêm thẻ `<link>` và `<script>` từ CDN vào `master.html`. Ta áp dụng cách này cho dự án:

````python
%%writefile my_tennis_club/members/templates/master.html
{% load static %}
<!DOCTYPE html>
<html>
<head>
  <meta name="viewport" content="width=device-width, initial-scale=1">
  <title>{% block title %}{% endblock %}</title>
  <link href="https://cdn.jsdelivr.net/npm/bootstrap@5.3.3/dist/css/bootstrap.min.css" rel="stylesheet">
  <link rel="stylesheet" href="{% static 'mystyles.css' %}">
</head>
<body>

<nav class="navbar navbar-expand navbar-dark bg-success px-3">
  <a class="navbar-brand" href="/">My Tennis Club</a>
  <div class="navbar-nav">
    <a class="nav-link" href="/members">Members</a>
  </div>
</nav>

<div class="container mt-4">
  {% block content %}
  {% endblock %}
</div>

<script src="https://cdn.jsdelivr.net/npm/bootstrap@5.3.3/dist/js/bootstrap.bundle.min.js"></script>
</body>
</html>
````

Dùng các class của Bootstrap trong trang danh sách thành viên:

````python
%%writefile my_tennis_club/members/templates/all_members.html
{% extends "master.html" %}

{% block title %}
  My Tennis Club - List of all members
{% endblock %}


{% block content %}
  <h1 class="mb-3">Members</h1>
  <div class="list-group">
    {% for x in mymembers %}
      <a class="list-group-item list-group-item-action" href="details/{{ x.slug }}">
        {{ x.firstname }} {{ x.lastname }}
      </a>
    {% endfor %}
  </div>
{% endblock %}
````
