## Create Template — Tạo template

Sau khi thêm dữ liệu vào cơ sở dữ liệu, ta muốn hiển thị dữ liệu lên **trang web** bằng một template.

Tạo template `all_members.html` trong thư mục `templates`:

```python
%%writefile my_tennis_club/members/templates/all_members.html
<!DOCTYPE html>
<html>
<body>

<h1>Members</h1>

<ul>
  {% for x in mymembers %}
    <li>{{ x.firstname }} {{ x.lastname }}</li>
  {% endfor %}
</ul>

</body>
</html>
```

> 📌 Phần trong `{% %}` là **thẻ template (template tag)** — ở đây là vòng lặp `for`. Phần trong `{{ }}` là **biến** — giá trị sẽ được chèn vào HTML. Bạn sẽ học kỹ ở chương [Django Syntax](029_django_variables.ipynb).

## Modify View — Sửa view

Sửa view `members` để lấy dữ liệu từ model và gửi cho template:

```python
%%writefile my_tennis_club/members/views.py
from django.http import HttpResponse
from django.template import loader
from .models import Member

def members(request):
    mymembers = Member.objects.all().values()
    template = loader.get_template('all_members.html')
    context = {
        'mymembers': mymembers,
    }
    return HttpResponse(template.render(context, request))
```

## Example Explained — Giải thích

View `members` làm các việc:

1. Tạo object `mymembers` chứa **mọi** giá trị của model `Member`.
2. Nạp template `all_members.html`.
3. Tạo **context** — một dictionary chứa `mymembers` để gửi cho template.
4. Gửi context cho template và trả về HTML cho trình duyệt.

## 🧠 Ghi nhớ nhanh

> - View lấy dữ liệu: `Member.objects.all().values()` → đưa vào **context** dict.
> - `template.render(context, request)` → template dùng `{% for x in mymembers %}` và `{{ x.firstname }}`.
> - `{% ... %}` = thẻ (logic) · `{{ ... }}` = biến (giá trị).
