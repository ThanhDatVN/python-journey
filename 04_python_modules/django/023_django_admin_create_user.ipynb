## Create User — Tạo tài khoản quản trị

Để đăng nhập trang admin, cần tạo một **superuser** (tài khoản quản trị cao nhất).

Trong terminal, chạy lệnh và nhập thông tin theo hướng dẫn:

```bash
python manage.py createsuperuser
```

```text
Username: admin
Email address: admin@example.com
Password:
Password (again):
Superuser created successfully.
```

> 📌 Mật khẩu **không hiện** khi gõ — đó là cơ chế bảo mật bình thường.

Lệnh này **hỏi đáp tương tác** nên không chạy được trực tiếp trong notebook. Thay vào đó, dùng chế độ **không tương tác** với mật khẩu lấy từ biến môi trường:

````python
import os

os.environ["DJANGO_SUPERUSER_PASSWORD"] = "Python@2026"   # ⚠️ chỉ dùng khi học!
````

````python
!cd my_tennis_club && python manage.py createsuperuser --noinput --username admin --email admin@example.com
````

Kiểm tra tài khoản đã được tạo:

````python
from django_helper import setup

setup()
from django.contrib.auth.models import User

print(User.objects.filter(is_superuser=True).values_list("username", "email"))
````

Giờ có thể mở `127.0.0.1:8000/admin/`, đăng nhập bằng `admin` / mật khẩu vừa đặt.

Sau khi đăng nhập, bạn thấy trang quản trị với **Users** và **Groups** — nhưng **chưa có** model `Member`. Bài tiếp theo sẽ thêm nó vào.

## 🧠 Ghi nhớ nhanh

> - Tạo superuser: `python manage.py createsuperuser` (hỏi đáp tương tác).
> - Không tương tác: đặt `DJANGO_SUPERUSER_PASSWORD` rồi `createsuperuser --noinput --username ... --email ...`.
> - Đăng nhập tại `/admin/`.
