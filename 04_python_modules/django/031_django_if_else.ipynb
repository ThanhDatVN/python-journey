## If Statement — Câu lệnh if

Câu lệnh `if` đánh giá một biến và thực thi một khối code nếu giá trị là `True`:

```text
{% if greeting == 1 %}
  <h1>Hello</h1>
{% endif %}
```

## Elif

`elif` = "nếu điều kiện trước sai, thử điều kiện này":

```text
{% if greeting == 1 %}
  <h1>Hello</h1>
{% elif greeting == 2 %}
  <h1>Welcome</h1>
{% endif %}
```

## Else

`else` thực thi khi mọi điều kiện trước đều sai:

```text
{% if greeting == 1 %}
  <h1>Hello</h1>
{% elif greeting == 2 %}
  <h1>Welcome</h1>
{% else %}
  <h1>Goodbye</h1>
{% endif %}
```

## Operators — Toán tử

Các ví dụ trên dùng toán tử `==` (so sánh bằng). Các toán tử khác có thể dùng trong `{% if %}`:

| Toán tử | Ví dụ |
|---|---|
| `==` | `{% if greeting == 2 %}` |
| `!=` | `{% if greeting != 1 %}` |
| `<` / `<=` | `{% if greeting < 3 %}` |
| `>` / `>=` | `{% if greeting > 1 %}` |
| `and` | `{% if greeting == 1 and day == "Friday" %}` |
| `or` | `{% if greeting == 1 or greeting == 5 %}` |
| `and` / `or` | `{% if greeting == 1 and day == "Friday" or greeting == 5 %}` |
| `in` | `{% if 'Banana' in fruits %}` |
| `not in` | `{% if 'Banana' not in fruits %}` |
| `is` | `{% if x is y %}` |
| `is not` | `{% if x is not y %}` |

> 📌 Không được dùng dấu ngoặc `()` trong `{% if %}` để nhóm điều kiện. Khi kết hợp `and` và `or`, `and` được ưu tiên trước.

## 👀 Thử nghiệm

````python
from django_helper import replace_view, preview

replace_view("testing", '''
def testing(request):
    template = loader.get_template('template.html')
    context = {
        'greeting': 2,
        'day': 'Friday',
        'fruits': ['Apple', 'Banana', 'Cherry'],
    }
    return HttpResponse(template.render(context, request))
''')
````

````python
%%writefile my_tennis_club/members/templates/template.html
<!DOCTYPE html>
<html>
<body>

{% if greeting == 1 %}
  <h1>Hello</h1>
{% elif greeting == 2 %}
  <h1>Welcome</h1>
{% else %}
  <h1>Goodbye</h1>
{% endif %}

{% if greeting > 1 and day == "Friday" %}
  <p>Have a nice weekend!</p>
{% endif %}

{% if 'Banana' in fruits %}
  <p>Yes, we have bananas.</p>
{% endif %}

</body>
</html>
````

````python
preview("/testing/")
````

## 🧠 Ghi nhớ nhanh

> - `{% if %}` · `{% elif %}` · `{% else %}` · `{% endif %}`.
> - Toán tử: `== != < > <= >= and or in not in is is not`.
> - ⚠️ Không dùng `()` để nhóm điều kiện trong `{% if %}`.
