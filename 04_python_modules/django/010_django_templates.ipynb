## Templates

Ở bài [Views](008_django_views.ipynb), kết quả chỉ là dòng chữ "Hello world!". Thông thường ta muốn trả về **trang HTML** — đó là việc của **template**.

Tạo thư mục `templates` trong thư mục `members`, và tạo file `myfirst.html`:

````python
import os

os.makedirs("my_tennis_club/members/templates", exist_ok=True)
````

````python
%%writefile my_tennis_club/members/templates/myfirst.html
<!DOCTYPE html>
<html>
<body>

<h1>Hello World!</h1>
<p>Welcome to my first Django project!</p>

</body>
</html>
````

## Modify the View — Sửa view

Mở `views.py` và thay view `members` để dùng template:

````python
%%writefile my_tennis_club/members/views.py
from django.http import HttpResponse
from django.template import loader

def members(request):
    template = loader.get_template('myfirst.html')
    return HttpResponse(template.render())
````

## Change Settings — Sửa cấu hình

Để làm việc với những thứ phức tạp hơn "Hello World!", phải **báo cho Django** biết có app mới.

Việc này làm trong `settings.py`, danh sách `INSTALLED_APPS` — thêm `'members'` vào:

```text
INSTALLED_APPS = [
    'django.contrib.admin',
    ...
    'django.contrib.staticfiles',
    'members',
]
```

Mở file bằng trình soạn thảo và sửa, hoặc chạy cell sau:

````python
from django_helper import edit_file, show_file

edit_file("my_tennis_club/my_tennis_club/settings.py",
          "    'django.contrib.staticfiles',\n]",
          "    'django.contrib.staticfiles',\n    'members',\n]")

show_file("my_tennis_club/my_tennis_club/settings.py", "INSTALLED_APPS", "]")
````

Sau đó chạy lệnh `migrate`:

````python
!cd my_tennis_club && python manage.py migrate
````

> 📌 `migrate` tạo các bảng mặc định của Django (người dùng, phiên đăng nhập…) trong cơ sở dữ liệu SQLite `db.sqlite3`.
