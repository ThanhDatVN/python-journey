# Django comment Tag

> 📖 **Học song song trên W3Schools:** [django_tags_comment.php](https://www.w3schools.com/django/django_tags_comment.php)
>
> 🎯 **Trong bài này:** Comments · Comment Description · Smaller Comments · Comment in Views · 👀 Thử nghiệm

## Comments — Chú thích

**Chú thích** cho phép có những đoạn code mà Django **bỏ qua**.

Dùng thẻ `{% comment %}`...`{% endcomment %}`:

```text
<h1>Welcome Everyone</h1>
{% comment %}
  <h1>Welcome ladies and gentlemen</h1>
{% endcomment %}
```

## Comment Description — Mô tả cho chú thích

Có thể thêm **thông điệp** mô tả chú thích:

```text
<h1>Welcome Everyone</h1>
{% comment "this was the original welcome message" %}
    <h1>Welcome ladies and gentlemen</h1>
{% endcomment %}
```

## Smaller Comments — Chú thích ngắn

Dùng `{# ... #}` cho chú thích ngắn — có thể đặt **giữa một dòng** để vô hiệu hoá một phần:

```text
<h1>Welcome{# Everyone#}</h1>
```

## Comment in Views — Chú thích trong view

View viết bằng Python, nên dùng chú thích Python `#`:

```text
def testing(request):
    template = loader.get_template('template.html')
    #context = {
    #  'var1': 'John',
    #}
    return HttpResponse(template.render())
```

## 👀 Thử nghiệm

````python
%%writefile my_tennis_club/members/templates/template.html
<!DOCTYPE html>
<html>
<body>

<h1>Welcome Everyone</h1>
{% comment "this was the original welcome message" %}
    <h1>Welcome ladies and gentlemen</h1>
{% endcomment %}

<h2>Hello{# Everyone#}!</h2>

</body>
</html>
````

````python
from django_helper import preview

preview("/testing/")
````

> 📌 Chú thích Django **không xuất hiện** trong HTML gửi tới trình duyệt — khác với chú thích HTML `<!-- -->` (vẫn nằm trong mã nguồn trang).

---

## 🧠 Ghi nhớ nhanh

> - Nhiều dòng: `{% comment "mô tả" %}...{% endcomment %}` · một dòng: `{# ... #}`.
> - Chú thích Django **không** được gửi tới trình duyệt (khác `<!-- -->` của HTML).
> - Trong view (Python): dùng `#`.
