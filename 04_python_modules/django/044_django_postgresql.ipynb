## PostgreSQL

**PostgreSQL** là hệ quản trị cơ sở dữ liệu quan hệ mã nguồn mở rất mạnh, thường dùng khi triển khai Django thật (thay cho SQLite).

## Install psycopg2 — Cài driver

Django cần driver `psycopg2` để kết nối PostgreSQL:

```bash
pip install psycopg2-binary
```

## Create Database in the Cloud — Tạo database trên cloud

W3Schools dùng dịch vụ **Amazon RDS** (AWS):

1. Tạo tài khoản AWS (AWS Free Tier).
2. Vào **RDS** → **Create database** → chọn **PostgreSQL**, gói **Free tier**.
3. Đặt tên database, tên đăng nhập (`master username`) và mật khẩu.
4. Bật **Public access** (để máy bạn kết nối được khi học).
5. Đợi trạng thái **Available**, ghi lại **Endpoint** (địa chỉ host) và **Port** (5432).

> 💡 Có thể dùng PostgreSQL **cài trên máy** (postgresql.org) hoặc các dịch vụ miễn phí khác — cấu hình Django giống hệt nhau.

## Connect to Database — Kết nối

Sửa `DATABASES` trong `settings.py`:

```text
DATABASES = {
    'default': {
        'ENGINE': 'django.db.backends.postgresql',
        'NAME': 'postgres',
        'USER': 'masterusername',
        'PASSWORD': 'mat-khau-cua-ban',
        'HOST': 'my-tennis-db.abc123xyz.ap-southeast-1.rds.amazonaws.com',
        'PORT': '5432',
    }
}
```

> ⚠️ Đừng ghi mật khẩu thật vào code! Đọc từ biến môi trường:

````python
import os

DATABASES = {
    'default': {
        'ENGINE': 'django.db.backends.postgresql',
        'NAME': os.environ.get('DB_NAME', 'postgres'),
        'USER': os.environ['DB_USER'],
        'PASSWORD': os.environ['DB_PASSWORD'],
        'HOST': os.environ['DB_HOST'],
        'PORT': os.environ.get('DB_PORT', '5432'),
    }
}
````

Database mới **trống** — chạy `migrate` để tạo bảng:

```bash
python manage.py migrate
```

## Add Members — Thêm dữ liệu

Database PostgreSQL mới chưa có thành viên nào. Tạo lại superuser và thêm thành viên qua trang admin (hoặc `python manage.py shell`) như các bài trước:

```bash
python manage.py createsuperuser
```

## ➕ Mở rộng: Chuyển dữ liệu từ SQLite sang PostgreSQL

```bash
python manage.py dumpdata members > members.json      # khi còn dùng SQLite
# đổi DATABASES sang PostgreSQL, chạy migrate, rồi:
python manage.py loaddata members.json
```

---

## 🧠 Ghi nhớ nhanh

> - PostgreSQL cho môi trường thật; driver: `pip install psycopg2-binary`.
> - Cấu hình `DATABASES` với `ENGINE = 'django.db.backends.postgresql'`, NAME, USER, PASSWORD, HOST, PORT.
> - ⚠️ Mật khẩu đọc từ **biến môi trường**; database mới → `migrate` + tạo lại superuser.
