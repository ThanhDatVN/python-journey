## Handle Static Files — Thu gom file tĩnh

File tĩnh trong dự án **không** được phục vụ tự động khi `DEBUG = False`, mà phải được **thu gom** vào **một thư mục** riêng.

## Modify Settings — Sửa cấu hình

Khai báo thư mục chứa file tĩnh đã thu gom bằng `STATIC_ROOT`, và đặt lại `DEBUG = False`:

```text
STATIC_URL = 'static/'
STATIC_ROOT = BASE_DIR / 'productionfiles'
```

````python
from django_helper import edit_file

settings = "my_tennis_club/my_tennis_club/settings.py"
edit_file(settings, "STATIC_URL = 'static/'", "STATIC_URL = 'static/'\nSTATIC_ROOT = BASE_DIR / 'productionfiles'")
edit_file(settings, "DEBUG = True", "DEBUG = False")
````

## Collect Static Files — Thu gom

Chạy lệnh `collectstatic`:

```bash
python manage.py collectstatic
```

````python
!cd my_tennis_club && python manage.py collectstatic --noinput
````

> 📌 Trong terminal, lệnh hỏi xác nhận *yes/no*; tuỳ chọn `--noinput` bỏ qua câu hỏi.

Hàng trăm file tĩnh được sao chép — phần lớn là của **trang admin**. File `myfirst.css` của ta cũng nằm trong số đó:

````python
import os

print(os.path.exists("my_tennis_club/productionfiles/myfirst.css"))
print(os.listdir("my_tennis_club/productionfiles"))
print(sum(len(files) for _, _, files in os.walk("my_tennis_club/productionfiles")), "file đã được thu gom")
````

## 👀 Xem kết quả

Giờ, kể cả với `DEBUG = False`, file CSS vẫn được phục vụ (bởi WhiteNoise):

````python
from django_helper import preview

preview("/static/myfirst.css")
````

---

## 🧠 Ghi nhớ nhanh

> - `STATIC_ROOT = BASE_DIR / 'productionfiles'` → nơi chứa file tĩnh đã thu gom.
> - `python manage.py collectstatic` → sao chép mọi file tĩnh (của app, admin…) vào `STATIC_ROOT`.
> - Mỗi lần thêm/sửa file tĩnh → chạy lại `collectstatic`.
