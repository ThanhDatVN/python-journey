## The extends Tag — Thẻ extends

Ở các bài trước, ta tạo hai template giống nhau ở phần lớn cấu trúc HTML (`<!DOCTYPE>`, `<html>`, `<body>`…). Django có thẻ `{% extends %}` để dùng **template cha (master template)** chung cho mọi trang.

Tạo `master.html` với hai **block** — `title` và `content`:

```python
%%writefile my_tennis_club/members/templates/master.html
<!DOCTYPE html>
<html>
<head>
  <title>{% block title %}{% endblock %}</title>
</head>
<body>

{% block content %}
{% endblock %}

</body>
</html>
```

Template cha chứa **chỗ trống** (block) để template con **điền vào** những phần riêng của mình.

Sửa `all_members.html` để **kế thừa** template cha:

```python
%%writefile my_tennis_club/members/templates/all_members.html
{% extends "master.html" %}

{% block title %}
  My Tennis Club - List of all members
{% endblock %}


{% block content %}
  <h1>Members</h1>

  <ul>
    {% for x in mymembers %}
      <li><a href="details/{{ x.id }}">{{ x.firstname }} {{ x.lastname }}</a></li>
    {% endfor %}
  </ul>
{% endblock %}
```

Sửa `details.html` tương tự:

```python
%%writefile my_tennis_club/members/templates/details.html
{% extends "master.html" %}

{% block title %}
  Details about {{ mymember.firstname }} {{ mymember.lastname }}
{% endblock %}


{% block content %}
  <h1>{{ mymember.firstname }} {{ mymember.lastname }}</h1>

  <p>Phone {{ mymember.phone }}</p>
  <p>Member since: {{ mymember.joined_date }}</p>

  <p>Back to <a href="/members">Members</a></p>

{% endblock %}
```

## 🧠 Ghi nhớ nhanh

> - Template cha: định nghĩa `{% block tên %}{% endblock %}`.
> - Template con: `{% extends "master.html" %}` + điền nội dung vào các block.
> - Lợi ích: không lặp HTML chung, sửa một chỗ áp dụng mọi trang.
