## Add Members — Thêm thành viên

Thêm thành viên mới qua trang admin:

1. Đăng nhập `127.0.0.1:8000/admin/`.
2. Ở mục **Members**, bấm nút **ADD MEMBER +** (góc phải).
3. Điền thông tin.
4. Bấm **SAVE**.

## Thử nghiệm bằng code

Form "Add" của admin nằm ở `/admin/members/member/add/`:

```python
from django_helper import setup

setup()
from django.conf import settings
from django.contrib.auth.models import User
from django.test import Client
from members.models import Member

settings.ALLOWED_HOSTS = ["*"]
client = Client()
client.force_login(User.objects.get(username="admin"))

response = client.post("/admin/members/member/add/", {
    "firstname": "Hà",
    "lastname": "Đặng",
    "phone": 922334455,
    "joined_date": "2024-09-01",
})
print("Status:", response.status_code)
print(list(Member.objects.all()))
```

---

## 🧠 Ghi nhớ nhanh

> - Admin → Members → **ADD MEMBER +** → điền → **SAVE**.
> - Phía sau: POST tới `/admin/<app>/<model>/add/`.
