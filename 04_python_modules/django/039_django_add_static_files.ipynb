## Create Static Folder — Tạo thư mục static

Khi xây dựng web, bạn sẽ cần thêm **file tĩnh (static files)**: ảnh, file CSS, file JavaScript…

Tạo thư mục `static` trong app `members`, và file CSS `myfirst.css`:

```python
import os

os.makedirs("my_tennis_club/members/static", exist_ok=True)
```

```python
%%writefile my_tennis_club/members/static/myfirst.css
body {
  background-color: lightblue;
  font-family: verdana;
}
```

## Modify the Template — Sửa template

Mở template `template.html`, thêm dòng `{% load static %}` ở **đầu** file, và tham chiếu tới file CSS bằng `{% static 'myfirst.css' %}`:

```python
%%writefile my_tennis_club/members/templates/template.html
{% load static %}
<!DOCTYPE html>
<html>
<link rel="stylesheet" href="{% static 'myfirst.css' %}">
<body>

{% for x in fruits %}
  <h1>{{ x }}</h1>
{% endfor %}

</body>
</html>
```

```python
from django_helper import replace_view

replace_view("testing", '''
def testing(request):
    template = loader.get_template('template.html')
    context = {
        'fruits': ['Apple', 'Banana', 'Cherry'],
    }
    return HttpResponse(template.render(context, request))
''')
```

## Didn't Work? — Không có tác dụng?

Ở bài [404 Template](020_django_404_template.ipynb), ta đã đặt `DEBUG = False`. Khi đó `runserver` **không** phục vụ file tĩnh nữa — Django coi việc này là của web server khi triển khai thật.

Tạm thời bật lại `DEBUG = True` để thấy CSS hoạt động — bài tiếp theo sẽ giải quyết triệt để bằng **WhiteNoise**:

```python
from django_helper import edit_file

edit_file("my_tennis_club/my_tennis_club/settings.py", "DEBUG = False", "DEBUG = True")
```

> 📌 Thẻ `{% static 'myfirst.css' %}` được dịch thành đường dẫn `/static/myfirst.css`. Chạy `runserver` và mở `127.0.0.1:8000/testing/` để thấy nền xanh nhạt.

---

## 🧠 Ghi nhớ nhanh

> - File tĩnh (CSS, JS, ảnh) đặt trong `members/static/`.
> - Template: `{% load static %}` ở đầu file + `{% static 'myfirst.css' %}` → `/static/myfirst.css`.
> - `runserver` chỉ phục vụ file tĩnh khi `DEBUG = True`.
