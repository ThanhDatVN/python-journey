## Virtual Environment — Môi trường ảo

Nên có một **môi trường ảo riêng** cho mỗi dự án Django.

Một cách quản lý môi trường ảo là `venv` — có sẵn trong Python.

Tạo môi trường ảo tên `myworld` (trong terminal, tại thư mục muốn chứa dự án):

```bash
python -m venv myworld
```

Lệnh này tạo thư mục `myworld` với các thư mục con và file:

```text
myworld
  Include
  Lib
  Scripts
  .gitignore
  pyvenv.cfg
```

Kích hoạt môi trường:

Windows:

```bash
myworld\Scripts\activate.bat
```

macOS / Linux:

```bash
source myworld/bin/activate
```

Khi đã kích hoạt, dấu nhắc lệnh có thêm tên môi trường:

```text
(myworld) C:\Users\Your Name>
```

> 📌 Mỗi lần mở terminal mới để làm việc với dự án, cần **kích hoạt lại** môi trường ảo.

## Với repo này

Nếu bạn đã làm theo README của repo (tạo `.venv` và cài `requirements.txt`), Django đã được cài sẵn trong môi trường của notebook. Kiểm tra notebook đang dùng Python nào:

````python
import sys

print(sys.executable)
print("Đang trong môi trường ảo:", sys.prefix != sys.base_prefix)
````
