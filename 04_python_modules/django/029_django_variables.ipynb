## Variables — Biến trong template

Biến trong template Django được hiển thị bằng **hai cặp ngoặc nhọn** `{{ }}`:

```text
<h1>Hello {{ firstname }}, how are you?</h1>
```

## Create Variable in View — Tạo biến trong view

Biến `firstname` ở trên được gửi tới template từ **view**. Sửa view `testing`:

````python
from django_helper import replace_view

replace_view("testing", '''
def testing(request):
    template = loader.get_template('template.html')
    context = {
        'firstname': 'Linus',
    }
    return HttpResponse(template.render(context, request))
''')
````

Và template:

````python
%%writefile my_tennis_club/members/templates/template.html
<!DOCTYPE html>
<html>
<body>

<h1>Hello {{ firstname }}, how are you?</h1>

</body>
</html>
````

## Create Variables in Template — Tạo biến ngay trong template

Có thể tạo biến **trực tiếp trong template** bằng thẻ `{% with %}`. Biến chỉ tồn tại tới `{% endwith %}`:

```text
{% with firstname="Tobias" %}
<h1>Hello {{ firstname }}, how are you?</h1>
{% endwith %}
```

> Bạn sẽ học thêm về **thẻ (tag)** ở bài tiếp theo.

## Data From a Model — Dữ liệu từ model

Ví dụ trên chỉ minh hoạ cách tạo và dùng biến. Thực tế, phần lớn dữ liệu đến từ **model**. Ta đã làm điều đó ở bài [Prep Template and View](016_django_prep_template_and_view.ipynb): view lấy `Member.objects.all().values()` và template duyệt qua bằng `{% for %}`.
