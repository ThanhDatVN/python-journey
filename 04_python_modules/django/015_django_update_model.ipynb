## Add Fields in the Model — Thêm trường

Để thêm trường vào bảng **sau khi** bảng đã được tạo, sửa model rồi chạy migration.

Thêm hai trường `phone` và `joined_date` vào model `Member`:

```python
%%writefile my_tennis_club/members/models.py
from django.db import models

class Member(models.Model):
    firstname = models.CharField(max_length=255)
    lastname = models.CharField(max_length=255)
    phone = models.IntegerField(null=True)
    joined_date = models.DateField(null=True)
```

> 📌 Vì bảng **đã có dữ liệu**, các trường mới phải cho phép rỗng: `null=True`. Nếu không, Django sẽ hỏi bạn giá trị mặc định cho các bản ghi cũ.

Tạo migration:

```python
!cd my_tennis_club && python manage.py makemigrations members
```

Chạy migrate:

```python
!cd my_tennis_club && python manage.py migrate
```

## Insert Data — Thêm dữ liệu cho trường mới

Giờ có thể nhập giá trị cho các trường mới:

```python
from django_helper import setup

setup()
from members.models import Member

data = {
    "An": (912345678, "2022-01-05"),
    "Bình": (987654321, "2022-04-01"),
    "Chi": (933445566, "2022-12-24"),
    "Dũng": (977889900, "2023-05-15"),
    "Em": (966112233, "2023-08-20"),
    "Giang": (944556677, "2024-02-02"),
}

for m in Member.objects.all():
    if m.firstname in data:
        m.phone, m.joined_date = data[m.firstname]
        m.save()

Member.objects.all().values()
```

---

## 🧠 Ghi nhớ nhanh

> - Thêm trường mới vào model → `makemigrations` → `migrate`.
> - Bảng đã có dữ liệu → trường mới nên `null=True` (hoặc có `default=`).
> - `IntegerField` số nguyên · `DateField` ngày.
