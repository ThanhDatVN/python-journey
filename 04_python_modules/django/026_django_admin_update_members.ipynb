# Django Admin - Update Members

## Update Members — Sửa thành viên

Với trang admin, việc **sửa dữ liệu** chỉ là thao tác trên giao diện:

1. Đăng nhập `127.0.0.1:8000/admin/`.
2. Chọn **Members**.
3. Bấm vào **tên thành viên** cần sửa.
4. Sửa các trường (ví dụ số điện thoại, ngày tham gia).
5. Bấm **SAVE**.

Thay đổi được lưu vào cơ sở dữ liệu và hiện ngay trên website.

## Thử nghiệm bằng code: mô phỏng form admin

Trang admin thực chất gửi một **POST request** tới đường dẫn `/admin/members/member/<id>/change/`. Ta mô phỏng đăng nhập và gửi form bằng test client:

```python
from django_helper import setup

setup()
from django.conf import settings
from django.contrib.auth.models import User
from django.test import Client
from members.models import Member

settings.ALLOWED_HOSTS = ["*"]
client = Client()
client.force_login(User.objects.get(username="admin"))

member = Member.objects.get(firstname="An")
response = client.post(f"/admin/members/member/{member.id}/change/", {
    "firstname": member.firstname,
    "lastname": member.lastname,
    "phone": 911222333,
    "joined_date": "2022-01-05",
})
print("Status:", response.status_code, "→ chuyển hướng về danh sách nếu thành công")

member.refresh_from_db()
print(member, member.phone)
```

---

## 🧠 Ghi nhớ nhanh

> - Admin → Members → bấm tên → sửa → **SAVE**.
> - Phía sau: POST tới `/admin/<app>/<model>/<id>/change/`.
