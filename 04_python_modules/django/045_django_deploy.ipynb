## Deploy to the World — Đưa website lên Internet

**Deploy (triển khai)** là đưa dự án lên một **server thật** để mọi người truy cập được.

W3Schools dùng **AWS Elastic Beanstalk (EB)** — dịch vụ tự lo server, cân bằng tải, tự mở rộng…

## Lock in Dependencies — Tạo requirements.txt

Liệt kê các package mà dự án cần:

```bash
pip freeze > requirements.txt
```

Ví dụ nội dung:

```text
asgiref==3.8.1
Django==5.2
psycopg2-binary==2.9.10
sqlparse==0.5.3
tzdata==2025.2
whitenoise==6.9.0
```

## Provide django.config — File cấu hình cho EB

Tạo thư mục `.ebextensions` trong dự án, và file `django.config`:

```text
option_settings:
  aws:elasticbeanstalk:container:python:
    WSGIPath: my_tennis_club.wsgi:application
```

> 📌 `WSGIPath` chỉ tới đối tượng `application` trong file `my_tennis_club/wsgi.py` — điểm vào của ứng dụng Django.

## Create a .zip File — Đóng gói

Nén **toàn bộ** nội dung thư mục dự án (không gồm thư mục môi trường ảo) thành file `.zip`:

````python
import shutil

shutil.make_archive("my_tennis_club", "zip", root_dir="my_tennis_club")
````

## Deploy with EB — Triển khai

1. Đăng nhập AWS Console → **Elastic Beanstalk** → **Create application**.
2. Chọn **Platform: Python**.
3. **Upload your code** → chọn file `.zip`.
4. Đợi môi trường chuyển sang **Health: OK**, mở **Domain** được cấp.

## Update Project — Cập nhật dự án

Mỗi lần sửa code:

1. Sửa và kiểm tra trên máy.
2. `collectstatic` nếu có thay đổi file tĩnh.
3. Tạo file `.zip` mới (đặt tên phiên bản khác).
4. Elastic Beanstalk → **Upload and deploy** → chọn file mới.

## Checklist trước khi deploy

```text
☐ DEBUG = False
☐ ALLOWED_HOSTS chứa tên miền thật
☐ SECRET_KEY và mật khẩu đọc từ biến môi trường
☐ Database production (PostgreSQL) đã migrate
☐ Đã chạy collectstatic
☐ requirements.txt đầy đủ
```

> - Deploy: `requirements.txt` + `.ebextensions/django.config` (WSGIPath) + nén `.zip` → upload lên Elastic Beanstalk.
