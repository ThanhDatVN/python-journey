## The Project - My Tennis Club — Làm đẹp dự án

Giờ ta thêm **CSS** cho toàn bộ website My Tennis Club, để mọi trang có giao diện thống nhất.

## Create the Stylesheet — Tạo file CSS

Tạo `mystyles.css` trong thư mục `static` của app `members`:

```python
%%writefile my_tennis_club/members/static/mystyles.css
* {
  box-sizing: border-box;
}

body {
  font-family: Arial, Helvetica, sans-serif;
  margin: 0;
}

.topnav {
  background-color: #04AA6D;
  overflow: hidden;
}

.topnav a {
  float: left;
  color: #f2f2f2;
  text-align: center;
  padding: 14px 16px;
  text-decoration: none;
  font-size: 17px;
}

.topnav a:hover {
  background-color: #ddd;
  color: black;
}

.main {
  padding: 16px;
}

ul {
  list-style-type: none;
  padding: 0;
}

li {
  padding: 8px 0;
}
```

## Modify Master — Sửa template cha

Nạp file CSS trong `master.html` và thêm **thanh điều hướng**:

```python
%%writefile my_tennis_club/members/templates/master.html
{% load static %}
<!DOCTYPE html>
<html>
<head>
  <title>{% block title %}{% endblock %}</title>
  <link rel="stylesheet" href="{% static 'mystyles.css' %}">
</head>
<body>

<div class="topnav">
  <a href="/">HOME</a>
  <a href="/members">Members</a>
</div>

<div class="main">
  {% block content %}
  {% endblock %}
</div>

</body>
</html>
```

## Collect Static Files — Thu gom file tĩnh

```python
!cd my_tennis_club && python manage.py collectstatic --noinput
```

## 👀 Xem kết quả

Mọi trang kế thừa `master.html` giờ đều có thanh điều hướng và giao diện mới:

```python
from django_helper import preview

preview("/members/")
```

> 💡 Chạy `runserver` và mở `127.0.0.1:8000` để xem giao diện thực tế trên trình duyệt.

---

## 🧠 Ghi nhớ nhanh

> - CSS chung đặt trong **template cha** (`master.html`) → mọi trang con đều được áp dụng.
> - `{% load static %}` + `<link rel="stylesheet" href="{% static 'mystyles.css' %}">`.
> - Sửa/thêm file tĩnh → `collectstatic`.
