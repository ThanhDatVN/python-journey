## Add a Global CSS File — File CSS dùng chung

Ở bài trước, file tĩnh nằm trong thư mục `static` của **app** `members`.

Muốn file tĩnh **dùng chung cho cả dự án**, tạo một thư mục ở **cấp dự án**, ví dụ `mystaticfiles`:

````python
import os

os.makedirs("my_tennis_club/mystaticfiles", exist_ok=True)
````

````python
%%writefile my_tennis_club/mystaticfiles/myglobal.css
body {
  color: violet;
}
````

## Modify Settings — Sửa cấu hình

Báo cho Django biết thư mục mới bằng `STATICFILES_DIRS`:

```text
STATICFILES_DIRS = [
    BASE_DIR / 'mystaticfiles'
]
```

````python
from django_helper import edit_file

edit_file("my_tennis_club/my_tennis_club/settings.py",
          "STATIC_ROOT = BASE_DIR / 'productionfiles'",
          "STATIC_ROOT = BASE_DIR / 'productionfiles'\n\nSTATICFILES_DIRS = [\n    BASE_DIR / 'mystaticfiles'\n]")
````

> 📌 Trong `STATICFILES_DIRS` có thể liệt kê **nhiều** thư mục. Nếu có các file **trùng tên**, Django dùng file tìm thấy **đầu tiên**.

## Modify the Template — Sửa template

````python
%%writefile my_tennis_club/members/templates/template.html
{% load static %}
<!DOCTYPE html>
<html>
<link rel="stylesheet" href="{% static 'myglobal.css' %}">
<body>

{% for x in fruits %}
  <h1>{{ x }}</h1>
{% endfor %}

</body>
</html>
````

## Collect Static Files — Thu gom lại

Vì `DEBUG = False`, phải chạy lại `collectstatic`:

````python
!cd my_tennis_club && python manage.py collectstatic --noinput
````

## 🧠 Ghi nhớ nhanh

> - File tĩnh dùng chung: thư mục cấp dự án + `STATICFILES_DIRS = [BASE_DIR / 'mystaticfiles']`.
> - Trùng tên file → Django dùng file tìm thấy **đầu tiên**.
> - Thêm file tĩnh mới → chạy lại `collectstatic`.
