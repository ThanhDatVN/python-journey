## Add Records — Thêm bản ghi

Bảng `Member` đang **rỗng**. Ta dùng **Python interpreter (Python shell)** của Django để thêm dữ liệu.

Trên W3Schools, mở shell bằng lệnh trong terminal:

```bash
python manage.py shell
```

Trong notebook, ta nạp Django **ngay trong kernel** bằng hàm `setup()`:

````python
from django_helper import setup

setup()
from members.models import Member
````

Xem bảng `Member` (hiện đang rỗng):

````python
Member.objects.all()
````

Kết quả là một **QuerySet** rỗng — QuerySet là tập hợp dữ liệu lấy từ cơ sở dữ liệu.

Thêm một bản ghi:

````python
member = Member(firstname='An', lastname='Nguyễn')
member.save()
````

Kiểm tra bản ghi đã được thêm:

````python
Member.objects.all().values()
````

## Add Multiple Records — Thêm nhiều bản ghi

Tạo một list các đối tượng `Member`, rồi gọi `.save()` cho từng đối tượng:

````python
member1 = Member(firstname='Bình', lastname='Trần')
member2 = Member(firstname='Chi', lastname='Lê')
member3 = Member(firstname='Dũng', lastname='Phạm')
member4 = Member(firstname='Em', lastname='Hoàng')
member5 = Member(firstname='Giang', lastname='Vũ')

members_list = [member1, member2, member3, member4, member5]

for x in members_list:
    x.save()
````

Giờ bảng có 6 thành viên:

````python
Member.objects.all().values()
````
