## What is an App? — App là gì?

**App** là một ứng dụng web có **ý nghĩa riêng** trong dự án — ví dụ trang chủ, form liên hệ, hay cơ sở dữ liệu thành viên.

Ở đây ta tạo app cho phép **liệt kê và đăng ký thành viên** câu lạc bộ.

## Create App — Tạo app

Đặt tên app là **`members`**.

Trong terminal, đứng ở thư mục dự án (`my_tennis_club`, nơi có `manage.py`):

```bash
python manage.py startapp members
```

Trong notebook:

````python
!cd my_tennis_club && python manage.py startapp members
````

Django tạo thư mục `members` trong dự án:

````python
from pathlib import Path

for p in sorted(Path("my_tennis_club/members").rglob("*")):
    print("    " * (len(p.parts) - 3) + p.name + ("/" if p.is_dir() else ""))
````

```text
members/
    __init__.py
    admin.py
    apps.py
    migrations/
        __init__.py
    models.py
    tests.py
    views.py
```

Đây là các file và thư mục với vai trò cụ thể — bạn sẽ dùng phần lớn chúng trong các bài sau.

Trước tiên, hãy xem cách tạo một **view** đơn giản để hiển thị gì đó lên trình duyệt.

## 🧠 Ghi nhớ nhanh

> - **Project** = cả website · **App** = một chức năng trong website.
> - Tạo app: `python manage.py startapp members` (đứng trong thư mục có `manage.py`).
> - App gồm: `models.py`, `views.py`, `admin.py`, `migrations/`…
