## Django QuerySet

**QuerySet** là một **tập hợp dữ liệu** lấy từ cơ sở dữ liệu.

QuerySet được tạo bằng các phương thức của model, và có thể **lọc, sắp xếp**… QuerySet giúp lấy đúng dữ liệu bạn cần.

Ở bài này, ta dùng bảng `Member` — hiện có các thành viên:

```python
from django_helper import setup

setup()
from members.models import Member

for m in Member.objects.all():
    print(m.id, m.firstname, m.lastname, m.phone, m.joined_date)
```

## Querying Data — Truy vấn dữ liệu

Trong `views.py`, ta tạo QuerySet bằng `Member.objects.all()`:

```python
mydata = Member.objects.all()
print(mydata)
```

`all()` trả về một QuerySet gồm các **object**. Bạn có thể dùng QuerySet trong template, hoặc duyệt trong Python:

```python
for x in mydata:
    print(x.firstname, "-", x.lastname)
```

## 👀 Dùng QuerySet trong template

```python
from django_helper import replace_view

replace_view("testing", '''
def testing(request):
    mydata = Member.objects.all()
    template = loader.get_template('template.html')
    context = {
        'mymembers': mydata,
    }
    return HttpResponse(template.render(context, request))
''')
```

```python
%%writefile my_tennis_club/members/templates/template.html
<!DOCTYPE html>
<html>
<body>

<table border='1'>
  <tr>
    <th>ID</th>
    <th>Firstname</th>
    <th>Lastname</th>
  </tr>
  {% for x in mymembers %}
    <tr>
      <td>{{ x.id }}</td>
      <td>{{ x.firstname }}</td>
      <td>{{ x.lastname }}</td>
    </tr>
  {% endfor %}
</table>

</body>
</html>
```

```python
from django_helper import preview

preview("/testing/")
```

> 📌 QuerySet là **"lười" (lazy)**: câu lệnh SQL chỉ thực sự chạy khi bạn **dùng** dữ liệu (duyệt, in, `len()`…).

---

## 🧠 Ghi nhớ nhanh

> - QuerySet = tập dữ liệu từ cơ sở dữ liệu: `Model.objects.all()`, `.filter()`, `.order_by()`…
> - Dùng trong Python (duyệt `for`) hoặc gửi cho template qua context.
> - QuerySet **lazy** — chỉ truy vấn khi cần dữ liệu.
