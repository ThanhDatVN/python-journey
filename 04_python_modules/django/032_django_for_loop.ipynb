## For Loops — Vòng lặp for

Vòng lặp `for` duyệt qua các phần tử của một mảng, list hoặc dictionary:

```text
{% for x in fruits %}
  <h1>{{ x }}</h1>
{% endfor %}
```

## Loop Through a List of Dictionaries — Duyệt list các dictionary

```text
{% for x in cars %}
  <h1>{{ x.brand }}</h1>
  <p>{{ x.model }}</p>
  <p>{{ x.year }}</p>
{% endfor %}
```

## Data From a Model — Dữ liệu từ model

Dữ liệu trong model giống một bảng gồm hàng và cột — ta duyệt qua QuerySet:

```text
<ul>
  {% for x in members %}
    <li>{{ x.firstname }}</li>
  {% endfor %}
</ul>
```

## Reversed — Đảo ngược

Từ khoá `reversed` duyệt **ngược** danh sách:

```text
{% for x in members reversed %}
```

## Empty — Danh sách rỗng

Thẻ `{% empty %}` hiển thị nội dung khi danh sách **rỗng**:

```text
<ul>
  {% for x in emptytestobject %}
    <li>{{ x.firstname }}</li>
  {% empty %}
    <li>No members</li>
  {% endfor %}
</ul>
```

## Loop Variables — Biến của vòng lặp

Django có các biến dùng được bên trong vòng lặp:

| Biến | Ý nghĩa |
|---|---|
| `forloop.counter` | Số thứ tự vòng lặp, **bắt đầu từ 1** |
| `forloop.counter0` | Số thứ tự, **bắt đầu từ 0** |
| `forloop.first` | `True` ở vòng **đầu tiên** |
| `forloop.last` | `True` ở vòng **cuối cùng** |
| `forloop.parentloop` | Vòng lặp **cha** (khi lồng nhau) |
| `forloop.revcounter` | Đếm ngược, kết thúc ở 1 |
| `forloop.revcounter0` | Đếm ngược, kết thúc ở 0 |

## 👀 Thử nghiệm

````python
from django_helper import replace_view, preview

replace_view("testing", '''
def testing(request):
    template = loader.get_template('template.html')
    context = {
        'fruits': ['Apple', 'Banana', 'Cherry'],
        'cars': [
            {'brand': 'Ford', 'model': 'Mustang', 'year': '1964'},
            {'brand': 'Ford', 'model': 'Bronco', 'year': '1970'},
            {'brand': 'Volvo', 'model': 'P1800', 'year': '1964'},
        ],
        'members': Member.objects.all().values(),
        'emptytestobject': Member.objects.filter(firstname='Nobody').values(),
    }
    return HttpResponse(template.render(context, request))
''')
````

````python
%%writefile my_tennis_club/members/templates/template.html
<!DOCTYPE html>
<html>
<body>

<ul>
  {% for x in fruits %}
    <li {% if forloop.first %}style="color:red"{% endif %}>{{ forloop.counter }}. {{ x }}</li>
  {% endfor %}
</ul>

{% for x in cars %}
  <p>{{ x.brand }} {{ x.model }} ({{ x.year }})</p>
{% endfor %}

<ul>
  {% for x in members reversed %}
    <li>{{ x.firstname }}</li>
  {% endfor %}
</ul>

<ul>
  {% for x in emptytestobject %}
    <li>{{ x.firstname }}</li>
  {% empty %}
    <li>No members</li>
  {% endfor %}
</ul>

</body>
</html>
````

````python
preview("/testing/")
````

## 🧠 Ghi nhớ nhanh

> - `{% for x in list %}...{% endfor %}`; `reversed` duyệt ngược; `{% empty %}` khi danh sách rỗng.
> - Biến vòng lặp: `forloop.counter` (từ 1) · `counter0` (từ 0) · `first` · `last` · `revcounter`.
