## Details Template — Template chi tiết

Tiếp theo, tạo trang hiển thị **thông tin chi tiết** của từng thành viên.

Tạo template `details.html`:

```python
%%writefile my_tennis_club/members/templates/details.html
<!DOCTYPE html>
<html>
<body>

<h1>{{ mymember.firstname }} {{ mymember.lastname }}</h1>

<p>Phone: {{ mymember.phone }}</p>
<p>Member since: {{ mymember.joined_date }}</p>

<p>Back to <a href="/members">Members</a></p>

</body>
</html>
```

## Add Link in all-members Template — Thêm link

Mỗi thành viên trong danh sách sẽ là một **liên kết** tới trang chi tiết:

```python
%%writefile my_tennis_club/members/templates/all_members.html
<!DOCTYPE html>
<html>
<body>

<h1>Members</h1>

<ul>
  {% for x in mymembers %}
    <li><a href="details/{{ x.id }}">{{ x.firstname }} {{ x.lastname }}</a></li>
  {% endfor %}
</ul>

</body>
</html>
```

## Create new View — Tạo view mới

Thêm view `details` nhận thêm tham số `id`:

```python
%%writefile my_tennis_club/members/views.py
from django.http import HttpResponse
from django.template import loader
from .models import Member

def members(request):
    mymembers = Member.objects.all().values()
    template = loader.get_template('all_members.html')
    context = {
        'mymembers': mymembers,
    }
    return HttpResponse(template.render(context, request))

def details(request, id):
    mymember = Member.objects.get(id=id)
    template = loader.get_template('details.html')
    context = {
        'mymember': mymember,
    }
    return HttpResponse(template.render(context, request))
```

## Example Explained — Giải thích

View `details`:

1. Nhận `id` làm đối số.
2. Dùng `id` để tìm đúng bản ghi trong bảng `Member`.
3. Nạp template `details.html`.
4. Tạo context chứa thành viên tìm được và gửi cho template.

## Add URLs — Thêm URL

Đường dẫn `members/details/<int:id>` bắt một **số nguyên** trong URL và truyền vào view với tên `id`:

```python
%%writefile my_tennis_club/members/urls.py
from django.urls import path
from . import views

urlpatterns = [
    path('members/', views.members, name='members'),
    path('members/details/<int:id>', views.details, name='details'),
]
```
