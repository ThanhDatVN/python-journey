## Models — Mô hình dữ liệu

Tới giờ, nội dung hiển thị vẫn là **tĩnh**, viết thẳng trong HTML. Ta muốn tạo một **bảng thành viên** trong cơ sở dữ liệu, và hiển thị dữ liệu từ đó.

Một **model** trong Django là **một bảng** trong cơ sở dữ liệu.

## Django Models

Cơ sở dữ liệu mặc định của Django là **SQLite** — file `db.sqlite3` được tạo khi chạy `migrate`. Trong khóa học, ta dùng SQLite (sau này có thể chuyển sang PostgreSQL — xem bài [PostgreSQL](044_django_postgresql.ipynb)).

## Create Table (Model) — Tạo bảng

Tạo model `Member` trong file `members/models.py`, có hai trường `firstname` và `lastname`, đều là **chuỗi tối đa 255 ký tự**:

```python
%%writefile my_tennis_club/members/models.py
from django.db import models

class Member(models.Model):
    firstname = models.CharField(max_length=255)
    lastname = models.CharField(max_length=255)
```

Chạy lệnh tạo **migration** (bản mô tả thay đổi cơ sở dữ liệu):

```python
!cd my_tennis_club && python manage.py makemigrations members
```

Django tạo file mô tả thay đổi, đặt trong `members/migrations/`. Xem nội dung file:

```python
from django_helper import show_file

show_file("my_tennis_club/members/migrations/0001_initial.py")
```

> 📌 Django tự thêm trường `id` làm **khoá chính** tự tăng — bạn có thể ghi đè bằng cách tự định nghĩa trường `id`.

Bảng **chưa** được tạo — phải chạy `migrate`:

```python
!cd my_tennis_club && python manage.py migrate
```

Giờ đã có bảng `members_member` trong cơ sở dữ liệu! 🎉

## View SQL — Xem câu lệnh SQL

Xem câu lệnh SQL mà migration tạo ra:

```python
!cd my_tennis_club && python manage.py sqlmigrate members 0001
```

## 🧠 Ghi nhớ nhanh

> - Model = **một bảng**: `class Member(models.Model)` với các trường `models.CharField(...)`…
> - Quy trình: sửa `models.py` → `makemigrations` → `migrate`.
> - Django tự thêm khoá chính `id`; bảng có tên `tênapp_tênmodel` (`members_member`).
