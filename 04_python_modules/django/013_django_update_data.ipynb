## Update Records — Cập nhật bản ghi

Để cập nhật bản ghi, trước hết **lấy** bản ghi cần sửa:

```python
from django_helper import setup

setup()
from members.models import Member

x = Member.objects.all()[4]
```

`x` giờ là thành viên ở **chỉ số 4** (bản ghi thứ năm) — "Em Hoàng". Kiểm tra:

```python
x.firstname
```

Đổi giá trị của bản ghi:

```python
x.firstname = "Ếch"
x.save()
```

Kiểm tra bảng:

```python
Member.objects.all().values()
```

## ➕ Mở rộng: Lấy bản ghi theo id và cập nhật hàng loạt

```python
m = Member.objects.get(id=5)
print(m.firstname, m.lastname)

# cập nhật (mọi) bản ghi thoả điều kiện bằng update() — không cần save():
count = Member.objects.filter(firstname="Ếch").update(firstname="Em")
print(count, "bản ghi được cập nhật")
print(Member.objects.get(id=5).firstname)
```

---

## 🧠 Ghi nhớ nhanh

> - Lấy bản ghi: `Member.objects.all()[i]` hoặc `Member.objects.get(id=...)`.
> - Sửa thuộc tính rồi `.save()`.
> - Cập nhật hàng loạt: `Member.objects.filter(...).update(field=...)`.
