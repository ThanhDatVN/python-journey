# Django Admin - Include Member

> 📖 **Học song song trên W3Schools:** [django_admin_include_members.php](https://www.w3schools.com/django/django_admin_include_members.php)
>
> 🎯 **Trong bài này:** Include Member in the Admin Interface · Make the List Display More Reader-Friendly · 👀 Kiểm tra

## Include Member in the Admin Interface — Đưa model vào trang admin

Để đưa model `Member` vào trang admin, phải **đăng ký (register)** nó trong file `members/admin.py`:

```python
%%writefile my_tennis_club/members/admin.py
from django.contrib import admin
from .models import Member

# Register your models here.
admin.site.register(Member)
```

Giờ trang admin có thêm mục **Members**. Mở mục này, bạn sẽ thấy danh sách thành viên — nhưng hiển thị khó hiểu: "Member object (1)", "Member object (2)"…

## Make the List Display More Reader-Friendly — Hiển thị dễ đọc hơn

Thêm phương thức `__str__()` vào model để Django biết cách hiển thị mỗi object (xem lại bài [__str__](../../02_python_classes/008_magic_method_str.ipynb)):

```python
%%writefile my_tennis_club/members/models.py
from django.db import models

class Member(models.Model):
    firstname = models.CharField(max_length=255)
    lastname = models.CharField(max_length=255)
    phone = models.IntegerField(null=True)
    joined_date = models.DateField(null=True)

    def __str__(self):
        return f"{self.firstname} {self.lastname}"
```

> 📌 Thêm phương thức **không** thay đổi cấu trúc bảng → **không** cần `makemigrations`.

## 👀 Kiểm tra

```python
from django_helper import setup

setup()
from django.contrib import admin
from members.models import Member

print("Member đã đăng ký trong admin:", admin.site.is_registered(Member))
print(list(Member.objects.all()))
```

---

## 🧠 Ghi nhớ nhanh

> - Đăng ký model: `admin.site.register(Member)` trong `members/admin.py`.
> - Thêm `__str__()` vào model → danh sách admin hiển thị tên thay vì "Member object (1)".
