## Template Tags — Thẻ template

Trong template Django, bạn có thể thực hiện **logic lập trình** như câu lệnh `if` hay vòng lặp `for`.

Các từ khoá đó gọi là **thẻ template (template tag)**, viết trong `{% %}`:

```text
{% if greeting == 1 %}
  <h1>Hello</h1>
{% else %}
  <h1>Bye</h1>
{% endif %}
```

## Django Code

Template **không phải** Python — Django dịch các thẻ thành HTML trước khi gửi tới trình duyệt. Hầu hết thẻ cần **thẻ đóng** (`{% endif %}`, `{% endfor %}`…).

## Template Tag Reference — Một số thẻ phổ biến

| Thẻ | Mô tả |
|---|---|
| `autoescape` | Bật/tắt tự động escape HTML |
| `block` | Định nghĩa một block (dùng với kế thừa template) |
| `comment` | Định nghĩa một đoạn chú thích |
| `csrf_token` | Chống tấn công CSRF trong form |
| `cycle` | Luân phiên giá trị qua mỗi vòng lặp |
| `extends` | Kế thừa template cha |
| `filter` | Áp dụng filter cho nội dung |
| `for` / `empty` | Vòng lặp / nội dung khi danh sách rỗng |
| `if` / `elif` / `else` | Rẽ nhánh |
| `include` | Chèn một template khác |
| `load` | Nạp thư viện thẻ (ví dụ `static`) |
| `now` | Hiển thị ngày giờ hiện tại |
| `url` | Tạo URL theo tên đường dẫn |
| `with` | Tạo biến tạm |

Xem đầy đủ: [Django Template Tag Reference](https://www.w3schools.com/django/django_ref_template_tags.php).

## 👀 Thử nghiệm

````python
from django_helper import replace_view, preview

replace_view("testing", '''
def testing(request):
    template = loader.get_template('template.html')
    context = {
        'greeting': 1,
    }
    return HttpResponse(template.render(context, request))
''')
````

````python
%%writefile my_tennis_club/members/templates/template.html
<!DOCTYPE html>
<html>
<body>

{% if greeting == 1 %}
  <h1>Hello</h1>
{% else %}
  <h1>Bye</h1>
{% endif %}

<p>Today: {% now "d/m/Y" %}</p>

</body>
</html>
````

````python
preview("/testing/")
````

## 🧠 Ghi nhớ nhanh

> - Thẻ template viết trong `{% %}` — dùng cho logic (`if`, `for`, `with`, `extends`, `include`…).
> - Biến viết trong `{{ }}`; phần lớn thẻ cần thẻ đóng (`endif`, `endfor`…).
