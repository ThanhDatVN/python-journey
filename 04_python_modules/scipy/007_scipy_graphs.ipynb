## Working with Graphs — Làm việc với đồ thị

**Đồ thị (graph)** là cấu trúc dữ liệu cơ bản. SciPy có module `scipy.sparse.csgraph` để làm việc với đồ thị.

## Adjacency Matrix — Ma trận kề

**Ma trận kề** là ma trận `n x n`, với `n` là số đỉnh (node).

Ví dụ đồ thị vô hướng có 3 đỉnh A, B, C với các cạnh có trọng số:

```text
A — B : trọng số 1
A — C : trọng số 2
```

Ma trận kề tương ứng:

```text
     A  B  C
A:  [0  1  2]
B:  [1  0  0]
C:  [2  0  0]
```

> 📌 Với đồ thị **vô hướng**, ma trận kề **đối xứng**. Với đồ thị **có hướng**, chỉ có giá trị ở các ô thể hiện chiều của cạnh.

## Connected Components — Thành phần liên thông

Tìm các thành phần liên thông bằng `connected_components()`:

````python
import numpy as np
from scipy.sparse.csgraph import connected_components
from scipy.sparse import csr_matrix

arr = np.array([
    [0, 1, 2],
    [1, 0, 0],
    [2, 0, 0]
])

newarr = csr_matrix(arr)

print(connected_components(newarr))
````

## Dijkstra

Dùng `dijkstra()` để tìm **đường đi ngắn nhất** từ một đỉnh tới các đỉnh khác. Tham số:

- `return_predecessors` — `True` để trả về cả đường đi.
- `indices` — chỉ số đỉnh xuất phát.
- `limit` — trọng số tối đa của đường đi.

Tìm đường ngắn nhất từ đỉnh 1 tới đỉnh 2:

````python
import numpy as np
from scipy.sparse.csgraph import dijkstra
from scipy.sparse import csr_matrix

arr = np.array([
    [0, 1, 2],
    [1, 0, 0],
    [2, 0, 0]
])

newarr = csr_matrix(arr)

print(dijkstra(newarr, return_predecessors=True, indices=0))
````

## Floyd Warshall

`floyd_warshall()` tìm đường ngắn nhất giữa **mọi cặp** đỉnh:

````python
import numpy as np
from scipy.sparse.csgraph import floyd_warshall
from scipy.sparse import csr_matrix

arr = np.array([
    [0, 1, 2],
    [1, 0, 0],
    [2, 0, 0]
])

newarr = csr_matrix(arr)

print(floyd_warshall(newarr, return_predecessors=True))
````

## Bellman Ford

`bellman_ford()` cũng tìm đường ngắn nhất giữa mọi cặp đỉnh, nhưng xử lý được cả **trọng số âm**:

````python
import numpy as np
from scipy.sparse.csgraph import bellman_ford
from scipy.sparse import csr_matrix

arr = np.array([
    [0, -1, 2],
    [1, 0, 0],
    [2, 0, 0]
])

newarr = csr_matrix(arr)

print(bellman_ford(newarr, return_predecessors=True, indices=0))
````

## Depth First Order — Duyệt theo chiều sâu

`depth_first_order()` trả về thứ tự **duyệt theo chiều sâu (DFS)** từ một đỉnh. Tham số: đồ thị và đỉnh bắt đầu:

````python
import numpy as np
from scipy.sparse.csgraph import depth_first_order
from scipy.sparse import csr_matrix

arr = np.array([
    [0, 1, 0, 1],
    [1, 1, 1, 1],
    [2, 1, 1, 0],
    [0, 1, 0, 1]
])

newarr = csr_matrix(arr)

print(depth_first_order(newarr, 1))
````

## Breadth First Order — Duyệt theo chiều rộng

`breadth_first_order()` trả về thứ tự **duyệt theo chiều rộng (BFS)**:

````python
import numpy as np
from scipy.sparse.csgraph import breadth_first_order
from scipy.sparse import csr_matrix

arr = np.array([
    [0, 1, 0, 1],
    [1, 1, 1, 1],
    [2, 1, 1, 0],
    [0, 1, 0, 1]
])

newarr = csr_matrix(arr)

print(breadth_first_order(newarr, 1))
````
