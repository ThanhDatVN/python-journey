## What is Statistical Significance Test? — Kiểm định ý nghĩa thống kê là gì?

Trong thống kê, **ý nghĩa thống kê (statistical significance)** cho biết kết quả quan sát được có **lý do** đằng sau, chứ không phải **ngẫu nhiên** hay tình cờ.

SciPy có module `scipy.stats` với các hàm kiểm định ý nghĩa thống kê.

Các khái niệm quan trọng:

## Hypothesis in Statistics — Giả thuyết trong thống kê

Giả thuyết là một **giả định** về một tham số của tổng thể.

### Null Hypothesis — Giả thuyết không (H₀)

Giả định rằng quan sát **không** có ý nghĩa thống kê.

### Alternate Hypothesis — Giả thuyết đối (H₁)

Giả định rằng quan sát **có** lý do đằng sau. Nó đối lập với giả thuyết không.

> Ví dụ: khi đánh giá một học sinh, ta thường giả định "học sinh **trung bình**" (H₀) và kiểm tra xem có đủ bằng chứng để nói "học sinh **giỏi hơn** trung bình" (H₁) hay không.

### One tailed test — Kiểm định một phía

Khi giả thuyết chỉ kiểm tra **một phía** của giá trị — ví dụ "trung bình **lớn hơn** k".

### Two tailed test — Kiểm định hai phía

Khi kiểm tra **cả hai phía** — ví dụ "trung bình **khác** k" (có thể lớn hơn hoặc nhỏ hơn).

### Alpha value — Mức ý nghĩa α

**Alpha** là mức ý nghĩa — cho biết dữ liệu phải "cực đoan" tới mức nào để **bác bỏ** giả thuyết không. Thường đặt **0.01, 0.05 hoặc 0.1**.

### P value — Giá trị p

**P value** cho biết dữ liệu thực sự "gần cực đoan" tới mức nào. So sánh p value với alpha:

> **Nếu p value ≤ alpha → bác bỏ giả thuyết không**, kết luận dữ liệu **có ý nghĩa thống kê**. Ngược lại thì không bác bỏ được.

## T-Test

**T-test** dùng để xác định có **sự khác biệt có ý nghĩa** giữa **trung bình** của hai biến hay không, tức là chúng có thuộc cùng một phân phối không.

Đây là kiểm định **hai phía**. Hàm `ttest_ind()` nhận hai mẫu cùng kích thước, trả về tuple gồm **t-statistic** và **p-value**.

Kiểm tra hai mẫu `v1`, `v2` có thuộc cùng phân phối không:

```python
import numpy as np
from scipy.stats import ttest_ind

v1 = np.random.normal(size=100)
v2 = np.random.normal(size=100)

res = ttest_ind(v1, v2)

print(res)
```

Chỉ lấy p-value:

```python
import numpy as np
from scipy.stats import ttest_ind

v1 = np.random.normal(size=100)
v2 = np.random.normal(size=100)

res = ttest_ind(v1, v2).pvalue

print(res)
```

## KS-Test

**KS test** (Kolmogorov–Smirnov) kiểm tra các giá trị có tuân theo một **phân phối cho trước** hay không.

Hàm nhận giá trị cần kiểm tra và **CDF** (hàm phân phối tích luỹ) dưới dạng chuỗi tên phân phối hoặc hàm. Có thể là kiểm định một hay hai phía (mặc định hai phía, đổi bằng `alternative`):

```python
import numpy as np
from scipy.stats import kstest

v = np.random.normal(size=100)

res = kstest(v, 'norm')

print(res)
```

## Statistical Description of Data — Mô tả thống kê dữ liệu

Hàm `describe()` tóm tắt thông tin của mảng:

- số quan sát (nobs)
- min và max
- trung bình (mean)
- phương sai (variance)
- độ lệch (skewness)
- độ nhọn (kurtosis)

```python
import numpy as np
from scipy.stats import describe

v = np.random.normal(size=100)
res = describe(v)

print(res)
```

## Normality Tests (Skewness and Kurtosis) — Kiểm định phân phối chuẩn

Kiểm định chuẩn dựa trên **độ lệch (skewness)** và **độ nhọn (kurtosis)**.

### Skewness — Độ lệch

Thước đo **độ đối xứng** của dữ liệu.

- Với phân phối chuẩn, skewness = **0**.
- Âm → dữ liệu **lệch trái**; dương → **lệch phải**.

### Kurtosis — Độ nhọn

Thước đo dữ liệu có đuôi **dày hay mỏng** so với phân phối chuẩn.

- Kurtosis dương → đuôi **dày**.
- Kurtosis âm → đuôi **mỏng**.

```python
import numpy as np
from scipy.stats import skew, kurtosis

v = np.random.normal(size=100)

print(skew(v))
print(kurtosis(v))
```

Kiểm tra dữ liệu có đến từ **phân phối chuẩn** hay không bằng `normaltest()` — trả về p-value của giả thuyết không "x đến từ phân phối chuẩn":

```python
import numpy as np
from scipy.stats import normaltest

v = np.random.normal(size=100)

print(normaltest(v))
```

## ➕ Mở rộng: Ví dụ có ý nghĩa thống kê

Hai nhóm có trung bình **thật sự khác nhau** → p-value rất nhỏ:

```python
import numpy as np
from scipy.stats import ttest_ind

rng = np.random.default_rng(0)
group_a = rng.normal(loc=7.0, scale=1, size=50)   # điểm lớp A
group_b = rng.normal(loc=7.8, scale=1, size=50)   # điểm lớp B

p = ttest_ind(group_a, group_b).pvalue
print(f"p-value = {p:.4f}")
print("Có ý nghĩa thống kê (α = 0.05)" if p <= 0.05 else "Không đủ bằng chứng")
```

## 🧠 Ghi nhớ nhanh

> - H₀ (giả thuyết không) vs H₁ (giả thuyết đối); **p ≤ α → bác bỏ H₀**.
> - `ttest_ind(a, b)` so sánh **trung bình** hai mẫu · `kstest(v, 'norm')` so với một phân phối.
> - `describe(v)` mô tả nhanh · `skew` (đối xứng) · `kurtosis` (độ dày đuôi) · `normaltest` (có chuẩn không).
