## What is Sparse Data — Dữ liệu thưa là gì?

**Dữ liệu thưa (sparse data)** là dữ liệu mà **phần lớn phần tử không được dùng** (không mang thông tin — thường là số 0).

Ví dụ mảng:

```text
[1, 0, 2, 0, 0, 3, 0, 0, 0, 0, 0, 0]
```

- **Dữ liệu thưa**: tập dữ liệu mà phần lớn giá trị bằng 0.
- **Dữ liệu dày (dense)**: ngược lại, phần lớn giá trị **khác 0**.

Trong tính toán khoa học, khi làm việc với **đạo hàm riêng** trong đại số tuyến tính, ta thường gặp dữ liệu thưa.

## How to Work With Sparse Data — Làm việc với dữ liệu thưa

SciPy có module `scipy.sparse` với các hàm xử lý dữ liệu thưa. Hai loại ma trận thưa chủ yếu:

- **CSC** — *Compressed Sparse Column*: nén theo cột, cắt cột nhanh.
- **CSR** — *Compressed Sparse Row*: nén theo hàng, cắt hàng nhanh.

Ta dùng **CSR** trong bài này.

## CSR Matrix — Ma trận CSR

Tạo ma trận CSR bằng cách truyền một mảng vào `scipy.sparse.csr_matrix()`:

````python
import numpy as np
from scipy.sparse import csr_matrix

arr = np.array([0, 0, 0, 0, 0, 1, 1, 0, 2])

print(csr_matrix(arr))
````

**Giải thích kết quả** — mỗi dòng có dạng `(hàng, cột)  giá_trị`:

- Phần tử đầu tiên khác 0 ở hàng `0`, cột `5`, giá trị `1`.
- Phần tử thứ hai ở hàng `0`, cột `6`, giá trị `1`.
- Phần tử thứ ba ở hàng `0`, cột `8`, giá trị `2`.

## Sparse Matrix Methods — Phương thức của ma trận thưa

Xem dữ liệu đã lưu (các phần tử khác 0) bằng thuộc tính `data`:

````python
import numpy as np
from scipy.sparse import csr_matrix

arr = np.array([[0, 0, 0], [0, 0, 1], [1, 0, 2]])

print(csr_matrix(arr).data)
````

Đếm số phần tử khác 0 bằng `count_nonzero()`:

````python
import numpy as np
from scipy.sparse import csr_matrix

arr = np.array([[0, 0, 0], [0, 0, 1], [1, 0, 2]])

print(csr_matrix(arr).count_nonzero())
````

Xoá các phần tử 0 khỏi ma trận bằng `eliminate_zeros()`:

````python
import numpy as np
from scipy.sparse import csr_matrix

arr = np.array([[0, 0, 0], [0, 0, 1], [1, 0, 2]])

mat = csr_matrix(arr)
mat.eliminate_zeros()

print(mat)
````

Gộp các phần tử trùng vị trí bằng `sum_duplicates()`:

````python
import numpy as np
from scipy.sparse import csr_matrix

arr = np.array([[0, 0, 0], [0, 0, 1], [1, 0, 2]])

mat = csr_matrix(arr)
mat.sum_duplicates()

print(mat)
````

Chuyển từ CSR sang CSC bằng `tocsc()`:

````python
import numpy as np
from scipy.sparse import csr_matrix

arr = np.array([[0, 0, 0], [0, 0, 1], [1, 0, 2]])

newarr = csr_matrix(arr).tocsc()

print(newarr)
````

> 📌 Ngoài các phương thức thưa riêng, ma trận thưa hỗ trợ hầu hết phép toán của ma trận thông thường: chuyển vị, cộng, nhân…

## ➕ Mở rộng: Tiết kiệm bộ nhớ

````python
import numpy as np
from scipy.sparse import csr_matrix

dense = np.zeros((1000, 1000))
dense[0, 0] = 1
dense[500, 500] = 2

sparse = csr_matrix(dense)
print("Dense :", dense.nbytes, "bytes")
print("Sparse:", sparse.data.nbytes + sparse.indices.nbytes + sparse.indptr.nbytes, "bytes")
````

> - `.data` · `.count_nonzero()` · `.eliminate_zeros()` · `.sum_duplicates()`.
