## Import SciPy

Sau khi cài, import **module** SciPy muốn dùng bằng câu lệnh `from scipy import module`:

````python
from scipy import constants
````

Giờ đã import module `constants`, sẵn sàng sử dụng:

````python
from scipy import constants

print(constants.liter)
````

> 📌 `constants`: SciPy có nhiều hằng số khoa học — ở đây ta lấy **lít** tính theo **mét khối** (0.001).

## Checking SciPy Version — Kiểm tra phiên bản

````python
import scipy

print(scipy.__version__)
````
