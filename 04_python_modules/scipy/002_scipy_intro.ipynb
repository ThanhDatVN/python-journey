## What is SciPy? — SciPy là gì?

- SciPy là thư viện **tính toán khoa học**, dùng NumPy bên dưới.
- SciPy là viết tắt của **Scientific Python**.
- Cung cấp thêm các hàm tiện ích cho **tối ưu hoá, thống kê và xử lý tín hiệu**.
- Giống NumPy, SciPy là mã nguồn mở, dùng miễn phí.
- SciPy được tạo bởi tác giả của NumPy — **Travis Oliphant**.

## Why Use SciPy? — Vì sao dùng SciPy?

Nếu SciPy dùng NumPy bên dưới, sao không chỉ dùng NumPy?

SciPy có các hàm được **tối ưu hoá** và **bổ sung** thường dùng trong NumPy và khoa học dữ liệu.

## Which Language is SciPy Written in? — Viết bằng ngôn ngữ gì?

SciPy chủ yếu được viết bằng **Python**, một số phần viết bằng **C** (và Fortran).

## Where is the SciPy Codebase? — Mã nguồn ở đâu?

Mã nguồn SciPy tại [github.com/scipy/scipy](https://github.com/scipy/scipy).

## ➕ Mở rộng: Các module con của SciPy

| Module | Chức năng |
|---|---|
| `scipy.constants` | Hằng số vật lý, toán học |
| `scipy.optimize` | Tối ưu hoá, tìm nghiệm |
| `scipy.sparse` | Ma trận thưa |
| `scipy.sparse.csgraph` | Thuật toán đồ thị |
| `scipy.spatial` | Dữ liệu không gian |
| `scipy.io` | Đọc/ghi file (Matlab…) |
| `scipy.interpolate` | Nội suy |
| `scipy.stats` | Thống kê |
