## Optimizers in SciPy — Bộ tối ưu hoá

**Optimizer** là tập các thủ tục trong SciPy để **tìm giá trị nhỏ nhất của một hàm**, hoặc **tìm nghiệm của một phương trình**.

## Optimizing Functions — Tối ưu hoá hàm

Về bản chất, mọi thuật toán học máy đều là những phương trình phức tạp cần được **tối thiểu hoá** dựa trên dữ liệu.

## Roots of an Equation — Nghiệm của phương trình

NumPy tìm được nghiệm của đa thức và phương trình tuyến tính, nhưng **không** tìm được nghiệm của phương trình **phi tuyến** như:

```text
x + cos(x)
```

Với những phương trình này, dùng hàm `optimize.root` của SciPy. Hàm nhận hai tham số:

- `fun` — hàm biểu diễn phương trình.
- `x0` — giá trị đoán ban đầu cho nghiệm.

Hàm trả về một object chứa thông tin về nghiệm. Nghiệm nằm ở thuộc tính `x`.

> ⚠️ **Lưu ý cập nhật:** W3Schools dùng `math.cos`, nhưng `root()` truyền vào hàm một **mảng NumPy** — từ NumPy 2.x, `math.cos(mảng)` báo lỗi. Hãy dùng `np.cos`, hàm hoạt động được với mảng.

````python
from scipy.optimize import root
import numpy as np

def eqn(x):
    return x + np.cos(x)

myroot = root(eqn, 0)

print(myroot.x)
````

In toàn bộ thông tin về nghiệm:

````python
from scipy.optimize import root
import numpy as np

def eqn(x):
    return x + np.cos(x)

myroot = root(eqn, 0)

print(myroot)
````

## Minimizing a Function — Tìm giá trị nhỏ nhất của hàm

Một hàm, trong ngữ cảnh này, biểu diễn một **đường cong** — có các **đỉnh** và **đáy**.

- Điểm cao nhất trên toàn đường cong: **cực đại toàn cục (global maxima)**; các đỉnh khác: **cực đại địa phương**.
- Điểm thấp nhất: **cực tiểu toàn cục (global minima)**; các đáy khác: **cực tiểu địa phương**.

### Finding Minima — Tìm cực tiểu

Dùng `scipy.optimize.minimize()`, nhận các tham số:

- `fun` — hàm cần tối thiểu hoá.
- `x0` — giá trị đoán ban đầu.
- `method` — tên phương pháp: `'CG'`, `'BFGS'`, `'Newton-CG'`, `'L-BFGS-B'`, `'TNC'`, `'COBYLA'`, `'SLSQP'`.
- `callback` — hàm được gọi sau mỗi vòng tối ưu.
- `options` — dictionary thêm tuỳ chọn, ví dụ `{"disp": True}`.

Tìm cực tiểu của `x^2 + x + 2` bằng `BFGS`:

````python
from scipy.optimize import minimize

def eqn(x):
    return x**2 + x + 2

mymin = minimize(eqn, 0, method='BFGS')

print(mymin)
````

## ➕ Mở rộng: Vẽ để kiểm chứng

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize

f = lambda x: x**2 + x + 2
res = minimize(f, 0, method="BFGS")

x = np.linspace(-3, 2, 100)
plt.plot(x, f(x))
plt.scatter(res.x, f(res.x), color="red", zorder=3, label=f"min tại x={res.x[0]:.2f}")
plt.legend()
plt.show()
````

---

## 🧠 Ghi nhớ nhanh

> - `optimize.root(fun, x0)` → nghiệm của phương trình (kể cả phi tuyến); nghiệm ở `.x`.
> - `optimize.minimize(fun, x0, method='BFGS')` → tìm **cực tiểu** của hàm.
> - Học máy về bản chất là **tối thiểu hoá** hàm mất mát.
