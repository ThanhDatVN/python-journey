## Constants in SciPy — Hằng số trong SciPy

Vì SciPy tập trung vào khoa học, nó có rất nhiều **hằng số khoa học** sẵn có — hữu ích khi làm việc với khoa học dữ liệu.

In giá trị hằng số PI:

```python
from scipy import constants

print(constants.pi)
```

## Constant Units — Danh sách đơn vị

Dùng hàm `dir()` để xem toàn bộ các đơn vị trong module constants:

```python
from scipy import constants

units = [u for u in dir(constants) if not u.startswith("_")]
print(len(units), "tên, ví dụ:", units[:20])
```

## Unit Categories — Các nhóm đơn vị

Các đơn vị được chia nhóm như sau. Mọi giá trị đều quy về **đơn vị chuẩn SI** của nhóm.

### Metric (SI) Prefixes — Tiền tố hệ mét

Trả về giá trị theo **mét**:

```python
from scipy import constants

print(constants.yotta)    # 1e+24
print(constants.kilo)     # 1000.0
print(constants.centi)    # 0.01
print(constants.milli)    # 0.001
print(constants.micro)    # 1e-06
print(constants.nano)     # 1e-09
```

### Binary Prefixes — Tiền tố nhị phân

Trả về giá trị theo **byte**:

```python
from scipy import constants

print(constants.kibi)    # 1024
print(constants.mebi)    # 1048576
print(constants.gibi)    # 1073741824
print(constants.tebi)
```

### Mass — Khối lượng

Trả về giá trị theo **kg**:

```python
from scipy import constants

print(constants.gram)
print(constants.metric_ton)
print(constants.pound)
print(constants.ounce)
print(constants.carat)
print(constants.atomic_mass)
```

### Angle — Góc

Trả về giá trị theo **radian**:

```python
from scipy import constants

print(constants.degree)
print(constants.arcmin)
print(constants.arcsec)
```

### Time — Thời gian

Trả về giá trị theo **giây**:

```python
from scipy import constants

print(constants.minute)
print(constants.hour)
print(constants.day)
print(constants.week)
print(constants.year)
print(constants.Julian_year)
```

### Length — Độ dài

Trả về giá trị theo **mét**:

```python
from scipy import constants

print(constants.inch)
print(constants.foot)
print(constants.mile)
print(constants.nautical_mile)
print(constants.angstrom)
print(constants.light_year)
print(constants.parsec)
```

### Pressure — Áp suất

Trả về giá trị theo **pascal**:

```python
from scipy import constants

print(constants.atm)
print(constants.bar)
print(constants.torr)
print(constants.psi)
```

### Area — Diện tích

Trả về giá trị theo **mét vuông**:

```python
from scipy import constants

print(constants.hectare)
print(constants.acre)
```

### Volume — Thể tích

Trả về giá trị theo **mét khối**:

```python
from scipy import constants

print(constants.liter)
print(constants.gallon)
print(constants.fluid_ounce)
print(constants.barrel)
```

### Speed — Tốc độ

Trả về giá trị theo **mét/giây**:

```python
from scipy import constants

print(constants.kmh)
print(constants.mph)
print(constants.mach)
print(constants.knot)
```

### Temperature — Nhiệt độ

Trả về giá trị theo **Kelvin**:

```python
from scipy import constants

print(constants.zero_Celsius)       # 0 °C = 273.15 K
print(constants.degree_Fahrenheit)
```

### Energy — Năng lượng

Trả về giá trị theo **joule**:

```python
from scipy import constants

print(constants.eV)
print(constants.calorie)
print(constants.erg)
print(constants.Btu)
print(constants.ton_TNT)
```

### Power — Công suất

Trả về giá trị theo **watt**:

```python
from scipy import constants

print(constants.hp)
print(constants.horsepower)
```

### Force — Lực

Trả về giá trị theo **newton**:

```python
from scipy import constants

print(constants.dyne)
print(constants.lbf)
print(constants.kgf)
```

## ➕ Mở rộng: Hằng số vật lý

```python
from scipy import constants

print("Tốc độ ánh sáng c =", constants.c, "m/s")
print("Hằng số hấp dẫn G =", constants.G)
print("Hằng số Planck h =", constants.h)
print("Gia tốc trọng trường g =", constants.g)
```

> 🧪 **Try it Yourself** — Đổi 5 dặm (mile) sang km bằng constants.

```python
from scipy import constants

miles = 5
print(miles * constants.mile / constants.kilo, "km")
```
