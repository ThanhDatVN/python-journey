## What is Pandas? — Pandas là gì?

**Pandas** là thư viện Python để làm việc với **tập dữ liệu (dataset)**.

Nó có các hàm để **phân tích, làm sạch, khám phá và xử lý** dữ liệu.

Tên "Pandas" xuất phát từ cả "**Pan**el **Da**ta" và "Python Data Analysis". Pandas được **Wes McKinney** tạo ra năm **2008**.

## Why Use Pandas? — Vì sao dùng Pandas?

- Pandas giúp phân tích dữ liệu lớn và rút ra kết luận dựa trên **lý thuyết thống kê**.
- Pandas có thể **làm sạch** dữ liệu lộn xộn, giúp dữ liệu dễ đọc và phù hợp hơn.

> Dữ liệu phù hợp (relevant) rất quan trọng trong khoa học dữ liệu.

> 📌 **Khoa học dữ liệu (data science)**: một nhánh của khoa học máy tính, nghiên cứu cách lưu trữ, sử dụng và phân tích dữ liệu để rút ra thông tin.

## What Can Pandas Do? — Pandas làm được gì?

Pandas trả lời các câu hỏi về dữ liệu, ví dụ:

- Có **tương quan** giữa hai hay nhiều cột không?
- Giá trị **trung bình** là bao nhiêu?
- Giá trị **lớn nhất**? **nhỏ nhất**?

Pandas còn **xoá các hàng không phù hợp** hoặc chứa giá trị sai — như giá trị rỗng hay NULL. Việc này gọi là **làm sạch (cleaning)** dữ liệu.

```python
import pandas as pd

df = pd.read_csv('data.csv')
print("Calories trung bình:", round(df["Calories"].mean(), 1))
print("Thời lượng dài nhất:", df["Duration"].max())
print("Tương quan Duration - Calories:", round(df["Duration"].corr(df["Calories"]), 2))
```
