# Pandas - Cleaning Empty Cells

> 📖 **Học song song trên W3Schools:** [pandas_cleaning_empty_cells.asp](https://www.w3schools.com/python/pandas/pandas_cleaning_empty_cells.asp)
>
> 🎯 **Trong bài này:** Empty Cells · Remove Rows · Replace Empty Values · Replace Only For Specified Columns · Replace Using Mean, Median, or Mode

## Empty Cells — Ô trống

Ô trống có thể khiến kết quả phân tích bị sai.

## Remove Rows — Xoá hàng

Một cách xử lý ô trống là **xoá các hàng chứa ô trống**.

Thường thì không sao, vì tập dữ liệu có thể rất lớn, xoá vài hàng không ảnh hưởng nhiều tới kết quả.

Tạo DataFrame mới **không chứa ô trống**:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

new_df = df.dropna()

print(new_df.to_string())
```

> 📌 Mặc định, `dropna()` trả về **DataFrame mới**, không thay đổi bản gốc.

Muốn thay đổi chính DataFrame gốc, dùng tham số `inplace=True`:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

df.dropna(inplace=True)

print(df.shape)
```

> 📌 `dropna(inplace=True)` **không** trả về DataFrame mới mà **xoá luôn** các hàng có giá trị rỗng trên DataFrame gốc.

## Replace Empty Values — Thay giá trị rỗng

Cách khác là **điền giá trị mới** vào ô trống — không phải xoá cả hàng chỉ vì vài ô trống.

Phương thức `fillna()` thay ô trống bằng một giá trị:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

df.fillna(130, inplace=True)

print(df.loc[[18, 22, 28]])
```

## Replace Only For Specified Columns — Chỉ thay ở cột chỉ định

Ví dụ trên thay **mọi** ô trống trong toàn bảng. Để chỉ thay ở một cột, chỉ định **tên cột**:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

df["Calories"] = df["Calories"].fillna(130)

print(df.loc[[18, 22, 28]])
```

> 💡 Pandas mới khuyến nghị gán lại `df["col"] = df["col"].fillna(...)` thay vì `df["col"].fillna(..., inplace=True)`.

## Replace Using Mean, Median, or Mode — Thay bằng trung bình, trung vị, yếu vị

Cách phổ biến là tính **trung bình (mean)**, **trung vị (median)** hoặc **yếu vị (mode)** của cột rồi điền vào ô trống.

Pandas có `mean()`, `median()`, `mode()`:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

x = df["Calories"].mean()
df["Calories"] = df["Calories"].fillna(x)

print(round(x, 2))
print(df.loc[[18, 28]])
```

> **Mean** = trung bình = tổng các giá trị chia cho số giá trị.

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

x = df["Calories"].median()
df["Calories"] = df["Calories"].fillna(x)

print(x)
```

> **Median** = trung vị = giá trị ở giữa sau khi sắp xếp tăng dần.

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

x = df["Calories"].mode()[0]
df["Calories"] = df["Calories"].fillna(x)

print(x)
```

> **Mode** = yếu vị = giá trị xuất hiện **nhiều nhất**.

---

## 🧠 Ghi nhớ nhanh

> - Xoá hàng có ô trống: `df.dropna()` (trả về bảng mới) / `inplace=True` (sửa bảng gốc).
> - Điền giá trị: `df.fillna(v)` · chỉ một cột: `df["c"] = df["c"].fillna(v)`.
> - Điền bằng thống kê: `mean()` trung bình · `median()` trung vị · `mode()[0]` yếu vị.
