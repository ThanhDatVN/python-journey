## Finding Relationships — Tìm mối liên hệ

Một phương thức rất mạnh của Pandas là `corr()`.

`corr()` tính **mối quan hệ (tương quan)** giữa từng cặp cột trong tập dữ liệu.

Ví dụ dùng `data.csv`:

```python
import pandas as pd

df = pd.read_csv('data.csv')

print(df.corr())
```

> 📌 `corr()` **bỏ qua** các cột không phải số.

## Result Explained — Giải thích kết quả

Kết quả là một bảng các số, thể hiện mức độ liên hệ giữa hai cột.

Giá trị nằm trong khoảng **-1 đến 1**:

- **1** — tương quan **hoàn hảo** (1:1): mỗi khi cột này tăng, cột kia cũng tăng.
- **0.9** — tương quan **mạnh**: khi một cột tăng, cột kia rất có thể cũng tăng.
- **-0.9** — tương quan mạnh nhưng **ngược chiều**: một cột tăng thì cột kia rất có thể giảm.
- **0.2** — tương quan **yếu**: một cột tăng không cho biết gì nhiều về cột kia.

> 💡 Thường coi là tương quan **tốt** khi giá trị tuyệt đối từ khoảng **0.6** trở lên.

## Perfect Correlation — Tương quan hoàn hảo

"Duration" và "Duration" có tương quan `1.000000` — điều hiển nhiên: một cột luôn tương quan hoàn hảo với chính nó.

## Good Correlation — Tương quan tốt

"Duration" và "Calories" có tương quan khoảng **0.98** — rất mạnh. Ta có thể dự đoán: **tập càng lâu, đốt càng nhiều calo** (và ngược lại).

```python
import pandas as pd

df = pd.read_csv('data.csv')
print(round(df["Duration"].corr(df["Calories"]), 3))
```

## Bad Correlation — Tương quan kém

"Duration" và "Maxpulse" có tương quan chỉ khoảng **0.1** — rất yếu. Ta **không** thể dự đoán nhịp tim tối đa chỉ dựa vào thời lượng tập, và ngược lại.

```python
import pandas as pd

df = pd.read_csv('data.csv')
print(round(df["Duration"].corr(df["Maxpulse"]), 3))
```

## ➕ Mở rộng: Trực quan hoá ma trận tương quan

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('data.csv')
sns.heatmap(df.corr(), annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - `df.corr()` → ma trận tương quan giữa các cột số, giá trị **-1 → 1**.
> - `1` hoàn hảo · `~0.9` mạnh · `~0.2` yếu · âm = **ngược chiều**.
> - Tương quan "tốt" khi |giá trị| ≥ ~0.6. Tương quan **không** có nghĩa là nhân quả!
