## Plotting — Vẽ đồ thị

Pandas dùng phương thức `plot()` để vẽ đồ thị.

Dùng `pyplot` — module con của **Matplotlib** — để hiển thị đồ thị.

> Xem thêm chương [Matplotlib](../../05_matplotlib/001_matplotlib_intro.ipynb).

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data.csv')

df.plot()

plt.show()
```

## Scatter Plot — Biểu đồ phân tán

Chỉ định `kind='scatter'`. Biểu đồ scatter cần trục x và trục y.

Ở ví dụ dưới, trục x là "Duration", trục y là "Calories". Vì đã biết hai cột này **tương quan mạnh** (0.98), ta kỳ vọng các điểm nằm gần một đường thẳng:

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data.csv')

df.plot(kind='scatter', x='Duration', y='Calories')

plt.show()
```

So sánh với hai cột **tương quan kém** — "Duration" và "Maxpulse" (~0.1):

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data.csv')

df.plot(kind='scatter', x='Duration', y='Maxpulse')

plt.show()
```

## Histogram

Dùng `kind='hist'`. Histogram chỉ cần **một cột** — nó cho biết **tần suất** của mỗi khoảng giá trị (ví dụ: có bao nhiêu buổi tập kéo dài 50-60 phút).

Ví dụ dùng cột "Duration":

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data.csv')

df["Duration"].plot(kind='hist')

plt.show()
```

> 📌 Histogram cho thấy phần lớn các buổi tập kéo dài khoảng 50-60 phút.

> 🧪 **Try it Yourself** — Vẽ histogram cột Pulse với 15 bins.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('data.csv')
df["Pulse"].plot(kind="hist", bins=15, edgecolor="white")
plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - `df.plot()` vẽ mọi cột số · cần `plt.show()` để hiển thị.
> - `kind='scatter', x=..., y=...` → biểu đồ phân tán (xem tương quan).
> - `df["col"].plot(kind='hist')` → histogram (tần suất).
