# Pandas - Cleaning Data of Wrong Format

## Data of Wrong Format — Dữ liệu sai định dạng

Ô có dữ liệu sai định dạng khiến việc phân tích khó khăn hoặc không thể thực hiện.

Có hai cách: **xoá** các hàng đó, hoặc **chuyển** mọi ô trong cột về cùng một định dạng.

## Convert Into a Correct Format — Chuyển về đúng định dạng

Trong tập dữ liệu, cột "Date" chứa ngày, nhưng có một ô sai định dạng (hàng 26: `20201226`) và một ô trống (hàng 22):

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')
print(df.loc[20:27])
```

Chuyển mọi ô của cột 'Date' thành kiểu ngày bằng `to_datetime()`:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

df['Date'] = pd.to_datetime(df['Date'], format='mixed')

print(df.loc[20:27])
```

> ⚠️ **Lưu ý cập nhật:** từ Pandas 2.0, khi cột chứa **nhiều định dạng ngày khác nhau** (`2020/12/01` và `20201226`), cần thêm `format='mixed'` — nếu không Pandas sẽ báo lỗi.

Kết quả: ngày ở hàng 26 đã được sửa, còn ô trống ở hàng 22 thành `NaT` (*Not a Time* — tức giá trị rỗng).

## Removing Rows — Xoá hàng

`NaT` vẫn là giá trị rỗng. Có thể xoá bằng `dropna()` với tham số `subset`:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')
df['Date'] = pd.to_datetime(df['Date'], format='mixed')

df.dropna(subset=['Date'], inplace=True)

print(df.shape)
print(22 in df.index)
```

---

## 🧠 Ghi nhớ nhanh

> - Sai định dạng → **chuyển** về đúng kiểu, hoặc **xoá** hàng.
> - Ngày: `pd.to_datetime(df['Date'], format='mixed')` (Pandas 2.x khi lẫn nhiều định dạng).
> - Ô không chuyển được → `NaT`; xoá bằng `df.dropna(subset=['Date'])`.
