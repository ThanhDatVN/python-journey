## Wrong Data — Dữ liệu sai

"Dữ liệu sai" không nhất thiết là ô trống hay sai định dạng — nó có thể chỉ là **giá trị không hợp lý**, như ai đó gõ "199" thay vì "1.99".

Đôi khi bạn phát hiện bằng cách nhìn tập dữ liệu, vì bạn có kỳ vọng về giá trị hợp lý.

Trong tập dữ liệu, hàng 7 có thời lượng **450**, trong khi các hàng khác chỉ 45-60:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')
print(df.loc[5:9])
```

Không phải lúc nào cũng là lỗi — nhưng với **thời lượng tập luyện** (tính bằng phút), 450 phút (7,5 giờ) gần như chắc chắn là gõ nhầm từ 45.

## Replacing Values — Thay giá trị

Cách sửa là thay bằng giá trị khác.

Trường hợp này có thể đoán là gõ nhầm, nên ta thay "450" bằng "45":

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

df.loc[7, 'Duration'] = 45

print(df.loc[7])
```

Với tập dữ liệu nhỏ, có thể sửa từng giá trị. Với dữ liệu lớn, nên đặt **quy tắc** — ví dụ: giá trị Duration không được vượt quá 120 — rồi duyệt để sửa:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

for x in df.index:
    if df.loc[x, "Duration"] > 120:
        df.loc[x, "Duration"] = 120

print(df["Duration"].max())
```

## Removing Rows — Xoá hàng

Cách khác là **xoá** các hàng có dữ liệu sai. Như vậy không cần tìm giá trị thay thế, và khả năng cao hàng đó không cần thiết cho việc phân tích:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

for x in df.index:
    if df.loc[x, "Duration"] > 120:
        df.drop(x, inplace=True)

print(df.shape)
```

## ➕ Mở rộng: Cách "vector hoá" không cần vòng lặp

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')
df["Duration"] = df["Duration"].clip(upper=120)     # thay giá trị > 120 bằng 120
clean = df[df["Duration"] <= 120]                   # hoặc lọc bỏ
print(df["Duration"].max(), clean.shape)
```

---

## 🧠 Ghi nhớ nhanh

> - Dữ liệu sai = giá trị **không hợp lý** (gõ nhầm, vượt ngưỡng).
> - Sửa một ô: `df.loc[hàng, 'cột'] = giá_trị`.
> - Đặt quy tắc + duyệt `df.index` để **thay** hoặc **xoá** (`df.drop(x)`).
> - Gọn hơn: `clip()` hoặc lọc boolean `df[df["c"] <= 120]`.
