## Read CSV Files — Đọc file CSV

Cách đơn giản để lưu tập dữ liệu lớn là dùng file **CSV** (*comma separated values* — giá trị ngăn cách bằng dấu phẩy).

File CSV chứa văn bản thuần, là định dạng phổ biến mà mọi chương trình — kể cả Pandas — đều đọc được.

Trong ví dụ, ta dùng file `data.csv`:

```python
import pandas as pd

df = pd.read_csv('data.csv')

print(df.to_string())
```

> 💡 Dùng `to_string()` để in **toàn bộ** DataFrame.

Nếu DataFrame lớn và **không** dùng `to_string()`, `print(df)` chỉ hiện 5 hàng đầu và 5 hàng cuối:

```python
import pandas as pd

df = pd.read_csv('data.csv')

print(df)
```

## max_rows — Số hàng tối đa được hiển thị

Số hàng được hiển thị do thiết lập `pd.options.display.max_rows` quyết định.

Kiểm tra giá trị hiện tại:

```python
import pandas as pd

print(pd.options.display.max_rows)
```

Trên đa số hệ thống, giá trị mặc định là **60**: nếu DataFrame có hơn 60 hàng, `print(df)` chỉ hiện phần đầu và cuối.

Tăng giá trị này để in toàn bộ:

```python
import pandas as pd

pd.options.display.max_rows = 9999

df = pd.read_csv('data.csv')

print(df.shape)
pd.options.display.max_rows = 60   # trả lại mặc định cho các cell sau
```

## ➕ Mở rộng: Các tham số hay dùng của read_csv

```python
import pandas as pd

df = pd.read_csv('data.csv', usecols=["Duration", "Calories"], nrows=5)
print(df)
```

- `sep=";"` — ký tự ngăn cách khác dấu phẩy
- `encoding="utf-8"` — bảng mã (tiếng Việt)
- `usecols=[...]` — chỉ đọc một số cột
- `nrows=n` — chỉ đọc `n` hàng đầu

---

## 🧠 Ghi nhớ nhanh

> - `pd.read_csv('file.csv')` → DataFrame; `df.to_string()` in toàn bộ.
> - `print(df)` với bảng lớn chỉ hiện đầu + cuối (`pd.options.display.max_rows`, mặc định 60).
> - Tham số hữu ích: `sep`, `encoding`, `usecols`, `nrows`.
