## Discovering Duplicates — Phát hiện dữ liệu trùng lặp

Hàng trùng lặp là các hàng được ghi lại **nhiều lần**.

Trong tập dữ liệu, hàng 10 và 11 có vẻ giống hệt nhau:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')
print(df.loc[9:12])
```

Phương thức `duplicated()` trả về giá trị **boolean** cho mỗi hàng — `True` nếu hàng đó là bản trùng của một hàng phía trước:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

print(df.duplicated())
```

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')
print(df[df.duplicated()])
```

## Removing Duplicates — Xoá dữ liệu trùng lặp

Dùng `drop_duplicates()`:

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

df.drop_duplicates(inplace=True)

print(df.shape)
print(11 in df.index)
```

> 📌 `inplace=True` đảm bảo phương thức **không** trả về DataFrame mới mà xoá luôn các hàng trùng trên DataFrame gốc.

## ➕ Mở rộng: Toàn bộ quy trình làm sạch

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')
df = df.drop_duplicates()
df["Date"] = pd.to_datetime(df["Date"], format="mixed")
df = df.dropna(subset=["Date"])
df.loc[df["Duration"] > 120, "Duration"] = 120
df["Calories"] = df["Calories"].fillna(df["Calories"].mean())

df.info()
```

---

## 🧠 Ghi nhớ nhanh

> - `df.duplicated()` → `True` ở các hàng **trùng** với hàng phía trước.
> - `df.drop_duplicates()` → xoá hàng trùng (`inplace=True` để sửa bảng gốc).
> - Quy trình làm sạch: trùng lặp → định dạng → ô trống → giá trị sai.
