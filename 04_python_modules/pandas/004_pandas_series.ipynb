## What is a Series? — Series là gì?

**Series** trong Pandas giống như **một cột** trong bảng.

Nó là mảng một chiều chứa dữ liệu thuộc bất kỳ kiểu nào.

Tạo Series đơn giản từ một list:

```python
import pandas as pd

a = [1, 7, 2]

myvar = pd.Series(a)

print(myvar)
```

## Labels — Nhãn

Nếu không chỉ định gì khác, các giá trị được gán nhãn bằng **chỉ số**: phần tử đầu tiên có chỉ số 0, phần tử thứ hai có chỉ số 1…

Nhãn dùng để truy cập giá trị:

```python
import pandas as pd

a = [1, 7, 2]
myvar = pd.Series(a)

print(myvar[0])
```

## Create Labels — Tự đặt nhãn

Dùng tham số `index` để đặt tên nhãn:

```python
import pandas as pd

a = [1, 7, 2]

myvar = pd.Series(a, index=["x", "y", "z"])

print(myvar)
```

Khi đã đặt nhãn, bạn truy cập phần tử bằng **tên nhãn**:

```python
import pandas as pd

a = [1, 7, 2]
myvar = pd.Series(a, index=["x", "y", "z"])

print(myvar["y"])
```

## Key/Value Objects as Series — Dùng dictionary

Có thể dùng đối tượng key/value như **dictionary** để tạo Series — **key** trở thành **nhãn**:

```python
import pandas as pd

calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories)

print(myvar)
```

Chỉ lấy một số phần tử của dictionary bằng tham số `index`:

```python
import pandas as pd

calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories, index=["day1", "day2"])

print(myvar)
```

## DataFrames

Dữ liệu trong Pandas thường là **bảng nhiều chiều**, gọi là **DataFrame**.

Series giống một **cột**, DataFrame là **cả bảng**.

Tạo DataFrame từ hai Series:

```python
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

myvar = pd.DataFrame(data)

print(myvar)
```

Bạn sẽ học về DataFrame ở bài tiếp theo.

> 🧪 **Try it Yourself** — Tạo Series nhiệt độ 4 ngày với nhãn là tên thứ, rồi in nhiệt độ ngày "T3".

```python
import pandas as pd

temps = pd.Series([31, 33, 30, 34], index=["T2", "T3", "T4", "T5"])
print(temps)
print("T3:", temps["T3"])
```

---

## 🧠 Ghi nhớ nhanh

> - **Series** = một **cột** (mảng 1 chiều có nhãn) · **DataFrame** = cả **bảng**.
> - `pd.Series(list, index=[...])` · từ dict: key → nhãn.
> - Truy cập: `s[0]` (chỉ số mặc định) hoặc `s["nhãn"]`.
