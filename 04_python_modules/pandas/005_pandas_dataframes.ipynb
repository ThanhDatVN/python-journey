## What is a DataFrame? — DataFrame là gì?

**DataFrame** là cấu trúc dữ liệu **hai chiều** — như mảng 2 chiều, hay một **bảng có hàng và cột**.

```python
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

# nạp dữ liệu vào DataFrame:
df = pd.DataFrame(data)

print(df)
```

## Locate Row — Lấy hàng

Như kết quả trên, DataFrame giống một bảng có hàng và cột.

Pandas dùng thuộc tính **`loc`** để lấy **một hoặc nhiều hàng**.

Lấy hàng 0:

```python
import pandas as pd

data = {"calories": [420, 380, 390], "duration": [50, 40, 45]}
df = pd.DataFrame(data)

print(df.loc[0])
```

> 📌 Kết quả là một **Series** (một hàng).

Lấy hàng 0 và 1 bằng **list chỉ số** (chú ý hai cặp ngoặc vuông):

```python
import pandas as pd

data = {"calories": [420, 380, 390], "duration": [50, 40, 45]}
df = pd.DataFrame(data)

print(df.loc[[0, 1]])
```

> 📌 Khi dùng `[[ ]]`, kết quả là một **DataFrame**.

## Named Indexes — Chỉ số có tên

Dùng tham số `index` để đặt tên cho các chỉ số:

```python
import pandas as pd

data = {
    "calories": [420, 380, 390],
    "duration": [50, 40, 45]
}

df = pd.DataFrame(data, index=["day1", "day2", "day3"])

print(df)
```

## Locate Named Indexes — Lấy hàng theo tên chỉ số

```python
import pandas as pd

data = {"calories": [420, 380, 390], "duration": [50, 40, 45]}
df = pd.DataFrame(data, index=["day1", "day2", "day3"])

print(df.loc["day2"])
```

## Load Files Into a DataFrame — Nạp file vào DataFrame

Nếu dữ liệu nằm trong file, Pandas có thể nạp thẳng vào DataFrame:

```python
import pandas as pd

df = pd.read_csv('data.csv')

print(df)
```

Bạn sẽ học thêm về đọc file ở các bài tiếp theo.

## ➕ Mở rộng: Lấy cột và hiển thị bảng đẹp

```python
import pandas as pd

data = {"calories": [420, 380, 390], "duration": [50, 40, 45]}
df = pd.DataFrame(data, index=["day1", "day2", "day3"])

print(df["calories"])          # một cột -> Series
print(df[["duration"]])        # list cột -> DataFrame
df                              # trong Jupyter: hiển thị dạng bảng HTML
```

> 🧪 **Try it Yourself** — Tạo DataFrame bảng điểm 3 học sinh (cột: name, math, english), rồi lấy hàng của học sinh thứ hai.

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["An", "Bình", "Chi"],
    "math": [8, 7, 9],
    "english": [6, 9, 8]
})
print(df)
print(df.loc[1])
```

---

## 🧠 Ghi nhớ nhanh

> - DataFrame = bảng **2 chiều** (hàng × cột): `pd.DataFrame(dict)`.
> - `df.loc[0]` một hàng (Series) · `df.loc[[0, 1]]` nhiều hàng (DataFrame).
> - `index=[...]` đặt tên hàng → `df.loc["day2"]`.
> - Cột: `df["col"]` (Series) · `df[["col"]]` (DataFrame).
