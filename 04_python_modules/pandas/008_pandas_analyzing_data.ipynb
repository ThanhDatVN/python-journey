## Viewing the Data — Xem nhanh dữ liệu

Một trong những phương thức hay dùng nhất để xem nhanh DataFrame là `head()`.

`head()` trả về **tiêu đề** và **số hàng chỉ định**, tính từ đầu bảng.

Lấy 10 hàng đầu:

````python
import pandas as pd

df = pd.read_csv('data.csv')

print(df.head(10))
````

> 📌 Nếu không chỉ định số hàng, `head()` trả về **5** hàng đầu.

````python
import pandas as pd

df = pd.read_csv('data.csv')

print(df.head())
````

Phương thức `tail()` trả về các hàng **cuối cùng**:

````python
import pandas as pd

df = pd.read_csv('data.csv')

print(df.tail())
````

## Info About the Data — Thông tin về dữ liệu

Phương thức `info()` cung cấp thêm thông tin về tập dữ liệu:

````python
import pandas as pd

df = pd.read_csv('data.csv')

df.info()
````

## Result Explained — Giải thích kết quả

Kết quả cho biết:

- Có **169 hàng** và **4 cột**:

```text
RangeIndex: 169 entries, 0 to 168
Data columns (total 4 columns):
```

- Tên mỗi cột và kiểu dữ liệu:

```text
 #   Column    Non-Null Count  Dtype
---  ------    --------------  -----
 0   Duration  169 non-null    int64
 1   Pulse     169 non-null    int64
 2   Maxpulse  169 non-null    int64
 3   Calories  164 non-null    float64
```

## Null Values — Giá trị rỗng

`info()` còn cho biết **mỗi cột có bao nhiêu giá trị không rỗng (Non-Null)**.

Cột "Calories" chỉ có **164 / 169** giá trị — tức là có **5 hàng** không có giá trị Calories.

Giá trị rỗng (Null) có thể làm sai kết quả phân tích. Nên cân nhắc **xoá** các hàng đó — bước này gọi là **làm sạch dữ liệu**, bạn sẽ học ở các bài sau.

## ➕ Mở rộng: describe() và isnull()

````python
import pandas as pd

df = pd.read_csv('data.csv')
print(df.describe().round(1))      # thống kê nhanh: count, mean, std, min, max...
print(df.isnull().sum())           # số giá trị rỗng mỗi cột
````

---

## 🧠 Ghi nhớ nhanh

> - `df.head(n)` n hàng đầu (mặc định 5) · `df.tail(n)` n hàng cuối.
> - `df.info()` → số hàng, tên cột, kiểu dữ liệu, số giá trị **Non-Null**.
> - `df.describe()` thống kê nhanh · `df.isnull().sum()` đếm giá trị rỗng.
