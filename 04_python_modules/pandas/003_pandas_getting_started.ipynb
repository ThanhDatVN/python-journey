## Import Pandas

````python
import pandas

mydataset = {
    'cars': ["BMW", "Volvo", "Ford"],
    'passings': [3, 7, 2]
}

myvar = pandas.DataFrame(mydataset)

print(myvar)
````

## Pandas as pd

Pandas thường được import với bí danh **`pd`**:

````python
import pandas as pd

mydataset = {
    'cars': ["BMW", "Volvo", "Ford"],
    'passings': [3, 7, 2]
}

myvar = pd.DataFrame(mydataset)

print(myvar)
````

## Checking Pandas Version — Kiểm tra phiên bản

````python
import pandas as pd

print(pd.__version__)
````
