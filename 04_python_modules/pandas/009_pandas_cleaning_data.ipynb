## Data Cleaning — Làm sạch dữ liệu

**Làm sạch dữ liệu** là sửa các dữ liệu "xấu" (bad data) trong tập dữ liệu.

Dữ liệu xấu có thể là:

- **Ô trống** (empty cells)
- Dữ liệu **sai định dạng** (wrong format)
- Dữ liệu **sai giá trị** (wrong data)
- Dữ liệu **trùng lặp** (duplicates)

Trong chương này, bạn sẽ học cách xử lý từng loại.

## Our Data Set — Tập dữ liệu thực hành

Trong các bài tiếp theo, ta dùng file `dirtydata.csv` (trên W3Schools file này cũng tên `data.csv`):

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')

print(df.to_string())
```

Tập dữ liệu chứa các ô trống ("Date" ở hàng 22, "Calories" ở hàng 18 và 28), sai định dạng ("Date" ở hàng 26), sai giá trị ("Duration" ở hàng 7), và trùng lặp (hàng 10 và 11).

```python
import pandas as pd

df = pd.read_csv('dirtydata.csv')
print(df.loc[[7, 10, 11, 18, 22, 26, 28]])
```
