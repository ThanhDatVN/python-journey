## Rayleigh Distribution — Phân phối Rayleigh

**Phân phối Rayleigh** được dùng trong **xử lý tín hiệu**.

Hai tham số:

- `scale` — độ lệch chuẩn, quyết định độ "dẹt" của phân phối. Mặc định 1.0.
- `size` — shape của mảng trả về.

Mẫu 2×3 với `scale` 2:

```python
from numpy import random

x = random.rayleigh(scale=2, size=(2, 3))

print(x)
```

## Visualization of Rayleigh Distribution — Trực quan hoá

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.rayleigh(size=1000), kind="kde")

plt.show()
```

## Similarity Between Rayleigh and Chi Square Distribution — Liên hệ với Chi bình phương

Với độ lệch chuẩn bằng 1 và bậc tự do bằng 2, Rayleigh và Chi bình phương biểu diễn **cùng một** phân phối (theo nghĩa: bình phương của biến Rayleigh có phân phối χ² với 2 bậc tự do).

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.kdeplot(random.rayleigh(size=1000) ** 2, label="rayleigh²")
sns.kdeplot(random.chisquare(df=2, size=1000), label="chi-square df=2")
plt.legend()
plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - Rayleigh: dùng trong **xử lý tín hiệu**.
> - `random.rayleigh(scale=..., size=...)`.
> - Rayleigh (scale=1) bình phương ≈ Chi bình phương với `df=2`.
