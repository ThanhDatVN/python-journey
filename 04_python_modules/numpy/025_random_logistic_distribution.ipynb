## Logistic Distribution — Phân phối logistic

**Phân phối logistic** dùng để mô tả **sự tăng trưởng**. Nó được dùng rộng rãi trong học máy: hồi quy logistic, mạng nơ-ron…

Ba tham số:

- `loc` — trung bình, nơi đỉnh nằm. Mặc định 0.
- `scale` — độ lệch chuẩn, độ "dẹt" của phân phối. Mặc định 1.
- `size` — shape của mảng trả về.

Mẫu 2×3 từ phân phối logistic trung bình 1, độ lệch chuẩn 2.0:

```python
from numpy import random

x = random.logistic(loc=1, scale=2, size=(2, 3))

print(x)
```

## Visualization of Logistic Distribution — Trực quan hoá

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.logistic(size=1000), kind="kde")

plt.show()
```

## Difference Between Logistic and Normal Distribution — So với phân phối chuẩn

Hai phân phối gần giống nhau, nhưng logistic có **diện tích ở hai đuôi lớn hơn** — tức là khả năng xuất hiện giá trị ở **xa trung bình** cao hơn so với phân phối chuẩn ("đuôi dày" hơn). Với `scale` phù hợp, hai phân phối gần như trùng khớp (trừ phần đỉnh):

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.kdeplot(random.normal(scale=2, size=1000), label="normal")
sns.kdeplot(random.logistic(size=1000), label="logistic")
plt.legend()
plt.show()
```

## 🧠 Ghi nhớ nhanh

> - Logistic: mô tả sự **tăng trưởng**, dùng nhiều trong học máy (hồi quy logistic, mạng nơ-ron).
> - `random.logistic(loc=..., scale=..., size=...)`.
> - Giống phân phối chuẩn nhưng **đuôi dày hơn**.
