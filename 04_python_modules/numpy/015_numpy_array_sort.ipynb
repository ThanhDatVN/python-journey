## Sorting Arrays — Sắp xếp mảng

**Sắp xếp** là đặt các phần tử theo một **thứ tự**: số, chữ cái, tăng hoặc giảm dần.

Hàm `sort()` sắp xếp mảng:

```python
import numpy as np

arr = np.array([3, 2, 0, 1])

print(np.sort(arr))
```

> 📌 `np.sort()` trả về **bản sao** đã sắp xếp; mảng gốc giữ nguyên.

Sắp xếp chuỗi theo thứ tự chữ cái:

```python
import numpy as np

arr = np.array(['banana', 'cherry', 'apple'])

print(np.sort(arr))
```

Sắp xếp mảng boolean:

```python
import numpy as np

arr = np.array([True, False, True])

print(np.sort(arr))
```

## Sorting a 2-D Array — Sắp xếp mảng 2 chiều

Với mảng 2-D, **mỗi hàng** được sắp xếp riêng:

```python
import numpy as np

arr = np.array([[3, 2, 4], [5, 0, 1]])

print(np.sort(arr))
```

## ➕ Mở rộng: Giảm dần và argsort

```python
import numpy as np

arr = np.array([3, 1, 2])
print(np.sort(arr)[::-1])     # giảm dần
print(np.argsort(arr))        # chỉ số theo thứ tự tăng dần

names = np.array(["An", "Bình", "Chi"])
scores = np.array([7, 9, 8])
print(names[np.argsort(scores)[::-1]])   # xếp hạng theo điểm
```

---

## 🧠 Ghi nhớ nhanh

> - `np.sort(arr)` → **bản sao** đã sắp xếp (số, chuỗi, boolean).
> - Mảng 2-D: sắp xếp **từng hàng**.
> - Giảm dần: `np.sort(arr)[::-1]`; lấy chỉ số thứ tự: `np.argsort(arr)`.
