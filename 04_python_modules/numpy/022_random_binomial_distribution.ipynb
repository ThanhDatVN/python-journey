## Binomial Distribution — Phân phối nhị thức

**Phân phối nhị thức** là **phân phối rời rạc (discrete)**. Nó mô tả kết quả của các tình huống **hai khả năng**, ví dụ tung đồng xu: sấp hoặc ngửa.

Ba tham số:

- `n` — số lần thử.
- `p` — xác suất xảy ra của mỗi lần thử (ví dụ tung đồng xu: 0.5).
- `size` — shape của mảng trả về.

> **Phân phối rời rạc**: chỉ có các giá trị rời rạc, ví dụ tung đồng xu 10 lần thì số lần ngửa chỉ có thể là 0, 1, 2…, 10 — không thể là 2.5.

Tung đồng xu 10 lần, lặp lại 10 thí nghiệm — đếm số lần ngửa mỗi thí nghiệm:

```python
from numpy import random

x = random.binomial(n=10, p=0.5, size=10)

print(x)
```

## Visualization of Binomial Distribution — Trực quan hoá

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(random.binomial(n=10, p=0.5, size=1000), discrete=True)

plt.show()
```

## Difference Between Normal and Binomial Distribution — So sánh với phân phối chuẩn

Khác biệt chính: phân phối chuẩn là **liên tục**, nhị thức là **rời rạc**. Nhưng khi có **đủ nhiều** dữ liệu, hai phân phối khá giống nhau (với `loc` và `scale` phù hợp):

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "normal": random.normal(loc=50, scale=5, size=1000),
    "binomial": random.binomial(n=100, p=0.5, size=1000),
}

sns.kdeplot(data["normal"], label="normal")
sns.kdeplot(data["binomial"], label="binomial")
plt.legend()
plt.show()
```

## 🧠 Ghi nhớ nhanh

> - Nhị thức = **rời rạc**, hai khả năng (thành công/thất bại).
> - `random.binomial(n=số_lần_thử, p=xác_suất, size=...)`.
> - Dữ liệu đủ lớn → gần giống phân phối chuẩn.
