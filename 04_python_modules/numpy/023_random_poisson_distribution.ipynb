## Poisson Distribution — Phân phối Poisson

**Poisson** là phân phối **rời rạc**. Nó ước lượng **số lần một sự kiện xảy ra** trong một khoảng thời gian xác định.

Ví dụ: một người ăn trung bình 2 bữa mỗi ngày → xác suất người đó ăn 3 bữa trong một ngày là bao nhiêu?

Hai tham số:

- `lam` — tốc độ hay số lần xảy ra đã biết (ví dụ: 2).
- `size` — shape của mảng trả về.

Sinh phân phối 1×10 cho sự kiện xảy ra trung bình 2 lần:

```python
from numpy import random

x = random.poisson(lam=2, size=10)

print(x)
```

## Visualization of Poisson Distribution — Trực quan hoá

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(random.poisson(lam=2, size=1000), discrete=True)

plt.show()
```

## Difference Between Normal and Poisson Distribution — So với phân phối chuẩn

Chuẩn là liên tục, Poisson là rời rạc. Nhưng với `lam` đủ lớn, Poisson trông giống phân phối chuẩn có `loc` và `scale` tương ứng:

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.kdeplot(random.normal(loc=50, scale=7, size=1000), label="normal")
sns.kdeplot(random.poisson(lam=50, size=1000), label="poisson")
plt.legend()
plt.show()
```

## Difference Between Binomial and Poisson Distribution — So với phân phối nhị thức

Nhị thức chỉ có **hai** kết quả; Poisson có **không giới hạn** kết quả. Nhưng với `n` rất lớn và `p` gần 0, nhị thức gần giống Poisson với `n * p ≈ lam`:

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.kdeplot(random.binomial(n=1000, p=0.01, size=1000), label="binomial")
sns.kdeplot(random.poisson(lam=10, size=1000), label="poisson")
plt.legend()
plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - Poisson = **rời rạc**, đếm số lần sự kiện xảy ra trong một khoảng thời gian.
> - `random.poisson(lam=trung_bình, size=...)`.
> - `lam` lớn → giống phân phối chuẩn; nhị thức (`n` lớn, `p` nhỏ) ≈ Poisson với `lam = n*p`.
