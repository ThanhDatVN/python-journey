## What are ufuncs? — ufunc là gì?

**ufunc** là viết tắt của *Universal Functions* — các **hàm toàn cục** của NumPy, hoạt động trên đối tượng `ndarray`.

## Why use ufuncs? — Vì sao dùng ufunc?

ufunc thực hiện **vector hoá (vectorization)** trong NumPy — nhanh hơn nhiều so với duyệt từng phần tử.

ufunc còn hỗ trợ **broadcasting** và các phương thức bổ sung như `reduce`, `accumulate`… rất hữu ích khi tính toán.

ufunc nhận thêm các tham số:

- `where` — mảng boolean hoặc điều kiện xác định nơi phép toán được thực hiện.
- `dtype` — kiểu dữ liệu trả về.
- `out` — mảng đầu ra nơi kết quả được ghi vào.

## What is Vectorization? — Vector hoá là gì?

Chuyển các câu lệnh lặp thành **phép toán trên vector** gọi là vector hoá. CPU hiện đại được tối ưu cho các phép toán này nên chúng chạy rất nhanh.

Cộng các phần tử của hai list:

```text
list 1: [1, 2, 3, 4]
list 2: [4, 5, 6, 7]
```

Một cách là duyệt cả hai list và cộng từng cặp. Không dùng ufunc, ta dùng hàm `zip()` có sẵn của Python:

````python
x = [1, 2, 3, 4]
y = [4, 5, 6, 7]
z = []

for i, j in zip(x, y):
    z.append(i + j)

print(z)
````

NumPy có ufunc `add(x, y)` cho kết quả tương tự:

````python
import numpy as np

x = [1, 2, 3, 4]
y = [4, 5, 6, 7]
z = np.add(x, y)

print(z)
````

## ➕ Mở rộng: Toán tử trên mảng cũng là ufunc

````python
import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([4, 5, 6, 7])
print(a + b)      # gọi np.add
print(a * 10)     # broadcasting: 10 được "trải" cho mọi phần tử
result = np.zeros(4, dtype=int)
np.add(a, b, out=result, where=a > 2)   # chỉ tính ở vị trí thoả điều kiện
print(result)
````

---

## 🧠 Ghi nhớ nhanh

> - ufunc = **Universal Function**, hàm NumPy chạy trên **cả mảng** (vector hoá) → rất nhanh.
> - Tham số thêm: `where`, `dtype`, `out`.
> - `np.add(x, y)` ⇔ `x + y` với mảng NumPy.
