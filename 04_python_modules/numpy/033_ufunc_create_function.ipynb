## How To Create Your Own ufunc — Tự tạo ufunc

Để tạo ufunc riêng, bạn định nghĩa một hàm Python bình thường, rồi đưa nó vào thư viện ufunc của NumPy bằng **`frompyfunc()`**.

`frompyfunc()` nhận các tham số:

- `function` — tên hàm.
- `inputs` — số đối số đầu vào (số mảng).
- `outputs` — số mảng đầu ra.

Tạo ufunc cộng riêng:

```python
import numpy as np

def myadd(x, y):
    return x + y

myadd = np.frompyfunc(myadd, 2, 1)

print(myadd([1, 2, 3, 4], [5, 6, 7, 8]))
```

## Check if a Function is a ufunc — Kiểm tra hàm có phải ufunc

Kiểm tra **kiểu** của hàm: ufunc trả về `<class 'numpy.ufunc'>`:

```python
import numpy as np

print(type(np.add))
```

Hàm không phải ufunc sẽ trả về kiểu khác, ví dụ hàm có sẵn `concatenate()`:

```python
import numpy as np

print(type(np.concatenate))
```

Hàm không tồn tại sẽ báo lỗi:

```python
import numpy as np

print(type(np.blahblah))
```

Dùng câu lệnh `if` để kiểm tra:

```python
import numpy as np

if type(np.add) == np.ufunc:
    print('add is ufunc')
else:
    print('add is not ufunc')
```

## ➕ Mở rộng: np.vectorize

`np.vectorize` là cách khác để "vector hoá" một hàm Python, cho phép chỉ định kiểu trả về:

```python
import numpy as np

def grade(score):
    return "A" if score >= 8 else "B" if score >= 6.5 else "C"

vgrade = np.vectorize(grade)
print(vgrade(np.array([9, 7, 5.5])))
```

> - Cách khác: `np.vectorize(hàm)`.
