## Finding GCD (Greatest Common Denominator) — Ước chung lớn nhất

**Ước chung lớn nhất (ƯCLN, GCD)** là số lớn nhất mà tất cả các số đã cho đều chia hết:

```python
import numpy as np

num1 = 6
num2 = 9

x = np.gcd(num1, num2)

print(x)
```

Kết quả `3` vì đó là số lớn nhất mà cả 6 và 9 đều chia hết (6/3 = 2, 9/3 = 3).

## Finding GCD in Arrays — ƯCLN của cả mảng

```python
import numpy as np

arr = np.array([20, 8, 32, 36, 16])

x = np.gcd.reduce(arr)

print(x)
```

Kết quả `4` vì đó là số lớn nhất mà mọi phần tử đều chia hết.

> 🧪 **Try it Yourself** — Rút gọn phân số 84/126 bằng ƯCLN.

```python
import numpy as np

a, b = 84, 126
g = np.gcd(a, b)
print(f"{a}/{b} = {a // g}/{b // g}")
```

---

## 🧠 Ghi nhớ nhanh

> - `np.gcd(a, b)` → ước chung lớn nhất.
> - `np.gcd.reduce(arr)` → ƯCLN của **cả mảng**.
