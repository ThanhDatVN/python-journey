## Chi Square Distribution — Phân phối Chi bình phương

**Phân phối Chi bình phương (χ²)** được dùng làm cơ sở để **kiểm định giả thuyết** thống kê.

Hai tham số:

- `df` — **bậc tự do (degree of freedom)**.
- `size` — shape của mảng trả về.

Mẫu 2×3 phân phối chi bình phương với bậc tự do 2:

```python
from numpy import random

x = random.chisquare(df=2, size=(2, 3))

print(x)
```

## Visualization of Chi Square Distribution — Trực quan hoá

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.chisquare(df=1, size=1000), kind="kde")

plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - Chi bình phương: nền tảng để **kiểm định giả thuyết**.
> - `random.chisquare(df=bậc_tự_do, size=...)`.
