## Uniform Distribution — Phân phối đều

**Phân phối đều** mô tả xác suất khi **mọi giá trị có cơ hội như nhau**. Thường dùng để sinh số ngẫu nhiên.

Ba tham số:

- `low` — cận dưới, mặc định `0.0`.
- `high` — cận trên, mặc định `1.0`.
- `size` — shape của mảng trả về.

Mẫu phân phối đều 2×3:

```python
from numpy import random

x = random.uniform(size=(2, 3))

print(x)
```

## Visualization of Uniform Distribution — Trực quan hoá

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.uniform(size=1000), kind="kde")

plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - Phân phối đều: mọi giá trị trong `[low, high)` có **xác suất như nhau**.
> - `random.uniform(low=0, high=1, size=...)`.
