## Import NumPy

````python
import numpy

arr = numpy.array([1, 2, 3, 4, 5])
print(arr)
````

## NumPy as np

NumPy thường được import với bí danh **`np`**:

````python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr)
````

## Checking NumPy Version — Kiểm tra phiên bản

````python
import numpy as np

print(np.__version__)
````
