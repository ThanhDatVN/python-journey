## Differences — Hiệu liên tiếp

**Hiệu rời rạc** là lấy phần tử sau trừ phần tử trước.

Ví dụ với `[1, 2, 3, 4]`, hiệu rời rạc là `[2-1, 3-2, 4-3]` = `[1, 1, 1]`.

Dùng hàm `diff()`:

```python
import numpy as np

arr = np.array([10, 15, 25, 5])

newarr = np.diff(arr)

print(newarr)
```

Kết quả: `[5 10 -20]` vì 15-10=5, 25-15=10, 5-25=-20.

Có thể lặp lại phép tính nhiều lần với tham số `n`.

Ví dụ với `[1, 2, 3, 4]`, `n=2` cho `[2-1, 3-2, 4-3]` = `[1, 1, 1]`, rồi lấy hiệu lần nữa: `[1-1, 1-1]` = `[0, 0]`:

```python
import numpy as np

arr = np.array([10, 15, 25, 5])

newarr = np.diff(arr, n=2)

print(newarr)
```

Kết quả: `[5 -30]` vì: 15-10=5, 25-15=10, 5-25=-20 → rồi 10-5=5, -20-10=-30.

> 🧪 **Try it Yourself** — Tính mức tăng/giảm doanh thu giữa các tháng liên tiếp.

```python
import numpy as np

revenue = np.array([120, 135, 128, 150, 170])
print(np.diff(revenue))
```

---

## 🧠 Ghi nhớ nhanh

> - `np.diff(arr)` → hiệu các cặp phần tử liên tiếp (`a[i+1] - a[i]`), kết quả ngắn hơn 1 phần tử.
> - `n=k` → lặp lại phép lấy hiệu `k` lần.
