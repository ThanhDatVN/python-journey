## Filtering Arrays — Lọc mảng

**Lọc** là lấy ra một số phần tử từ mảng có sẵn để tạo mảng mới.

Trong NumPy, ta lọc bằng **mảng boolean** (boolean index list): `True` → giữ, `False` → bỏ.

```python
import numpy as np

arr = np.array([41, 42, 43, 44])

x = [True, False, True, False]

newarr = arr[x]

print(newarr)
```

Ví dụ trên trả về `[41, 43]` vì chỉ số 0 và 2 là `True`.

## Creating the Filter Array — Tạo mảng lọc

Tạo mảng lọc chỉ giữ giá trị **lớn hơn 42**:

```python
import numpy as np

arr = np.array([41, 42, 43, 44])

filter_arr = []

for element in arr:
    if element > 42:
        filter_arr.append(True)
    else:
        filter_arr.append(False)

newarr = arr[filter_arr]

print(filter_arr)
print(newarr)
```

Chỉ giữ số **chẵn**:

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6, 7])

filter_arr = []

for element in arr:
    if element % 2 == 0:
        filter_arr.append(True)
    else:
        filter_arr.append(False)

newarr = arr[filter_arr]

print(filter_arr)
print(newarr)
```

## Creating Filter Directly From Array — Tạo bộ lọc trực tiếp

Cách trên khá dài. NumPy cho phép **dùng thẳng điều kiện** trên mảng để tạo mảng lọc:

```python
import numpy as np

arr = np.array([41, 42, 43, 44])

filter_arr = arr > 42

newarr = arr[filter_arr]

print(filter_arr)
print(newarr)
```

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6, 7])

filter_arr = arr % 2 == 0

newarr = arr[filter_arr]

print(filter_arr)
print(newarr)
```

## ➕ Mở rộng: Kết hợp nhiều điều kiện

Dùng `&` (và), `|` (hoặc), `~` (phủ định) — mỗi điều kiện **đặt trong ngoặc**:

```python
import numpy as np

arr = np.arange(1, 21)
print(arr[(arr > 5) & (arr % 3 == 0)])
print(arr[(arr < 3) | (arr > 18)])
```

> 🧪 **Try it Yourself** — Lọc ra các điểm thi từ 8 trở lên.

```python
import numpy as np

scores = np.array([6.5, 8, 9.5, 7, 8.5, 5])
print(scores[scores >= 8])
```

## 🧠 Ghi nhớ nhanh

> - Lọc bằng mảng boolean: `arr[[True, False, ...]]`.
> - Gọn nhất: `arr[arr > 42]`, `arr[arr % 2 == 0]`.
> - Nhiều điều kiện: `arr[(cond1) & (cond2)]`, `|`, `~` — nhớ **ngoặc**.
