## Zipf distributions — Phân phối Zipf

Phân phối Zipf dùng để lấy mẫu dữ liệu theo **định luật Zipf**.

> **Định luật Zipf**: trong một tập hợp, phần tử phổ biến thứ *n* xuất hiện với tần suất bằng **1/n** lần phần tử phổ biến nhất. Ví dụ: từ phổ biến thứ 5 trong tiếng Anh xuất hiện khoảng 1/5 lần so với từ phổ biến nhất.

Hai tham số:

- `a` — tham số phân phối.
- `size` — shape của mảng trả về.

Mẫu 2×3 với tham số 2:

```python
from numpy import random

x = random.zipf(a=2, size=(2, 3))

print(x)
```

## Visualization of Zipf Distribution — Trực quan hoá

Lấy mẫu 1000 điểm nhưng chỉ vẽ những điểm nhỏ hơn 10 cho dễ nhìn:

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

x = random.zipf(a=2, size=1000)
sns.histplot(x[x < 10], discrete=True)

plt.show()
```
