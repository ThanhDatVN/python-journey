# NumPy Joining Array

> 📖 **Học song song trên W3Schools:** [numpy_array_join.asp](https://www.w3schools.com/python/numpy/numpy_array_join.asp)
>

## Joining NumPy Arrays — Nối mảng

**Nối (join)** là đặt nội dung của hai hay nhiều mảng vào một mảng.

Trong SQL, ta nối bảng theo khoá; trong NumPy, ta nối mảng theo **trục (axis)**.

Truyền một **tuple** các mảng vào hàm `concatenate()`, kèm trục. Nếu không truyền `axis`, mặc định là `0`:

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.concatenate((arr1, arr2))

print(arr)
```

Nối hai mảng 2-D theo **hàng** (`axis=1`):

```python
import numpy as np

arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

arr = np.concatenate((arr1, arr2), axis=1)

print(arr)
```

## Joining Arrays Using Stack Functions — Nối bằng stack

**Stack** giống concatenate, chỉ khác là stack nối theo một **trục mới**:

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.stack((arr1, arr2), axis=1)

print(arr)
```

## Stacking Along Rows — Xếp theo hàng

`hstack()` xếp nối tiếp theo hàng (ngang):

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.hstack((arr1, arr2))

print(arr)
```

## Stacking Along Columns — Xếp theo cột

`vstack()` xếp chồng theo cột (dọc):

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.vstack((arr1, arr2))

print(arr)
```

## Stacking Along Height (depth) — Xếp theo chiều sâu

`dstack()` xếp theo chiều sâu:

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

arr = np.dstack((arr1, arr2))

print(arr)
```

---

## 🧠 Ghi nhớ nhanh

> - `np.concatenate((a, b), axis=...)` — nối theo trục **có sẵn** (mặc định `axis=0`).
> - `np.stack((a, b), axis=...)` — nối theo trục **mới**.
> - `hstack` ngang · `vstack` dọc · `dstack` theo chiều sâu.
