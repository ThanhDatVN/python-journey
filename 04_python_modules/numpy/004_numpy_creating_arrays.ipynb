## Create a NumPy ndarray Object — Tạo mảng

Đối tượng mảng trong NumPy gọi là `ndarray`. Tạo bằng hàm `array()`:

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print(arr)
print(type(arr))
```

`array()` nhận list, tuple hoặc bất kỳ đối tượng giống mảng nào:

```python
import numpy as np

arr = np.array((1, 2, 3, 4, 5))
print(arr)
```

## Dimensions in Arrays — Số chiều

Một **chiều (dimension)** là một tầng lồng nhau của mảng. Mảng lồng nhau = mảng chứa mảng.

## 0-D Arrays — Mảng 0 chiều

Mảng 0 chiều (*scalar*) là mỗi phần tử đơn lẻ trong mảng:

```python
import numpy as np

arr = np.array(42)
print(arr)
```

## 1-D Arrays — Mảng 1 chiều

Mảng chứa các phần tử 0-D gọi là mảng 1 chiều — loại mảng phổ biến nhất:

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr)
```

## 2-D Arrays — Mảng 2 chiều

Mảng chứa các mảng 1-D gọi là mảng 2 chiều, thường dùng để biểu diễn **ma trận**:

```python
import numpy as np

arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
```

## 3-D arrays — Mảng 3 chiều

Mảng chứa các mảng 2-D:

```python
import numpy as np

arr = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])
print(arr)
```

## Check Number of Dimensions? — Kiểm tra số chiều

Thuộc tính `ndim` cho biết mảng có bao nhiêu chiều:

```python
import numpy as np

a = np.array(42)
b = np.array([1, 2, 3, 4, 5])
c = np.array([[1, 2, 3], [4, 5, 6]])
d = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

## Higher Dimensional Arrays — Mảng nhiều chiều hơn

Mảng có thể có bao nhiêu chiều cũng được. Dùng tham số `ndmin` để chỉ định số chiều tối thiểu:

```python
import numpy as np

arr = np.array([1, 2, 3, 4], ndmin=5)

print(arr)
print('number of dimensions :', arr.ndim)
```

## ➕ Mở rộng: Các hàm tạo mảng nhanh

```python
import numpy as np

print(np.zeros(5))
print(np.ones((2, 3)))
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
print(np.full((2, 2), 7))
print(np.eye(3))
```

> 🧪 **Try it Yourself** — Tạo mảng 2 chiều biểu diễn bảng điểm 3 học sinh × 4 môn, in ra số chiều.

```python
import numpy as np

scores = np.array([[8, 7, 9, 6], [5, 8, 7, 9], [9, 9, 8, 10]])
print(scores)
print("Số chiều:", scores.ndim)
```

---

## 🧠 Ghi nhớ nhanh

> - Tạo mảng: `np.array(list_hoặc_tuple)` → kiểu `ndarray`.
> - 0-D scalar · 1-D vector · 2-D ma trận · 3-D…; xem số chiều: `arr.ndim`.
> - `ndmin=n` → ép số chiều tối thiểu.
> - Tạo nhanh: `zeros ones arange linspace full eye`.
