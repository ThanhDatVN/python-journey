## Simple Arithmetic — Số học cơ bản

Có thể dùng trực tiếp các toán tử `+ - * /` giữa các mảng NumPy. Phần này giới thiệu các **hàm** tương ứng, nhận được mọi đối tượng giống mảng (list, tuple…) và hỗ trợ **tham số điều kiện** `where`.

## Addition — Phép cộng

```python
import numpy as np

arr1 = np.array([10, 11, 12, 13, 14, 15])
arr2 = np.array([20, 21, 22, 23, 24, 25])

newarr = np.add(arr1, arr2)

print(newarr)
```

## Subtraction — Phép trừ

```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50, 60])
arr2 = np.array([20, 21, 22, 23, 24, 25])

newarr = np.subtract(arr1, arr2)

print(newarr)
```

## Multiplication — Phép nhân

```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50, 60])
arr2 = np.array([20, 21, 22, 23, 24, 25])

newarr = np.multiply(arr1, arr2)

print(newarr)
```

## Division — Phép chia

```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50, 60])
arr2 = np.array([3, 5, 10, 8, 2, 33])

newarr = np.divide(arr1, arr2)

print(newarr)
```

## Power — Luỹ thừa

```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50, 60])
arr2 = np.array([3, 5, 6, 8, 2, 33])

newarr = np.power(arr1, arr2)

print(newarr)
```

> ⚠️ Kết quả cuối cùng bị **tràn số** (overflow) vì 60³³ vượt quá giới hạn của kiểu `int64` — một điểm khác biệt quan trọng giữa NumPy và số nguyên Python (không giới hạn).

## Remainder — Phần dư

`mod()` và `remainder()` đều trả về phần dư:

```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50, 60])
arr2 = np.array([3, 7, 9, 8, 2, 33])

print(np.mod(arr1, arr2))
print(np.remainder(arr1, arr2))
```

## Quotient and Mod — Thương và dư

`divmod()` trả về **hai mảng**: thương và phần dư:

```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50, 60])
arr2 = np.array([3, 7, 9, 8, 2, 33])

newarr = np.divmod(arr1, arr2)

print(newarr)
```

## Absolute Values — Giá trị tuyệt đối

`absolute()` và `abs()` cho cùng kết quả. Nên dùng `absolute()` để tránh nhầm với hàm `abs()` có sẵn của Python:

```python
import numpy as np

arr = np.array([-1, -2, 1, 2, 3, -4])

newarr = np.absolute(arr)

print(newarr)
```

## 🧠 Ghi nhớ nhanh

> - `add subtract multiply divide power` ⇔ `+ - * / **`.
> - Phần dư: `mod`/`remainder` · thương và dư: `divmod` · trị tuyệt đối: `absolute`.
> - ⚠️ Số nguyên NumPy có giới hạn → có thể **tràn số**.
