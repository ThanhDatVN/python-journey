## Multinomial Distribution — Phân phối đa thức

**Phân phối đa thức** là dạng tổng quát của phân phối nhị thức.

Nó mô tả kết quả của các tình huống **nhiều khả năng** — khác với nhị thức chỉ có hai. Ví dụ: nhóm máu, kết quả tung xúc xắc.

Ba tham số:

- `n` — số lần thử.
- `pvals` — danh sách xác suất của từng kết quả (ví dụ xúc xắc: `[1/6] * 6`).
- `size` — shape của mảng trả về.

Tung xúc xắc 6 lần, đếm số lần ra mỗi mặt:

```python
from numpy import random

x = random.multinomial(n=6, pvals=[1/6, 1/6, 1/6, 1/6, 1/6, 1/6])

print(x)
```

> 📌 Mẫu đa thức **không** sinh ra một giá trị đơn lẻ — nó sinh ra **một giá trị cho mỗi** phần tử của `pvals`.
>
> 📌 Vì là dạng tổng quát của nhị thức, biểu đồ và độ tương đồng với phân phối chuẩn giống như nhị thức khi có nhiều biến.

```python
from numpy import random

rolls = random.multinomial(n=600, pvals=[1/6] * 6, size=3)
print(rolls)
print(rolls.sum(axis=1))   # mỗi hàng luôn có tổng = n
```

---

## 🧠 Ghi nhớ nhanh

> - Đa thức = nhị thức tổng quát, **nhiều** kết quả có thể.
> - `random.multinomial(n=..., pvals=[...], size=...)` → mỗi lần trả về **số lần đếm cho từng kết quả**.
