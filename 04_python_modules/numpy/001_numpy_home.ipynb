## Learn NumPy — Học NumPy

**NumPy** là thư viện Python dùng để làm việc với **mảng (array)** số.

Đây là nền tảng của gần như mọi thư viện khoa học dữ liệu trong Python: Pandas, SciPy, Matplotlib, scikit-learn… đều xây dựng trên NumPy.

## Learning by Examples — Học qua ví dụ

```python
import numpy

arr = numpy.array([1, 2, 3, 4, 5])

print(arr)
```
