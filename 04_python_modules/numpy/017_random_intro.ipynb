## What is a Random Number? — Số ngẫu nhiên là gì?

Số ngẫu nhiên **không** có nghĩa là mỗi lần một số khác nhau. Ngẫu nhiên nghĩa là thứ **không thể dự đoán một cách logic**.

## Pseudo Random and True Random — Giả ngẫu nhiên và ngẫu nhiên thật

Máy tính chạy theo chương trình — một tập lệnh xác định. Nên để sinh số ngẫu nhiên phải có một **thuật toán**. Nếu có thuật toán thì số đó **dự đoán được** → gọi là **giả ngẫu nhiên (pseudo random)**.

Muốn có số **ngẫu nhiên thật**, máy tính phải lấy dữ liệu từ bên ngoài: thao tác bàn phím, chuột, dữ liệu mạng…

Thông thường ta **không cần** ngẫu nhiên thật — trừ khi liên quan tới bảo mật (khoá mã hoá…). Trong bài học này, ta dùng số giả ngẫu nhiên.

## Generate Random Number — Sinh số nguyên ngẫu nhiên

NumPy có module `random` để làm việc với số ngẫu nhiên.

Sinh một số nguyên ngẫu nhiên từ 0 đến 100:

```python
from numpy import random

x = random.randint(100)

print(x)
```

## Generate Random Float — Sinh số thực ngẫu nhiên

`rand()` trả về số thực ngẫu nhiên trong khoảng 0 đến 1:

```python
from numpy import random

x = random.rand()

print(x)
```

## Generate Random Array — Sinh mảng ngẫu nhiên

### Integers — Số nguyên

`randint()` nhận tham số `size` để chỉ định shape của mảng.

Mảng 1-D gồm 5 số nguyên ngẫu nhiên từ 0 đến 100:

```python
from numpy import random

x = random.randint(100, size=(5))

print(x)
```

Mảng 2-D gồm 3 hàng, mỗi hàng 5 số:

```python
from numpy import random

x = random.randint(100, size=(3, 5))

print(x)
```

### Floats — Số thực

`rand()` nhận kích thước các chiều làm đối số:

```python
from numpy import random

x = random.rand(5)
print(x)

y = random.rand(3, 5)
print(y)
```

## Generate Random Number From Array — Chọn ngẫu nhiên từ mảng

`choice()` chọn ngẫu nhiên một giá trị từ mảng cho trước:

```python
from numpy import random

x = random.choice([3, 5, 7, 9])

print(x)
```

Thêm `size` để tạo mảng gồm các giá trị chọn từ mảng cho trước:

```python
from numpy import random

x = random.choice([3, 5, 7, 9], size=(3, 5))

print(x)
```

## ➕ Mở rộng: seed — kết quả có thể lặp lại

Đặt **seed** để mỗi lần chạy cho cùng một dãy số "ngẫu nhiên" (rất cần khi thí nghiệm, kiểm thử):

```python
import numpy as np

np.random.seed(42)
print(np.random.randint(100, size=5))

np.random.seed(42)
print(np.random.randint(100, size=5))   # giống hệt dòng trên

rng = np.random.default_rng(seed=42)    # API hiện đại được khuyến nghị
print(rng.integers(100, size=5))
```

> 🧪 **Try it Yourself** — Mô phỏng tung xúc xắc 10 lần (giá trị từ 1 đến 6).

```python
from numpy import random

print(random.randint(1, 7, size=10))
```

---

## 🧠 Ghi nhớ nhanh

> - Máy tính sinh số **giả ngẫu nhiên** (pseudo random) bằng thuật toán.
> - `random.randint(n, size=...)` số nguyên · `random.rand(...)` số thực 0–1 · `random.choice(arr, size=...)` chọn từ mảng.
> - `np.random.seed(n)` → kết quả lặp lại được.
