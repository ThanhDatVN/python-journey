## Exponential Distribution — Phân phối mũ

**Phân phối mũ** mô tả **thời gian cho tới sự kiện kế tiếp**, ví dụ: thời gian tới lần hỏng hóc, thời gian tới lần thành công…

Hai tham số:

- `scale` — nghịch đảo của tốc độ (xem `lam` ở Poisson). Mặc định 1.0.
- `size` — shape của mảng trả về.

Mẫu 2×3 phân phối mũ với `scale` 2.0:

```python
from numpy import random

x = random.exponential(scale=2, size=(2, 3))

print(x)
```

## Visualization of Exponential Distribution — Trực quan hoá

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.exponential(size=1000), kind="kde")

plt.show()
```

## Relation Between Poisson and Exponential Distribution — Liên hệ với Poisson

- Poisson trả lời: **bao nhiêu lần** sự kiện xảy ra trong một khoảng thời gian.
- Phân phối mũ trả lời: **bao lâu** giữa hai lần xảy ra sự kiện.

---

## 🧠 Ghi nhớ nhanh

> - Phân phối mũ: **thời gian chờ** tới sự kiện kế tiếp.
> - `random.exponential(scale=..., size=...)` — `scale` = 1 / tốc độ.
> - Poisson = "bao nhiêu lần" · Mũ = "bao lâu giữa các lần".
