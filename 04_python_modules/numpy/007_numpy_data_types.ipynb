## Data Types in Python — Kiểu dữ liệu trong Python

Mặc định Python có các kiểu: `str` (chuỗi), `int` (số nguyên), `float` (số thực), `bool` (logic), `complex` (số phức).

## Data Types in NumPy — Kiểu dữ liệu trong NumPy

NumPy có thêm nhiều kiểu, ký hiệu bằng **một ký tự**:

| Ký tự | Kiểu |
|---|---|
| `i` | integer — số nguyên |
| `b` | boolean |
| `u` | unsigned integer — số nguyên không dấu |
| `f` | float — số thực |
| `c` | complex float — số phức |
| `m` | timedelta — khoảng thời gian |
| `M` | datetime — ngày giờ |
| `O` | object |
| `S` | string (bytes) |
| `U` | unicode string |
| `V` | void — khối bộ nhớ cố định |

## Checking the Data Type of an Array — Xem kiểu dữ liệu

Thuộc tính `dtype` trả về kiểu dữ liệu của mảng:

```python
import numpy as np

arr = np.array([1, 2, 3, 4])
print(arr.dtype)
```

```python
import numpy as np

arr = np.array(['apple', 'banana', 'cherry'])
print(arr.dtype)
```

## Creating Arrays With a Defined Data Type — Chỉ định kiểu khi tạo

Dùng tham số `dtype`:

```python
import numpy as np

arr = np.array([1, 2, 3, 4], dtype='S')

print(arr)
print(arr.dtype)
```

Với `i`, `u`, `f`, `S`, `U` có thể chỉ định thêm **kích thước** (byte):

```python
import numpy as np

arr = np.array([1, 2, 3, 4], dtype='i4')

print(arr)
print(arr.dtype)
```

## What if a Value Can Not Be Converted? — Không chuyển đổi được thì sao?

Nếu giá trị không thể ép sang kiểu chỉ định, NumPy báo lỗi `ValueError`:

```python
import numpy as np

arr = np.array(['a', '2', '3'], dtype='i')
```

## Converting Data Type on Existing Arrays — Đổi kiểu mảng có sẵn

Cách tốt nhất là tạo **bản sao** với phương thức `astype()`:

```python
import numpy as np

arr = np.array([1.1, 2.1, 3.1])

newarr = arr.astype('i')

print(newarr)
print(newarr.dtype)
```

Có thể truyền kiểu Python thay vì ký tự:

```python
import numpy as np

arr = np.array([1.1, 2.1, 3.1])
newarr = arr.astype(int)

print(newarr)
print(newarr.dtype)
```

Chuyển số nguyên sang boolean:

```python
import numpy as np

arr = np.array([1, 0, 3])
newarr = arr.astype(bool)

print(newarr)
print(newarr.dtype)
```

## 🧠 Ghi nhớ nhanh

> - Xem kiểu: `arr.dtype` · chỉ định khi tạo: `np.array(..., dtype='i4')`.
> - Ký tự: `i` int · `u` uint · `f` float · `b` bool · `S`/`U` chuỗi · `M` datetime.
> - Đổi kiểu: `arr.astype('i')` / `arr.astype(int)` → tạo **bản sao**.
> - ⚠️ Không chuyển được → `ValueError`.
