## What is a Set — Tập hợp là gì?

**Tập hợp (set)** trong toán học là một nhóm các phần tử **duy nhất** (không trùng lặp).

Tập hợp dùng cho các phép toán giao, hợp, hiệu.

## Create Sets in NumPy — Tạo tập hợp

Dùng `unique()` để tìm các phần tử duy nhất của một mảng, ví dụ chuyển mảng có phần tử lặp thành tập hợp:

```python
import numpy as np

arr = np.array([1, 1, 1, 2, 3, 4, 5, 5, 6, 7])

x = np.unique(arr)

print(x)
```

## Finding Union — Hợp

Dùng `union1d()`:

```python
import numpy as np

arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([3, 4, 5, 6])

newarr = np.union1d(arr1, arr2)

print(newarr)
```

## Finding Intersection — Giao

Dùng `intersect1d()`:

```python
import numpy as np

arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([3, 4, 5, 6])

newarr = np.intersect1d(arr1, arr2, assume_unique=True)

print(newarr)
```

> 📌 Tham số tuỳ chọn `assume_unique=True` giúp tính nhanh hơn — **chỉ** đặt khi chắc chắn các mảng đã là tập hợp (không trùng lặp).

## Finding Difference — Hiệu

Tìm các phần tử có trong tập thứ nhất mà **không** có trong tập thứ hai, dùng `setdiff1d()`:

```python
import numpy as np

set1 = np.array([1, 2, 3, 4])
set2 = np.array([3, 4, 5, 6])

newarr = np.setdiff1d(set1, set2, assume_unique=True)

print(newarr)
```

## Finding Symmetric Difference — Hiệu đối xứng

Tìm các phần tử **không** có mặt ở **cả hai** tập, dùng `setxor1d()`:

```python
import numpy as np

set1 = np.array([1, 2, 3, 4])
set2 = np.array([3, 4, 5, 6])

newarr = np.setxor1d(set1, set2, assume_unique=True)

print(newarr)
```

---

## 🧠 Ghi nhớ nhanh

> - `np.unique()` → loại trùng, tạo tập hợp (đã sắp xếp).
> - Hợp `union1d` · Giao `intersect1d` · Hiệu `setdiff1d` · Hiệu đối xứng `setxor1d`.
> - `assume_unique=True` → nhanh hơn khi chắc chắn không có phần tử trùng.
