# NumPy LCM Lowest Common Multiple

> 📖 **Học song song trên W3Schools:** [numpy_ufunc_lcm.asp](https://www.w3schools.com/python/numpy/numpy_ufunc_lcm.asp)
>
> 🎯 **Trong bài này:** Finding LCM (Lowest Common Multiple) · Finding LCM in Arrays

## Finding LCM (Lowest Common Multiple) — Bội chung nhỏ nhất

**Bội chung nhỏ nhất (BCNN, LCM)** là số nhỏ nhất chia hết cho tất cả các số đã cho:

```python
import numpy as np

num1 = 4
num2 = 6

x = np.lcm(num1, num2)

print(x)
```

Kết quả `12` vì đó là bội chung nhỏ nhất của 4 và 6 (4×3 = 6×2 = 12).

## Finding LCM in Arrays — BCNN của cả mảng

Dùng phương thức `reduce()` — nó áp dụng ufunc (ở đây là `lcm()`) lên **từng phần tử**, giảm mảng đi một chiều:

```python
import numpy as np

arr = np.array([3, 6, 9])

x = np.lcm.reduce(arr)

print(x)
```

Kết quả `18` vì đó là BCNN của cả ba số (3×6 = 6×3 = 9×2 = 18).

BCNN của mọi số từ 1 đến 10:

```python
import numpy as np

arr = np.arange(1, 11)

x = np.lcm.reduce(arr)

print(x)
```

---

## 🧠 Ghi nhớ nhanh

> - `np.lcm(a, b)` → bội chung nhỏ nhất của hai số.
> - `np.lcm.reduce(arr)` → BCNN của **cả mảng**.
