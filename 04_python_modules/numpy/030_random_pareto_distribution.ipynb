## Pareto Distribution — Phân phối Pareto

Phân phối theo **nguyên lý Pareto**, hay **quy tắc 80-20**: 20% nguyên nhân tạo ra 80% kết quả (ví dụ: 20% dân số nắm 80% tài sản).

Hai tham số:

- `a` — tham số hình dạng.
- `size` — shape của mảng trả về.

Mẫu 2×3 với hình dạng 2:

```python
from numpy import random

x = random.pareto(a=2, size=(2, 3))

print(x)
```

## Visualization of Pareto Distribution — Trực quan hoá

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(random.pareto(a=2, size=1000))

plt.show()
```

## 🧠 Ghi nhớ nhanh

> - `random.pareto(a=hình_dạng, size=...)`.
