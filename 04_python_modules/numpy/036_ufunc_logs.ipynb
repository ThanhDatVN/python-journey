# NumPy Logs

> 📖 **Học song song trên W3Schools:** [numpy_ufunc_logs.asp](https://www.w3schools.com/python/numpy/numpy_ufunc_logs.asp)
>
> 🎯 **Trong bài này:** Logs · Log at Base 2 · Log at Base 10 · Natural Log, or Log at Base e · Log at Any Base

## Logs — Logarit

NumPy có các hàm tính logarit cơ số 2, cơ số `e` và cơ số 10. Phần này cũng hướng dẫn tính log cơ số bất kỳ bằng một ufunc tự tạo.

> 📌 Mọi hàm log sẽ cho `-inf` hoặc `inf` khi không tính được.

## Log at Base 2 — Log cơ số 2

```python
import numpy as np

arr = np.arange(1, 10)

print(np.log2(arr))
```

> 📌 `arange(1, 10)` trả về mảng số nguyên từ 1 (gồm) đến 10 (không gồm).

## Log at Base 10 — Log cơ số 10

```python
import numpy as np

arr = np.arange(1, 10)

print(np.log10(arr))
```

## Natural Log, or Log at Base e — Log tự nhiên

```python
import numpy as np

arr = np.arange(1, 10)

print(np.log(arr))
```

## Log at Any Base — Log cơ số bất kỳ

NumPy không có hàm log cơ số bất kỳ, nên ta dùng `frompyfunc()` cùng `math.log()` (có 2 tham số đầu vào, 1 đầu ra):

```python
from math import log
import numpy as np

nplog = np.frompyfunc(log, 2, 1)

print(nplog(100, 15))
```

> 💡 Cách khác dùng công thức đổi cơ số: `log_b(x) = log(x) / log(b)`.

```python
import numpy as np

print(np.log(100) / np.log(15))
```

---

## 🧠 Ghi nhớ nhanh

> - `np.log2()` · `np.log10()` · `np.log()` (cơ số e).
> - Cơ số bất kỳ: `np.frompyfunc(math.log, 2, 1)` hoặc `np.log(x) / np.log(b)`.
