# Random Permutations

> 📖 **Học song song trên W3Schools:** [numpy_random_permutation.asp](https://www.w3schools.com/python/numpy/numpy_random_permutation.asp)
>
> 🎯 **Trong bài này:** Random Permutations of Elements · Shuffling Arrays · Generating Permutation of Arrays

## Random Permutations of Elements — Hoán vị ngẫu nhiên

**Hoán vị (permutation)** là một cách sắp xếp lại các phần tử. Ví dụ `[3, 2, 1]` là một hoán vị của `[1, 2, 3]`.

Module random của NumPy có hai phương thức: `shuffle()` và `permutation()`.

## Shuffling Arrays — Xáo trộn mảng

**Shuffle** đổi chỗ các phần tử **ngay trên mảng gốc**:

```python
from numpy import random
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

random.shuffle(arr)

print(arr)
```

> 📌 `shuffle()` **thay đổi mảng gốc**.

## Generating Permutation of Arrays — Tạo hoán vị

`permutation()` trả về **mảng mới** đã xáo trộn, mảng gốc giữ nguyên:

```python
from numpy import random
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print(random.permutation(arr))
print(arr)
```

> 🧪 **Try it Yourself** — Xáo trộn danh sách học sinh để chia nhóm ngẫu nhiên.

```python
import numpy as np

students = np.array(["An", "Bình", "Chi", "Dũng", "Em", "Giang"])
shuffled = np.random.permutation(students)
print("Nhóm 1:", shuffled[:3])
print("Nhóm 2:", shuffled[3:])
```

---

## 🧠 Ghi nhớ nhanh

> - `random.shuffle(arr)` → xáo trộn **tại chỗ** (sửa mảng gốc).
> - `random.permutation(arr)` → trả về **mảng mới** đã xáo trộn.
