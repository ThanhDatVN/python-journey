## Hyperbolic Functions — Hàm hyperbolic

NumPy có các ufunc `sinh()`, `cosh()`, `tanh()` nhận giá trị radian và trả về sinh, cosh, tanh tương ứng.

Tính sinh của PI/2:

```python
import numpy as np

x = np.sinh(np.pi / 2)

print(x)
```

Tính cosh cho mọi giá trị trong mảng:

```python
import numpy as np

arr = np.array([np.pi / 2, np.pi / 3, np.pi / 4, np.pi / 5])

x = np.cosh(arr)

print(x)
```

## Finding Angles — Tìm góc

Dùng `arcsinh()`, `arccosh()`, `arctanh()`:

```python
import numpy as np

x = np.arcsinh(1.0)

print(x)
```

### Angles of Each Value in Arrays — Góc của từng giá trị

```python
import numpy as np

arr = np.array([0.1, 0.2, 0.5])

x = np.arctanh(arr)

print(x)
```

## ➕ Mở rộng: tanh trong học máy

Hàm `tanh` "nén" mọi giá trị về khoảng (-1, 1) — được dùng làm **hàm kích hoạt** trong mạng nơ-ron:

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 100)
plt.plot(x, np.tanh(x))
plt.title("tanh(x)")
plt.grid()
plt.show()
```

## 🧠 Ghi nhớ nhanh
