## Normal Distribution — Phân phối chuẩn

**Phân phối chuẩn** là một trong những phân phối quan trọng nhất, còn gọi là **phân phối Gauss** (theo tên nhà toán học Carl Friedrich Gauss).

Nó phù hợp với xác suất của rất nhiều hiện tượng: điểm IQ, nhịp tim, chiều cao…

Dùng `random.normal()` với ba tham số:

- `loc` — **trung bình (mean)**: nơi đỉnh "chuông" nằm.
- `scale` — **độ lệch chuẩn (standard deviation)**: đồ thị "béo" hay "gầy".
- `size` — shape của mảng trả về.

Sinh phân phối chuẩn kích thước 2×3:

```python
from numpy import random

x = random.normal(size=(2, 3))

print(x)
```

Phân phối chuẩn kích thước 2×3, trung bình 1, độ lệch chuẩn 2:

```python
from numpy import random

x = random.normal(loc=1, scale=2, size=(2, 3))

print(x)
```

## Visualization of Normal Distribution — Trực quan hoá

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(random.normal(size=1000), kind="kde")

plt.show()
```

> 📌 Đường cong phân phối chuẩn còn được gọi là **đường cong hình chuông (Bell Curve)**.

## ➕ Mở rộng: So sánh các độ lệch chuẩn

```python
from numpy import random
import matplotlib.pyplot as plt
import seaborn as sns

for s in (0.5, 1, 2):
    sns.kdeplot(random.normal(loc=0, scale=s, size=1000), label=f"scale={s}")

plt.legend()
plt.show()
```

## 🧠 Ghi nhớ nhanh

> - `random.normal(loc=trung_bình, scale=độ_lệch_chuẩn, size=...)`.
> - Đồ thị hình **chuông**; `scale` càng lớn → chuông càng "béo".
> - Mô tả nhiều hiện tượng tự nhiên: chiều cao, IQ, nhịp tim…
