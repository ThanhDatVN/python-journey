## Distplots

Distplot (*distribution plot*) nhận một mảng và vẽ đường cong tương ứng với **phân phối** của các điểm trong mảng.

> ⚠️ **Lưu ý cập nhật:** W3Schools dùng hàm `sns.distplot()` — hàm này đã **lỗi thời (deprecated)** trong Seaborn mới. Khóa học dùng các hàm thay thế hiện đại: `sns.displot()`, `sns.histplot()` và `sns.kdeplot()`.

## Import Matplotlib

````python
import matplotlib.pyplot as plt
````

## Import Seaborn

````python
import seaborn as sns
````

## Plotting a Distplot — Vẽ distplot

````python
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot([0, 1, 2, 3, 4, 5], kde=True)

plt.show()
````

## Plotting a Distplot Without the Histogram — Chỉ vẽ đường cong

Chỉ vẽ đường cong mật độ (KDE), không có cột histogram:

````python
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot([0, 1, 2, 3, 4, 5], kind="kde")

plt.show()
````

## ➕ Mở rộng: histplot và kdeplot

Hai hàm "cấp thấp" hay dùng nhất, vẽ được nhiều đường lên cùng một trục:

````python
import matplotlib.pyplot as plt
import seaborn as sns
from numpy import random

data = random.normal(size=1000)
sns.histplot(data, kde=True, color="skyblue")
plt.show()
````

---

## 🧠 Ghi nhớ nhanh

> - Seaborn = thư viện vẽ thống kê, dựa trên Matplotlib: `import seaborn as sns`.
> - `sns.distplot()` đã lỗi thời → dùng `sns.displot(data, kde=True)` / `kind="kde"`.
> - `sns.histplot()` (histogram) · `sns.kdeplot()` (đường cong mật độ).
