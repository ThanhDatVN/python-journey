# NumPy Trigonometric Functions

## Trigonometric Functions — Hàm lượng giác

NumPy có các ufunc `sin()`, `cos()`, `tan()` nhận giá trị **radian** và trả về sin, cos, tan tương ứng.

Tính sin của PI/2:

```python
import numpy as np

x = np.sin(np.pi / 2)

print(x)
```

Tính sin cho mọi giá trị trong mảng:

```python
import numpy as np

arr = np.array([np.pi / 2, np.pi / 3, np.pi / 4, np.pi / 5])

x = np.sin(arr)

print(x)
```

## Convert Degrees Into Radians — Đổi độ sang radian

Mặc định các hàm lượng giác nhận **radian**. Cần đổi độ sang radian:

> Radian = độ × π / 180

```python
import numpy as np

arr = np.array([90, 180, 270, 360])

x = np.deg2rad(arr)

print(x)
```

## Radians to Degrees — Đổi radian sang độ

```python
import numpy as np

arr = np.array([np.pi / 2, np.pi, 1.5 * np.pi, 2 * np.pi])

x = np.rad2deg(arr)

print(x)
```

## Finding Angles — Tìm góc

Tìm góc từ giá trị sin, cos, tan (ví dụ sin, cos, tan **nghịch đảo**) bằng `arcsin()`, `arccos()`, `arctan()`:

```python
import numpy as np

x = np.arcsin(1.0)

print(x)
```

### Angles of Each Value in Arrays — Góc của từng giá trị trong mảng

```python
import numpy as np

arr = np.array([1, -1, 0.1])

x = np.arcsin(arr)

print(x)
```

## Hypotenues — Cạnh huyền

Tính cạnh huyền theo định lý Pythagoras bằng `hypot()` (nhận hai cạnh góc vuông):

```python
import numpy as np

base = 3
perp = 4

x = np.hypot(base, perp)

print(x)
```

---

## 🧠 Ghi nhớ nhanh

> - `np.sin cos tan` nhận **radian**; nghịch đảo: `arcsin arccos arctan`.
> - Đổi đơn vị: `np.deg2rad()` / `np.rad2deg()`.
> - Cạnh huyền: `np.hypot(a, b)`.
