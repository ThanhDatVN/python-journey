## What is NumPy? — NumPy là gì?

- NumPy là thư viện Python để làm việc với **mảng**.
- Nó cũng có các hàm cho **đại số tuyến tính**, **biến đổi Fourier** và **ma trận**.
- NumPy được **Travis Oliphant** tạo ra năm **2005**, là mã nguồn mở, dùng miễn phí.
- NumPy là viết tắt của **Numerical Python**.

## Why Use NumPy? — Vì sao dùng NumPy?

Python có **list** dùng như mảng, nhưng list xử lý **chậm**.

NumPy cung cấp đối tượng mảng **nhanh hơn list truyền thống tới khoảng 50 lần**.

Đối tượng mảng trong NumPy gọi là **`ndarray`**, kèm rất nhiều hàm hỗ trợ giúp làm việc với mảng dễ dàng.

Mảng được dùng rất nhiều trong khoa học dữ liệu, nơi **tốc độ và tài nguyên** cực kỳ quan trọng.

## Why is NumPy Faster Than Lists? — Vì sao NumPy nhanh hơn list?

Khác với list, mảng NumPy được lưu ở **một vùng nhớ liên tục**, nên chương trình truy cập và xử lý rất hiệu quả (*locality of reference*). NumPy cũng được tối ưu cho CPU hiện đại.

So sánh thử thời gian tính tổng bình phương một triệu số:

```python
import time
import numpy as np

n = 1_000_000
py_list = list(range(n))
np_arr = np.arange(n)

t = time.perf_counter()
total = sum(x * x for x in py_list)
print(f"Python list: {time.perf_counter() - t:.4f} giây")

t = time.perf_counter()
total = (np_arr * np_arr).sum()
print(f"NumPy array: {time.perf_counter() - t:.4f} giây")
```

## Which Language is NumPy written in? — NumPy viết bằng ngôn ngữ gì?

NumPy là thư viện Python, viết **một phần bằng Python**, còn phần lớn các đoạn cần tính toán nhanh được viết bằng **C hoặc C++**.

## Where is the NumPy Codebase? — Mã nguồn ở đâu?

Mã nguồn NumPy nằm tại [github.com/numpy/numpy](https://github.com/numpy/numpy).
