## Scatter Plot — Biểu đồ phân tán

**Biểu đồ phân tán** là biểu đồ mà mỗi giá trị trong tập dữ liệu được biểu diễn bằng **một chấm**.

Matplotlib có phương thức vẽ scatter, cần hai mảng cùng độ dài: một cho trục x, một cho trục y:

```python
x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6]
y = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]
```

Mảng `x` là **tuổi** của từng xe, mảng `y` là **tốc độ** của từng xe.

```python
import matplotlib.pyplot as plt

x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6]
y = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

plt.scatter(x, y)
plt.show()
```

## Scatter Plot Explained — Giải thích

Trục x là tuổi xe, trục y là tốc độ.

Có vẻ như hai xe nhanh nhất đều 2 tuổi, xe chậm nhất 12 tuổi.

> 📌 Có vẻ **xe càng mới thì càng nhanh** — nhưng đó có thể là trùng hợp, vì ta chỉ mới ghi nhận 13 xe.

## Random Data Distributions — Phân phối dữ liệu ngẫu nhiên

Trong học máy, tập dữ liệu có thể chứa hàng nghìn — thậm chí hàng triệu — giá trị.

Khi thử nghiệm thuật toán, bạn có thể chưa có dữ liệu thật — hãy dùng **giá trị sinh ngẫu nhiên**.

Tạo hai mảng 1000 số ngẫu nhiên phân phối chuẩn:

- Mảng thứ nhất: trung bình 5.0, độ lệch chuẩn 1.0.
- Mảng thứ hai: trung bình 10.0, độ lệch chuẩn 2.0.

```python
import numpy
import matplotlib.pyplot as plt

x = numpy.random.normal(5.0, 1.0, 1000)
y = numpy.random.normal(10.0, 2.0, 1000)

plt.scatter(x, y)
plt.show()
```

## Scatter Plot Explained — Giải thích

Các chấm **tập trung quanh** giá trị 5 trên trục x và 10 trên trục y.

Độ phân tán theo trục y **rộng hơn** trục x (độ lệch chuẩn 2.0 so với 1.0).

---

## 🧠 Ghi nhớ nhanh

> - Scatter plot: mỗi điểm dữ liệu là một chấm — `plt.scatter(x, y)`.
> - Dùng để **nhìn mối liên hệ** giữa hai biến (tuổi xe ↔ tốc độ).
> - Thử nghiệm thuật toán bằng dữ liệu ngẫu nhiên: `numpy.random.normal(...)`.
