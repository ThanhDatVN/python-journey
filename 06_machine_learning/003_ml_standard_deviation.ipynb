## What is Standard Deviation? — Độ lệch chuẩn là gì?

**Độ lệch chuẩn (standard deviation)** là con số mô tả các giá trị **phân tán** thế nào.

- Độ lệch chuẩn **thấp** → phần lớn giá trị **gần trung bình**.
- Độ lệch chuẩn **cao** → các giá trị **trải rộng** hơn.

Ví dụ: tốc độ 7 chiếc xe:

```python
speed = [86, 87, 88, 86, 87, 85, 86]
```

Độ lệch chuẩn là `0.9` — phần lớn giá trị trong khoảng 0.9 quanh trung bình 86.4.

Với một tập khác:

```python
speed = [32, 111, 138, 28, 59, 77, 97]
```

Độ lệch chuẩn là `37.85` — giá trị trải rộng trong khoảng 37.85 quanh trung bình 77.4.

NumPy có `std()`:

```python
import numpy

speed = [86, 87, 88, 86, 87, 85, 86]

x = numpy.std(speed)

print(x)
```

```python
import numpy

speed = [32, 111, 138, 28, 59, 77, 97]

x = numpy.std(speed)

print(x)
```

## Variance — Phương sai

**Phương sai** cũng cho biết mức độ phân tán. Thực tế, **căn bậc hai của phương sai chính là độ lệch chuẩn**!

Cách tính phương sai:

1. Tính trung bình: `(32+111+138+28+59+77+97) / 7 = 77.4`
2. Với mỗi giá trị, tính hiệu với trung bình: `32 - 77.4 = -45.4`, `111 - 77.4 = 33.6`…
3. Bình phương từng hiệu: `(-45.4)² = 2061.16`, `(33.6)² = 1128.96`…
4. Phương sai = trung bình của các bình phương đó: `(2061.16 + 1128.96 + ...) / 7 = 1432.2`

NumPy có `var()`:

```python
import numpy

speed = [32, 111, 138, 28, 59, 77, 97]

x = numpy.var(speed)

print(x)
```

## Standard Deviation — Từ phương sai tới độ lệch chuẩn

Độ lệch chuẩn = căn bậc hai của phương sai: `√1432.25 = 37.85`

```python
import numpy

speed = [32, 111, 138, 28, 59, 77, 97]

print(numpy.sqrt(numpy.var(speed)))
print(numpy.std(speed))
```

## Symbols — Ký hiệu

- Độ lệch chuẩn thường ký hiệu là **σ** (Sigma).
- Phương sai thường ký hiệu là **σ²** (Sigma bình phương).

---

## 🧠 Ghi nhớ nhanh

> - **Độ lệch chuẩn σ** = mức phân tán quanh trung bình: `numpy.std()`.
> - **Phương sai σ²** = trung bình bình phương độ lệch: `numpy.var()`.
> - σ = √(σ²).
