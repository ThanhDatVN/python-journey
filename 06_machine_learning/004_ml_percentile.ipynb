## What are Percentiles? — Phân vị là gì?

**Phân vị (percentile)** dùng trong thống kê để cho biết giá trị mà **một tỉ lệ phần trăm** các giá trị **nhỏ hơn** nó.

Ví dụ: có danh sách tuổi của những người sống trên một con phố:

```python
ages = [5, 31, 43, 48, 50, 41, 7, 11, 15, 39, 80, 82, 32, 2, 8, 6, 25, 36, 27, 61, 31]
```

Phân vị thứ 75 là gì? Kết quả là `43` — nghĩa là **75%** số người có tuổi **từ 43 trở xuống**.

NumPy có `percentile()`:

```python
import numpy

ages = [5, 31, 43, 48, 50, 41, 7, 11, 15, 39, 80, 82, 32, 2, 8, 6, 25, 36, 27, 61, 31]

x = numpy.percentile(ages, 75)

print(x)
```

90% số người có tuổi **nhỏ hơn hoặc bằng** bao nhiêu?

```python
import numpy

ages = [5, 31, 43, 48, 50, 41, 7, 11, 15, 39, 80, 82, 32, 2, 8, 6, 25, 36, 27, 61, 31]

x = numpy.percentile(ages, 90)

print(x)
```

## ➕ Mở rộng: Tứ phân vị

Phân vị 25, 50, 75 gọi là **tứ phân vị (quartiles)**. Phân vị 50 chính là **trung vị**:

```python
import numpy

ages = [5, 31, 43, 48, 50, 41, 7, 11, 15, 39, 80, 82, 32, 2, 8, 6, 25, 36, 27, 61, 31]
print(numpy.percentile(ages, [25, 50, 75]))
print(numpy.median(ages))
```

---

## 🧠 Ghi nhớ nhanh

> - Phân vị thứ *p* = giá trị mà **p%** dữ liệu nhỏ hơn hoặc bằng nó: `numpy.percentile(data, p)`.
> - Phân vị 50 = **trung vị**; 25/50/75 = tứ phân vị.
