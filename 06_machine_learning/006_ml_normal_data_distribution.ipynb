## Normal Data Distribution — Phân phối chuẩn

Ở bài trước, ta tạo mảng ngẫu nhiên **đều** — mọi giá trị có cơ hội như nhau.

Trong thống kê, một kiểu phân phối rất quan trọng là **phân phối chuẩn**, hay **phân phối Gauss** (theo nhà toán học Carl Friedrich Gauss).

Một phân phối chuẩn điển hình:

```python
import numpy
import matplotlib.pyplot as plt

x = numpy.random.normal(5.0, 1.0, 100000)

plt.hist(x, 100)
plt.show()
```

> 📌 Phân phối chuẩn còn gọi là **đường cong hình chuông** vì hình dạng giống cái chuông.

## Histogram Explained — Giải thích

Ta dùng `numpy.random.normal()` tạo mảng 100 000 giá trị, rồi vẽ histogram 100 cột.

Ta chỉ định **trung bình là 5.0**, **độ lệch chuẩn là 1.0**.

Nghĩa là các giá trị **tập trung quanh 5.0**, và hiếm khi cách xa trung bình quá 1.0.

Từ histogram có thể thấy phần lớn giá trị nằm trong khoảng **4.0 đến 6.0**, đỉnh ở khoảng **5.0**.

## ➕ Mở rộng: Quy tắc 68-95-99.7

Với phân phối chuẩn: ~68% giá trị nằm trong ±1σ, ~95% trong ±2σ, ~99.7% trong ±3σ:

```python
import numpy

x = numpy.random.normal(5.0, 1.0, 100000)
for k in (1, 2, 3):
    inside = numpy.mean(numpy.abs(x - 5.0) <= k * 1.0)
    print(f"±{k}σ: {inside:.1%}")
```

---

## 🧠 Ghi nhớ nhanh

> - `numpy.random.normal(trung_bình, độ_lệch_chuẩn, số_lượng)` → dữ liệu phân phối chuẩn.
> - Đồ thị hình **chuông**, đỉnh ở trung bình.
> - Quy tắc **68-95-99.7**: tỉ lệ giá trị trong ±1σ, ±2σ, ±3σ.
