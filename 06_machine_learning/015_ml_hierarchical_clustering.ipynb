## Hierarchical Clustering — Phân cụm phân cấp

**Phân cụm phân cấp** là phương pháp **học không giám sát (unsupervised learning)** để phân cụm các điểm dữ liệu. Thuật toán xây dựng các cụm bằng cách đo **sự khác biệt** giữa dữ liệu.

> **Học không giám sát**: mô hình không cần được huấn luyện, và ta không cần biến "mục tiêu".

Phương pháp này dùng được cho mọi loại dữ liệu để trực quan hoá và diễn giải **mối quan hệ** giữa các điểm.

Ở bài này, ta dùng phân cụm phân cấp để nhóm các điểm dữ liệu, và trực quan hoá các cụm bằng cả **dendrogram** và **scatter plot**.

## How does it work? — Hoạt động thế nào?

Ta dùng **Agglomerative Clustering** — một dạng phân cụm phân cấp theo hướng **từ dưới lên**:

1. Ban đầu, **mỗi điểm** dữ liệu là một cụm riêng.
2. Tìm hai cụm **gần nhau nhất** theo khoảng cách, rồi **gộp** chúng lại.
3. Lặp lại cho tới khi chỉ còn **một cụm** lớn.

Phân cụm phân cấp yêu cầu chọn **cách đo khoảng cách** và **cách liên kết (linkage)**. Ở đây dùng khoảng cách **euclidean** và liên kết **Ward** — tìm cách gộp cụm sao cho **phương sai** giữa các cụm tăng ít nhất.

Bắt đầu bằng cách trực quan hoá một vài điểm dữ liệu:

```python
import numpy as np
import matplotlib.pyplot as plt

x = [4, 5, 10, 4, 3, 11, 14, 6, 10, 12]
y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21]

plt.scatter(x, y)
plt.show()
```

Tính liên kết Ward bằng khoảng cách euclidean, rồi vẽ **dendrogram**. Dendrogram cho thấy thứ bậc của các cụm, từ dưới (từng điểm) lên trên (cụm cha):

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

x = [4, 5, 10, 4, 3, 11, 14, 6, 10, 12]
y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21]

data = list(zip(x, y))

linkage_data = linkage(data, method='ward', metric='euclidean')
dendrogram(linkage_data)

plt.show()
```

Làm tương tự với `AgglomerativeClustering` của scikit-learn, và vẽ kết quả lên scatter plot với **2 cụm**:

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import AgglomerativeClustering

x = [4, 5, 10, 4, 3, 11, 14, 6, 10, 12]
y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21]

data = list(zip(x, y))

hierarchical_cluster = AgglomerativeClustering(n_clusters=2, metric='euclidean', linkage='ward')
labels = hierarchical_cluster.fit_predict(data)

plt.scatter(x, y, c=labels)
plt.show()
```

> ⚠️ **Lưu ý cập nhật:** W3Schools dùng tham số `affinity='euclidean'` — tham số này đã bị **xoá** ở scikit-learn mới, thay bằng `metric='euclidean'`.

## Example Explained — Giải thích

- `zip()` ghép `x` và `y` thành các cặp toạ độ.
- `linkage()` tính liên kết Ward; `dendrogram()` vẽ cây phân cấp.
- `AgglomerativeClustering(...).fit_predict(data)` trả về **nhãn cụm** cho từng điểm; ta dùng nhãn làm màu khi vẽ.

## 🧠 Ghi nhớ nhanh

> - Phân cụm phân cấp = **học không giám sát**, gộp dần các cụm gần nhau nhất (agglomerative).
> - SciPy: `linkage(data, method='ward', metric='euclidean')` + `dendrogram()`.
> - scikit-learn: `AgglomerativeClustering(n_clusters=2, metric='euclidean', linkage='ward').fit_predict(data)`.
