## Mean, Median, and Mode — Trung bình, trung vị, yếu vị

Có thể học được gì từ một tập số?

Trong học máy (và toán học) thường có ba giá trị ta quan tâm:

- **Mean** — giá trị trung bình
- **Median** — giá trị ở giữa
- **Mode** — giá trị xuất hiện nhiều nhất

Ví dụ: ta ghi lại tốc độ của 13 chiếc xe:

````python
speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]
````

Tốc độ trung bình, ở giữa, hay phổ biến nhất là bao nhiêu?

## Mean — Trung bình

Cộng tất cả giá trị rồi chia cho số lượng giá trị:

```text
(99+86+87+88+111+86+103+87+94+78+77+85+86) / 13 = 89.77
```

NumPy có phương thức `mean()`:

````python
import numpy

speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

x = numpy.mean(speed)

print(x)
````

## Median — Trung vị

Giá trị **ở giữa** sau khi sắp xếp tất cả giá trị:

```text
77, 78, 85, 86, 86, 86, 87, 87, 88, 94, 99, 103, 111
```

Cần sắp xếp trước khi tìm trung vị. NumPy có `median()`:

````python
import numpy

speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

x = numpy.median(speed)

print(x)
````

Nếu có **hai** số ở giữa, lấy trung bình của hai số đó:

```text
77, 78, 85, 86, 86, 86, 87, 87, 94, 98, 99, 103
(86 + 87) / 2 = 86.5
```

````python
import numpy

speed = [99, 86, 87, 88, 86, 103, 87, 94, 78, 77, 85, 86]

x = numpy.median(speed)

print(x)
````

## Mode — Yếu vị

Giá trị **xuất hiện nhiều nhất**:

```text
99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86 = 86
```

Module `stats` của SciPy có `mode()`:

````python
from scipy import stats

speed = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

x = stats.mode(speed)

print(x)
print(x.mode, "xuất hiện", x.count, "lần")
````

## Chapter Summary — Tóm tắt

Mean, Median, Mode là các kỹ thuật thường dùng trong học máy, nên cần hiểu khái niệm đằng sau chúng.

> 🧪 **Try it Yourself** — Tính mean, median, mode của điểm thi trong lớp.

````python
import numpy
from scipy import stats

scores = [7, 8, 6, 9, 8, 10, 5, 8, 7, 9]
print("Mean:", numpy.mean(scores))
print("Median:", numpy.median(scores))
print("Mode:", stats.mode(scores).mode)
````

---

## 🧠 Ghi nhớ nhanh

> - **Mean** trung bình: `numpy.mean()`.
> - **Median** giá trị giữa (sau khi sắp xếp): `numpy.median()`.
> - **Mode** giá trị nhiều nhất: `scipy.stats.mode()`.

**❓ Câu hỏi:** Trung vị (median) của `[1, 3, 2, 10]` là bao nhiêu?

- **A.** 2.5
- **B.** 4
- **C.** 2
- **D.** 3

<details>
<summary>👉 Xem đáp án</summary>

✅ **Đáp án: A** — 2.5

Sắp xếp: 1, 2, 3, 10 → hai số giữa là 2 và 3 → (2 + 3) / 2 = 2.5.

</details>
