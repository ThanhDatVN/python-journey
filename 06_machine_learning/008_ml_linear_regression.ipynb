## Regression — Hồi quy

Thuật ngữ **hồi quy (regression)** dùng khi bạn cố tìm **mối quan hệ giữa các biến**.

Trong học máy và mô hình thống kê, mối quan hệ đó được dùng để **dự đoán** kết quả của các sự kiện tương lai.

## Linear Regression — Hồi quy tuyến tính

**Hồi quy tuyến tính** dùng mối quan hệ giữa các điểm dữ liệu để vẽ **một đường thẳng** đi qua chúng.

Đường thẳng đó có thể dùng để dự đoán các giá trị tương lai.

> 📌 Trong học máy, **dự đoán tương lai** là rất quan trọng.

## How Does it Work? — Hoạt động thế nào?

Python có các phương thức tìm mối quan hệ giữa các điểm dữ liệu và vẽ đường hồi quy tuyến tính — bạn không cần tự tính bằng công thức toán.

Ví dụ: trục x là tuổi xe, trục y là tốc độ (13 xe chạy qua trạm thu phí):

```python
import matplotlib.pyplot as plt

x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6]
y = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

plt.scatter(x, y)
plt.show()
```

Import `scipy` và vẽ đường hồi quy tuyến tính:

```python
import matplotlib.pyplot as plt
from scipy import stats

x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6]
y = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

slope, intercept, r, p, std_err = stats.linregress(x, y)

def myfunc(x):
    return slope * x + intercept

mymodel = list(map(myfunc, x))

plt.scatter(x, y)
plt.plot(x, mymodel)
plt.show()
```

## Example Explained — Giải thích ví dụ

- `stats.linregress(x, y)` trả về các giá trị quan trọng của hồi quy tuyến tính: **slope** (hệ số góc), **intercept** (hệ số chặn), **r**, **p**, **std_err**.
- `myfunc(x)` dùng slope và intercept để trả về giá trị `y` mới — giá trị này cho biết điểm `x` tương ứng nằm ở đâu trên đường thẳng.
- `map()` chạy từng giá trị của `x` qua hàm, tạo mảng `y` mới.
- Cuối cùng vẽ scatter gốc và đường hồi quy.

## R for Relationship — R thể hiện mức độ quan hệ

Cần biết mối quan hệ giữa x và y **mạnh tới đâu** — nếu không có quan hệ, hồi quy tuyến tính không dự đoán được gì.

Mức độ quan hệ đo bằng **hệ số tương quan r**, trong khoảng **-1 đến 1**: 0 là không quan hệ, -1 và 1 là quan hệ 100%.

```python
from scipy import stats

x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6]
y = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

slope, intercept, r, p, std_err = stats.linregress(x, y)

print(r)
```

> 📌 Kết quả **-0.76** cho thấy có mối quan hệ (ngược chiều) — chưa hoàn hảo nhưng đủ để dùng hồi quy tuyến tính cho dự đoán.

## Predict Future Values — Dự đoán giá trị tương lai

Dự đoán tốc độ của một chiếc xe **10 năm tuổi**:

```python
from scipy import stats

x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6]
y = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

slope, intercept, r, p, std_err = stats.linregress(x, y)

def myfunc(x):
    return slope * x + intercept

speed = myfunc(10)

print(speed)
```

Dự đoán tốc độ là khoảng **85.6** — cũng có thể đọc được từ đồ thị.

## Bad Fit? — Không phù hợp?

Ví dụ dữ liệu **không phù hợp** với hồi quy tuyến tính:

```python
import matplotlib.pyplot as plt
from scipy import stats

x = [89, 43, 36, 36, 95, 10, 66, 34, 38, 20, 26, 29, 48, 64, 6, 5, 36, 66, 72, 40]
y = [21, 46, 3, 35, 67, 95, 53, 72, 58, 10, 26, 34, 90, 33, 38, 20, 56, 2, 47, 15]

slope, intercept, r, p, std_err = stats.linregress(x, y)

def myfunc(x):
    return slope * x + intercept

mymodel = list(map(myfunc, x))

plt.scatter(x, y)
plt.plot(x, mymodel)
plt.show()

print("r =", r)
```

r rất nhỏ (gần 0) → dữ liệu **không** phù hợp với hồi quy tuyến tính.

---

## 🧠 Ghi nhớ nhanh

> - Hồi quy tuyến tính: vẽ **đường thẳng** `y = slope * x + intercept` qua dữ liệu.
> - `slope, intercept, r, p, std_err = stats.linregress(x, y)`.
> - **r** (-1 → 1) đo mức độ quan hệ; |r| gần 0 → **không phù hợp**.
> - Dự đoán: thay `x` mới vào hàm `slope * x + intercept`.
