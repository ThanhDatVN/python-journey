## Multiple Regression — Hồi quy bội

Hồi quy bội giống hồi quy tuyến tính, nhưng có **nhiều hơn một** biến độc lập — tức là ta dự đoán một giá trị dựa trên **hai hay nhiều** biến.

Xét tập dữ liệu xe dưới đây (file `cars.csv`):

```python
import pandas

df = pandas.read_csv("cars.csv")
df.head(10)
```

Ta có thể dự đoán lượng phát thải **CO2** của xe dựa trên **dung tích động cơ (Volume)**, nhưng với hồi quy bội ta đưa thêm biến **trọng lượng (Weight)** để dự đoán chính xác hơn.

## How Does it Work? — Hoạt động thế nào?

Trong Python có các module làm sẵn việc này. Bắt đầu bằng import Pandas, đọc file CSV thành DataFrame:

```python
import pandas

df = pandas.read_csv("cars.csv")
```

Tạo danh sách các **biến độc lập** đặt tên `X`, và **biến phụ thuộc** đặt tên `y`:

```python
X = df[['Weight', 'Volume']]
y = df['CO2']
```

> 📌 Quy ước: biến độc lập viết **hoa** `X`, biến phụ thuộc viết **thường** `y`.

Dùng phương thức trong module `sklearn`. Từ `sklearn`, dùng `LinearRegression()` để tạo đối tượng hồi quy tuyến tính. Đối tượng có phương thức `fit()` nhận dữ liệu và mô tả mối quan hệ:

```python
from sklearn import linear_model

regr = linear_model.LinearRegression()
regr.fit(X, y)
```

Giờ có một đối tượng hồi quy sẵn sàng **dự đoán CO2** dựa trên trọng lượng và dung tích xe:

```python
# dự đoán CO2 của xe nặng 2300 kg, động cơ 1300 cm3:
predictedCO2 = regr.predict(pandas.DataFrame([[2300, 1300]], columns=['Weight', 'Volume']))

print(predictedCO2)
```

> 💡 Truyền DataFrame có tên cột (thay vì list `[[2300, 1300]]` như W3Schools) để scikit-learn không cảnh báo về tên đặc trưng.

## Coefficient — Hệ số

**Hệ số (coefficient)** là thừa số mô tả mối quan hệ với một biến chưa biết.

Ví dụ: nếu `x` là biến, thì `2x` là `x` hai lần — `2` là hệ số.

In giá trị hệ số của đối tượng hồi quy:

```python
print(regr.coef_)
```

## Result Explained — Giải thích kết quả

Mảng kết quả thể hiện hệ số của **trọng lượng** và **dung tích**:

- Hệ số Weight: nếu trọng lượng tăng **1 kg**, CO2 tăng thêm khoảng *(hệ số thứ nhất)* g.
- Hệ số Volume: nếu dung tích tăng **1 cm³**, CO2 tăng thêm khoảng *(hệ số thứ hai)* g.

Kiểm chứng: tăng trọng lượng thêm 1000 kg và quan sát CO2 thay đổi:

```python
for weight in (2300, 3300):
    co2 = regr.predict(pandas.DataFrame([[weight, 1300]], columns=['Weight', 'Volume']))
    print(weight, "kg ->", co2.round(2))

print("Chênh lệch ≈ 1000 × hệ số Weight =", round(1000 * regr.coef_[0], 2))
```

---

## 🧠 Ghi nhớ nhanh

> - Hồi quy bội: dự đoán `y` từ **nhiều** biến `X`.
> - `regr = linear_model.LinearRegression()` → `regr.fit(X, y)` → `regr.predict(...)`.
> - `regr.coef_` → mức thay đổi của `y` khi mỗi biến tăng 1 đơn vị.
