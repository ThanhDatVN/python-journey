# Machine Learning - Cross Validation

>
> 🎯 **Trong bài này:** Cross Validation · K-Fold · Stratified K-Fold · Leave-One-Out (LOO) · Leave-P-Out (LPO) · Shuffle Split · Ending Notes

## Cross Validation — Kiểm định chéo

Khi điều chỉnh mô hình, ta muốn **tăng hiệu năng tổng thể trên dữ liệu chưa thấy**. Điều chỉnh tham số có thể cho kết quả tốt hơn nhiều trên tập test — nhưng tối ưu tham số **theo tập test** sẽ làm **rò rỉ thông tin**, khiến mô hình kém hơn trên dữ liệu thực.

Để tránh, nên đánh giá **trước khi** dùng tập test. **Cross validation (CV)** là cách làm điều đó: chia dữ liệu huấn luyện thành nhiều phần, lần lượt dùng từng phần để kiểm tra.

Ta dùng tập dữ liệu **iris**:

```python
from sklearn import datasets

X, y = datasets.load_iris(return_X_y=True)
```

Có nhiều phương pháp cross validation — bắt đầu với **k-fold**.

## K-Fold

Dữ liệu được chia thành **k** tập (fold) bằng nhau. Mô hình được huấn luyện trên **k-1** fold và kiểm tra trên fold còn lại. Lặp lại **k** lần, mỗi lần một fold làm tập kiểm tra:

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import KFold, cross_val_score

clf = DecisionTreeClassifier(random_state=42)

k_folds = KFold(n_splits=5)

scores = cross_val_score(clf, X, y, cv=k_folds)

print("Cross Validation Scores: ", scores)
print("Average CV Score: ", scores.mean())
print("Number of CV Scores used in Average: ", len(scores))
```

## Stratified K-Fold

Khi các lớp **mất cân bằng**, cần tính đến điều đó khi chia. **Stratified K-Fold** đảm bảo mỗi fold có **tỉ lệ các lớp giống nhau** như toàn bộ dữ liệu:

```python
from sklearn.model_selection import StratifiedKFold

sk_folds = StratifiedKFold(n_splits=5)

scores = cross_val_score(clf, X, y, cv=sk_folds)

print("Cross Validation Scores: ", scores)
print("Average CV Score: ", scores.mean())
print("Number of CV Scores used in Average: ", len(scores))
```

## Leave-One-Out (LOO)

Thay vì chọn số fold, **Leave-One-Out** dùng **1** mẫu để kiểm tra và **n-1** mẫu còn lại để huấn luyện — lặp lại cho **từng** mẫu:

```python
from sklearn.model_selection import LeaveOneOut

loo = LeaveOneOut()

scores = cross_val_score(clf, X, y, cv=loo)

print("Average CV Score: ", scores.mean())
print("Number of CV Scores used in Average: ", len(scores))
```

> Số điểm số bằng số mẫu trong tập dữ liệu (150).

## Leave-P-Out (LPO)

Giống LOO nhưng chọn **p** mẫu để kiểm tra. Số lần lặp tăng rất nhanh theo p:

```python
from sklearn.model_selection import LeavePOut

lpo = LeavePOut(p=2)

scores = cross_val_score(clf, X, y, cv=lpo)

print("Average CV Score: ", scores.mean())
print("Number of CV Scores used in Average: ", len(scores))
```

> 📌 Với p = 2 và 150 mẫu, có tới **11 175** lần huấn luyện!

## Shuffle Split

Khác k-fold, **ShuffleSplit** chừa lại một **tỉ lệ phần trăm** dữ liệu không dùng cho train/test, và lấy mẫu ngẫu nhiên mỗi lần:

```python
from sklearn.model_selection import ShuffleSplit

ss = ShuffleSplit(train_size=0.6, test_size=0.3, n_splits=5, random_state=0)

scores = cross_val_score(clf, X, y, cv=ss)

print("Cross Validation Scores: ", scores)
print("Average CV Score: ", scores.mean())
print("Number of CV Scores used in Average: ", len(scores))
```

## Ending Notes — Ghi chú

Đây chỉ là vài phương pháp CV phổ biến. Phương pháp phù hợp phụ thuộc vào **tập dữ liệu** của bạn.
