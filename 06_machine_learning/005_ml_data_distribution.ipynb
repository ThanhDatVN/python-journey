## Data Distribution — Phân phối dữ liệu

Trước đó trong khóa học, ta chỉ dùng lượng dữ liệu rất nhỏ trong ví dụ để hiểu các khái niệm.

Thực tế, tập dữ liệu **lớn hơn nhiều**, nhưng có thể rất khó thu thập dữ liệu thật — ít nhất là ở giai đoạn đầu dự án.

## How Can we Get Big Data Sets? — Làm sao có tập dữ liệu lớn?

Để tạo tập dữ liệu lớn để thử nghiệm, ta dùng module NumPy — nó có nhiều phương thức tạo **tập dữ liệu ngẫu nhiên** kích thước bất kỳ.

Tạo mảng 250 số thực ngẫu nhiên trong khoảng 0 đến 5:

```python
import numpy

x = numpy.random.uniform(0.0, 5.0, 250)

print(x[:20])
```

## Histogram

Để trực quan hoá tập dữ liệu, ta vẽ **histogram**. Dùng module Matplotlib:

```python
import numpy
import matplotlib.pyplot as plt

x = numpy.random.uniform(0.0, 5.0, 250)

plt.hist(x, 5)
plt.show()
```

## Histogram Explained — Giải thích histogram

Ta dùng mảng ở trên để vẽ histogram với **5 cột**:

- Cột đầu cho biết bao nhiêu giá trị trong khoảng 0 đến 1
- Cột thứ hai: bao nhiêu giá trị trong khoảng 1 đến 2
- …

Các cột có chiều cao gần như nhau vì dữ liệu được sinh theo **phân phối đều**.

> 📌 Giá trị trong mảng là ngẫu nhiên, nên mỗi lần chạy kết quả có thể hơi khác.

## Big Data Distributions — Phân phối với dữ liệu lớn

Một mảng 250 giá trị chưa phải là lớn, nhưng giờ bạn đã biết cách tạo tập giá trị ngẫu nhiên — và chỉ cần đổi tham số là có tập dữ liệu lớn.

Tạo mảng **100 000** số ngẫu nhiên và vẽ histogram **100 cột**:

```python
import numpy
import matplotlib.pyplot as plt

x = numpy.random.uniform(0.0, 5.0, 100000)

plt.hist(x, 100)
plt.show()
```

## 🧠 Ghi nhớ nhanh

> - Tạo dữ liệu thử nghiệm lớn: `numpy.random.uniform(low, high, size)`.
> - Trực quan hoá phân phối: `plt.hist(data, số_cột)`.
> - Phân phối **đều** → các cột histogram cao gần bằng nhau.
