## Data Types — Các kiểu dữ liệu

Để phân tích, cần biết mình đang làm việc với kiểu dữ liệu nào. Có ba nhóm chính:

- **Numerical (số)** — chia làm hai loại:
  - **Discrete (rời rạc)**: số đếm được, giới hạn ở số nguyên. Ví dụ: số xe chạy qua.
  - **Continuous (liên tục)**: số đo được, có vô số giá trị. Ví dụ: giá một món hàng, kích thước.
- **Categorical (phân loại)**: giá trị **không so sánh lớn nhỏ** được. Ví dụ: màu sắc, có/không.
- **Ordinal (thứ bậc)**: giống phân loại nhưng **có thứ tự**. Ví dụ: xếp loại A tốt hơn B.

Biết kiểu dữ liệu giúp bạn chọn đúng kỹ thuật phân tích.
