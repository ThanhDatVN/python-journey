## Scale Features — Chuẩn hoá đặc trưng

Khi dữ liệu có các giá trị khác nhau, thậm chí **đơn vị đo** khác nhau, rất khó so sánh. Kilogram so với mét thế nào? Độ cao so với thời gian?

Câu trả lời là **scale (chuẩn hoá)**: đưa dữ liệu về các giá trị mới dễ so sánh hơn.

Xét bảng xe, nhưng lần này cột **Volume** đổi sang **lít** thay vì cm³ (1.0 thay vì 1000):

````python
import pandas

df = pandas.read_csv("cars.csv")
df["Volume"] = df["Volume"] / 1000
df.head(8)
````

Rất khó so sánh dung tích 1.0 với trọng lượng 790, nhưng nếu scale cả hai về giá trị so sánh được, ta dễ thấy một giá trị lớn/nhỏ thế nào so với giá trị khác.

Có nhiều phương pháp scale. Ở đây ta dùng **standardization (chuẩn hoá z-score)**:

```text
z = (x - u) / s
```

Trong đó `z` là giá trị mới, `x` là giá trị gốc, `u` là trung bình, `s` là độ lệch chuẩn.

Cột trọng lượng trong `cars.csv` có trung bình ~1300 và độ lệch chuẩn ~250, nên giá trị scale của một xe nặng 790 kg là:

```text
(790 - 1300) / 250 ≈ -2.0
```

Module `sklearn` có `StandardScaler()` trả về đối tượng Scaler với phương thức `fit_transform()`:

````python
import pandas
from sklearn.preprocessing import StandardScaler

scale = StandardScaler()

df = pandas.read_csv("cars.csv")
df["Volume"] = df["Volume"] / 1000

X = df[['Weight', 'Volume']]

scaledX = scale.fit_transform(X)

print(scaledX[:8])
````

Hai cột giờ có cùng "thang đo": trung bình 0, độ lệch chuẩn 1.

## Predict CO2 Values — Dự đoán CO2

Khi dữ liệu đã được scale, **giá trị dùng để dự đoán cũng phải được scale** theo cùng cách:

````python
import pandas
from sklearn import linear_model
from sklearn.preprocessing import StandardScaler

scale = StandardScaler()

df = pandas.read_csv("cars.csv")
df["Volume"] = df["Volume"] / 1000

X = df[['Weight', 'Volume']]
y = df['CO2']

scaledX = scale.fit_transform(X)

regr = linear_model.LinearRegression()
regr.fit(scaledX, y)

scaled = scale.transform(pandas.DataFrame([[2300, 1.3]], columns=['Weight', 'Volume']))

predictedCO2 = regr.predict(scaled)
print(predictedCO2)
````

> 📌 Dùng `fit_transform()` trên dữ liệu huấn luyện, và chỉ `transform()` trên dữ liệu mới.

---

## 🧠 Ghi nhớ nhanh

> - Scale đưa các cột khác đơn vị về **cùng thang đo**: `z = (x - u) / s`.
> - `StandardScaler().fit_transform(X)` → trung bình 0, độ lệch chuẩn 1.
> - Dữ liệu mới dự đoán phải `scale.transform(...)` theo cùng scaler.
