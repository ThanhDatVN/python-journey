## What is a confusion matrix? — Ma trận nhầm lẫn là gì?

**Ma trận nhầm lẫn (confusion matrix)** là bảng dùng trong bài toán **phân loại** để đánh giá **lỗi ở đâu** trong mô hình.

Hàng là các lớp **thực tế**; cột là các **dự đoán**. Nhờ bảng này, dễ thấy dự đoán nào sai.

## Creating a Confusion Matrix — Tạo ma trận nhầm lẫn

Ma trận nhầm lẫn có thể tạo từ dự đoán của một mô hình **logistic regression**… Hiện tại ta tạo giá trị **thực tế** và **dự đoán** bằng NumPy:

````python
import numpy

numpy.random.seed(0)
actual = numpy.random.binomial(1, 0.9, size=1000)
predicted = numpy.random.binomial(1, 0.9, size=1000)
````

Để tạo ma trận, import `metrics` từ `sklearn`:

````python
from sklearn import metrics

confusion_matrix = metrics.confusion_matrix(actual, predicted)
print(confusion_matrix)
````

Để hiển thị trực quan, chuyển bảng thành **confusion matrix display**:

````python
import matplotlib.pyplot as plt

cm_display = metrics.ConfusionMatrixDisplay(confusion_matrix=confusion_matrix,
                                            display_labels=[0, 1])
cm_display.plot()
plt.show()
````

## Results Explained — Giải thích kết quả

Ma trận có **4 góc phần tư**:

- **True Negative (TN)** — trên trái: thực tế âm, dự đoán âm (đúng).
- **False Positive (FP)** — trên phải: thực tế âm, dự đoán dương (sai).
- **False Negative (FN)** — dưới trái: thực tế dương, dự đoán âm (sai).
- **True Positive (TP)** — dưới phải: thực tế dương, dự đoán dương (đúng).

*True* nghĩa là dự đoán **chính xác**, *False* nghĩa là có **lỗi**.

## Created Metrics — Các chỉ số đánh giá

Ma trận cho ra nhiều chỉ số hữu ích: Accuracy, Precision, Sensitivity (Recall), Specificity và F-score.

### Accuracy — Độ chính xác

Tỉ lệ dự đoán đúng trên tổng số:

```text
(True Positive + True Negative) / Total Predictions
```

````python
Accuracy = metrics.accuracy_score(actual, predicted)
print(Accuracy)
````

### Precision — Độ chính xác dương

Trong các trường hợp dự đoán **dương**, bao nhiêu phần trăm thực sự dương?

```text
True Positive / (True Positive + False Positive)
```

> Precision **không** đánh giá các trường hợp âm được dự đoán đúng.

````python
Precision = metrics.precision_score(actual, predicted)
print(Precision)
````

### Sensitivity (Recall) — Độ nhạy

Trong các trường hợp **thực sự dương**, bao nhiêu phần trăm được dự đoán dương?

Độ nhạy cho biết mô hình **phát hiện dương tính** tốt thế nào:

```text
True Positive / (True Positive + False Negative)
```

````python
Sensitivity_recall = metrics.recall_score(actual, predicted)
print(Sensitivity_recall)
````

### Specificity — Độ đặc hiệu

Mô hình dự đoán **âm tính** tốt thế nào? Giống độ nhạy nhưng từ góc nhìn âm tính:

```text
True Negative / (True Negative + False Positive)
```

Vì là ngược lại của Recall, ta dùng `recall_score` với nhãn dương là 0:

````python
Specificity = metrics.recall_score(actual, predicted, pos_label=0)
print(Specificity)
````

### F-score

F-score là **trung bình điều hoà** của precision và sensitivity — cân bằng cả FP và FN, phù hợp với dữ liệu mất cân bằng:

```text
2 * ((Precision * Sensitivity) / (Precision + Sensitivity))
```

> Chỉ số này không xét đến các giá trị True Negative.

````python
F1_score = metrics.f1_score(actual, predicted)
print(F1_score)
````

Tất cả trong một:

````python
print({"Accuracy": Accuracy, "Precision": Precision, "Sensitivity_recall": Sensitivity_recall,
       "Specificity": Specificity, "F1_score": F1_score})
````

---

## 🧠 Ghi nhớ nhanh

> - Ma trận nhầm lẫn: hàng = **thực tế**, cột = **dự đoán** → TN, FP, FN, TP.
> - Accuracy = (TP+TN)/Tổng · Precision = TP/(TP+FP) · Recall = TP/(TP+FN) · Specificity = TN/(TN+FP).
> - F1 = trung bình điều hoà của Precision và Recall.
