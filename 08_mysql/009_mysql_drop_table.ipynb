# Python MySQL Drop Table

> 📖 **Học song song trên W3Schools:** [python_mysql_drop_table.asp](https://www.w3schools.com/python/python_mysql_drop_table.asp)
>
> 🎯 **Trong bài này:** Delete a Table · Drop Only if Exist · ➕ Thực hành ngay với SQLite

## Delete a Table — Xoá bảng

Xoá một bảng có sẵn bằng câu lệnh `DROP TABLE`:

```python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="yourusername",
    password="yourpassword",
    database="mydatabase"
)

mycursor = mydb.cursor()

sql = "DROP TABLE customers"

mycursor.execute(sql)
```

## Drop Only if Exist — Chỉ xoá nếu tồn tại

Nếu bảng **không tồn tại** (đã bị xoá trước đó), `DROP TABLE` sẽ báo lỗi. Thêm `IF EXISTS` để tránh lỗi:

```python
sql = "DROP TABLE IF EXISTS customers"

mycursor.execute(sql)
```

## ➕ Thực hành ngay với SQLite

```python
import sqlite3

conn = sqlite3.connect(":memory:")
cur = conn.cursor()
cur.execute("CREATE TABLE customers (name TEXT, address TEXT)")

cur.execute("DROP TABLE customers")
print("Đã xoá bảng")

cur.execute("DROP TABLE IF EXISTS customers")
print("Không lỗi dù bảng không còn")
```

```python
import sqlite3

conn = sqlite3.connect(":memory:")
conn.execute("DROP TABLE customers")   # bảng không tồn tại
```

---

## 🧠 Ghi nhớ nhanh

> - `DROP TABLE tên` → xoá **cả bảng** (cấu trúc + dữ liệu).
> - `DROP TABLE IF EXISTS tên` → không lỗi nếu bảng không tồn tại.
> - So sánh: `DELETE FROM` xoá **dữ liệu**, `DROP TABLE` xoá **bảng**.
