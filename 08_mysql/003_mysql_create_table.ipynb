## Creating a Table — Tạo bảng

Để tạo bảng trong MySQL, dùng câu lệnh `CREATE TABLE`.

> 📌 Nhớ chỉ định **tên cơ sở dữ liệu** khi tạo kết nối.

Tạo bảng `customers`:

```python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="yourusername",
    password="yourpassword",
    database="mydatabase"
)

mycursor = mydb.cursor()

mycursor.execute("CREATE TABLE customers (name VARCHAR(255), address VARCHAR(255))")
```

## Check if Table Exists — Kiểm tra bảng

Liệt kê các bảng bằng `SHOW TABLES`:

```python
mycursor.execute("SHOW TABLES")

for x in mycursor:
    print(x)
```

## Primary Key — Khoá chính

Khi tạo bảng, nên tạo thêm một cột **khoá duy nhất (unique key)** cho mỗi bản ghi — gọi là **khoá chính (primary key)**.

Dùng `INT AUTO_INCREMENT PRIMARY KEY` để mỗi bản ghi tự nhận một số duy nhất, bắt đầu từ 1 và tăng dần:

```python
mycursor.execute("CREATE TABLE customers (id INT AUTO_INCREMENT PRIMARY KEY, name VARCHAR(255), address VARCHAR(255))")
```

Nếu bảng đã tồn tại, dùng `ALTER TABLE` để thêm khoá chính:

```python
mycursor.execute("ALTER TABLE customers ADD COLUMN id INT AUTO_INCREMENT PRIMARY KEY")
```

## ➕ Thực hành ngay với SQLite

```python
import sqlite3

conn = sqlite3.connect(":memory:")
cur = conn.cursor()

cur.execute("""
    CREATE TABLE customers (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name VARCHAR(255),
        address VARCHAR(255)
    )
""")

# SHOW TABLES trong SQLite:
cur.execute("SELECT name FROM sqlite_master WHERE type='table'")
print(cur.fetchall())

# Xem cấu trúc bảng:
cur.execute("PRAGMA table_info(customers)")
for column in cur.fetchall():
    print(column)
```

## 🧠 Ghi nhớ nhanh

> - `CREATE TABLE tên (cột KIỂU, ...)` · liệt kê bảng: `SHOW TABLES`.
> - Khoá chính tự tăng: `id INT AUTO_INCREMENT PRIMARY KEY` (SQLite: `INTEGER PRIMARY KEY AUTOINCREMENT`).
> - Thêm cột cho bảng có sẵn: `ALTER TABLE ... ADD COLUMN ...`.
