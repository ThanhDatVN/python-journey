## Delete Record — Xoá bản ghi

Xoá bản ghi khỏi bảng bằng câu lệnh `DELETE FROM`.

Xoá các bản ghi có địa chỉ "Hai Bà Trưng 21":

```python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="yourusername",
    password="yourpassword",
    database="mydatabase"
)

mycursor = mydb.cursor()

sql = "DELETE FROM customers WHERE address = 'Hai Bà Trưng 21'"

mycursor.execute(sql)

mydb.commit()

print(mycursor.rowcount, "record(s) deleted")
```

> ⚠️ **Quan trọng:** gọi `mydb.commit()` để lưu thay đổi.
>
> ⚠️ Chú ý mệnh đề `WHERE`: nó xác định bản ghi nào bị xoá. **Bỏ `WHERE` sẽ xoá TẤT CẢ bản ghi!**

## Prevent SQL Injection — Phòng chống SQL Injection

Dùng placeholder `%s` để escape giá trị:

```python
sql = "DELETE FROM customers WHERE address = %s"
adr = ("Lý Thường Kiệt 34", )

mycursor.execute(sql, adr)

mydb.commit()

print(mycursor.rowcount, "record(s) deleted")
```

## ➕ Thực hành ngay với SQLite

```python
import sqlite3

customers = [("An", "Lê Lợi 12"), ("Bình", "Trần Hưng Đạo 4"), ("Dũng", "Hai Bà Trưng 21"),
             ("Em", "Lý Thường Kiệt 34")]
conn = sqlite3.connect(":memory:")
cur = conn.cursor()
cur.execute("CREATE TABLE customers (id INTEGER PRIMARY KEY AUTOINCREMENT, name TEXT, address TEXT)")
cur.executemany("INSERT INTO customers (name, address) VALUES (?, ?)", customers)

cur.execute("DELETE FROM customers WHERE address = 'Hai Bà Trưng 21'")
conn.commit()
print(cur.rowcount, "record(s) deleted")

cur.execute("DELETE FROM customers WHERE address = ?", ("Lý Thường Kiệt 34",))
conn.commit()
print(cur.rowcount, "record(s) deleted")

cur.execute("SELECT * FROM customers")
print(cur.fetchall())
```

> - ⚠️ **Thiếu `WHERE` → xoá toàn bộ bảng!**
> - Dùng placeholder cho giá trị từ người dùng.
