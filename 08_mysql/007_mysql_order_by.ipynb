## Sort the Result — Sắp xếp kết quả

Dùng câu lệnh `ORDER BY` để sắp xếp kết quả **tăng dần** hoặc **giảm dần**.

Mặc định, `ORDER BY` sắp xếp **tăng dần**. Để sắp giảm dần, dùng từ khoá `DESC`.

Sắp xếp theo tên (bảng chữ cái):

```python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="yourusername",
    password="yourpassword",
    database="mydatabase"
)

mycursor = mydb.cursor()

sql = "SELECT * FROM customers ORDER BY name"

mycursor.execute(sql)

myresult = mycursor.fetchall()

for x in myresult:
    print(x)
```

## ORDER BY DESC — Sắp giảm dần

```python
sql = "SELECT * FROM customers ORDER BY name DESC"

mycursor.execute(sql)

myresult = mycursor.fetchall()

for x in myresult:
    print(x)
```

## ➕ Thực hành ngay với SQLite

```python
import sqlite3

customers = [("Chi", "Nguyễn Huệ 65"), ("An", "Lê Lợi 12"), ("Em", "Lý Thường Kiệt 34"),
             ("Bình", "Trần Hưng Đạo 4"), ("Dũng", "Hai Bà Trưng 21")]
conn = sqlite3.connect(":memory:")
cur = conn.cursor()
cur.execute("CREATE TABLE customers (id INTEGER PRIMARY KEY AUTOINCREMENT, name TEXT, address TEXT)")
cur.executemany("INSERT INTO customers (name, address) VALUES (?, ?)", customers)

cur.execute("SELECT * FROM customers ORDER BY name")
print(cur.fetchall())

cur.execute("SELECT * FROM customers ORDER BY name DESC")
print(cur.fetchall())
```

> 📌 SQLite (và MySQL với một số collation) so sánh chuỗi theo mã ký tự — tên có dấu như "Bình" có thể không đúng thứ tự từ điển tiếng Việt.

## 🧠 Ghi nhớ nhanh

> - `ORDER BY cột` → tăng dần (mặc định, `ASC`) · `ORDER BY cột DESC` → giảm dần.
