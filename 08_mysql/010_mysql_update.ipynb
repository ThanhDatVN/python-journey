## Update Table — Cập nhật bảng

Cập nhật bản ghi có sẵn bằng câu lệnh `UPDATE`.

Đổi địa chỉ "Trần Hưng Đạo 4" thành "Pasteur 123":

```python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="yourusername",
    password="yourpassword",
    database="mydatabase"
)

mycursor = mydb.cursor()

sql = "UPDATE customers SET address = 'Pasteur 123' WHERE address = 'Trần Hưng Đạo 4'"

mycursor.execute(sql)

mydb.commit()

print(mycursor.rowcount, "record(s) affected")
```

> ⚠️ Nhớ `commit()`. Và chú ý mệnh đề `WHERE`: **bỏ `WHERE` sẽ cập nhật TẤT CẢ bản ghi!**

## Prevent SQL Injection — Phòng chống SQL Injection

```python
sql = "UPDATE customers SET address = %s WHERE address = %s"
val = ("Pasteur 123", "Trần Hưng Đạo 4")

mycursor.execute(sql, val)

mydb.commit()

print(mycursor.rowcount, "record(s) affected")
```

## ➕ Thực hành ngay với SQLite

```python
import sqlite3

customers = [("An", "Lê Lợi 12"), ("Bình", "Trần Hưng Đạo 4"), ("Chi", "Nguyễn Huệ 65")]
conn = sqlite3.connect(":memory:")
cur = conn.cursor()
cur.execute("CREATE TABLE customers (id INTEGER PRIMARY KEY AUTOINCREMENT, name TEXT, address TEXT)")
cur.executemany("INSERT INTO customers (name, address) VALUES (?, ?)", customers)

cur.execute("UPDATE customers SET address = ? WHERE address = ?", ("Pasteur 123", "Trần Hưng Đạo 4"))
conn.commit()
print(cur.rowcount, "record(s) affected")

cur.execute("SELECT * FROM customers")
print(cur.fetchall())
```

## 🧠 Ghi nhớ nhanh

> - `UPDATE bảng SET cột = giá_trị WHERE điều_kiện` + `commit()`.
> - ⚠️ **Thiếu `WHERE` → cập nhật toàn bộ bảng!**
> - Dùng placeholder: `SET address = %s WHERE address = %s`.
