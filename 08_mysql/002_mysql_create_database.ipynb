## Creating a Database — Tạo cơ sở dữ liệu

Để tạo cơ sở dữ liệu trong MySQL, dùng câu lệnh `CREATE DATABASE`:

````python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="yourusername",
    password="yourpassword"
)

mycursor = mydb.cursor()

mycursor.execute("CREATE DATABASE mydatabase")
````

Nếu chạy không lỗi, bạn đã tạo thành công cơ sở dữ liệu `mydatabase`.

> 📌 **Cursor** là đối tượng dùng để **thực thi câu lệnh SQL** và đọc kết quả.

## Check if Database Exists — Kiểm tra cơ sở dữ liệu

Liệt kê mọi cơ sở dữ liệu bằng câu lệnh `SHOW DATABASES`:

````python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="yourusername",
    password="yourpassword"
)

mycursor = mydb.cursor()

mycursor.execute("SHOW DATABASES")

for x in mycursor:
    print(x)
````

Kết quả mẫu:

```text
('information_schema',)
('mydatabase',)
('mysql',)
('performance_schema',)
('sys',)
```

Hoặc kết nối **thẳng** vào một cơ sở dữ liệu bằng tham số `database` — nếu cơ sở dữ liệu không tồn tại sẽ báo lỗi:

````python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="yourusername",
    password="yourpassword",
    database="mydatabase"
)
````

## ➕ Thực hành ngay với SQLite

Với SQLite, **mỗi file là một cơ sở dữ liệu** — `connect()` tự tạo file nếu chưa có:

````python
import os
import sqlite3

conn = sqlite3.connect("mydatabase.db")
print(os.path.exists("mydatabase.db"))

cur = conn.cursor()
cur.execute("PRAGMA database_list")   # tương tự SHOW DATABASES
print(cur.fetchall())

conn.close()
os.remove("mydatabase.db")            # dọn dẹp
````

## 🧠 Ghi nhớ nhanh

> - `mycursor = mydb.cursor()` → `mycursor.execute("SQL...")`.
> - Tạo: `CREATE DATABASE tên` · liệt kê: `SHOW DATABASES`.
> - Kết nối thẳng vào database: thêm `database="tên"` khi `connect()`.
