# Python MySQL Insert Into Table

> 🎯 **Trong bài này:** Insert Into Table · Insert Multiple Rows · Get Inserted ID · ➕ Thực hành ngay với SQLite

## Insert Into Table — Chèn dữ liệu

Để thêm bản ghi vào bảng, dùng câu lệnh `INSERT INTO`.

Chèn một bản ghi vào bảng `customers`:

```python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="yourusername",
    password="yourpassword",
    database="mydatabase"
)

mycursor = mydb.cursor()

sql = "INSERT INTO customers (name, address) VALUES (%s, %s)"
val = ("An", "Lê Lợi 12")
mycursor.execute(sql, val)

mydb.commit()

print(mycursor.rowcount, "record inserted.")
```

> ⚠️ **Quan trọng:** phải gọi `mydb.commit()` — nếu không, thay đổi sẽ **không được lưu** vào bảng.

## Insert Multiple Rows — Chèn nhiều dòng

Dùng phương thức `executemany()` với tham số thứ hai là **list các tuple**:

```python
sql = "INSERT INTO customers (name, address) VALUES (%s, %s)"
val = [
    ("Bình", "Trần Hưng Đạo 4"),
    ("Chi", "Nguyễn Huệ 65"),
    ("Dũng", "Hai Bà Trưng 21"),
    ("Em", "Lý Thường Kiệt 34"),
]

mycursor.executemany(sql, val)

mydb.commit()

print(mycursor.rowcount, "was inserted.")
```

## Get Inserted ID — Lấy ID vừa chèn

Lấy `id` của bản ghi vừa chèn bằng thuộc tính `lastrowid`:

```python
sql = "INSERT INTO customers (name, address) VALUES (%s, %s)"
val = ("Giang", "Nguyễn Trãi 2")
mycursor.execute(sql, val)

mydb.commit()

print("1 record inserted, ID:", mycursor.lastrowid)
```

> 📌 Nếu chèn **nhiều** dòng, `lastrowid` trả về id của dòng **cuối cùng**.

## ➕ Thực hành ngay với SQLite

```python
import sqlite3

conn = sqlite3.connect(":memory:")
cur = conn.cursor()
cur.execute("CREATE TABLE customers (id INTEGER PRIMARY KEY AUTOINCREMENT, name TEXT, address TEXT)")

# Một dòng — placeholder của SQLite là ?
cur.execute("INSERT INTO customers (name, address) VALUES (?, ?)", ("An", "Lê Lợi 12"))
conn.commit()
print(cur.rowcount, "record inserted.")

# Nhiều dòng
val = [
    ("Bình", "Trần Hưng Đạo 4"),
    ("Chi", "Nguyễn Huệ 65"),
    ("Dũng", "Hai Bà Trưng 21"),
    ("Em", "Lý Thường Kiệt 34"),
]
cur.executemany("INSERT INTO customers (name, address) VALUES (?, ?)", val)
conn.commit()
print(cur.rowcount, "was inserted.")

# ID vừa chèn
cur.execute("INSERT INTO customers (name, address) VALUES (?, ?)", ("Giang", "Nguyễn Trãi 2"))
conn.commit()
print("1 record inserted, ID:", cur.lastrowid)
```

---

## 🧠 Ghi nhớ nhanh

> - `INSERT INTO bảng (cột, ...) VALUES (%s, %s)` + `execute(sql, val)`.
> - ⚠️ Luôn `mydb.commit()` để lưu thay đổi.
> - Nhiều dòng: `executemany(sql, list_tuple)` · ID vừa chèn: `cursor.lastrowid` · số dòng: `cursor.rowcount`.
