## Join Two or More Tables — Kết hợp bảng

Kết hợp các dòng từ hai hay nhiều bảng dựa trên **cột liên quan** giữa chúng, bằng câu lệnh `JOIN`.

Giả sử có bảng `users` và bảng `products`:

**users**

```text
{ id: 1, name: 'An',   fav: 154 },
{ id: 2, name: 'Bình', fav: 154 },
{ id: 3, name: 'Chi',  fav: 155 },
{ id: 4, name: 'Dũng', fav: NULL },
{ id: 5, name: 'Em',   fav: NULL }
```

**products**

```text
{ id: 154, name: 'Bánh mì' },
{ id: 155, name: 'Phở' },
{ id: 156, name: 'Cà phê' }
```

Hai bảng có thể kết hợp bằng trường `fav` của users và trường `id` của products.

Kết hợp users và products để xem **món yêu thích** của từng người:

````python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="yourusername",
    password="yourpassword",
    database="mydatabase"
)

mycursor = mydb.cursor()

sql = "SELECT \
  users.name AS user, \
  products.name AS favorite \
  FROM users \
  INNER JOIN products ON users.fav = products.id"

mycursor.execute(sql)

myresult = mycursor.fetchall()

for x in myresult:
    print(x)
````

> 📌 Có thể dùng `JOIN` thay cho `INNER JOIN` — kết quả như nhau.

## LEFT JOIN

Ở ví dụ trên, Dũng và Em **không** có trong kết quả, vì `INNER JOIN` chỉ hiển thị các bản ghi **khớp ở cả hai bảng**.

Muốn hiển thị **tất cả** users, kể cả không có món yêu thích, dùng `LEFT JOIN`:

````python
sql = "SELECT \
  users.name AS user, \
  products.name AS favorite \
  FROM users \
  LEFT JOIN products ON users.fav = products.id"
````

## RIGHT JOIN

Muốn hiển thị **tất cả** products, và người dùng thích chúng (kể cả khi không ai thích), dùng `RIGHT JOIN`:

````python
sql = "SELECT \
  users.name AS user, \
  products.name AS favorite \
  FROM users \
  RIGHT JOIN products ON users.fav = products.id"
````

> 📌 Dũng và Em — những người không có món yêu thích — **không** có trong kết quả RIGHT JOIN.

## ➕ Thực hành ngay với SQLite

````python
import sqlite3

conn = sqlite3.connect(":memory:")
cur = conn.cursor()
cur.execute("CREATE TABLE users (id INTEGER PRIMARY KEY, name TEXT, fav INTEGER)")
cur.execute("CREATE TABLE products (id INTEGER PRIMARY KEY, name TEXT)")
cur.executemany("INSERT INTO users VALUES (?, ?, ?)",
                [(1, "An", 154), (2, "Bình", 154), (3, "Chi", 155), (4, "Dũng", None), (5, "Em", None)])
cur.executemany("INSERT INTO products VALUES (?, ?)", [(154, "Bánh mì"), (155, "Phở"), (156, "Cà phê")])

base = "SELECT users.name AS user, products.name AS favorite FROM users {} products ON users.fav = products.id"

for join in ("INNER JOIN", "LEFT JOIN", "RIGHT JOIN"):
    cur.execute(base.format(join))
    print(f"{join:<11}", cur.fetchall())
````

> 💡 `RIGHT JOIN` được SQLite hỗ trợ từ phiên bản 3.39.

---

## 🧠 Ghi nhớ nhanh

> - `INNER JOIN` (hoặc `JOIN`) → chỉ các dòng **khớp ở cả hai** bảng.
> - `LEFT JOIN` → **mọi** dòng bảng trái (+ dữ liệu khớp ở bảng phải, nếu có).
> - `RIGHT JOIN` → **mọi** dòng bảng phải.
> - Cú pháp: `FROM a JOIN b ON a.cột = b.cột`.
