Python có thể dùng trong các ứng dụng **cơ sở dữ liệu**.

Một trong những cơ sở dữ liệu phổ biến nhất là **MySQL**.

## MySQL Database — Cơ sở dữ liệu MySQL

Để thực hành các ví dụ trong chương này, máy bạn cần có MySQL. Tải MySQL miễn phí tại [mysql.com/downloads](https://www.mysql.com/downloads/).

> 💡 **Không muốn cài MySQL?** Mỗi bài trong chương có thêm mục **➕ Thực hành ngay với SQLite** — dùng module `sqlite3` có sẵn trong Python, cú pháp SQL gần như giống hệt. Bạn chạy được ngay trong notebook.

## Install MySQL Driver — Cài driver

Python cần một **driver** để truy cập MySQL. Ta dùng driver "MySQL Connector":

```bash
pip install mysql-connector-python
```

## Test MySQL Connector — Kiểm tra driver

Nếu câu lệnh sau chạy **không báo lỗi**, driver đã được cài đúng:

````python
import mysql.connector
````

## Create Connection — Tạo kết nối

Bắt đầu bằng việc tạo kết nối tới cơ sở dữ liệu, dùng **tên đăng nhập** và **mật khẩu** MySQL của bạn:

````python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="yourusername",
    password="yourpassword"
)

print(mydb)
````

Kết quả mẫu:

```text
<mysql.connector.connection_cext.CMySQLConnection object at 0x...>
```

Giờ bạn có thể truy vấn cơ sở dữ liệu bằng câu lệnh **SQL**.

> ⚠️ Đừng ghi mật khẩu thật vào notebook rồi đưa lên GitHub! Hãy đọc từ biến môi trường: `password=os.environ["MYSQL_PASSWORD"]`.

## ➕ Thực hành ngay với SQLite

SQLite lưu cả cơ sở dữ liệu trong **một file** (hoặc trong bộ nhớ) — không cần server:

````python
import sqlite3

conn = sqlite3.connect(":memory:")   # cơ sở dữ liệu tạm trong RAM
print(conn)
print("SQLite version:", sqlite3.sqlite_version)
conn.close()
````

| | MySQL | SQLite |
|---|---|---|
| Cài đặt | Cần server | Có sẵn trong Python |
| Kết nối | `mysql.connector.connect(host, user, password)` | `sqlite3.connect("file.db")` |
| Placeholder | `%s` | `?` |
| Tự tăng khoá | `INT AUTO_INCREMENT PRIMARY KEY` | `INTEGER PRIMARY KEY AUTOINCREMENT` |
