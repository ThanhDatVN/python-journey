## Select With a Filter — Chọn có điều kiện

Khi chọn bản ghi, có thể **lọc** bằng câu lệnh `WHERE`.

Chọn các bản ghi có địa chỉ là "Lê Lai 38":

```python
import mysql.connector

mydb = mysql.connector.connect(
    host="localhost",
    user="yourusername",
    password="yourpassword",
    database="mydatabase"
)

mycursor = mydb.cursor()

sql = "SELECT * FROM customers WHERE address = 'Lê Lai 38'"

mycursor.execute(sql)

myresult = mycursor.fetchall()

for x in myresult:
    print(x)
```

## Wildcard Characters — Ký tự đại diện

Có thể chọn bản ghi **bắt đầu, chứa, hoặc kết thúc** bằng một chuỗi. Dùng `%` đại diện cho **ký tự bất kỳ** và toán tử `LIKE`.

Chọn các bản ghi có địa chỉ **chứa** chữ "Lê":

```python
sql = "SELECT * FROM customers WHERE address LIKE '%Lê%'"

mycursor.execute(sql)

myresult = mycursor.fetchall()

for x in myresult:
    print(x)
```

## Prevent SQL Injection — Phòng chống SQL Injection

Khi giá trị truy vấn do **người dùng nhập**, bạn phải **escape** giá trị đó.

Việc này ngăn chặn **SQL injection** — một kỹ thuật tấn công web phổ biến nhằm phá hoại hoặc chiếm quyền cơ sở dữ liệu.

MySQL Connector dùng placeholder `%s` để escape giá trị:

```python
sql = "SELECT * FROM customers WHERE address = %s"
adr = ("Nguyễn Huệ 65", )

mycursor.execute(sql, adr)

myresult = mycursor.fetchall()

for x in myresult:
    print(x)
```

## ➕ Thực hành ngay với SQLite

```python
import sqlite3

customers = [
    ("An", "Lê Lợi 12"), ("Bình", "Trần Hưng Đạo 4"), ("Chi", "Nguyễn Huệ 65"),
    ("Dũng", "Hai Bà Trưng 21"), ("Em", "Lý Thường Kiệt 34"), ("Giang", "Nguyễn Trãi 2"),
    ("Hà", "Lê Duẩn 1"), ("Khoa", "Pasteur 33"), ("Lan", "Điện Biên Phủ 98"),
    ("Minh", "Võ Văn Tần 2"), ("Nam", "Lê Lai 38"), ("Oanh", "Phan Xích Long 95"),
]
conn = sqlite3.connect(":memory:")
cur = conn.cursor()
cur.execute("CREATE TABLE customers (id INTEGER PRIMARY KEY AUTOINCREMENT, name TEXT, address TEXT)")
cur.executemany("INSERT INTO customers (name, address) VALUES (?, ?)", customers)

cur.execute("SELECT * FROM customers WHERE address = 'Lê Lai 38'")
print(cur.fetchall())

cur.execute("SELECT * FROM customers WHERE address LIKE '%Lê%'")
print(cur.fetchall())

cur.execute("SELECT * FROM customers WHERE address = ?", ("Nguyễn Huệ 65",))
print(cur.fetchall())
```

Vì sao **không bao giờ** ghép chuỗi trực tiếp từ dữ liệu người dùng? Xem "tấn công" sau:

```python
user_input = "x' OR '1'='1"

# ❌ Nguy hiểm: ghép chuỗi
cur.execute(f"SELECT * FROM customers WHERE address = '{user_input}'")
print("Ghép chuỗi trả về", len(cur.fetchall()), "dòng — lộ toàn bộ bảng!")

# ✅ An toàn: dùng placeholder
cur.execute("SELECT * FROM customers WHERE address = ?", (user_input,))
print("Placeholder trả về", len(cur.fetchall()), "dòng")
```

## 🧠 Ghi nhớ nhanh

> - Lọc: `SELECT ... WHERE cột = 'giá trị'`.
> - `LIKE '%chữ%'` — `%` đại diện cho chuỗi ký tự bất kỳ.
> - ⚠️ Dữ liệu người dùng → luôn dùng **placeholder** (`%s` MySQL, `?` SQLite), **không** ghép chuỗi → chống SQL injection.
