## Loop Through a Dictionary — Duyệt dictionary

Duyệt dictionary bằng vòng lặp `for`. Mặc định, biến lặp nhận lần lượt các **key**.

In tất cả các key:

````python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
for x in thisdict:
    print(x)
````

In tất cả các value:

````python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
for x in thisdict:
    print(thisdict[x])
````

Dùng phương thức `values()` để lấy value:

````python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
for x in thisdict.values():
    print(x)
````

Dùng phương thức `keys()` để lấy key:

````python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
for x in thisdict.keys():
    print(x)
````

Duyệt **cả key và value** bằng `items()`:

````python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
for x, y in thisdict.items():
    print(x, y)
````

## ➕ Mở rộng: Dictionary comprehension

Giống list comprehension, nhưng tạo ra dictionary:

````python
prices = {"apple": 10, "banana": 5, "cherry": 20}
discounted = {k: v * 0.9 for k, v in prices.items()}
print(discounted)

expensive = {k: v for k, v in prices.items() if v > 8}
print(expensive)
````

> 🧪 **Try it Yourself** — In bảng điểm theo dạng "Tên: điểm" và tính điểm trung bình.

````python
scores = {"An": 8.5, "Bình": 7.0, "Chi": 9.5}
for name, score in scores.items():
    print(f"{name}: {score}")
print("Trung bình:", sum(scores.values()) / len(scores))
````

---

## 🧠 Ghi nhớ nhanh

> - `for k in d:` → duyệt **key** (mặc định).
> - `for v in d.values():` → duyệt **value**.
> - `for k, v in d.items():` → duyệt **cả hai** (hay dùng nhất).
> - Dict comprehension: `{k: v for k, v in d.items() if ...}`.
