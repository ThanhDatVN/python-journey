## Dictionary

Dictionary (từ điển) dùng để lưu dữ liệu theo cặp **key: value** (khoá: giá trị).

Dictionary là tập hợp **có thứ tự**\*, **thay đổi được** và **không cho phép key trùng lặp**.

> \* Từ Python 3.7, dictionary **có thứ tự**. Ở Python 3.6 trở về trước, dictionary không có thứ tự.

Dictionary được viết bằng dấu ngoặc nhọn, mỗi phần tử là một cặp `key: value`:

````python
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}
print(thisdict)
````

## Dictionary Items — Phần tử của dictionary

Phần tử dictionary **có thứ tự**, **thay đổi được**, và **không cho phép trùng key**. Mỗi phần tử được truy cập bằng **tên key**:

````python
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}
print(thisdict["brand"])
````

## Ordered or Unordered? — Có thứ tự hay không?

*Có thứ tự* nghĩa là các phần tử giữ đúng thứ tự khi được thêm vào, và thứ tự đó không tự thay đổi. Tuy vậy, bạn **không** truy cập được bằng chỉ số như list (`thisdict[0]` sẽ tìm key tên là `0`).

## Changeable — Thay đổi được

Có thể sửa, thêm, xoá phần tử sau khi dictionary đã được tạo.

## Duplicates Not Allowed — Không trùng key

Một dictionary không thể có hai phần tử cùng key. Key trùng sẽ **ghi đè** giá trị cũ:

````python
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964,
    "year": 2020
}
print(thisdict)
````

## Dictionary Length — Độ dài dictionary

````python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(len(thisdict))
````

## Dictionary Items - Data Types — Kiểu dữ liệu của value

Value có thể thuộc bất kỳ kiểu dữ liệu nào:

````python
thisdict = {
    "brand": "Ford",
    "electric": False,
    "year": 1964,
    "colors": ["red", "white", "blue"]
}
print(thisdict)
````

> 📌 **Key** phải là kiểu **bất biến** (chuỗi, số, tuple…). List không làm key được.

## type()

````python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(type(thisdict))
````

## The dict() Constructor — Hàm khởi tạo dict()

````python
thisdict = dict(name="John", age=36, country="Norway")
print(thisdict)
````

## Python Collections (Arrays) — So sánh 4 kiểu tập hợp

| Kiểu | Thứ tự | Thay đổi được | Trùng lặp |
|---|---|---|---|
| **List** `[ ]` | Có | Có | Cho phép |
| **Tuple** `( )` | Có | Không | Cho phép |
| **Set** `{ }` | Không | Thêm/xoá được | Không |
| **Dictionary** `{k: v}` | Có | Có | Không trùng key |

> 🧪 **Try it Yourself** — Tạo dictionary mô tả bản thân bạn: tên, tuổi, thành phố, sở thích (list).

````python
me = {
    "name": "An",
    "age": 20,
    "city": "Đà Nẵng",
    "hobbies": ["đọc sách", "bóng đá"]
}
print(me)
print(len(me))
````

---

## 🧠 Ghi nhớ nhanh

> - Dictionary: `{key: value}` — **có thứ tự** (từ 3.7) · **thay đổi được** · **không trùng key**.
> - Truy cập bằng **key**: `d["brand"]` (không dùng chỉ số).
> - Key trùng → giá trị sau **ghi đè** giá trị trước.
> - Key phải **bất biến** (str, int, tuple…); value thì kiểu gì cũng được.
