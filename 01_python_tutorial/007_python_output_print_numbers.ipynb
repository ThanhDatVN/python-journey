## Print Numbers — In số

Bạn cũng dùng `print()` để in số. Khác với văn bản, **số không đặt trong dấu nháy**:

```python
print(3)
print(358)
print(50000)
```

### Calculations — Tính toán trực tiếp

Có thể làm phép tính ngay bên trong `print()`; Python tính xong rồi mới in kết quả:

```python
print(3 + 3)
print(2 * 5)
print(10 - 4)
```

So sánh sự khác biệt khi đặt phép tính trong dấu nháy:

```python
print(3 + 3)     # tính toán -> 6
print("3 + 3")   # chỉ là văn bản -> 3 + 3
```

## Mix Text and Numbers — Kết hợp chữ và số

Dùng **dấu phẩy** để in nhiều giá trị cùng lúc. Python tự chèn một khoảng trắng giữa các giá trị:

```python
print("I am", 35, "years old.")
```

Muốn đổi ký tự ngăn cách, dùng tham số `sep`:

```python
print(2026, 10, 3, sep="-")
print("a", "b", "c", sep="")
```

> ⚠️ Không thể nối trực tiếp chuỗi và số bằng dấu `+` — bạn sẽ học cách xử lý trong bài *Concatenate Strings* và *Format Strings*.

```python
print("I am " + 35 + " years old.")
```

> 🧪 **Try it Yourself** — Hãy in ra tuổi của bạn và năm hiện tại theo dạng: Tôi 20 tuổi, năm nay là 2026

```python
print("Tôi", 20, "tuổi, năm nay là", 2026)
```

---

## 🧠 Ghi nhớ nhanh

> - Số **không** đặt trong dấu nháy: `print(3 + 3)` → `6`, còn `print("3 + 3")` → `3 + 3`.
> - Dùng **dấu phẩy** để in nhiều giá trị: `print("I am", 35)` — tự thêm khoảng trắng.
> - `sep="-"` đổi ký tự ngăn cách giữa các giá trị.
> - ⚠️ `"text" + 35` → `TypeError` (không cộng chuỗi với số).
