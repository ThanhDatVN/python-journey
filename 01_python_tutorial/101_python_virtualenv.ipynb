## What is a Virtual Environment? — Môi trường ảo là gì?

**Môi trường ảo (virtual environment)** là một "không gian" **tách biệt** cho từng dự án Python: có trình thông dịch, package và phiên bản package **riêng**.

Nhờ vậy, dự án A dùng thư viện phiên bản 1.0 và dự án B dùng phiên bản 2.0 mà **không xung đột** nhau.

> 📌 Python có sẵn module **`venv`** để tạo môi trường ảo.

## Creating a Virtual Environment — Tạo môi trường ảo

Mở terminal tại thư mục dự án và chạy:

```bash
python -m venv myfirstproject
```

Lệnh này tạo thư mục `myfirstproject` với cấu trúc:

```text
myfirstproject/
├── Include/
├── Lib/
├── Scripts/        (macOS/Linux: bin/)
├── .gitignore
└── pyvenv.cfg
```

## Activate Virtual Environment — Kích hoạt

Windows:

```bash
myfirstproject\Scripts\activate
```

macOS/Linux:

```bash
source myfirstproject/bin/activate
```

Sau khi kích hoạt, dấu nhắc lệnh có thêm tên môi trường ở đầu:

```text
(myfirstproject) C:\Users\Your Name>
```

## Install Packages — Cài package

Khi môi trường đang được kích hoạt, `pip install` sẽ cài package **vào môi trường này**:

```bash
pip install cowsay
```

## Using Package — Dùng package

Tạo file `test.py`:

```text
import cowsay

cowsay.cow("Good Mooooorning!")
```

Chạy `python test.py` — một chú bò ASCII sẽ "nói" câu chào. 🐮

## Deactivate Virtual Environment — Thoát môi trường

```bash
deactivate
```

Sau khi thoát, chạy lại `python test.py` sẽ báo lỗi `ModuleNotFoundError: No module named 'cowsay'` — vì `cowsay` chỉ được cài **trong** môi trường ảo.

## Delete Virtual Environment — Xoá môi trường ảo

Môi trường ảo chỉ là một thư mục — xoá thư mục là xong:

Windows:

```bash
rmdir /s /q myfirstproject
```

macOS/Linux:

```bash
rm -rf myfirstproject
```

## ➕ Mở rộng: Kiểm tra notebook đang chạy trong môi trường nào

````python
import sys

print(sys.executable)          # đường dẫn trình thông dịch đang dùng
print(sys.prefix != sys.base_prefix and "Đang trong môi trường ảo" or "Không dùng môi trường ảo")
````

## 🧠 Ghi nhớ nhanh

> - Môi trường ảo = không gian **riêng** cho từng dự án → tránh xung đột phiên bản package.
> - Tạo: `python -m venv ten_moi_truong`.
> - Kích hoạt: `ten\Scripts\activate` (Windows) · `source ten/bin/activate` (macOS/Linux).
> - Thoát: `deactivate` · Xoá: xoá thư mục môi trường.
