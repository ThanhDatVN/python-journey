## Access Items — Truy cập phần tử

Phần tử của list được đánh chỉ số; truy cập bằng cách ghi chỉ số trong ngoặc vuông:

````python
thislist = ["apple", "banana", "cherry"]
print(thislist[1])
````

> 📌 Phần tử đầu tiên có chỉ số `0`.

## Negative Indexing — Chỉ số âm

Chỉ số âm đếm **từ cuối**: `-1` là phần tử cuối cùng, `-2` là phần tử kế cuối…

````python
thislist = ["apple", "banana", "cherry"]
print(thislist[-1])
````

## Range of Indexes — Khoảng chỉ số

Lấy một khoảng phần tử bằng cách chỉ định vị trí bắt đầu và kết thúc. Kết quả là **một list mới**:

````python
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(thislist[2:5])
````

> Khoảng lấy từ chỉ số 2 (**bao gồm**) đến chỉ số 5 (**không bao gồm**).

Bỏ trống vị trí bắt đầu → lấy từ phần tử đầu tiên:

````python
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(thislist[:4])
````

Bỏ trống vị trí kết thúc → lấy đến hết list:

````python
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(thislist[2:])
````

## Range of Negative Indexes — Khoảng chỉ số âm

Lấy từ `"orange"` (-4) đến trước `"mango"` (-1):

````python
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(thislist[-4:-1])
````

## Check if Item Exists — Kiểm tra phần tử có tồn tại

Dùng từ khoá `in`:

````python
thislist = ["apple", "banana", "cherry"]
if "apple" in thislist:
    print("Yes, 'apple' is in the fruits list")
````

## ➕ Mở rộng: Truy cập ngoài phạm vi

Chỉ số vượt quá độ dài list sẽ gây lỗi `IndexError`:

````python
thislist = ["apple", "banana", "cherry"]
print(thislist[3])
````

> 🧪 **Try it Yourself** — Lấy ra 3 phần tử ở giữa list (không lấy phần tử đầu và cuối).

````python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:4])
````

---

## 🧠 Ghi nhớ nhanh

> - `a[0]` phần tử đầu · `a[-1]` phần tử cuối.
> - `a[2:5]` lấy từ chỉ số 2 đến **trước** 5 → trả về **list mới**.
> - `x in a` kiểm tra tồn tại.
> - ⚠️ Chỉ số vượt phạm vi → `IndexError`.
