## Unpacking a Tuple — Giải nén tuple

Khi tạo tuple, ta thường gán giá trị cho nó. Việc này gọi là **"đóng gói" (packing)**:

```python
fruits = ("apple", "banana", "cherry")
```

Python cũng cho phép **"giải nén" (unpacking)**: lấy các giá trị ra thành từng biến riêng:

```python
fruits = ("apple", "banana", "cherry")

(green, yellow, red) = fruits

print(green)
print(yellow)
print(red)
```

> 📌 Số biến phải **bằng** số phần tử của tuple, nếu không bạn phải dùng dấu `*` để gom phần còn lại vào một list.

```python
fruits = ("apple", "banana", "cherry")
(green, yellow) = fruits
```

## Using Asterisk `*` — Dùng dấu hoa thị

Nếu số biến **ít hơn** số phần tử, thêm `*` vào trước một biến — các giá trị còn lại sẽ được gom thành **list**:

```python
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")

(green, yellow, *red) = fruits

print(green)
print(yellow)
print(red)
```

Nếu dấu `*` đặt ở biến khác (không phải biến cuối), Python sẽ gán giá trị sao cho số phần tử còn lại khớp với số biến phía sau:

```python
fruits = ("apple", "mango", "papaya", "pineapple", "cherry")

(green, *tropic, red) = fruits

print(green)
print(tropic)
print(red)
```

## ➕ Mở rộng: Ứng dụng thực tế

Hàm trả về nhiều giá trị (thực chất là trả về một tuple):

```python
def min_max(numbers):
    return min(numbers), max(numbers)

low, high = min_max([4, 8, 1, 9])
print(low, high)
```

Duyệt list các tuple:

```python
students = [("An", 8.5), ("Bình", 7.0), ("Chi", 9.2)]
for name, score in students:
    print(f"{name}: {score}")
```

Bỏ qua giá trị không cần bằng biến `_` (quy ước):

```python
point = (3, 7, 10)
x, _, z = point
print(x, z)
```

> 🧪 **Try it Yourself** — Giải nén tuple thông tin sinh viên thành 3 biến rồi in câu giới thiệu.

```python
student = ("Nguyễn An", 20, "CNTT")
name, age, major = student
print(f"{name}, {age} tuổi, ngành {major}")
```

---

## 🧠 Ghi nhớ nhanh

> - Packing: `t = (1, 2, 3)` · Unpacking: `a, b, c = t`.
> - ⚠️ Số biến phải khớp số phần tử, hoặc dùng `*`: `a, *rest = t` (`rest` là **list**).
> - `_` để bỏ qua giá trị không cần: `x, _, z = point`.
