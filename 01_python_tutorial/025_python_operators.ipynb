## Python Operators — Toán tử

**Toán tử (operator)** dùng để thực hiện phép toán trên biến và giá trị.

Ví dụ dưới đây dùng toán tử `+` để cộng hai giá trị:

```python
print(10 + 5)
```

Toán tử hoạt động với cả giá trị trực tiếp lẫn biến, và có thể kết hợp nhiều phép:

```python
sum1 = 100 + 50       # 150 (100 + 50)
sum2 = sum1 + 250     # 400 (150 + 250)
sum3 = sum2 + sum2    # 800 (400 + 400)

print(sum1, sum2, sum3)
```

## Operator Groups — Các nhóm toán tử

Python chia toán tử thành các nhóm sau — mỗi nhóm có một bài riêng:

| Nhóm | Ví dụ | Bài học |
|---|---|---|
| Arithmetic — Số học | `+  -  *  /  %  **  //` | [Arithmetic Operators](026_arithmetic_operators.ipynb) |
| Assignment — Gán | `=  +=  -=  *=  :=` | [Assignment Operators](027_assignment_operators.ipynb) |
| Ternary — Điều kiện rút gọn | `a if điều_kiện else b` | [Ternary Operator](028_ternary_operator.ipynb) |
| Comparison — So sánh | `==  !=  >  <  >=  <=` | [Comparison Operators](029_comparison_operators.ipynb) |
| Logical — Logic | `and  or  not` | [Logical Operators](030_logical_operators.ipynb) |
| Identity — Định danh | `is  is not` | [Identity Operators](031_identity_operators.ipynb) |
| Membership — Thành viên | `in  not in` | [Membership Operators](032_membership_operators.ipynb) |
| Bitwise — Thao tác bit | `&  \|  ^  ~  <<  >>` | [Bitwise Operators](033_bitwise_operators.ipynb) |

Khi nhiều toán tử xuất hiện trong cùng một biểu thức, thứ tự tính được quyết định bởi [Operator Precedence](034_operator_precedence.ipynb).

### Xem nhanh mỗi nhóm một ví dụ

```python
x = 7
y = 3

print(x + y)            # arithmetic
x += 1                  # assignment -> x = 8
print(x)
print(x > y)            # comparison
print(x > 5 and y > 5)  # logical
print(3 in [1, 2, 3])   # membership
print(x is x)           # identity
print(6 & 3)            # bitwise
```

> 🧪 **Try it Yourself** — Thay đổi giá trị x, y rồi chạy lại để thấy kết quả thay đổi.

```python
x = 15
y = 4

print(x + y, x - y, x * y, x / y)
```
