## Bitwise Operators — Toán tử thao tác bit

Toán tử bitwise làm việc trên **biểu diễn nhị phân** của số nguyên, xử lý từng bit một:

| Toán tử | Tên | Mô tả | Ví dụ |
|---|---|---|---|
| `&` | AND | Bit = 1 nếu **cả hai** bit là 1 | `x & y` |
| `\|` | OR | Bit = 1 nếu **một trong hai** bit là 1 | `x \| y` |
| `^` | XOR | Bit = 1 nếu **chỉ một** trong hai bit là 1 | `x ^ y` |
| `~` | NOT | Đảo tất cả các bit | `~x` |
| `<<` | Zero fill left shift | Dịch trái, chèn 0 vào bên phải | `x << 2` |
| `>>` | Signed right shift | Dịch phải, giữ bit dấu | `x >> 2` |

Dùng `bin()` để xem dạng nhị phân của một số:

````python
print(bin(6))
print(bin(3))
````

## AND — `&`

```text
6 = 110
3 = 011
--------
2 = 010
```

````python
print(6 & 3)
````

## OR — `|`

```text
6 = 110
3 = 011
--------
7 = 111
```

````python
print(6 | 3)
````

## XOR — `^`

```text
6 = 110
3 = 011
--------
5 = 101
```

````python
print(6 ^ 3)
````

## NOT — `~`

`~x` cho kết quả `-(x + 1)` (do cách máy tính biểu diễn số âm — *bù hai*):

````python
print(~3)
````

## Shift — Dịch bit

Dịch trái `n` bit tương đương **nhân** với `2**n`; dịch phải tương đương **chia lấy phần nguyên** cho `2**n`:

````python
print(3 << 2)   # 0011 -> 1100 = 12
print(8 >> 2)   # 1000 -> 0010 = 2
````

## ➕ Mở rộng: Ứng dụng cờ quyền (flags)

````python
READ = 0b100
WRITE = 0b010
EXECUTE = 0b001

permission = READ | WRITE        # gộp quyền
print(bin(permission))
print(bool(permission & WRITE))  # có quyền ghi?
print(bool(permission & EXECUTE))
````

> 🧪 **Try it Yourself** — Đổi giá trị a, b và quan sát kết quả ở dạng nhị phân.

````python
a = 12
b = 10

print(bin(a), bin(b))
print("AND:", bin(a & b))
print("OR: ", bin(a | b))
print("XOR:", bin(a ^ b))
````

---

## 🧠 Ghi nhớ nhanh

> - `&` AND · `|` OR · `^` XOR · `~` NOT · `<<` dịch trái · `>>` dịch phải.
> - Xem dạng nhị phân: `bin(6)` → `'0b110'`.
> - `x << n` = `x * 2**n` · `x >> n` = `x // 2**n` · `~x` = `-(x + 1)`.
