## Python Logical Operators — Toán tử logic trong điều kiện

Toán tử logic dùng để **kết hợp nhiều điều kiện** trong câu lệnh `if`. Python có ba toán tử logic:

- `and` — `True` nếu **cả hai** điều kiện đúng
- `or` — `True` nếu **ít nhất một** điều kiện đúng
- `not` — **đảo ngược** kết quả

## The and Operator — Toán tử and

Kiểm tra `a` lớn hơn `b`, **và** `c` lớn hơn `a`:

```python
a = 200
b = 33
c = 500
if a > b and c > a:
    print("Both conditions are True")
```

## The or Operator — Toán tử or

Kiểm tra `a` lớn hơn `b`, **hoặc** `a` lớn hơn `c`:

```python
a = 200
b = 33
c = 500
if a > b or a > c:
    print("At least one of the conditions is True")
```

## The not Operator — Toán tử not

Đảo ngược kết quả của điều kiện:

```python
a = 33
b = 200
if not a > b:
    print("a is NOT greater than b")
```

## Combining Multiple Operators — Kết hợp nhiều toán tử

Có thể kết hợp `and`, `or`, `not` trong một điều kiện. Thứ tự ưu tiên: `not` → `and` → `or`:

```python
age = 25
is_student = False
has_discount_code = True

if (age < 30 and is_student) or has_discount_code:
    print("You get a discount!")
```

## Truth Tables — Bảng chân trị

| A | B | `A and B` | `A or B` |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

```python
print(True and False)
print(True or False)
print(not True)
```

## Using Parentheses for Clarity — Dùng ngoặc cho rõ ràng

Với điều kiện phức tạp, **dấu ngoặc** giúp code dễ đọc và tránh nhầm thứ tự ưu tiên:

```python
temperature = 25
is_raining = False
is_weekend = True

if (temperature > 20 and not is_raining) or is_weekend:
    print("Great day for outdoor activities!")
```

## More Examples — Thêm ví dụ

Đăng nhập:

```python
username = "admin"
password = "secret123"

if username == "admin" and password == "secret123":
    print("Login successful")
else:
    print("Invalid credentials")
```

Kiểm tra khoảng giá trị:

```python
number = 15
if number > 0 and number < 100:
    print("Number is between 0 and 100")
```

Kiểm tra nhiều khả năng:

```python
day = "Saturday"
if day == "Saturday" or day == "Sunday":
    print("It's the weekend!")
```

> 🧪 **Try it Yourself** — Chỉ cho vào rạp nếu tuổi >= 18 HOẶC đi cùng phụ huynh, VÀ phải có vé.

```python
age = 15
with_parent = True
has_ticket = True

if (age >= 18 or with_parent) and has_ticket:
    print("Mời vào!")
else:
    print("Không được vào.")
```

## 🧠 Ghi nhớ nhanh

> - `and` cả hai đúng · `or` một trong hai đúng · `not` đảo ngược.
> - Ưu tiên: `not` → `and` → `or`; điều kiện phức tạp → **thêm ngoặc**.
> - ⚠️ Viết `if day == "Sat" or day == "Sun"`, **không** viết `if day == "Sat" or "Sun"` (luôn `True`!).

- **B.** Chỉ đúng khi `day` là `"Sun"`

<summary>👉 Xem đáp án</summary>

✅ **Đáp án: D** — Luôn được đánh giá là `True` vì chuỗi `"Sun"` là truthy

Python hiểu là `(day == "Sat") or ("Sun")`. Viết đúng: `day == "Sat" or day == "Sun"`, hoặc `day in ("Sat", "Sun")`.
