## Removing Items — Xoá phần tử

Có nhiều cách để xoá phần tử khỏi dictionary.

`pop()` xoá phần tử có **key chỉ định** (và trả về value của nó):

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
removed = thisdict.pop("model")
print(thisdict)
print(removed)
```

`popitem()` xoá phần tử **được thêm vào cuối cùng** (ở Python trước 3.7, nó xoá một phần tử ngẫu nhiên):

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict.popitem()
print(thisdict)
```

Từ khoá `del` xoá phần tử có key chỉ định:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
del thisdict["model"]
print(thisdict)
```

`del` cũng xoá được **toàn bộ** dictionary:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
del thisdict
print(thisdict)  # lỗi vì "thisdict" không còn tồn tại
```

`clear()` làm rỗng dictionary:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict.clear()
print(thisdict)
```

## ➕ Mở rộng: pop() với giá trị mặc định

Xoá key không tồn tại sẽ gây lỗi — trừ khi bạn truyền giá trị mặc định cho `pop()`:

```python
d = {"a": 1}
d.pop("b")
```

```python
d = {"a": 1}
print(d.pop("b", "không có"))
```

> 🧪 **Try it Yourself** — Xoá một sản phẩm đã hết hàng khỏi giỏ hàng và in ra tên sản phẩm bị xoá.

```python
cart = {"áo": 2, "quần": 1, "mũ": 3}
item = "quần"
qty = cart.pop(item)
print(f"Đã xoá {qty} {item}")
print(cart)
```

---

## 🧠 Ghi nhớ nhanh

> - `pop(key)` → xoá theo key, **trả về value**; `pop(key, mặc_định)` để không lỗi.
> - `popitem()` → xoá phần tử **thêm vào cuối cùng**.
> - `del d[key]` xoá một phần tử · `del d` xoá cả biến · `clear()` làm rỗng.
