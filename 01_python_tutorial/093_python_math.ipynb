# Python Math

> 📖 **Học song song trên W3Schools:** [python_math.asp](https://www.w3schools.com/python/python_math.asp)
>
> 🎯 **Trong bài này:** Built-in Math Functions · ➕ Mở rộng: round(), sum(), divmod() · The Math Module

Python có sẵn một số **hàm toán học**, cùng với module **`math`** mở rộng giúp thực hiện các phép toán trên số.

## Built-in Math Functions — Hàm toán học có sẵn

`min()` và `max()` tìm giá trị nhỏ nhất/lớn nhất trong một iterable:

```python
x = min(5, 10, 25)
y = max(5, 10, 25)

print(x)
print(y)
```

`abs()` trả về giá trị tuyệt đối (luôn dương):

```python
x = abs(-7.25)
print(x)
```

`pow(x, y)` trả về `x` mũ `y`:

```python
x = pow(4, 3)
print(x)
```

## ➕ Mở rộng: round(), sum(), divmod()

```python
print(round(3.14159, 2))
print(round(2.5), round(3.5))   # làm tròn "về số chẵn" (banker's rounding)
print(sum([1, 2, 3, 4]))
print(divmod(17, 5))            # (thương, dư)
```

## The Math Module — Module math

Python có module `math` mở rộng danh sách hàm toán học. Cần import trước khi dùng:

```python
import math

x = math.sqrt(64)
print(x)
```

`math.ceil()` làm tròn **lên**, `math.floor()` làm tròn **xuống** tới số nguyên gần nhất:

```python
import math

x = math.ceil(1.4)
y = math.floor(1.4)

print(x)
print(y)
```

Hằng số `math.pi` trả về giá trị số PI (3.14…):

```python
import math

x = math.pi
print(x)
```

Một vài hàm/hằng số khác:

```python
import math

print(math.e)              # số e
print(math.factorial(5))   # 5!
print(math.gcd(12, 18))    # ước chung lớn nhất
print(math.log10(1000))
print(math.sin(math.pi / 2))
```

Danh sách đầy đủ: [Math Module Reference](../10_reference/math_module.md).

> 🧪 **Try it Yourself** — Tính diện tích hình tròn bán kính r, làm tròn 2 chữ số thập phân.

```python
import math

r = 5
area = math.pi * r ** 2
print(round(area, 2))
```

## 🧠 Ghi nhớ nhanh

> - Module `math`: `sqrt()` căn bậc hai · `ceil()` làm tròn **lên** · `floor()` làm tròn **xuống** · `pi` · `e` · `factorial()`.
> - ⚠️ `round(2.5)` → `2` (làm tròn về số chẵn gần nhất).
