## The Else Keyword — Từ khoá else

Từ khoá `else` bắt **mọi trường hợp** không thoả các điều kiện phía trước:

````python
a = 200
b = 33
if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
else:
    print("a is greater than b")
````

## Else Without Elif — else không cần elif

Có thể dùng `else` ngay sau `if` để tạo hai nhánh:

````python
a = 200
b = 33
if b > a:
    print("b is greater than a")
else:
    print("b is not greater than a")
````

## How Else Works — else hoạt động thế nào?

`else` là nhánh **"còn lại"**: nó chạy khi **tất cả** điều kiện trước đều sai. `else` không có điều kiện riêng:

````python
number = 7

if number % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")
````

## Complete If-Elif-Else Chain — Chuỗi if-elif-else đầy đủ

````python
temperature = 22

if temperature > 30:
    print("It's hot outside!")
elif temperature > 20:
    print("It's warm outside")
elif temperature > 10:
    print("It's cool outside")
else:
    print("It's cold outside!")
````

## Else as Fallback — else làm phương án mặc định

`else` hữu ích để xử lý dữ liệu không hợp lệ hoặc gán giá trị mặc định:

````python
username = "Al"

if len(username) >= 3:
    print("Username accepted")
else:
    print("Error: Username must be at least 3 characters")
````

> 🧪 **Try it Yourself** — Viết chương trình kiểm tra mật khẩu: đúng "python123" thì chào mừng, sai thì báo lỗi.

````python
password = "python123"

if password == "python123":
    print("Đăng nhập thành công!")
else:
    print("Sai mật khẩu!")
````

## 🧠 Ghi nhớ nhanh

> - `else:` → nhánh "còn lại", **không có điều kiện**, đặt **cuối cùng**.
> - Cấu trúc đầy đủ: `if` → `elif` (0 hoặc nhiều) → `else` (0 hoặc 1).
> - Chuỗi `if/elif/else` đảm bảo **đúng một** nhánh được chạy.
