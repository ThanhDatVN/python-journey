## Copy a Dictionary — Sao chép dictionary

Bạn **không thể** sao chép dictionary chỉ bằng `dict2 = dict1`: khi đó `dict2` chỉ là một **tham chiếu** tới `dict1`, thay đổi ở `dict1` sẽ ảnh hưởng tới `dict2`.

```python
dict1 = {"brand": "Ford", "model": "Mustang"}
dict2 = dict1
dict1["year"] = 1964
print(dict2)
```

Có nhiều cách tạo bản sao thực sự. Cách thứ nhất là phương thức `copy()`:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
mydict = thisdict.copy()
print(mydict)
```

Cách thứ hai là hàm có sẵn `dict()`:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
mydict = dict(thisdict)
print(mydict)
```

Kiểm chứng tính độc lập:

```python
original = {"a": 1}
copied = original.copy()
original["b"] = 2
print(original)
print(copied)
```

## ➕ Mở rộng: Deep copy

`copy()` và `dict()` là **sao chép nông** — value là list/dict con vẫn dùng chung. Dùng `copy.deepcopy()` khi có cấu trúc lồng nhau:

```python
import copy

config = {"db": {"host": "localhost", "port": 5432}}
shallow = config.copy()
deep = copy.deepcopy(config)

config["db"]["port"] = 3306
print(shallow["db"]["port"])
print(deep["db"]["port"])
```

> 🧪 **Try it Yourself** — Tạo bản sao của cài đặt mặc định rồi sửa bản sao — bản gốc phải không đổi.

```python
default_settings = {"theme": "light", "font_size": 14}
my_settings = default_settings.copy()
my_settings["theme"] = "dark"
print(default_settings)
print(my_settings)
```

## 🧠 Ghi nhớ nhanh

> - ⚠️ `d2 = d1` không sao chép — chỉ tạo thêm tên cho cùng một dict.
> - Sao chép: `d1.copy()` hoặc `dict(d1)`.
> - Có dict/list lồng nhau → `copy.deepcopy(d1)`.
