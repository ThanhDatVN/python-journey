## Arguments — Đối số

Thông tin có thể được truyền vào hàm dưới dạng **đối số (argument)**.

Đối số được ghi sau tên hàm, trong cặp ngoặc tròn. Có thể có nhiều đối số, ngăn cách bằng dấu phẩy:

```python
def my_function(fname):
    print(fname + " Refsnes")

my_function("Emil")
my_function("Tobias")
my_function("Linus")
```

## Parameters vs Arguments — Tham số và đối số

- **Tham số (parameter)**: biến ghi trong ngoặc khi **định nghĩa** hàm.
- **Đối số (argument)**: giá trị thực tế truyền vào khi **gọi** hàm.

```python
def greet(name):          # name là tham số
    print("Hello", name)

greet("Emil")             # "Emil" là đối số
```

> Hai từ này thường được dùng lẫn cho nhau; cả hai đều chỉ "thông tin truyền vào hàm".

## Number of Arguments — Số lượng đối số

Mặc định, hàm phải được gọi với **đúng số** đối số mà nó yêu cầu:

```python
def my_function(fname, lname):
    print(fname + " " + lname)

my_function("Emil", "Refsnes")
```

Gọi thiếu (hoặc thừa) đối số sẽ báo lỗi:

```python
def my_function(fname, lname):
    print(fname + " " + lname)

my_function("Emil")
```

## Default Parameter Values — Giá trị mặc định

Tham số có thể có **giá trị mặc định**, dùng khi gọi hàm mà không truyền đối số đó:

```python
def my_function(name="friend"):
    print("Hello", name)

my_function("Emil")
my_function()
```

```python
def my_function(country="Norway"):
    print("I am from " + country)

my_function("Sweden")
my_function("India")
my_function()
my_function("Brazil")
```

## Keyword Arguments — Đối số từ khoá

Bạn có thể truyền đối số theo cú pháp `key = value`. Khi đó **thứ tự không quan trọng**:

```python
def my_function(animal, name):
    print("I have a", animal)
    print("My", animal + "'s name is", name)

my_function(animal="dog", name="Buddy")
my_function(name="Buddy", animal="dog")
```

## Positional Arguments — Đối số theo vị trí

Khi gọi hàm **không** ghi tên, đối số được gán **theo thứ tự**:

```python
def my_function(animal, name):
    print("I have a", animal)
    print("My", animal + "'s name is", name)

my_function("dog", "Buddy")
```

Đổi thứ tự → kết quả khác:

```python
def my_function(animal, name):
    print("I have a", animal)
    print("My", animal + "'s name is", name)

my_function("Buddy", "dog")
```

## Mixing Positional and Keyword Arguments — Kết hợp hai loại

Được phép trộn, nhưng đối số **vị trí phải đứng trước** đối số từ khoá:

```python
def my_function(animal, name, age):
    print("I have a", age, "year old", animal, "named", name)

my_function("dog", name="Buddy", age=5)
```

```python
def my_function(animal, name):
    print(animal, name)

my_function(name="Buddy", "dog")
```

## Passing Different Data Types — Truyền các kiểu dữ liệu khác nhau

Đối số có thể là **bất kỳ kiểu dữ liệu nào**: chuỗi, số, list, dictionary…

```python
def my_function(fruits):
    for fruit in fruits:
        print(fruit)

my_fruits = ["apple", "banana", "cherry"]
my_function(my_fruits)
```

```python
def my_function(person):
    print("Name:", person["name"])
    print("Age:", person["age"])

my_person = {"name": "Emil", "age": 25}
my_function(my_person)
```

## Return Values — Giá trị trả về

```python
def my_function(x, y):
    return x + y

result = my_function(5, 3)
print(result)
```

## Returning Different Data Types — Trả về nhiều kiểu dữ liệu

Hàm có thể trả về list, tuple, dictionary…:

```python
def my_function():
    return ["apple", "banana", "cherry"]

fruits = my_function()
print(fruits[0])
```

Trả về **tuple** rồi unpack — cách trả về "nhiều giá trị":

```python
def my_function():
    return (10, 20)

x, y = my_function()
print("x:", x)
print("y:", y)
```

## Positional-Only Arguments — Đối số chỉ theo vị trí

Thêm `, /` sau các tham số để bắt buộc chúng chỉ được truyền **theo vị trí**:

```python
def my_function(name, /):
    print("Hello", name)

my_function("Emil")
```

```python
def my_function(name, /):
    print("Hello", name)

my_function(name="Emil")
```

## Keyword-Only Arguments — Đối số chỉ theo từ khoá

Thêm `*, ` trước các tham số để bắt buộc chúng chỉ được truyền **theo tên**:

```python
def my_function(*, name):
    print("Hello", name)

my_function(name="Emil")
```

```python
def my_function(*, name):
    print("Hello", name)

my_function("Emil")
```

## Combining Positional-Only and Keyword-Only — Kết hợp cả hai

Tham số **trước `/`** chỉ theo vị trí; tham số **sau `*`** chỉ theo từ khoá:

```python
def my_function(a, b, /, *, c, d):
    return a + b + c + d

result = my_function(5, 10, c=15, d=20)
print(result)
```

> 🧪 **Try it Yourself** — Viết hàm tính tiền với giảm giá mặc định 0%; thử gọi có và không có discount.

```python
def total_price(price, quantity, discount=0):
    return price * quantity * (1 - discount / 100)

print(total_price(100, 3))
print(total_price(100, 3, discount=10))
```

## 🧠 Ghi nhớ nhanh

> - **Tham số** = biến khi định nghĩa · **Đối số** = giá trị khi gọi.
> - Mặc định phải truyền **đủ** số đối số; tham số có giá trị mặc định `name="friend"` thì có thể bỏ.
> - Positional: theo **thứ tự** · Keyword: `name="Buddy"`, thứ tự **không quan trọng**.
> - ⚠️ Trộn hai loại → positional **đứng trước** keyword.
> - `/` → các tham số trước nó chỉ theo vị trí · `*` → các tham số sau nó chỉ theo tên.
