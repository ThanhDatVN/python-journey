## Specify a Variable Type — Chỉ định kiểu cho biến

Đôi khi bạn muốn chỉ định rõ kiểu dữ liệu cho một biến. Việc này gọi là **casting** (ép kiểu). Python là ngôn ngữ hướng đối tượng, nên các kiểu dữ liệu — kể cả kiểu cơ bản — đều được định nghĩa bằng **class**, và casting thực chất là gọi **hàm khởi tạo (constructor)** của class đó:

- `int()` — tạo số nguyên từ số nguyên, số thực (bỏ phần thập phân) hoặc chuỗi (chuỗi phải biểu diễn một số nguyên).
- `float()` — tạo số thực từ số nguyên, số thực hoặc chuỗi (chuỗi biểu diễn số nguyên hoặc số thực).
- `str()` — tạo chuỗi từ hầu hết mọi kiểu dữ liệu: chuỗi, số nguyên, số thực…

### Example — Integers

````python
x = int(1)     # x sẽ là 1
y = int(2.8)   # y sẽ là 2
z = int("3")   # z sẽ là 3

print(x, y, z)
````

### Example — Floats

````python
x = float(1)      # x sẽ là 1.0
y = float(2.8)    # y sẽ là 2.8
z = float("3")    # z sẽ là 3.0
w = float("4.2")  # w sẽ là 4.2

print(x, y, z, w)
````

### Example — Strings

````python
x = str("s1")  # x sẽ là 's1'
y = str(2)     # y sẽ là '2'
z = str(3.0)   # z sẽ là '3.0'

print(x, y, z)
````

## ➕ Mở rộng: Khi nào casting thất bại?

Chuỗi phải biểu diễn đúng một con số thì mới chuyển được:

````python
int("hello")
````

`int()` không nhận chuỗi có dấu chấm thập phân — hãy chuyển sang `float` trước:

````python
int("3.5")
````

````python
print(int(float("3.5")))
````

## ➕ Mở rộng: Ứng dụng thường gặp

Casting rất hay dùng khi **kết hợp chuỗi và số**:

````python
age = 20
print("Tôi " + str(age) + " tuổi")
````

Hoặc khi tính toán với số được lưu ở dạng chuỗi (ví dụ dữ liệu nhập từ bàn phím):

````python
a = "10"
b = "20"
print(a + b)            # nối chuỗi
print(int(a) + int(b))  # cộng số
````

> 🧪 **Try it Yourself** — Đổi giá trị của price thành "19.99" — dòng nào cần sửa để chương trình vẫn chạy?

````python
price = "20"
quantity = 3

total = int(price) * quantity
print("Tổng tiền:", total)
````

---

## 🧠 Ghi nhớ nhanh

> - Casting = gọi constructor: `int()`, `float()`, `str()`.
> - `int("3")` ✅ · `int("3.5")` ❌ → dùng `int(float("3.5"))`.
> - ⚠️ `int("hello")` → `ValueError`.
> - Hay dùng nhất: `str(so)` để nối với chuỗi; `int(chuoi)` để tính toán.
