# Python - Change List Items

> 📖 **Học song song trên W3Schools:** [python_lists_change.asp](https://www.w3schools.com/python/python_lists_change.asp)
>
> 🎯 **Trong bài này:** Change Item Value · Change a Range of Item Values · Insert Items

## Change Item Value — Đổi giá trị một phần tử

Để thay đổi giá trị của một phần tử, gán giá trị mới vào chỉ số tương ứng:

````python
thislist = ["apple", "banana", "cherry"]
thislist[1] = "blackcurrant"
print(thislist)
````

## Change a Range of Item Values — Đổi một khoảng phần tử

Gán một **list mới** cho một khoảng chỉ số:

````python
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "mango"]
thislist[1:3] = ["blackcurrant", "watermelon"]
print(thislist)
````

Nếu số phần tử mới **nhiều hơn** số phần tử bị thay, các phần tử mới được chèn vào và list dài thêm:

````python
thislist = ["apple", "banana", "cherry"]
thislist[1:2] = ["blackcurrant", "watermelon"]
print(thislist)
````

> 📌 Độ dài list thay đổi khi số phần tử chèn vào khác số phần tử bị thay.

Nếu số phần tử mới **ít hơn**, list sẽ ngắn lại:

````python
thislist = ["apple", "banana", "cherry"]
thislist[1:3] = ["watermelon"]
print(thislist)
````

## Insert Items — Chèn phần tử

Để chèn một phần tử mới **mà không thay thế** phần tử nào, dùng phương thức `insert()`. Phần tử được chèn vào đúng vị trí chỉ định:

````python
thislist = ["apple", "banana", "cherry"]
thislist.insert(2, "watermelon")
print(thislist)
````

> 🧪 **Try it Yourself** — Đổi phần tử cuối cùng của list thành "Python" (gợi ý: chỉ số -1).

````python
languages = ["C", "Java", "PHP"]
languages[-1] = "Python"
print(languages)
````

---

## 🧠 Ghi nhớ nhanh

> - Sửa một phần tử: `a[1] = "new"`.
> - Sửa một khoảng: `a[1:3] = [...]` — list có thể **dài ra hoặc ngắn lại**.
> - Chèn mà không thay thế: `a.insert(i, x)`.
