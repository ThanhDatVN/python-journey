## Generators

**Generator** là hàm có thể **tạm dừng rồi tiếp tục** quá trình chạy.

Khi được gọi, generator trả về một đối tượng **iterator**. Code bên trong chỉ chạy khi bạn **duyệt** qua iterator đó.

Generator dùng từ khoá `yield` thay cho `return`:

```python
def my_generator():
    yield 1
    yield 2
    yield 3

for value in my_generator():
    print(value)
```

## The yield Keyword — Từ khoá yield

`yield` khác `return` ở chỗ: nó **trả về một giá trị và tạm dừng** hàm, **lưu lại trạng thái** (biến cục bộ, vị trí đang chạy). Lần lấy giá trị tiếp theo, hàm chạy tiếp **từ chỗ vừa dừng**:

```python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1

for num in count_up_to(5):
    print(num)
```

## Generators Saves Memory — Generator tiết kiệm bộ nhớ

Generator **không** tạo sẵn mọi giá trị trong bộ nhớ, mà tạo ra **từng giá trị khi cần**. Rất phù hợp với dữ liệu lớn:

```python
def large_sequence(n):
    for i in range(n):
        yield i

# Không tạo list một triệu phần tử trong bộ nhớ
gen = large_sequence(1000000)
print(next(gen))
print(next(gen))
print(next(gen))
```

## Using next() with Generators — Dùng next()

Hàm `next()` lấy **từng giá trị** của generator một cách thủ công:

```python
def simple_gen():
    yield "Emil"
    yield "Tobias"
    yield "Linus"

gen = simple_gen()
print(next(gen))
print(next(gen))
print(next(gen))
```

Khi hết giá trị, `next()` báo lỗi `StopIteration`:

```python
def simple_gen():
    yield 1
    yield 2

gen = simple_gen()
print(next(gen))
print(next(gen))
print(next(gen))  # hết giá trị -> StopIteration
```

> 📌 Vòng lặp `for` tự động xử lý `StopIteration` — đó là lý do `for` dừng đúng lúc.

## Generator Expressions — Biểu thức generator

Giống list comprehension nhưng dùng **ngoặc tròn** `()` thay cho ngoặc vuông:

```python
# List comprehension — tạo list ngay
list_comp = [x * x for x in range(5)]
print(list_comp)

# Generator expression — tạo generator
gen_exp = (x * x for x in range(5))
print(gen_exp)
print(list(gen_exp))
```

Rất tiện khi đưa vào các hàm như `sum()`:

```python
total = sum(x * x for x in range(10))
print(total)
```

## Fibonacci Sequence Generator — Generator Fibonacci

Generator có thể **vô hạn** — chỉ tạo giá trị khi được yêu cầu:

```python
def fibonacci():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

gen = fibonacci()
for _ in range(10):
    print(next(gen), end=" ")
```

## Generator Methods — Các phương thức của generator

### send() Method

`send()` **gửi một giá trị vào** generator — giá trị đó trở thành kết quả của biểu thức `yield`:

```python
def echo_generator():
    while True:
        received = yield
        print("Received:", received)

gen = echo_generator()
next(gen)  # "mồi" (prime) generator trước
gen.send("Hello")
gen.send("World")
```

### close() Method

`close()` **dừng** generator:

```python
def my_gen():
    try:
        yield 1
        yield 2
        yield 3
    finally:
        print("Generator closed")

gen = my_gen()
print(next(gen))
gen.close()
```

> 🧪 **Try it Yourself** — Viết generator even_numbers(n) sinh ra các số chẵn từ 0 đến n.

```python
def even_numbers(n):
    for i in range(0, n + 1, 2):
        yield i

print(list(even_numbers(10)))
```

## 🧠 Ghi nhớ nhanh

> - Generator = hàm dùng `yield`; gọi hàm trả về **iterator**, code chạy **dần dần**.
> - `yield` trả giá trị + **tạm dừng** + nhớ trạng thái; lần sau chạy tiếp từ đó.
> - **Tiết kiệm bộ nhớ**: tạo từng giá trị khi cần — dùng được cả cho dãy vô hạn.
> - `next(gen)` lấy giá trị kế; hết thì `StopIteration`. Biểu thức generator: `(x*x for x in ...)`.
