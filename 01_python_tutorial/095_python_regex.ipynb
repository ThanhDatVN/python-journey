**RegEx** (Regular Expression — biểu thức chính quy) là một chuỗi ký tự tạo thành **mẫu tìm kiếm**.

RegEx dùng để kiểm tra một chuỗi có chứa mẫu tìm kiếm nào đó hay không.

## RegEx Module

Python có module có sẵn tên `re` để làm việc với biểu thức chính quy:

```python
import re
```

## RegEx in Python

Sau khi import, bạn có thể dùng RegEx. Tìm xem chuỗi có bắt đầu bằng "The" và kết thúc bằng "Spain" không:

```python
import re

txt = "The rain in Spain"
x = re.search("^The.*Spain$", txt)
print(x)
```

## RegEx Functions — Các hàm RegEx

| Hàm | Mô tả |
|---|---|
| `findall` | Trả về list chứa **tất cả** kết quả khớp |
| `search` | Trả về `Match` object nếu có khớp **ở bất kỳ đâu** |
| `split` | Trả về list, chuỗi được tách tại mỗi chỗ khớp |
| `sub` | Thay thế các chỗ khớp bằng chuỗi khác |

## Metacharacters — Ký tự đặc biệt

| Ký tự | Mô tả | Ví dụ |
|---|---|---|
| `[]` | Một tập ký tự | `"[a-m]"` |
| `\` | Báo hiệu chuỗi đặc biệt (hoặc thoát ký tự đặc biệt) | `"\d"` |
| `.` | Bất kỳ ký tự nào (trừ xuống dòng) | `"he..o"` |
| `^` | Bắt đầu bằng | `"^hello"` |
| `$` | Kết thúc bằng | `"planet$"` |
| `*` | 0 lần trở lên | `"he.*o"` |
| `+` | 1 lần trở lên | `"he.+o"` |
| `?` | 0 hoặc 1 lần | `"he.?o"` |
| `{}` | Đúng số lần chỉ định | `"he.{2}o"` |
| `\|` | Hoặc | `"falls\|stays"` |
| `()` | Nhóm | |

## Flags — Cờ

| Cờ | Viết tắt | Mô tả |
|---|---|---|
| `re.ASCII` | `re.A` | `\w \W \b \B \d \D \s \S` chỉ khớp ký tự ASCII |
| `re.DEBUG` | | Hiển thị thông tin debug |
| `re.DOTALL` | `re.S` | `.` khớp cả ký tự xuống dòng |
| `re.IGNORECASE` | `re.I` | Không phân biệt hoa thường |
| `re.MULTILINE` | `re.M` | `^` và `$` khớp ở đầu/cuối mỗi dòng |
| `re.NOFLAG` | | Không dùng cờ nào |
| `re.UNICODE` | `re.U` | Khớp ký tự Unicode (mặc định trong Python 3) |
| `re.VERBOSE` | `re.X` | Cho phép khoảng trắng và comment trong mẫu |

## Special Sequences — Chuỗi đặc biệt

| Chuỗi | Mô tả |
|---|---|
| `\A` | Khớp nếu ký tự chỉ định ở **đầu** chuỗi |
| `\b` | Khớp ở đầu hoặc cuối một **từ** |
| `\B` | Khớp nhưng **không** ở đầu/cuối từ |
| `\d` | Chữ số (0-9) |
| `\D` | **Không** phải chữ số |
| `\s` | Khoảng trắng |
| `\S` | **Không** phải khoảng trắng |
| `\w` | Ký tự "từ" (a-Z, 0-9, `_`) |
| `\W` | **Không** phải ký tự "từ" |
| `\Z` | Khớp nếu ký tự chỉ định ở **cuối** chuỗi |

## Sets — Tập ký tự

| Tập | Khớp với |
|---|---|
| `[arn]` | Một trong các ký tự `a`, `r`, `n` |
| `[a-n]` | Bất kỳ chữ thường nào từ `a` đến `n` |
| `[^arn]` | Bất kỳ ký tự nào **trừ** `a`, `r`, `n` |
| `[0123]` | Một trong các chữ số `0`, `1`, `2`, `3` |
| `[0-9]` | Bất kỳ chữ số nào |
| `[0-5][0-9]` | Số có hai chữ số từ `00` đến `59` |
| `[a-zA-Z]` | Bất kỳ chữ cái nào, hoa hoặc thường |
| `[+]` | Ký tự `+` (trong `[]`, ký tự đặc biệt mất ý nghĩa) |

> 💡 Nên viết mẫu RegEx bằng **raw string** `r"..."` để `\` không bị hiểu là ký tự thoát.

## The findall() Function

Trả về **list** tất cả kết quả khớp:

```python
import re

txt = "The rain in Spain"
x = re.findall("ai", txt)
print(x)
```

Không khớp → trả về list rỗng:

```python
import re

txt = "The rain in Spain"
x = re.findall("Portugal", txt)
print(x)
```

## The search() Function

Tìm trong chuỗi, trả về `Match` object nếu có khớp. Nếu có nhiều chỗ khớp, chỉ trả về **chỗ đầu tiên**:

```python
import re

txt = "The rain in Spain"
x = re.search(r"\s", txt)

print("The first white-space character is located in position:", x.start())
```

Không khớp → trả về `None`:

```python
import re

txt = "The rain in Spain"
x = re.search("Portugal", txt)
print(x)
```

## The split() Function

Trả về list, chuỗi được tách tại mỗi chỗ khớp:

```python
import re

txt = "The rain in Spain"
x = re.split(r"\s", txt)
print(x)
```

Giới hạn số lần tách bằng tham số `maxsplit`:

```python
import re

txt = "The rain in Spain"
x = re.split(r"\s", txt, maxsplit=1)
print(x)
```

## The sub() Function

Thay thế các chỗ khớp bằng chuỗi bạn chọn:

```python
import re

txt = "The rain in Spain"
x = re.sub(r"\s", "9", txt)
print(x)
```

Giới hạn số lần thay thế bằng tham số `count`:

```python
import re

txt = "The rain in Spain"
x = re.sub(r"\s", "9", txt, count=2)
print(x)
```

## Match Object

`Match` object chứa thông tin về lần tìm kiếm và kết quả. Nếu không khớp, kết quả là `None` thay vì `Match` object.

```python
import re

txt = "The rain in Spain"
x = re.search("ai", txt)
print(x)
```

Các thuộc tính/phương thức của `Match` object:

- `.span()` — tuple vị trí bắt đầu và kết thúc của chỗ khớp
- `.string` — chuỗi gốc đã truyền vào
- `.group()` — phần chuỗi khớp

```python
import re

txt = "The rain in Spain"
x = re.search(r"\bS\w+", txt)
print(x.span())
print(x.string)
print(x.group())
```

## ➕ Mở rộng: Ví dụ thực tế

```python
import re

text = "Liên hệ: an.nguyen@gmail.com hoặc 0901-234-567, binh@company.vn"
emails = re.findall(r"[\w.]+@[\w.]+\.\w+", text)
phones = re.findall(r"\d{4}-\d{3}-\d{3}", text)
print(emails)
print(phones)
```

> 🧪 **Try it Yourself** — Tìm tất cả các số trong chuỗi bằng \d+.

```python
import re

txt = "Lớp có 35 học sinh, 18 nam và 17 nữ"
print(re.findall(r"\d+", txt))
```
