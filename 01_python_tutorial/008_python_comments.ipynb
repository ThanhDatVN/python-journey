# Python Comments

> 📖 **Học song song trên W3Schools:** [python_comments.asp](https://www.w3schools.com/python/python_comments.asp)
>
> 🎯 **Trong bài này:** Comments · Creating a Comment · Multiline Comments

## Comments — Chú thích

Comment dùng để:

- Giải thích code, giúp người khác (và chính bạn sau này) dễ hiểu hơn.
- Làm code dễ đọc hơn.
- Tạm thời **ngăn không cho một đoạn code chạy** khi đang thử nghiệm.

## Creating a Comment — Tạo comment

Comment bắt đầu bằng dấu `#`, Python bỏ qua mọi thứ phía sau dấu `#` trên dòng đó:

````python
# This is a comment
print("Hello, World!")
````

Comment cũng có thể đặt ở **cuối dòng** code:

````python
print("Hello, World!")  # This is a comment
````

Comment không nhất thiết là lời giải thích — bạn có thể "comment out" một dòng code để nó không chạy:

````python
# print("Hello, World!")
print("Cheers, Mate!")
````

## Multiline Comments — Comment nhiều dòng

Python không có cú pháp riêng cho comment nhiều dòng. Cách phổ biến là đặt `#` ở đầu **mỗi dòng**:

````python
# This is a comment
# written in
# more than just one line
print("Hello, World!")
````

Một cách khác là dùng **chuỗi nhiều dòng** (ba dấu nháy). Vì chuỗi này không được gán cho biến nào, Python đọc rồi bỏ qua:

````python
"""
This is a comment
written in
more than just one line
"""
print("Hello, World!")
````

> 💡 Trong notebook, chọn nhiều dòng rồi nhấn `Ctrl + /` để comment hoặc bỏ comment cùng lúc.

> 🧪 **Try it Yourself** — Comment dòng print thứ hai để chỉ còn một dòng được in ra.

````python
print("Dòng 1")
print("Dòng 2")
print("Dòng 3")
````

---

## 🧠 Ghi nhớ nhanh

> - Comment bắt đầu bằng `#`, Python bỏ qua phần sau `#` trên dòng đó.
> - Dùng comment để **giải thích code** hoặc **tạm vô hiệu hoá** một dòng.
> - Comment nhiều dòng: `#` ở mỗi dòng, hoặc chuỗi `"""..."""` không gán biến.
> - Phím tắt: `Ctrl + /`.
