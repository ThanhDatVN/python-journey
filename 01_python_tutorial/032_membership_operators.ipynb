## Membership Operators — Toán tử thành viên

Toán tử thành viên kiểm tra một giá trị có **nằm trong** một dãy (chuỗi, list, tuple, set, dict…) hay không:

| Toán tử | Mô tả | Ví dụ |
|---|---|---|
| `in` | `True` nếu giá trị có trong dãy | `x in y` |
| `not in` | `True` nếu giá trị không có trong dãy | `x not in y` |

### Examples

````python
fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)
print("pineapple" not in fruits)
````

## Membership in Strings — Trong chuỗi

Với chuỗi, `in` kiểm tra **chuỗi con**:

````python
text = "Hello World"

print("H" in text)
print("hello" in text)   # phân biệt hoa thường
print("World" in text)
````

## ➕ Mở rộng: Membership in Dictionaries — Trong dictionary

Với dict, `in` kiểm tra **key** (không phải value):

````python
person = {"name": "John", "age": 36}

print("name" in person)
print("John" in person)
print("John" in person.values())
````

## ➕ Mở rộng: Dùng với `if`

````python
allowed_users = ["an", "binh", "chi"]
user = "binh"

if user in allowed_users:
    print("Welcome,", user)
else:
    print("Access denied")
````

> 🧪 **Try it Yourself** — Kiểm tra một ký tự có phải nguyên âm không.

````python
letter = "e"
print(letter in "aeiou")
````

---

## 🧠 Ghi nhớ nhanh

> - `in` / `not in` → kiểm tra phần tử có trong chuỗi, list, tuple, set, dict.
> - Với chuỗi → kiểm tra **chuỗi con**; phân biệt hoa thường.
> - Với dict → kiểm tra **key** (muốn kiểm tra value: `x in d.values()`).
