## Recursion — Đệ quy

**Đệ quy (recursion)** là khi một hàm **gọi lại chính nó**.

Đệ quy giúp chia một bài toán lớn thành các bài toán con **giống hệt nhưng nhỏ hơn**. Một số bài toán viết bằng đệ quy rất tự nhiên — ví dụ duyệt cấu trúc cây, hoặc các công thức toán định nghĩa theo chính nó.

Hàm đếm ngược đơn giản:

````python
def countdown(n):
    if n <= 0:
        print("Done!")
    else:
        print(n)
        countdown(n - 1)

countdown(5)
````

## Base Case and Recursive Case — Điều kiện dừng và bước đệ quy

Mọi hàm đệ quy cần **hai phần**:

- **Base case** (điều kiện dừng): trường hợp đơn giản nhất, trả kết quả **không** cần gọi đệ quy.
- **Recursive case** (bước đệ quy): hàm gọi lại chính nó với đầu vào **nhỏ hơn**, tiến dần tới base case.

> ⚠️ Thiếu base case → hàm gọi mãi không dừng → lỗi `RecursionError`.

Tính giai thừa: `5! = 5 × 4 × 3 × 2 × 1`:

````python
def factorial(n):
    # Base case
    if n == 0 or n == 1:
        return 1
    # Recursive case
    else:
        return n * factorial(n - 1)

print(factorial(5))
````

```text
factorial(5) = 5 * factorial(4)
             = 5 * 4 * factorial(3)
             = 5 * 4 * 3 * factorial(2)
             = 5 * 4 * 3 * 2 * factorial(1)
             = 5 * 4 * 3 * 2 * 1 = 120
```

## Fibonacci Sequence — Dãy Fibonacci

Trong dãy Fibonacci, mỗi số bằng tổng hai số đứng trước: 0, 1, 1, 2, 3, 5, 8, 13…

````python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(7))
````

## Recursion with Lists — Đệ quy với list

Tính tổng các phần tử bằng cách xử lý **một phần tử mỗi lần gọi**:

````python
def sum_list(numbers):
    if len(numbers) == 0:
        return 0
    else:
        return numbers[0] + sum_list(numbers[1:])

my_list = [1, 2, 3, 4, 5]
print(sum_list(my_list))
````

Tìm giá trị lớn nhất:

````python
def find_max(numbers):
    if len(numbers) == 1:
        return numbers[0]
    else:
        max_of_rest = find_max(numbers[1:])
        return numbers[0] if numbers[0] > max_of_rest else max_of_rest

my_list = [3, 7, 2, 9, 1]
print(find_max(my_list))
````

## Recursion Depth Limit — Giới hạn độ sâu đệ quy

Python giới hạn số lần gọi đệ quy lồng nhau (mặc định khoảng **1000**) để tránh tràn bộ nhớ:

````python
import sys
print(sys.getrecursionlimit())
````

Có thể tăng giới hạn bằng `sys.setrecursionlimit()`, nhưng cần cẩn thận — đặt quá cao có thể làm chương trình bị treo:

````python
import sys
sys.setrecursionlimit(2000)
print(sys.getrecursionlimit())
````

Vượt giới hạn sẽ báo lỗi:

````python
def forever(n):
    return forever(n + 1)

forever(1)
````

> 💡 Với bài toán cần độ sâu rất lớn, nên viết lại bằng **vòng lặp**.

> 🧪 **Try it Yourself** — Viết hàm đệ quy tính tổng các số từ 1 đến n.

````python
def sum_to(n):
    if n == 0:
        return 0
    return n + sum_to(n - 1)

print(sum_to(10))
````

## 🧠 Ghi nhớ nhanh

> - Đệ quy = hàm **gọi lại chính nó** với bài toán **nhỏ hơn**.
> - Bắt buộc có **base case** (điều kiện dừng) + **recursive case** (tiến dần tới base case).
> - ⚠️ Thiếu điều kiện dừng → `RecursionError` (giới hạn mặc định ~1000 lần).
> - Ví dụ kinh điển: giai thừa, Fibonacci, tổng list.
