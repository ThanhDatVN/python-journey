# Python - String Concatenation

> 📖 **Học song song trên W3Schools:** [python_strings_concatenate.asp](https://www.w3schools.com/python/python_strings_concatenate.asp)
>
> 🎯 **Trong bài này:** String Concatenation · ➕ Mở rộng: Lặp chuỗi với `*` · ➕ Mở rộng: Nối chuỗi với số · ➕ Mở rộng: Nối nhiều chuỗi với `join()`

## String Concatenation — Nối chuỗi

Để nối (ghép) hai chuỗi lại với nhau, dùng toán tử `+`.

Ghép biến `a` với biến `b` vào biến `c`:

````python
a = "Hello"
b = "World"
c = a + b
print(c)
````

Muốn có khoảng trắng ở giữa, hãy nối thêm `" "`:

````python
a = "Hello"
b = "World"
c = a + " " + b
print(c)
````

## ➕ Mở rộng: Lặp chuỗi với `*`

Toán tử `*` lặp lại một chuỗi nhiều lần:

````python
print("Ha" * 3)
print("-" * 20)
````

## ➕ Mở rộng: Nối chuỗi với số

Chỉ có thể nối **chuỗi với chuỗi**. Nối chuỗi với số sẽ gây lỗi:

````python
age = 36
txt = "My name is John, I am " + age
print(txt)
````

Cách sửa: chuyển số sang chuỗi bằng `str()`, hoặc tốt hơn là dùng **f-string** (bài tiếp theo):

````python
age = 36
txt = "My name is John, I am " + str(age)
print(txt)
````

## ➕ Mở rộng: Nối nhiều chuỗi với `join()`

Khi cần nối **nhiều** chuỗi trong một list, phương thức `join()` gọn và nhanh hơn dùng `+` liên tục:

````python
words = ["Python", "is", "fun"]
print(" ".join(words))
print("-".join(words))
````

> 🧪 **Try it Yourself** — Ghép họ, tên đệm và tên thành họ tên đầy đủ, có khoảng trắng ở giữa.

````python
ho = "Nguyễn"
dem = "Văn"
ten = "An"

print(ho + " " + dem + " " + ten)
````

---

## 🧠 Ghi nhớ nhanh

> - Nối chuỗi bằng `+`; nhớ tự thêm `" "` nếu cần khoảng trắng.
> - `"ab" * 3` → `ababab`.
> - Nối nhiều chuỗi trong list: `" ".join(words)`.
> - ⚠️ Chỉ nối được **chuỗi với chuỗi**.
