## Join Two Lists — Nối hai list

Có nhiều cách để nối (ghép) hai hoặc nhiều list.

Cách dễ nhất là dùng toán tử `+`:

```python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]

list3 = list1 + list2
print(list3)
```

Cách khác: thêm **lần lượt từng phần tử** của `list2` vào `list1` bằng `append()`:

```python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]

for x in list2:
    list1.append(x)

print(list1)
```

Hoặc dùng phương thức `extend()` — thêm toàn bộ phần tử của list này vào cuối list kia:

```python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]

list1.extend(list2)
print(list1)
```

## ➕ Mở rộng: Khác biệt giữa các cách

- `list1 + list2` tạo ra **list mới**, hai list gốc giữ nguyên.
- `append()` trong vòng lặp và `extend()` **thay đổi** `list1`.

## ➕ Mở rộng: Unpacking với `*`

Từ Python 3.5, có thể dùng toán tử `*` để "trải" các list vào một list mới:

```python
list1 = ["a", "b"]
list2 = [1, 2]
list3 = [True]

combined = [*list1, *list2, *list3]
print(combined)
```

## ➕ Mở rộng: Nhân list với `*`

```python
print([0] * 5)
print(["ha"] * 3)
```

> 🧪 **Try it Yourself** — Ghép danh sách lớp A và lớp B thành danh sách toàn khối.

```python
class_a = ["An", "Bình"]
class_b = ["Chi", "Dũng", "Em"]

all_students = class_a + class_b
print(all_students)
print("Sĩ số:", len(all_students))
```

---

## 🧠 Ghi nhớ nhanh

> - `a + b` → tạo **list mới** (a, b giữ nguyên).
> - `a.extend(b)` hoặc vòng lặp `append` → **sửa** `a`.
> - `[*a, *b]` → unpacking thành list mới.
