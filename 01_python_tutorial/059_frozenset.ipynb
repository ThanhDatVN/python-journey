## Python frozenset

`frozenset` là phiên bản **bất biến (immutable)** của set.

Giống set, frozenset chứa các phần tử **duy nhất**, **không có thứ tự**. Khác set, frozenset **không thể thay đổi** sau khi tạo: không thêm, không xoá được phần tử.

## Creating a frozenset — Tạo frozenset

Dùng hàm khởi tạo `frozenset()` với một iterable bất kỳ:

```python
x = frozenset({"apple", "banana", "cherry"})
print(x)
print(type(x))
```

Phần tử trùng lặp bị loại bỏ như với set:

```python
x = frozenset(["a", "b", "a", "c"])
print(x)
```

## ➕ Mở rộng: Không thể thay đổi

frozenset không có các phương thức `add()`, `remove()`, `update()`…:

```python
x = frozenset({"apple", "banana", "cherry"})
x.add("orange")
```

## Frozenset Methods — Phương thức

frozenset hỗ trợ các phương thức **không làm thay đổi** dữ liệu:

| Phương thức | Toán tử | Mô tả |
|---|---|---|
| `copy()` | | Trả về bản sao |
| `difference()` | `-` | Trả về frozenset hiệu |
| `intersection()` | `&` | Trả về frozenset giao |
| `isdisjoint()` | | `True` nếu hai tập không có phần tử chung |
| `issubset()` | `<=` / `<` | `True` nếu là tập con / tập con thực sự |
| `issuperset()` | `>=` / `>` | `True` nếu là tập cha / tập cha thực sự |
| `symmetric_difference()` | `^` | Trả về frozenset hiệu đối xứng |
| `union()` | `\|` | Trả về frozenset hợp |

```python
a = frozenset({1, 2, 3})
b = frozenset({3, 4, 5})

print(a | b)
print(a & b)
print(a - b)
print(a.isdisjoint(b))
print(frozenset({1, 2}).issubset(a))
```

## ➕ Mở rộng: Khi nào dùng frozenset?

Vì không đổi được, frozenset **băm (hash) được** — nên có thể dùng làm **key của dictionary** hoặc **phần tử của set khác**. Set thường thì không làm được:

```python
groups = {
    frozenset({"An", "Bình"}): "Nhóm 1",
    frozenset({"Chi", "Dũng"}): "Nhóm 2",
}
print(groups[frozenset({"Bình", "An"})])
```

```python
bad = {{"An", "Bình"}: "Nhóm 1"}
```

> 🧪 **Try it Yourself** — Tạo frozenset các ngày nghỉ lễ cố định, rồi kiểm tra một ngày có phải ngày nghỉ không.

```python
holidays = frozenset({"01-01", "30-04", "01-05", "02-09"})
print("30-04" in holidays)
```

> - Băm được → dùng làm **key của dict** hoặc phần tử của set khác.
