## List Methods — Các phương thức của list

Python có các phương thức có sẵn sau đây để dùng với list:

| Phương thức | Mô tả |
|---|---|
| `append()` | Thêm một phần tử vào cuối list |
| `clear()` | Xoá tất cả phần tử |
| `copy()` | Trả về bản sao của list |
| `count()` | Đếm số phần tử có giá trị chỉ định |
| `extend()` | Thêm các phần tử của một iterable vào cuối list |
| `index()` | Trả về chỉ số của phần tử đầu tiên có giá trị chỉ định |
| `insert()` | Chèn phần tử vào vị trí chỉ định |
| `pop()` | Xoá (và trả về) phần tử tại vị trí chỉ định |
| `remove()` | Xoá phần tử đầu tiên có giá trị chỉ định |
| `reverse()` | Đảo ngược thứ tự list |
| `sort()` | Sắp xếp list |

Chi tiết tham số và ví dụ: [List Methods Reference](../10_reference/list_methods.md).

## ➕ Mở rộng: Ví dụ các phương thức chưa gặp

### count()

```python
fruits = ["apple", "banana", "cherry", "banana"]
print(fruits.count("banana"))
```

### index()

```python
fruits = ["apple", "banana", "cherry"]
print(fruits.index("cherry"))
```

`index()` báo lỗi nếu không tìm thấy giá trị:

```python
fruits = ["apple", "banana", "cherry"]
print(fruits.index("mango"))
```

## ➕ Mở rộng: Hàm có sẵn hay dùng với list

Ngoài phương thức, các **hàm built-in** sau cũng rất hữu ích với list:

```python
numbers = [4, 8, 15, 16, 23, 42]

print(len(numbers))
print(min(numbers))
print(max(numbers))
print(sum(numbers))
print(sorted(numbers, reverse=True))
print(list(reversed(numbers)))
```

## ➕ Mở rộng: Phương thức thay đổi list trả về None

Các phương thức như `append()`, `sort()`, `reverse()` thay đổi list **tại chỗ** và trả về `None`. Đây là lỗi thường gặp:

```python
numbers = [3, 1, 2]
result = numbers.sort()
print(result)    # None!
print(numbers)   # list đã được sắp xếp
```

> 🧪 **Try it Yourself** — Tìm vị trí của số lớn nhất trong list (gợi ý: kết hợp index() và max()).

```python
numbers = [12, 45, 7, 89, 23]
print(numbers.index(max(numbers)))
```

## 🧠 Ghi nhớ nhanh

> - 11 phương thức: `append clear copy count extend index insert pop remove reverse sort`.
> - Hàm built-in hay dùng: `len min max sum sorted reversed`.
> - ⚠️ `append`, `sort`, `reverse`… sửa list **tại chỗ** và trả về `None` — đừng gán `a = a.sort()`.
