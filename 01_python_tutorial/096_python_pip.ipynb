## What is PIP? — PIP là gì?

**PIP** là trình quản lý **package** (gói) cho Python — dùng để cài đặt và quản lý các package/module.

> 📌 Từ Python 3.4, PIP được cài sẵn cùng Python.

## What is a Package? — Package là gì?

Một package chứa tất cả các file cần thiết cho một module. Module là các thư viện code Python bạn có thể đưa vào dự án.

## Check if PIP is Installed — Kiểm tra PIP

Mở terminal (Command Prompt) và gõ:

```bash
pip --version
```

Trong notebook, thêm dấu `!` để chạy lệnh terminal:

````python
!pip --version
````

## Install PIP — Cài PIP

Nếu máy chưa có PIP, tải và cài tại: [https://pypi.org/project/pip/](https://pypi.org/project/pip/)

## Download a Package — Tải package

Dùng lệnh `pip install` trong terminal để tải package, ví dụ package `camelcase`:

```bash
pip install camelcase
```

Trong Jupyter, nên dùng **magic `%pip`** — nó đảm bảo package được cài đúng vào môi trường của notebook đang chạy:

````python
%pip install camelcase
````

## Using a Package — Dùng package

Sau khi cài, import package vào chương trình:

````python
import camelcase

c = camelcase.CamelCase()
txt = "hello world"
print(c.hump(txt))
````

> Kết quả: `Hello World`.

## Find Packages — Tìm package

Tìm thêm package tại [https://pypi.org/](https://pypi.org/) — kho package chính thức của Python.

## Remove a Package — Gỡ package

```bash
pip uninstall camelcase
```

PIP sẽ hỏi xác nhận việc gỡ; gõ `y` để đồng ý.

## List Packages — Liệt kê package

```bash
pip list
```

````python
%pip list
````

## ➕ Mở rộng: requirements.txt

Dự án thực tế liệt kê các package cần thiết trong file `requirements.txt` (như repo này):

```bash
pip freeze > requirements.txt      # xuất danh sách package hiện có
pip install -r requirements.txt    # cài tất cả package trong file
```

Kiểm tra một package đã được cài hay chưa ngay trong Python:

````python
import importlib.util

for name in ["json", "requests", "camelcase"]:
    found = importlib.util.find_spec(name) is not None
    print(f"{name}: {'đã có' if found else 'chưa cài'}")
````
