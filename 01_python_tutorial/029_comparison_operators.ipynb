## Comparison Operators — Toán tử so sánh

Toán tử so sánh dùng để so sánh hai giá trị. Kết quả luôn là `True` hoặc `False`:

| Toán tử | Tên | Ví dụ |
|---|---|---|
| `==` | Bằng | `x == y` |
| `!=` | Khác | `x != y` |
| `>` | Lớn hơn | `x > y` |
| `<` | Nhỏ hơn | `x < y` |
| `>=` | Lớn hơn hoặc bằng | `x >= y` |
| `<=` | Nhỏ hơn hoặc bằng | `x <= y` |

### Examples

````python
x = 5
y = 3

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
````

> ⚠️ Đừng nhầm `=` (gán giá trị) với `==` (so sánh bằng)!

## Chaining Comparison Operators — So sánh nối tiếp

Python cho phép **nối** các phép so sánh, giống cách viết trong toán học:

````python
x = 5

print(1 < x < 10)
print(1 < x and x < 10)   # cách viết tương đương
````

## ➕ Mở rộng: So sánh chuỗi

Chuỗi được so sánh theo thứ tự từ điển (dựa trên mã Unicode của từng ký tự), và **phân biệt hoa thường**:

````python
print("apple" == "apple")
print("apple" == "Apple")
print("apple" < "banana")
print("Zebra" < "apple")   # chữ hoa có mã nhỏ hơn chữ thường
````

## ➕ Mở rộng: So sánh số thực

Số thực có thể bị sai số nhỏ khi tính toán, nên so sánh bằng `==` đôi khi cho kết quả bất ngờ:

````python
print(0.1 + 0.2 == 0.3)
print(0.1 + 0.2)

import math
print(math.isclose(0.1 + 0.2, 0.3))
````

> 🧪 **Try it Yourself** — Kiểm tra một điểm số có nằm trong khoảng hợp lệ 0..10 hay không bằng so sánh nối tiếp.

````python
score = 8.5
print(0 <= score <= 10)
````

---

## 🧠 Ghi nhớ nhanh

> - `==  !=  >  <  >=  <=` → luôn trả về `True`/`False`.
> - ⚠️ `=` là **gán**, `==` là **so sánh**.
> - So sánh nối tiếp: `1 < x < 10`.
> - ⚠️ `0.1 + 0.2 == 0.3` → `False`; dùng `math.isclose()`.
