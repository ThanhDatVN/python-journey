## Variables — Biến

Biến (variable) là "chiếc hộp" có tên dùng để **lưu trữ giá trị dữ liệu**.

## Creating Variables — Tạo biến

Python không có lệnh khai báo biến riêng. Biến được tạo ra **ngay lần đầu bạn gán giá trị** cho nó.

### Example

````python
x = 5
y = "John"

print(x)
print(y)
````

Biến không cần khai báo kiểu dữ liệu, và có thể **đổi kiểu** sau khi đã gán:

````python
x = 4        # x đang là kiểu int
x = "Sally"  # bây giờ x là kiểu str
print(x)
````

> 🧪 **Try it Yourself** — Đổi giá trị của name và age thành thông tin của bạn rồi chạy lại.

````python
name = "John"
age = 25

print(name)
print(age)
````

## Casting — Ép kiểu

Muốn chỉ định rõ kiểu dữ liệu của biến, bạn có thể dùng *casting*:

````python
x = str(3)    # x sẽ là '3'
y = int(3)    # y sẽ là 3
z = float(3)  # z sẽ là 3.0

print(x)
print(y)
print(z)
````

## Get the Type — Xem kiểu dữ liệu

Hàm `type()` cho biết kiểu dữ liệu của một biến:

````python
x = 5
y = "John"

print(type(x))
print(type(y))
````

Bạn sẽ học thêm về kiểu dữ liệu và casting ở các bài sau.

## Single or Double Quotes? — Nháy đơn hay nháy kép?

Biến kiểu chuỗi có thể khai báo bằng nháy đơn hoặc nháy kép — kết quả như nhau:

````python
x = "John"
y = 'John'

print(x)
print(y)
````

## Case-Sensitive — Phân biệt hoa thường

Tên biến **phân biệt chữ hoa và chữ thường**:

````python
a = 4
A = "Sally"
# A sẽ không ghi đè lên a — đây là hai biến khác nhau

print(a)
print(A)
````

---

## 🧠 Ghi nhớ nhanh

> - Biến được tạo **ngay khi gán**: `x = 5` — không cần khai báo kiểu.
> - Một biến có thể **đổi kiểu** khi gán giá trị mới.
> - `type(x)` cho biết kiểu dữ liệu; `str()`, `int()`, `float()` để ép kiểu.
> - Tên biến **phân biệt hoa thường**: `a` và `A` là hai biến khác nhau.
