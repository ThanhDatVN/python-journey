## Python Install — Cài đặt Python

Nhiều máy macOS và Linux đã có sẵn Python. Để kiểm tra, mở **Command Prompt** (Windows) hoặc **Terminal** (macOS/Linux) rồi gõ:

```bash
python --version
```

Trên macOS/Linux, lệnh có thể là `python3 --version`.

Nếu máy chưa có Python, hãy tải bản mới nhất tại [python.org/downloads](https://www.python.org/downloads/).

> ⚠️ **Lưu ý cho Windows:** khi cài đặt, nhớ tích chọn **"Add python.exe to PATH"** để có thể gõ lệnh `python` ở bất kỳ đâu.

## Python Quickstart — Chương trình đầu tiên

Python là ngôn ngữ thông dịch: bạn viết code vào file `.py`, rồi đưa file đó cho trình thông dịch Python chạy.

Cách truyền thống: tạo file `helloworld.py` bằng trình soạn thảo, rồi chạy trong terminal:

```bash
python helloworld.py
```

Trong notebook, ta có thể tạo file bằng *magic command* `%%writefile` (dòng đầu tiên của cell):

````python
%%writefile helloworld.py
print("Hello, World!")
````

Rồi chạy file đó — dấu `!` ở đầu dòng nghĩa là "chạy lệnh này trong terminal":

````python
!python helloworld.py
````

Chúc mừng! Bạn vừa viết và chạy chương trình Python đầu tiên. 🎉

## The Python Command Line — Chế độ tương tác

Muốn thử nhanh vài dòng code, bạn có thể mở chế độ tương tác bằng cách gõ `python` trong terminal:

```text
C:\Users\Your Name>python
Python 3.12.0 ...
Type "help", "copyright", "credits" or "license" for more information.
>>> print("Hello, World!")
Hello, World!
>>> exit()
```

Gõ `exit()` (hoặc nhấn `Ctrl + Z` rồi `Enter` trên Windows, `Ctrl + D` trên macOS/Linux) để thoát.

> 💡 Mỗi cell trong notebook giống như một "phiên tương tác" nâng cao: chạy lúc nào cũng được, sửa lúc nào cũng được.
