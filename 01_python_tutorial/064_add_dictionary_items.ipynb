## Adding Items — Thêm phần tử

Thêm phần tử bằng cách gán giá trị cho một **key mới**:

````python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict["color"] = "red"
print(thisdict)
````

> 📌 Cú pháp thêm và sửa **giống hệt nhau**: nếu key đã có → sửa giá trị; nếu chưa có → thêm mới.

## Update Dictionary — Thêm bằng update()

`update()` cũng thêm được phần tử — nếu key chưa tồn tại, phần tử sẽ được **thêm vào**:

````python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict.update({"color": "red"})
print(thisdict)
````

## ➕ Mở rộng: setdefault()

`setdefault(key, value)` chỉ thêm khi key **chưa có**, và luôn trả về giá trị hiện tại của key:

````python
stock = {"apple": 10}
stock.setdefault("apple", 0)    # đã có -> không đổi
stock.setdefault("banana", 0)   # chưa có -> thêm
print(stock)
````

## ➕ Mở rộng: Đếm tần suất bằng dictionary

````python
text = "the cat and the dog and the bird"
counts = {}
for word in text.split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
````

> 🧪 **Try it Yourself** — Xây dựng danh bạ: thêm 3 người vào dictionary rỗng.

````python
phonebook = {}
phonebook["An"] = "0901 111 222"
phonebook["Bình"] = "0902 333 444"
phonebook.update({"Chi": "0903 555 666"})
print(phonebook)
````

---

## 🧠 Ghi nhớ nhanh

> - Thêm phần tử: `d["key_moi"] = value` (cùng cú pháp với sửa).
> - `update({...})` → key có rồi thì sửa, chưa có thì thêm.
> - Mẹo đếm: `d[k] = d.get(k, 0) + 1`.
