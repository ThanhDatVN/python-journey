## Lambda Functions — Hàm lambda

**Lambda** là một hàm nhỏ **không có tên** (anonymous function).

Lambda nhận **bao nhiêu đối số cũng được**, nhưng chỉ có **một biểu thức**.

### Syntax — Cú pháp

```text
lambda arguments : expression
```

Biểu thức được tính và **kết quả được trả về**.

Lambda cộng 10 vào đối số `a`, rồi trả về kết quả:

````python
x = lambda a: a + 10
print(x(5))
````

Lambda có thể nhận nhiều đối số. Nhân `a` với `b`:

````python
x = lambda a, b: a * b
print(x(5, 6))
````

Cộng ba đối số:

````python
x = lambda a, b, c: a + b + c
print(x(5, 6, 2))
````

## Why Use Lambda Functions? — Vì sao dùng lambda?

Sức mạnh của lambda thể hiện rõ nhất khi dùng **bên trong một hàm khác**.

Ví dụ: hàm nhận một đối số `n` và trả về một lambda nhân đối số của nó với `n`:

````python
def myfunc(n):
    return lambda a: a * n

mydoubler = myfunc(2)
print(mydoubler(11))
````

Dùng cùng hàm đó để tạo cả hàm nhân đôi và nhân ba:

````python
def myfunc(n):
    return lambda a: a * n

mydoubler = myfunc(2)
mytripler = myfunc(3)

print(mydoubler(11))
print(mytripler(11))
````

> 💡 Dùng lambda khi cần một hàm **ngắn, dùng tạm** trong thời gian ngắn.

## Lambda with Built-in Functions — Lambda với hàm có sẵn

Lambda thường được dùng với `map()`, `filter()` và `sorted()`.

### Using Lambda with map() — Dùng với map()

`map()` áp dụng một hàm lên **từng phần tử**:

````python
numbers = [1, 2, 3, 4, 5]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)
````

### Using Lambda with filter() — Dùng với filter()

`filter()` giữ lại các phần tử mà hàm trả về `True`:

````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
odd_numbers = list(filter(lambda x: x % 2 != 0, numbers))
print(odd_numbers)
````

### Using Lambda with sorted() — Dùng với sorted()

Sắp xếp list các tuple theo phần tử thứ hai:

````python
students = [("Emil", 25), ("Tobias", 22), ("Linus", 28)]
sorted_students = sorted(students, key=lambda x: x[1])
print(sorted_students)
````

Sắp xếp chuỗi theo độ dài:

````python
words = ["apple", "pie", "banana", "cherry"]
sorted_words = sorted(words, key=lambda x: len(x))
print(sorted_words)
````

> 🧪 **Try it Yourself** — Dùng sorted + lambda để sắp xếp sản phẩm theo giá giảm dần.

````python
products = [("Bút", 5000), ("Vở", 12000), ("Thước", 8000)]
print(sorted(products, key=lambda p: p[1], reverse=True))
````

## 🧠 Ghi nhớ nhanh

> - `lambda đối_số: biểu_thức` → hàm vô danh, **một biểu thức**, tự trả về kết quả.
> - Hay dùng làm `key=` cho `sorted()`, hoặc với `map()` và `filter()`.
> - Hàm trả về lambda → tạo "nhà máy hàm": `myfunc(2)` thành hàm nhân đôi.
> - Logic dài/phức tạp → dùng `def` cho dễ đọc.
