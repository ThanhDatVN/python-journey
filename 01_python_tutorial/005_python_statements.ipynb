## Statements — Câu lệnh

Một chương trình máy tính là một danh sách các **"chỉ thị"** để máy tính thực hiện. Trong lập trình, mỗi chỉ thị như vậy gọi là một **câu lệnh (statement)**.

Câu lệnh sau yêu cầu Python in ra dòng chữ "Python is fun!":

```python
print("Python is fun!")
```

Trong Python, mỗi câu lệnh thường nằm trên **một dòng** và kết thúc khi xuống dòng — không cần dấu `;` như C hay Java.

## Many Statements — Nhiều câu lệnh

Hầu hết chương trình Python gồm nhiều câu lệnh. Chúng được thực hiện **lần lượt, từ trên xuống dưới**, đúng theo thứ tự bạn viết:

```python
print("Hello World!")
print("Have a good day.")
print("Learning Python is fun!")
```

Thử đổi thứ tự các dòng và quan sát kết quả thay đổi theo:

> 🧪 **Try it Yourself** — Đảo vị trí các dòng print rồi chạy lại.

```python
print("1. Thức dậy")
print("2. Đánh răng")
print("3. Đi học Python")
```

## Semicolons (Optional, Rarely Used) — Dấu chấm phẩy

Dấu chấm phẩy `;` là **tuỳ chọn**. Bạn *có thể* dùng nó để đặt nhiều câu lệnh trên cùng một dòng:

```python
print("Hello"); print("How are you?"); print("Bye bye!")
```

Nếu đặt hai câu lệnh trên cùng một dòng mà **không** có `;`, Python sẽ báo lỗi:

```python
print("Python is fun!") print("Really!")
```

> 💡 Dù hợp lệ, việc nhồi nhiều lệnh vào một dòng làm code khó đọc. Quy ước của Python (PEP 8) khuyên **mỗi dòng một câu lệnh**.

---

## 🧠 Ghi nhớ nhanh

> - Mỗi chỉ thị là một **statement**; chương trình chạy **từ trên xuống dưới**.
> - Mỗi dòng một câu lệnh — không cần `;`.
> - `;` cho phép nhiều lệnh trên một dòng nhưng **hiếm khi dùng** (khó đọc).
> - ⚠️ Hai lệnh cùng dòng mà không có `;` → `SyntaxError`.
