## Access Items — Truy cập phần tử

Bạn **không thể** truy cập phần tử set bằng chỉ số hoặc key, vì set không có thứ tự:

```python
thisset = {"apple", "banana", "cherry"}
print(thisset[0])
```

Thay vào đó, bạn có thể **duyệt** qua set bằng vòng lặp `for`, hoặc **kiểm tra** một giá trị có trong set không bằng từ khoá `in`.

Duyệt qua set:

```python
thisset = {"apple", "banana", "cherry"}

for x in thisset:
    print(x)
```

Kiểm tra `"banana"` có trong set không:

```python
thisset = {"apple", "banana", "cherry"}
print("banana" in thisset)
```

Kiểm tra `"banana"` **không** có trong set:

```python
thisset = {"apple", "banana", "cherry"}
print("banana" not in thisset)
```

## Change Items — Sửa phần tử

Sau khi tạo set, bạn **không thể sửa** phần tử, nhưng có thể **thêm** phần tử mới (bài tiếp theo).

## ➕ Mở rộng: Vì sao kiểm tra `in` với set rất nhanh?

Set được cài đặt bằng *bảng băm (hash table)*, nên phép kiểm tra `in` gần như tức thì — kể cả khi set có hàng triệu phần tử. List thì phải dò từng phần tử.

```python
banned_words = {"spam", "scam", "hack"}
message = "this is not a scam"

for word in message.split():
    if word in banned_words:
        print("Phát hiện từ cấm:", word)
```

> 🧪 **Try it Yourself** — Kiểm tra một số có nằm trong tập số nguyên tố nhỏ hơn 20 hay không.

```python
primes = {2, 3, 5, 7, 11, 13, 17, 19}
n = 15
print(n in primes)
```

> - Kiểm tra `in` với set **rất nhanh** (hash table).
