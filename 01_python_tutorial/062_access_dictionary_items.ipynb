# Python - Access Dictionary Items

>
> 🎯 **Trong bài này:** Accessing Items · Get Keys · Get Values · Get Items · Check if Key Exists

## Accessing Items — Truy cập phần tử

Truy cập phần tử bằng cách ghi **tên key** trong ngoặc vuông:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
x = thisdict["model"]
print(x)
```

Phương thức `get()` cho kết quả tương tự:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
x = thisdict.get("model")
print(x)
```

Khác biệt quan trọng: key **không tồn tại** thì `[]` báo lỗi, còn `get()` trả về `None` (hoặc giá trị mặc định bạn chỉ định):

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(thisdict["color"])
```

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(thisdict.get("color"))
print(thisdict.get("color", "unknown"))
```

## Get Keys — Lấy danh sách key

Phương thức `keys()` trả về danh sách tất cả các key:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
x = thisdict.keys()
print(x)
```

Danh sách này là một **view** (khung nhìn) của dictionary — thay đổi dictionary thì danh sách key cũng được cập nhật theo:

```python
car = {"brand": "Ford", "model": "Mustang", "year": 1964}

x = car.keys()
print(x)  # trước khi thay đổi

car["color"] = "white"
print(x)  # sau khi thay đổi
```

## Get Values — Lấy danh sách value

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
x = thisdict.values()
print(x)
```

Cũng là view — cập nhật theo dictionary:

```python
car = {"brand": "Ford", "model": "Mustang", "year": 1964}
x = car.values()
car["year"] = 2020
print(x)
```

## Get Items — Lấy các cặp key-value

Phương thức `items()` trả về từng phần tử dưới dạng **tuple** `(key, value)`:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
x = thisdict.items()
print(x)
```

## Check if Key Exists — Kiểm tra key có tồn tại

Dùng từ khoá `in`:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
if "model" in thisdict:
    print("Yes, 'model' is one of the keys in the thisdict dictionary")
```

> 🧪 **Try it Yourself** — Tra cứu điểm của một học sinh; nếu không có tên thì in ra "Không tìm thấy".

```python
scores = {"An": 8.5, "Bình": 7.0, "Chi": 9.5}
name = "Dũng"
print(scores.get(name, "Không tìm thấy"))
```

---

## 🧠 Ghi nhớ nhanh

> - `d["key"]` → ⚠️ lỗi `KeyError` nếu key không tồn tại.
> - `d.get("key", mặc_định)` → an toàn, trả về `None`/mặc định.
> - `keys()` · `values()` · `items()` → các **view** tự cập nhật theo dictionary.
> - `"key" in d` → kiểm tra key tồn tại.
