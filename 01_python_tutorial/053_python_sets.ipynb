## Set

Set dùng để lưu **nhiều phần tử** trong một biến.

Set là một trong 4 kiểu tập hợp có sẵn của Python (cùng với List, Tuple và Dictionary).

Set là tập hợp **không có thứ tự**, **không đánh chỉ số**, và phần tử **không sửa được** (nhưng có thể **thêm** hoặc **xoá** phần tử).

Set được viết bằng **dấu ngoặc nhọn**:

```python
thisset = {"apple", "banana", "cherry"}
print(thisset)
```

> 📌 Vì set không có thứ tự, các phần tử có thể hiện ra theo **thứ tự khác** mỗi lần chạy.

## Set Items — Phần tử của set

Phần tử trong set: **không có thứ tự**, **không sửa được**, và **không cho phép trùng lặp**.

### Unordered — Không có thứ tự

Phần tử không có thứ tự cố định và không thể truy cập bằng chỉ số.

### Unchangeable — Không sửa được

Không thể sửa một phần tử sau khi set được tạo — nhưng có thể xoá phần tử cũ và thêm phần tử mới.

### Duplicates Not Allowed — Không trùng lặp

Giá trị trùng sẽ bị bỏ qua:

```python
thisset = {"apple", "banana", "cherry", "apple"}
print(thisset)
```

> ⚠️ `True` và `1` được coi là **cùng giá trị**; `False` và `0` cũng vậy:

```python
thisset = {"apple", "banana", "cherry", True, 1, 2}
print(thisset)

thisset = {"apple", "banana", "cherry", False, True, 0}
print(thisset)
```

## Get the Length of a Set — Độ dài set

```python
thisset = {"apple", "banana", "cherry"}
print(len(thisset))
```

## Set Items - Data Types — Kiểu dữ liệu của phần tử

```python
set1 = {"apple", "banana", "cherry"}
set2 = {1, 5, 7, 9, 3}
set3 = {True, False, False}
set4 = {"abc", 34, True, 40, "male"}

print(set1, set2, set3, set4)
```

## type() & The set() Constructor

```python
myset = {"apple", "banana", "cherry"}
print(type(myset))

thisset = set(("apple", "banana", "cherry"))  # chú ý hai cặp ngoặc
print(thisset)
```

> ⚠️ `{}` tạo ra **dictionary rỗng**, không phải set rỗng. Muốn tạo set rỗng, dùng `set()`:

```python
print(type({}))
print(type(set()))
```

## ➕ Mở rộng: Loại bỏ phần tử trùng

```python
emails = ["a@x.com", "b@x.com", "a@x.com", "c@x.com", "b@x.com"]
unique = set(emails)
print(unique)
print(len(unique), "email khác nhau")
```

> 🧪 **Try it Yourself** — Đếm xem có bao nhiêu chữ cái khác nhau trong một từ.

```python
word = "mississippi"
letters = set(word)
print(letters, len(letters))
```

## 🧠 Ghi nhớ nhanh

> - Set: `{ }` — **không thứ tự · không chỉ số · không trùng lặp**; phần tử không sửa được nhưng **thêm/xoá được**.
> - ⚠️ `True == 1`, `False == 0` → bị coi là trùng.
> - ⚠️ `{}` là **dict rỗng**; set rỗng phải viết `set()`.
> - Ứng dụng: `set(list)` để loại trùng.
