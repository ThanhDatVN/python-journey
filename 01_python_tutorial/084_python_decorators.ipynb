# Python Decorators

> 📖 **Học song song trên W3Schools:** [python_decorators.asp](https://www.w3schools.com/python/python_decorators.asp)
>
> 🎯 **Trong bài này:** Basic Decorator · Multiple Decorator Calls · Arguments in the Decorated Function · *args and **kwargs · Decorator With Arguments · Multiple Decorators · Preserving Function Metadata · ➕ Mở rộng: Decorator đo thời gian chạy

**Decorator** cho phép thêm hành vi mới cho một hàm **mà không sửa code bên trong hàm đó**.

Về bản chất, decorator là một hàm **nhận một hàm khác làm đối số**, và **trả về một hàm mới**.

## Basic Decorator — Decorator cơ bản

Định nghĩa decorator trước, sau đó áp dụng bằng cú pháp `@ten_decorator` đặt ngay trên định nghĩa hàm:

```python
def changecase(func):
    def myinner():
        return func().upper()
    return myinner

@changecase
def myfunction():
    return "Hello Sally"

print(myfunction())
```

Bằng cách đặt `@changecase` ngay trên định nghĩa, hàm `myfunction` được "trang trí" (decorate) bởi `changecase`. Python thực chất làm việc này:

```python
def changecase(func):
    def myinner():
        return func().upper()
    return myinner

def myfunction():
    return "Hello Sally"

myfunction = changecase(myfunction)   # @changecase tương đương dòng này
print(myfunction())
```

## Multiple Decorator Calls — Dùng lại decorator

Một decorator có thể dùng cho **nhiều hàm**:

```python
def changecase(func):
    def myinner():
        return func().upper()
    return myinner

@changecase
def myfunction():
    return "Hello Sally"

@changecase
def otherfunction():
    return "I am speed!"

print(myfunction())
print(otherfunction())
```

## Arguments in the Decorated Function — Hàm được trang trí có đối số

Hàm bên trong (`myinner`) phải nhận và **chuyển tiếp** đối số cho hàm gốc:

```python
def changecase(func):
    def myinner(x):
        return func(x).upper()
    return myinner

@changecase
def myfunction(nam):
    return "Hello " + nam

print(myfunction("John"))
```

## *args and **kwargs

Dùng `*args, **kwargs` để decorator chạy được với **mọi** hàm, dù hàm có bao nhiêu đối số:

```python
def changecase(func):
    def myinner(*args, **kwargs):
        return func(*args, **kwargs).upper()
    return myinner

@changecase
def myfunction(nam):
    return "Hello " + nam

@changecase
def greet(first, last="Doe"):
    return f"Hi {first} {last}"

print(myfunction("John"))
print(greet("Jane", last="Smith"))
```

## Decorator With Arguments — Decorator có tham số

Muốn truyền tham số cho decorator, cần **thêm một tầng hàm** bao ngoài:

```python
def changecase(n):
    def changecase(func):
        def myinner():
            if n == 1:
                a = func().lower()
            else:
                a = func().upper()
            return a
        return myinner
    return changecase

@changecase(1)
def myfunction():
    return "Hello Linus"

print(myfunction())
```

## Multiple Decorators — Nhiều decorator

Có thể chồng nhiều decorator lên một hàm. Decorator **gần hàm nhất** được áp dụng **trước**:

```python
def changecase(func):
    def myinner():
        return func().upper()
    return myinner

def addgreeting(func):
    def myinner():
        return "Hello " + func() + " Have a good day!"
    return myinner

@changecase
@addgreeting
def myfunction():
    return "Tobias"

print(myfunction())
```

## Preserving Function Metadata — Giữ thông tin của hàm gốc

Hàm có các thông tin như `__name__` (tên) và `__doc__` (docstring):

```python
def myfunction():
    return "Have a great day!"

print(myfunction.__name__)
```

Sau khi bị decorate, các thông tin này bị **thay** bằng thông tin của hàm `myinner`:

```python
def changecase(func):
    def myinner():
        return func().upper()
    return myinner

@changecase
def myfunction():
    return "Have a great day!"

print(myfunction.__name__)
```

Dùng `functools.wraps` để **giữ lại** thông tin của hàm gốc:

```python
import functools

def changecase(func):
    @functools.wraps(func)
    def myinner():
        return func().upper()
    return myinner

@changecase
def myfunction():
    return "Have a great day!"

print(myfunction.__name__)
```

## ➕ Mở rộng: Decorator đo thời gian chạy

```python
import functools
import time

def timer(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        start = time.perf_counter()
        result = func(*args, **kwargs)
        print(f"{func.__name__} chạy mất {time.perf_counter() - start:.5f} giây")
        return result
    return wrapper

@timer
def slow_sum(n):
    return sum(range(n))

print(slow_sum(1_000_000))
```

> 🧪 **Try it Yourself** — Viết decorator log_call in ra "Gọi hàm <tên>" trước khi chạy hàm.

```python
import functools

def log_call(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        print("Gọi hàm", func.__name__)
        return func(*args, **kwargs)
    return wrapper

@log_call
def add(a, b):
    return a + b

print(add(2, 3))
```

## 🧠 Ghi nhớ nhanh

> - Decorator = hàm **nhận hàm** và **trả về hàm mới** → thêm hành vi mà không sửa hàm gốc.
> - `@deco` trên `def f` ⇔ `f = deco(f)`.
> - Dùng `*args, **kwargs` trong wrapper để decorator hợp với mọi hàm.
> - Decorator có tham số → thêm **một tầng hàm** bao ngoài.
> - Nhiều decorator: cái **gần hàm nhất chạy trước**; luôn dùng `@functools.wraps(func)`.
