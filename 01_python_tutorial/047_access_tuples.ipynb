# Python - Access Tuple Items

> 📖 **Học song song trên W3Schools:** [python_tuples_access.asp](https://www.w3schools.com/python/python_tuples_access.asp)
>
> 🎯 **Trong bài này:** Access Tuple Items · Negative Indexing · Range of Indexes · Range of Negative Indexes · Check if Item Exists

## Access Tuple Items — Truy cập phần tử

Truy cập phần tử tuple bằng chỉ số trong ngoặc vuông, giống list:

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple[1])
```

## Negative Indexing — Chỉ số âm

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple[-1])
```

## Range of Indexes — Khoảng chỉ số

Kết quả là một **tuple mới**:

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple[2:5])
```

Bỏ trống vị trí bắt đầu hoặc kết thúc:

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple[:4])
print(thistuple[2:])
```

## Range of Negative Indexes — Khoảng chỉ số âm

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(thistuple[-4:-1])
```

## Check if Item Exists — Kiểm tra phần tử

```python
thistuple = ("apple", "banana", "cherry")
if "apple" in thistuple:
    print("Yes, 'apple' is in the fruits tuple")
```

> 🧪 **Try it Yourself** — Lấy ra các ngày cuối tuần (2 phần tử cuối) bằng chỉ số âm.

```python
days = ("Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun")
print(days[-2:])
```

---

## 🧠 Ghi nhớ nhanh

> - Truy cập giống list: `t[0]`, `t[-1]`, `t[2:5]`.
> - Cắt tuple → **tuple mới**.
> - `x in t` → kiểm tra tồn tại.
