## Escape Character — Ký tự thoát

Để chèn những ký tự "không hợp lệ" vào chuỗi, ta dùng **ký tự thoát (escape character)**: dấu gạch chéo ngược `\` theo sau là ký tự muốn chèn.

Ví dụ, dấu nháy kép bên trong chuỗi được bao bởi nháy kép sẽ gây lỗi:

````python
txt = "We are the so-called "Vikings" from the north."
````

Sửa bằng cách dùng `\"`:

````python
txt = "We are the so-called \"Vikings\" from the north."
print(txt)
````

## Escape Characters — Bảng ký tự thoát

| Code | Kết quả |
|---|---|
| `\'` | Nháy đơn |
| `\"` | Nháy kép |
| `\\` | Dấu gạch chéo ngược |
| `\n` | Xuống dòng (new line) |
| `\r` | Về đầu dòng (carriage return) |
| `\t` | Tab |
| `\b` | Xoá lùi (backspace) |
| `\f` | Sang trang (form feed) |
| `\ooo` | Ký tự theo giá trị bát phân (octal) |
| `\xhh` | Ký tự theo giá trị thập lục phân (hex) |

### Examples

````python
print('It\'s alright.')
print("This will insert one \\ (backslash).")
print("Hello\nWorld!")
print("Hello\tWorld!")
````

````python
# Giá trị bát phân và thập lục phân
print("\110\145\154\154\157")
print("\x48\x65\x6c\x6c\x6f")
````

## ➕ Mở rộng: Raw String — Chuỗi thô

Thêm chữ `r` trước chuỗi để Python **không** xử lý ký tự thoát. Rất tiện khi viết đường dẫn Windows hoặc biểu thức RegEx:

````python
print("C:\\Users\\name\\new_folder")
print(r"C:\Users\name\new_folder")
````

> 🧪 **Try it Yourself** — Dùng \n và \t để in ra một bảng nhỏ gồm hai cột Tên và Tuổi.

````python
print("Tên\tTuổi\nAn\t20\nBình\t22")
````

---

## 🧠 Ghi nhớ nhanh

> - Escape character = `\` + ký tự: `\"` `\'` `\\` `\n` (xuống dòng) `\t` (tab).
> - Raw string `r"C:\new"` → không xử lý ký tự thoát (tiện cho đường dẫn, RegEx).
