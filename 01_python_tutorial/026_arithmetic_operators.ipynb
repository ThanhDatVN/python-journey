## Arithmetic Operators — Toán tử số học

Toán tử số học dùng với giá trị số để thực hiện các phép toán thông thường:

| Toán tử | Tên | Ví dụ |
|---|---|---|
| `+` | Cộng | `x + y` |
| `-` | Trừ | `x - y` |
| `*` | Nhân | `x * y` |
| `/` | Chia | `x / y` |
| `%` | Chia lấy dư (modulus) | `x % y` |
| `**` | Luỹ thừa | `x ** y` |
| `//` | Chia lấy phần nguyên (floor division) | `x // y` |

## Examples

````python
x = 15
y = 4

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x % y)
print(x ** y)
print(x // y)
````

## Division in Python — Phép chia trong Python

Python có **hai** toán tử chia:

- `/` — chia thường, **luôn trả về float**.
- `//` — chia lấy phần nguyên, **làm tròn xuống** về số nguyên gần nhất.

````python
print(12 / 3)    # 4.0 — vẫn là float dù chia hết
print(12 // 3)   # 4
print(7 / 2)     # 3.5
print(7 // 2)    # 3
````

> ⚠️ `//` làm tròn **xuống** (về phía âm vô cùng), nên với số âm kết quả có thể khiến bạn bất ngờ:

````python
print(-7 // 2)   # -4, không phải -3
````

## ➕ Mở rộng: Modulus — Chia lấy dư

`%` trả về **phần dư** của phép chia. Rất hay dùng để kiểm tra chẵn/lẻ hoặc chia hết:

````python
print(10 % 3)   # 1
print(10 % 2)   # 0 -> 10 là số chẵn
print(7 % 2)    # 1 -> 7 là số lẻ
````

## ➕ Mở rộng: Exponentiation — Luỹ thừa

````python
print(2 ** 10)     # 1024
print(9 ** 0.5)    # căn bậc hai -> 3.0
````

## ➕ Mở rộng: Chia cho 0

````python
print(10 / 0)
````

> 🧪 **Try it Yourself** — Đổi tổng số giây rồi chạy lại: chương trình đổi giây sang phút và giây.

````python
total_seconds = 250

minutes = total_seconds // 60
seconds = total_seconds % 60
print(minutes, "phút", seconds, "giây")
````

---

## 🧠 Ghi nhớ nhanh

> - `+ - * /` · `%` (chia lấy dư) · `**` (luỹ thừa) · `//` (chia lấy nguyên).
> - `/` **luôn** trả về `float`: `12 / 3` → `4.0`.
> - `//` làm tròn **xuống**: `7 // 2` → `3`, `-7 // 2` → `-4`.
> - `n % 2 == 0` → kiểm tra số chẵn.
