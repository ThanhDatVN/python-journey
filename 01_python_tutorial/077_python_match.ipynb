## The Python Match Statement — Câu lệnh match

Thay vì viết nhiều câu `if..elif`, bạn có thể dùng câu lệnh `match` (có từ **Python 3.10**).

Câu lệnh `match` chọn **một** trong nhiều khối code để thực thi.

### Syntax — Cú pháp

```text
match expression:
    case x:
        code block
    case y:
        code block
    case z:
        code block
```

Cách hoạt động:

- Biểu thức sau `match` được đánh giá **một lần**.
- Giá trị đó được so sánh lần lượt với từng `case`.
- Nếu khớp, khối code của `case` đó được thực thi.

In tên thứ trong tuần dựa trên số thứ tự:

````python
day = 4
match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
````

## Default Value — Giá trị mặc định

Dùng dấu gạch dưới `_` làm `case` cuối cùng để xử lý **mọi trường hợp còn lại** (giống `else`):

````python
day = 4
match day:
    case 6:
        print("Today is Saturday")
    case 7:
        print("Today is Sunday")
    case _:
        print("Looking forward to the Weekend")
````

> 📌 `case _:` luôn khớp nên phải đặt ở **cuối cùng**.

## Combine Values — Kết hợp nhiều giá trị

Dùng ký tự `|` (nghĩa là **hoặc**) để kiểm tra nhiều giá trị trong một `case`:

````python
day = 4
match day:
    case 1 | 2 | 3 | 4 | 5:
        print("Today is a weekday")
    case 6 | 7:
        print("I love weekends!")
````

## If Statements as Guards — Điều kiện bảo vệ (guard)

Thêm `if` vào `case` để kiểm tra điều kiện bổ sung:

````python
month = 5
day = 4
match day:
    case 1 | 2 | 3 | 4 | 5 if month == 4:
        print("A weekday in April")
    case 1 | 2 | 3 | 4 | 5 if month == 5:
        print("A weekday in May")
    case _:
        print("No match")
````

## ➕ Mở rộng: match với chuỗi và cấu trúc dữ liệu

`match` còn so khớp được **cấu trúc** (structural pattern matching):

````python
command = "go north"

match command.split():
    case ["go", direction]:
        print("Đi về hướng", direction)
    case ["look"]:
        print("Nhìn xung quanh")
    case ["quit" | "exit"]:
        print("Thoát game")
    case _:
        print("Không hiểu lệnh")
````

> 🧪 **Try it Yourself** — Viết máy tính đơn giản: match theo phép toán "+", "-", "*", "/".

````python
a, b = 10, 4
op = "*"

match op:
    case "+":
        print(a + b)
    case "-":
        print(a - b)
    case "*":
        print(a * b)
    case "/":
        print(a / b)
    case _:
        print("Phép toán không hợp lệ")
````

---

## 🧠 Ghi nhớ nhanh

> - `match giá_trị:` + các `case` — thay cho chuỗi `if/elif` dài (Python **3.10+**).
> - `case _:` → trường hợp mặc định, đặt **cuối cùng**.
> - `case 1 | 2:` → khớp một trong nhiều giá trị.
> - `case x if điều_kiện:` → thêm điều kiện bảo vệ (guard).
