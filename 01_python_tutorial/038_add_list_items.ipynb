# Python - Add List Items

## Append Items — Thêm vào cuối

Thêm một phần tử vào **cuối** list bằng phương thức `append()`:

````python
thislist = ["apple", "banana", "cherry"]
thislist.append("orange")
print(thislist)
````

## Insert Items — Chèn vào vị trí bất kỳ

Chèn phần tử vào vị trí chỉ định bằng `insert()`:

````python
thislist = ["apple", "banana", "cherry"]
thislist.insert(1, "orange")
print(thislist)
````

> 📌 Sau các ví dụ trên, list có 4 phần tử.

## Extend List — Nối list

Thêm các phần tử của **list khác** vào list hiện tại bằng `extend()`:

````python
thislist = ["apple", "banana", "cherry"]
tropical = ["mango", "pineapple", "papaya"]
thislist.extend(tropical)
print(thislist)
````

## Add Any Iterable — Thêm từ bất kỳ iterable nào

`extend()` không chỉ nhận list, mà nhận **mọi iterable** (tuple, set, dictionary…):

````python
thislist = ["apple", "banana", "cherry"]
thistuple = ("kiwi", "orange")
thislist.extend(thistuple)
print(thislist)
````

## ➕ Mở rộng: append() khác extend() thế nào?

````python
a = [1, 2]
a.append([3, 4])   # thêm cả list như MỘT phần tử
print(a)

b = [1, 2]
b.extend([3, 4])   # thêm TỪNG phần tử
print(b)
````

> 🧪 **Try it Yourself** — Xây dựng danh sách việc cần làm: thêm 2 việc bằng append và chèn 1 việc khẩn cấp lên đầu.

````python
todo = ["Học Python"]
todo.append("Làm bài tập")
todo.append("Đọc sách")
todo.insert(0, "Nộp báo cáo")
print(todo)
````

---

## 🧠 Ghi nhớ nhanh

> - `append(x)` → thêm **một** phần tử vào cuối.
> - `insert(i, x)` → chèn vào vị trí `i`.
> - `extend(iterable)` → thêm **từng** phần tử của iterable khác.
> - ⚠️ `append([3, 4])` thêm cả list như **một** phần tử.
