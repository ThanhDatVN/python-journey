## Python range

`range()` là hàm có sẵn, trả về một **dãy số bất biến (immutable)**, thường dùng để lặp một số lần nhất định trong vòng `for`.

## Creating ranges — Tạo range

Cú pháp:

```text
range(start, stop, step)
```

| Tham số | Mô tả |
|---|---|
| `start` | Tuỳ chọn. Số bắt đầu, mặc định `0` |
| `stop` | **Bắt buộc**. Dãy dừng **trước** số này |
| `step` | Tuỳ chọn. Bước nhảy, mặc định `1` |

## Call range() With One Argument — Một đối số

Chỉ truyền `stop`: dãy bắt đầu từ 0 và dừng trước `stop`:

````python
x = range(10)
print(list(x))
````

## Call range() With Two Arguments — Hai đối số

Truyền `start` và `stop`:

````python
x = range(3, 10)
print(list(x))
````

## Call range() With Three Arguments — Ba đối số

Thêm `step` để chỉ định bước nhảy:

````python
x = range(3, 10, 2)
print(list(x))
````

`step` âm → đếm ngược:

````python
print(list(range(10, 0, -2)))
````

## Using ranges — Dùng range

`range` thường được dùng trong vòng lặp `for`:

````python
for i in range(5):
    print(i)
````

## Using List to Display Ranges — Hiển thị range bằng list

In trực tiếp một range chỉ cho ra "công thức" của nó. Muốn xem các giá trị, chuyển sang list:

````python
print(range(5))
print(list(range(5)))
````

## Slicing Ranges — Truy cập và cắt range

Range hỗ trợ chỉ số và slice như list — kết quả cắt là một **range mới**:

````python
r = range(10)
print(r[2])
print(r[:3])
print(list(r[:3]))
````

## Membership Testing — Kiểm tra thành viên

````python
r = range(0, 10, 2)
print(6 in r)
print(7 in r)
````

## Length — Độ dài

````python
r = range(0, 10, 2)
print(len(r))
````

## ➕ Mở rộng: Vì sao range tiết kiệm bộ nhớ?

`range` không lưu tất cả các số mà chỉ lưu `start`, `stop`, `step` — các giá trị được tính khi cần:

````python
import sys
print(sys.getsizeof(range(1_000_000)))        # vài chục byte
print(sys.getsizeof(list(range(1_000_000))))  # hàng MB
````

> 🧪 **Try it Yourself** — Dùng range để in các số chia hết cho 5 từ 0 đến 50.

````python
for n in range(0, 51, 5):
    print(n, end=" ")
````

---

## 🧠 Ghi nhớ nhanh

> - `range(stop)` · `range(start, stop)` · `range(start, stop, step)` — **không gồm** `stop`.
> - `step` âm → đếm ngược: `range(10, 0, -1)`.
> - Xem giá trị: `list(range(...))`; hỗ trợ `r[i]`, slice, `in`, `len()`.
> - Range **bất biến** và **tiết kiệm bộ nhớ**.
