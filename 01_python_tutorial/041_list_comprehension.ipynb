## List Comprehension

List comprehension là cú pháp **ngắn gọn** để tạo list mới từ các giá trị của một list (hoặc iterable) có sẵn.

Ví dụ: từ một list trái cây, tạo list mới chỉ gồm các loại có chữ `"a"` trong tên.

Không dùng list comprehension, bạn phải viết vòng `for` với câu lệnh điều kiện:

````python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = []

for x in fruits:
    if "a" in x:
        newlist.append(x)

print(newlist)
````

Với list comprehension, chỉ cần **một dòng**:

````python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

newlist = [x for x in fruits if "a" in x]

print(newlist)
````

## The Syntax — Cú pháp

```text
newlist = [expression for item in iterable if condition == True]
```

Kết quả trả về là một **list mới**, list cũ không bị thay đổi.

## Condition — Điều kiện

*Điều kiện* giống bộ lọc, chỉ giữ lại các phần tử cho kết quả `True`:

````python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x for x in fruits if x != "apple"]
print(newlist)
````

Điều kiện là **tuỳ chọn**, có thể bỏ:

````python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x for x in fruits]
print(newlist)
````

## Iterable

*Iterable* có thể là bất kỳ đối tượng duyệt được nào: list, tuple, set…

Dùng `range()` để tạo iterable:

````python
newlist = [x for x in range(10)]
print(newlist)
````

Kết hợp với điều kiện — chỉ lấy số nhỏ hơn 5:

````python
newlist = [x for x in range(10) if x < 5]
print(newlist)
````

## Expression — Biểu thức

*Expression* là phần tử ở vị trí hiện tại trong vòng lặp, nhưng cũng là **kết quả** — bạn có thể biến đổi nó trước khi đưa vào list mới:

````python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x.upper() for x in fruits]
print(newlist)
````

Có thể đặt bất kỳ giá trị nào:

````python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = ["hello" for x in fruits]
print(newlist)
````

Expression cũng có thể chứa điều kiện — không phải để lọc, mà để **quyết định giá trị**:

````python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x if x != "banana" else "orange" for x in fruits]
print(newlist)
````

> Đọc là: *"lấy x nếu x không phải banana, còn nếu là banana thì thay bằng orange"*.

> 🧪 **Try it Yourself** — Tạo list bình phương của các số chẵn từ 0 đến 20.

````python
squares = [x ** 2 for x in range(21) if x % 2 == 0]
print(squares)
````

---

## 🧠 Ghi nhớ nhanh

> - Cú pháp: `[expression for item in iterable if condition]`.
> - `if` ở **cuối** = **lọc** phần tử.
> - `x if ... else ...` ở **đầu** = **biến đổi** giá trị.
> - Luôn tạo **list mới**, list gốc không đổi.
