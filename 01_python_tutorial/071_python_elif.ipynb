## The Elif Keyword — Từ khoá elif

Từ khoá `elif` nghĩa là *"nếu các điều kiện trước không đúng, thì thử điều kiện này"*:

````python
a = 33
b = 33
if b > a:
    print("b is greater than a")
elif a == b:
    print("a and b are equal")
````

Vì `a` bằng `b`, điều kiện đầu là `False` nhưng điều kiện `elif` là `True`, nên Python in ra "a and b are equal".

## Multiple Elif Statements — Nhiều elif

Bạn có thể dùng bao nhiêu `elif` tuỳ ý:

````python
score = 75

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
elif score >= 60:
    print("Grade: D")
````

## How Elif Works — elif hoạt động thế nào?

Python kiểm tra các điều kiện **từ trên xuống dưới**, và **chỉ chạy khối đầu tiên** có điều kiện `True` — các khối còn lại bị bỏ qua, dù điều kiện của chúng cũng đúng:

````python
age = 25

if age < 13:
    print("You are a child")
elif age < 20:
    print("You are a teenager")
elif age < 65:
    print("You are an adult")
elif age >= 65:
    print("You are a senior")
````

> ⚠️ Thứ tự điều kiện rất quan trọng! Nếu đặt `score >= 60` lên đầu, mọi điểm từ 60 trở lên đều thành "D".

## When to Use Elif — Khi nào dùng elif?

Dùng `elif` khi các điều kiện **loại trừ nhau** — chỉ một nhánh được chạy. So với nhiều câu `if` riêng lẻ, `elif` hiệu quả hơn vì dừng ngay khi tìm được điều kiện đúng:

````python
day = 3

if day == 1:
    print("Monday")
elif day == 2:
    print("Tuesday")
elif day == 3:
    print("Wednesday")
elif day == 4:
    print("Thursday")
elif day == 5:
    print("Friday")
elif day == 6:
    print("Saturday")
elif day == 7:
    print("Sunday")
````

> 🧪 **Try it Yourself** — Phân loại chỉ số BMI: < 18.5 gầy, < 25 bình thường, < 30 thừa cân, còn lại béo phì.

````python
bmi = 22.4

if bmi < 18.5:
    print("Gầy")
elif bmi < 25:
    print("Bình thường")
elif bmi < 30:
    print("Thừa cân")
elif bmi >= 30:
    print("Béo phì")
````

## 🧠 Ghi nhớ nhanh

> - `elif` = "else if" → kiểm tra thêm điều kiện khi các điều kiện trước sai.
> - Kiểm tra **từ trên xuống**, chỉ chạy **nhánh đúng đầu tiên**.
> - ⚠️ Sắp xếp điều kiện từ **hẹp → rộng** (hoặc cao → thấp) để tránh nhánh sai "nuốt" mất.
