> 🎯 **Trong bài này:** Join Two Tuples · Multiply Tuples · ➕ Mở rộng: Unpacking với `*`

## Join Two Tuples — Nối hai tuple

Dùng toán tử `+` để nối hai hoặc nhiều tuple:

```python
tuple1 = ("a", "b", "c")
tuple2 = (1, 2, 3)

tuple3 = tuple1 + tuple2
print(tuple3)
```

## Multiply Tuples — Nhân tuple

Muốn lặp lại nội dung của tuple nhiều lần, dùng toán tử `*`:

```python
fruits = ("apple", "banana", "cherry")
mytuple = fruits * 2

print(mytuple)
```

## ➕ Mở rộng: Unpacking với `*`

```python
t1 = (1, 2)
t2 = (3, 4)
print((*t1, *t2, 5))
```

> 🧪 **Try it Yourself** — Ghép tuple các ngày trong tuần và tuple các ngày cuối tuần.

```python
weekdays = ("Mon", "Tue", "Wed", "Thu", "Fri")
weekend = ("Sat", "Sun")
week = weekdays + weekend
print(week, len(week))
```
