## Python Numbers — Kiểu số

Python có ba kiểu số:

- `int` — số nguyên
- `float` — số thực (số có phần thập phân)
- `complex` — số phức

Biến kiểu số được tạo khi bạn gán một giá trị số:

```python
x = 1     # int
y = 2.8   # float
z = 1j    # complex

print(type(x))
print(type(y))
print(type(z))
```

## Int — Số nguyên

`int` là số nguyên dương hoặc âm, không có phần thập phân, **độ dài không giới hạn**:

```python
x = 1
y = 35656222554887711
z = -3255522

print(type(x))
print(type(y))
print(type(z))
```

> 💡 Với số lớn, bạn có thể dùng dấu `_` để dễ đọc: `1_000_000` chính là `1000000`.

```python
population = 8_100_000_000
print(population)
```

## Float — Số thực

`float` là số dương hoặc âm có **một hoặc nhiều chữ số thập phân**:

```python
x = 1.10
y = 1.0
z = -35.59

print(type(x))
print(type(y))
print(type(z))
```

Float cũng có thể viết dạng **số khoa học** với chữ `e`, biểu thị luỹ thừa của 10:

```python
x = 35e3
y = 12E4
z = -87.7e100

print(x)
print(type(x))
print(type(y))
print(type(z))
```

## Complex — Số phức

Số phức được viết với chữ `j` biểu thị phần ảo:

```python
x = 3 + 5j
y = 5j
z = -5j

print(type(x))
print(type(y))
print(type(z))
```

## Type Conversion — Chuyển đổi kiểu

Dùng `int()`, `float()` và `complex()` để chuyển đổi giữa các kiểu số:

```python
x = 1     # int
y = 2.8   # float
z = 1j    # complex

a = float(x)   # int -> float
b = int(y)     # float -> int (cắt bỏ phần thập phân)
c = complex(x) # int -> complex

print(a)
print(b)
print(c)

print(type(a))
print(type(b))
print(type(c))
```

> ⚠️ Không thể chuyển số phức sang kiểu số khác:

```python
int(1j)
```

## Random Number — Số ngẫu nhiên

Python không có hàm `random()` có sẵn, nhưng có **module** `random` để tạo số ngẫu nhiên:

```python
import random

print(random.randrange(1, 10))
```

Mỗi lần chạy lại cell, bạn sẽ được một số khác nhau. Xem thêm tại [Random Module](../10_reference/random_module.md).

> 🧪 **Try it Yourself** — Hãy tạo một "xúc xắc" in ra số ngẫu nhiên từ 1 đến 6 (gợi ý: randrange(1, 7)).

```python
import random

print(random.randrange(1, 7))
```

## 🧠 Ghi nhớ nhanh

> - `int` (số nguyên, không giới hạn độ dài) · `float` (có phần thập phân, hoặc dạng `35e3`) · `complex` (`3 + 5j`).
> - Chuyển đổi bằng `int()`, `float()`, `complex()`; ⚠️ `int(2.8)` → `2` (**cắt**, không làm tròn).
> - ⚠️ Không chuyển được số phức sang `int`/`float`.
> - Số ngẫu nhiên: `import random` → `random.randrange(1, 10)`.
