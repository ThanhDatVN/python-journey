## Variable Names — Đặt tên biến

Tên biến có thể ngắn (như `x`, `y`) hoặc mô tả rõ ý nghĩa (như `age`, `carname`, `total_volume`).

**Quy tắc đặt tên biến trong Python:**

- Phải bắt đầu bằng **chữ cái** hoặc **dấu gạch dưới** `_`.
- **Không** được bắt đầu bằng chữ số.
- Chỉ chứa chữ cái, chữ số và dấu gạch dưới (`A-z`, `0-9`, `_`).
- Phân biệt hoa thường: `age`, `Age` và `AGE` là ba biến khác nhau.
- Không được trùng với **từ khoá** của Python (như `if`, `for`, `class`, `True`…).

### Example — Tên hợp lệ

```python
myvar = "John"
my_var = "John"
_my_var = "John"
myVar = "John"
MYVAR = "John"
myvar2 = "John"

print(myvar, my_var, _my_var, myVar, MYVAR, myvar2)
```

### Example — Tên không hợp lệ

```python
2myvar = "John"
my-var = "John"
my var = "John"
```

```python
import keyword

print(keyword.kwlist)
```

## Multi Words Variable Names — Tên biến nhiều từ

Tên gồm nhiều từ sẽ khó đọc nếu viết liền. Có ba cách viết phổ biến:

### Camel Case

Mỗi từ, trừ từ đầu tiên, viết hoa chữ cái đầu:

```python
myVariableName = "John"
```

### Pascal Case

Mỗi từ đều viết hoa chữ cái đầu:

```python
MyVariableName = "John"
```

### Snake Case

Các từ ngăn cách bằng dấu gạch dưới:

```python
my_variable_name = "John"
```

> 💡 Quy ước chính thức của Python (PEP 8): dùng **snake_case** cho biến và hàm, **PascalCase** cho tên class, và **UPPER_CASE** cho hằng số.

> 🧪 **Try it Yourself** — Hãy tạo biến lưu tên trường học của bạn theo kiểu snake_case rồi in ra.

```python
school_name = "Đại học ABC"
print(school_name)
```

## 🧠 Ghi nhớ nhanh

> - Bắt đầu bằng **chữ cái hoặc `_`**; chỉ chứa chữ, số, `_`.
> - ⚠️ Không bắt đầu bằng số, không có `-` hay khoảng trắng, không trùng **từ khoá**.
> - Quy ước Python: `snake_case` cho biến/hàm, `PascalCase` cho class, `UPPER_CASE` cho hằng số.
