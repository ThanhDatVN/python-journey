## The Ternary Operator — Toán tử điều kiện rút gọn

Toán tử *ternary* (còn gọi là **biểu thức điều kiện**) cho phép viết câu lệnh `if...else` đơn giản trên **một dòng**:

```text
giá_trị_khi_đúng if điều_kiện else giá_trị_khi_sai
```

### Example

````python
a = 2
b = 330
print("A") if a > b else print("B")
````

Thường gặp hơn là dùng ternary để **chọn giá trị** gán cho biến:

````python
age = 20
status = "adult" if age >= 18 else "minor"
print(status)
````

So sánh với cách viết đầy đủ:

````python
age = 20
if age >= 18:
    status = "adult"
else:
    status = "minor"
print(status)
````

### Ternary trong biểu thức

Vì là một biểu thức, ternary có thể đặt ngay trong `print()`, f-string, hay đối số của hàm:

````python
x = -7
print(f"{x} là số {'dương' if x > 0 else 'không dương'}")
print(abs(x) if x < 0 else x)
````

## Instead of Elif — Thay cho elif

Có thể lồng nhiều ternary để xử lý 3 khả năng trở lên:

````python
a = 330
b = 330
print("A") if a > b else print("=") if a == b else print("B")
````

````python
score = 75
grade = "A" if score >= 90 else "B" if score >= 70 else "C"
print(grade)
````

> 💡 Ternary lồng nhau nhanh chóng trở nên khó đọc. Nếu có hơn 2 nhánh, nên dùng `if...elif...else` đầy đủ.

> 🧪 **Try it Yourself** — Viết biểu thức ternary trả về "chẵn" hoặc "lẻ" cho số n.

````python
n = 7
result = "chẵn" if n % 2 == 0 else "lẻ"
print(n, "là số", result)
````

## 🧠 Ghi nhớ nhanh

> - Công thức: `giá_trị_đúng if điều_kiện else giá_trị_sai`.
> - Dùng để **chọn giá trị** gọn trên một dòng.
> - Thay `elif`: `"A" if s >= 90 else "B" if s >= 70 else "C"` — nhưng quá 2 nhánh nên dùng `if/elif`.
