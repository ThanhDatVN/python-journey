## Set Methods — Phương thức của set

Python có các phương thức có sẵn sau đây để dùng với set:

| Phương thức | Toán tử | Mô tả |
|---|---|---|
| `add()` | | Thêm một phần tử |
| `clear()` | | Xoá tất cả phần tử |
| `copy()` | | Trả về bản sao |
| `difference()` | `-` | Trả về set hiệu của hai hoặc nhiều set |
| `difference_update()` | `-=` | Xoá khỏi set hiện tại các phần tử có trong set khác |
| `discard()` | | Xoá phần tử (không lỗi nếu không tồn tại) |
| `intersection()` | `&` | Trả về set giao |
| `intersection_update()` | `&=` | Chỉ giữ lại phần tử chung |
| `isdisjoint()` | | `True` nếu hai set không có phần tử chung |
| `issubset()` | `<=` | `True` nếu set này là tập con của set kia |
| | `<` | `True` nếu là tập con thực sự |
| `issuperset()` | `>=` | `True` nếu set này là tập cha của set kia |
| | `>` | `True` nếu là tập cha thực sự |
| `pop()` | | Xoá và trả về một phần tử bất kỳ |
| `remove()` | | Xoá phần tử (lỗi nếu không tồn tại) |
| `symmetric_difference()` | `^` | Trả về set hiệu đối xứng |
| `symmetric_difference_update()` | `^=` | Cập nhật bằng hiệu đối xứng |
| `union()` | `\|` | Trả về set hợp |
| `update()` | `\|=` | Cập nhật bằng hợp với set khác |

Chi tiết: [Set Methods Reference](../10_reference/set_methods.md).

## ➕ Mở rộng: Tập con, tập cha, rời nhau

```python
a = {1, 2}
b = {1, 2, 3, 4}
c = {7, 8}

print(a.issubset(b))     # a ⊆ b
print(a <= b)
print(a < b)             # tập con thực sự
print(b.issuperset(a))   # b ⊇ a
print(a.isdisjoint(c))   # không có phần tử chung
```

## ➕ Mở rộng: Toán tử gán kết hợp

```python
s = {1, 2, 3}
s |= {4, 5}
print(s)
s -= {1}
print(s)
s &= {2, 3, 4, 9}
print(s)
```

> 🧪 **Try it Yourself** — Kiểm tra một người đã hoàn thành tất cả môn bắt buộc chưa (gợi ý: issubset).

```python
required = {"Toán", "Lý", "Tin"}
completed = {"Toán", "Tin", "Anh", "Lý"}

print(required.issubset(completed))
print("Còn thiếu:", required - completed)
```

## 🧠 Ghi nhớ nhanh

> - Thêm/xoá: `add update remove discard pop clear`.
> - Phép tập hợp: `union intersection difference symmetric_difference` (+ các bản `_update`).
> - Quan hệ: `issubset` `<=` · `issuperset` `>=` · `isdisjoint`.
