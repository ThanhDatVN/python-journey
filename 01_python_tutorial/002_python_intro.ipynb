## What is Python? — Python là gì?

Python là một ngôn ngữ lập trình bậc cao, được **Guido van Rossum** tạo ra và phát hành lần đầu năm **1991**.

Python thường được dùng cho:

- Phát triển web phía server (Django, Flask, FastAPI…)
- Phát triển phần mềm, công cụ dòng lệnh
- Toán học, khoa học dữ liệu, trí tuệ nhân tạo
- Viết script tự động hoá hệ thống

## What can Python do? — Python làm được gì?

- Chạy trên server để tạo ứng dụng web.
- Dùng kèm phần mềm khác để tạo *workflow* tự động.
- Kết nối cơ sở dữ liệu, đọc và sửa file.
- Xử lý dữ liệu lớn và tính toán toán học phức tạp.
- Làm nguyên mẫu (prototype) nhanh, hoặc viết phần mềm hoàn chỉnh để đưa vào sản phẩm.

## Why Python? — Vì sao chọn Python?

- Chạy trên nhiều nền tảng: Windows, macOS, Linux, Raspberry Pi…
- Cú pháp đơn giản, gần với tiếng Anh.
- Viết được chương trình với **ít dòng code hơn** nhiều ngôn ngữ khác.
- Là ngôn ngữ *thông dịch (interpreted)*: code chạy ngay khi viết xong, nên làm thử nghiệm rất nhanh.
- Hỗ trợ nhiều phong cách: lập trình thủ tục, hướng đối tượng, hoặc hàm.

## Python Syntax compared to other languages

- Python được thiết kế để **dễ đọc**, có nhiều điểm giống tiếng Anh và chịu ảnh hưởng từ toán học.
- Python dùng **xuống dòng** để kết thúc một lệnh, không cần dấu `;` như nhiều ngôn ngữ khác.
- Python dùng **thụt lề (indentation)** để xác định phạm vi của vòng lặp, hàm, class… Các ngôn ngữ khác thường dùng dấu ngoặc nhọn `{ }`.
