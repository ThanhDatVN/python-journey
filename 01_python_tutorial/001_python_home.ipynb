## ➕ Mở rộng: Phím tắt Jupyter cần nhớ

| Phím | Tác dụng |
|---|---|
| `Shift + Enter` | Chạy cell hiện tại và chuyển xuống cell kế tiếp |
| `Ctrl + Enter` | Chạy cell hiện tại, con trỏ đứng yên |
| `Esc` rồi `A` / `B` | Thêm cell mới phía trên / phía dưới |
| `Esc` rồi `D` `D` | Xoá cell đang chọn |
| `Esc` rồi `M` / `Y` | Chuyển cell sang Markdown / Code |
| `Ctrl + /` | Comment hoặc bỏ comment dòng code |

> 💡 Nếu kết quả trông "lạ", hãy chọn menu **Kernel → Restart Kernel and Run All Cells** để chạy lại toàn bộ notebook từ đầu.
