# Python - Copy Lists

## Copy a List — Sao chép list

Bạn **không thể** sao chép list chỉ bằng `list2 = list1`. Khi đó `list2` chỉ là một **tham chiếu** tới `list1` — thay đổi `list1` thì `list2` cũng thay đổi theo:

````python
list1 = ["apple", "banana", "cherry"]
list2 = list1
list1.append("orange")
print(list2)
````

Có nhiều cách tạo **bản sao thực sự**.

## Use the copy() Method — Dùng copy()

````python
thislist = ["apple", "banana", "cherry"]
mylist = thislist.copy()
print(mylist)
````

## Use the list() Method — Dùng list()

````python
thislist = ["apple", "banana", "cherry"]
mylist = list(thislist)
print(mylist)
````

## Use the slice Operator — Dùng toán tử slice

Dùng `[:]` (cắt toàn bộ) cũng tạo ra bản sao:

````python
thislist = ["apple", "banana", "cherry"]
mylist = thislist[:]
print(mylist)
````

Kiểm chứng: thay đổi bản gốc không ảnh hưởng bản sao:

````python
original = [1, 2, 3]
copied = original.copy()
original.append(4)

print(original)
print(copied)
````

## ➕ Mở rộng: Shallow copy vs Deep copy

Ba cách trên đều là **sao chép nông**: nếu list chứa list con, các list con vẫn **dùng chung**:

````python
matrix = [[1, 2], [3, 4]]
shallow = matrix.copy()
matrix[0].append(99)
print(shallow)
````

Muốn sao chép hoàn toàn độc lập, dùng `copy.deepcopy()`:

````python
import copy

matrix = [[1, 2], [3, 4]]
deep = copy.deepcopy(matrix)
matrix[0].append(99)
print(deep)
````

> 🧪 **Try it Yourself** — Thay copy() bằng phép gán = rồi chạy lại — kết quả khác thế nào?

````python
a = ["x", "y"]
b = a.copy()
a.append("z")
print("a =", a)
print("b =", b)
````

---

## 🧠 Ghi nhớ nhanh

> - ⚠️ `b = a` **không** sao chép — `b` chỉ là tên khác của cùng một list.
> - 3 cách sao chép: `a.copy()` · `list(a)` · `a[:]`.
> - List lồng nhau → dùng `copy.deepcopy(a)` để sao chép hoàn toàn.
