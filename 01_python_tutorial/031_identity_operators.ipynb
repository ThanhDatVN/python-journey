## Identity Operators — Toán tử định danh

Toán tử định danh **không** so sánh hai giá trị có bằng nhau hay không, mà kiểm tra chúng có phải **cùng một object** (cùng vị trí trong bộ nhớ) hay không:

| Toán tử | Mô tả | Ví dụ |
|---|---|---|
| `is` | `True` nếu hai biến trỏ tới cùng một object | `x is y` |
| `is not` | `True` nếu hai biến trỏ tới hai object khác nhau | `x is not y` |

## Examples

```python
x = ["apple", "banana"]
y = ["apple", "banana"]
z = x

print(x is z)
print(x is y)
print(x == y)
```

- `x is z` → `True` vì `z` chính là `x` (cùng object).
- `x is y` → `False` vì `y` là một list khác, dù có cùng nội dung.
- `x == y` → `True` vì nội dung bằng nhau.

```python
x = ["apple", "banana"]
y = ["apple", "banana"]

print(x is not y)
```

## Difference Between is and == — Phân biệt `is` và `==`

- `is` kiểm tra hai biến có trỏ tới **cùng một object** không.
- `==` kiểm tra hai biến có **giá trị bằng nhau** không.

Dùng hàm `id()` để xem "địa chỉ" (định danh) của object:

```python
x = [1, 2, 3]
y = [1, 2, 3]
z = x

print(id(x), id(y), id(z))
```

Khi thay đổi `x`, `z` cũng thay đổi theo — vì chúng là **một**:

```python
x = [1, 2, 3]
z = x
x.append(4)
print(z)
```

## ➕ Mở rộng: Khi nào dùng `is`?

Quy ước phổ biến nhất: dùng `is` / `is not` để so sánh với `None`:

```python
result = None

if result is None:
    print("Chưa có kết quả")
```

> 🧪 **Try it Yourself** — Thử dùng copy() để tạo bản sao — kết quả của is và == sẽ ra sao?

```python
a = [1, 2, 3]
b = a.copy()

print(a == b)
print(a is b)
```

---

## 🧠 Ghi nhớ nhanh

> - `is` / `is not` → so sánh **cùng object** (cùng vùng nhớ, xem bằng `id()`).
> - `==` → so sánh **giá trị**.
> - Hai list cùng nội dung: `==` là `True`, `is` là `False`.
> - Quy ước: so sánh với `None` bằng `is None`.
