## Strings — Chuỗi

Chuỗi (string) trong Python được đặt trong **nháy đơn** hoặc **nháy kép**. `'hello'` và `"hello"` là như nhau.

Bạn hiển thị chuỗi bằng hàm `print()`:

````python
print("Hello")
print('Hello')
````

## Quotes Inside Quotes — Dấu nháy bên trong chuỗi

Có thể dùng dấu nháy bên trong chuỗi, miễn là nó **khác** loại nháy bao quanh chuỗi:

````python
print("It's alright")
print("He is called 'Johnny'")
print('He is called "Johnny"')
````

## Assign String to a Variable — Gán chuỗi cho biến

````python
a = "Hello"
print(a)
````

## Multiline Strings — Chuỗi nhiều dòng

Dùng **ba dấu nháy** (kép `"""` hoặc đơn `'''`) để tạo chuỗi nhiều dòng:

````python
a = """Lorem ipsum dolor sit amet,
consectetur adipiscing elit,
sed do eiusmod tempor incididunt
ut labore et dolore magna aliqua."""
print(a)
````

Các ký tự xuống dòng được giữ nguyên đúng vị trí như trong code.

## Strings are Arrays — Chuỗi là một dãy ký tự

Giống nhiều ngôn ngữ khác, chuỗi trong Python là một **dãy các ký tự**. Python không có kiểu "ký tự" riêng — một ký tự chỉ là chuỗi có độ dài 1.

Dùng dấu ngoặc vuông `[]` để truy cập từng ký tự. **Chỉ số (index) bắt đầu từ 0**:

````python
a = "Hello, World!"
print(a[1])
````

## Looping Through a String — Duyệt qua chuỗi

Vì chuỗi là một dãy, ta có thể duyệt từng ký tự bằng vòng lặp `for`:

````python
for x in "banana":
    print(x)
````

Bạn sẽ học thêm về vòng lặp `for` trong bài *Python For Loops*.

## String Length — Độ dài chuỗi

Hàm `len()` trả về độ dài của chuỗi:

````python
a = "Hello, World!"
print(len(a))
````

## Check String — Kiểm tra chuỗi con

Dùng từ khoá `in` để kiểm tra một cụm từ có nằm trong chuỗi không:

````python
txt = "The best things in life are free!"
print("free" in txt)
````

Kết hợp với câu lệnh `if`:

````python
txt = "The best things in life are free!"
if "free" in txt:
    print("Yes, 'free' is present.")
````

## Check if NOT — Kiểm tra KHÔNG chứa

Dùng `not in` để kiểm tra một cụm từ **không** có trong chuỗi:

````python
txt = "The best things in life are free!"
print("expensive" not in txt)
````

````python
txt = "The best things in life are free!"
if "expensive" not in txt:
    print("No, 'expensive' is NOT present.")
````

> 🧪 **Try it Yourself** — Thay đổi câu và từ cần tìm, rồi kiểm tra kết quả.

````python
sentence = "Tôi đang học Python trên Jupyter Notebook"
word = "Python"

print(len(sentence))
print(word in sentence)
````

## 🧠 Ghi nhớ nhanh

> - Chuỗi đặt trong `'...'` hoặc `"..."`; nhiều dòng dùng `"""..."""`.
> - Chuỗi là **dãy ký tự**: `a[0]` là ký tự đầu tiên (chỉ số từ **0**).
> - `len(a)` → độ dài; duyệt bằng `for ch in a`.
> - `"x" in a` / `"x" not in a` → kiểm tra chuỗi con.
