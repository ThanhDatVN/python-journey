## Python Loops — Vòng lặp

Python có hai lệnh lặp cơ bản:

- vòng lặp `while`
- vòng lặp `for`

## The while Loop — Vòng lặp while

Vòng lặp `while` thực thi một khối lệnh **chừng nào điều kiện còn đúng**.

In `i` khi `i` còn nhỏ hơn 6:

````python
i = 1
while i < 6:
    print(i)
    i += 1
````

> ⚠️ **Nhớ tăng `i`**, nếu không vòng lặp sẽ chạy **mãi mãi** (vòng lặp vô hạn).
>
> Vòng lặp `while` cần một biến điều khiển sẵn sàng từ trước — ở đây là biến đếm `i`, khởi tạo bằng 1.

## The break Statement — Câu lệnh break

`break` **dừng vòng lặp ngay lập tức**, kể cả khi điều kiện vẫn đúng.

Thoát khỏi vòng lặp khi `i` bằng 3:

````python
i = 1
while i < 6:
    print(i)
    if i == 3:
        break
    i += 1
````

## The continue Statement — Câu lệnh continue

`continue` **bỏ qua phần còn lại của vòng hiện tại** và chuyển sang vòng tiếp theo.

Bỏ qua khi `i` bằng 3:

````python
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue
    print(i)
````

## The else Statement — Câu lệnh else

Với `else`, bạn có thể chạy một khối lệnh **một lần** khi điều kiện không còn đúng nữa:

````python
i = 1
while i < 6:
    print(i)
    i += 1
else:
    print("i is no longer less than 6")
````

> 📌 Khối `else` **không** chạy nếu vòng lặp bị dừng bằng `break`.

## ➕ Mở rộng: Vòng lặp "chờ điều kiện"

`while` phù hợp khi **không biết trước số vòng lặp**, ví dụ: gấp đôi số tiền tiết kiệm đến khi vượt mục tiêu:

````python
money = 1_000_000
target = 10_000_000
months = 0

while money < target:
    money *= 1.5
    months += 1

print(f"Cần {months} tháng, số tiền: {money:,.0f}")
````

> 🧪 **Try it Yourself** — Đếm ngược từ 10 về 1 rồi in "Phóng tàu! 🚀".

````python
n = 10
while n > 0:
    print(n, end=" ")
    n -= 1
print("Phóng tàu! 🚀")
````

---

## 🧠 Ghi nhớ nhanh

> - `while điều_kiện:` → lặp **chừng nào** điều kiện còn `True`.
> - ⚠️ Nhớ **cập nhật biến điều khiển**, nếu không sẽ lặp vô hạn.
> - `break` thoát vòng lặp · `continue` bỏ qua phần còn lại, sang vòng mới.
> - `else` chạy khi điều kiện thành `False` (không chạy nếu `break`).
