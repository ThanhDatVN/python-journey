## What is a Module? — Module là gì?

Có thể hiểu module giống như một **thư viện code**: một file chứa các hàm, biến… mà bạn muốn dùng lại trong chương trình.

## Create a Module — Tạo module

Để tạo module, chỉ cần lưu code vào một file có đuôi `.py`.

Trong notebook, ta dùng `%%writefile` để tạo file `mymodule.py`:

````python
%%writefile mymodule.py
def greeting(name):
    print("Hello, " + name)

person1 = {
    "name": "John",
    "age": 36,
    "country": "Norway"
}
````

## Use a Module — Dùng module

Dùng module vừa tạo bằng câu lệnh `import`:

````python
import mymodule

mymodule.greeting("Jonathan")
````

> 📌 Khi gọi hàm trong module, dùng cú pháp `ten_module.ten_ham`.

## Variables in Module — Biến trong module

Module có thể chứa hàm, và cả **biến** mọi kiểu (list, dictionary, object…):

````python
import mymodule

a = mymodule.person1["age"]
print(a)
````

## Naming a Module — Đặt tên module

Tên module là tên file (không có đuôi `.py`). Đặt tên tuỳ ý, nhưng **không trùng** tên module có sẵn (ví dụ đừng đặt `math.py`, `random.py`).

## Re-naming a Module — Đặt bí danh cho module

Dùng từ khoá `as` để tạo **bí danh (alias)**:

````python
import mymodule as mx

a = mx.person1["age"]
print(a)
````

## Built-in Modules — Module có sẵn

Python có rất nhiều module có sẵn, import lúc nào cũng được:

````python
import platform

x = platform.system()
print(x)
````

## Using the dir() Function — Hàm dir()

Hàm `dir()` liệt kê tất cả tên hàm/biến có trong một module:

````python
import platform

x = dir(platform)
print(x[:15], "...")
````

> 📌 `dir()` dùng được với mọi module, kể cả module bạn tự tạo.

## Import From Module — Import một phần

Chỉ import một phần của module bằng `from`:

````python
from mymodule import person1

print(person1["age"])
````

> ⚠️ Khi import bằng `from`, dùng tên trực tiếp (`person1["age"]`), **không** viết `mymodule.person1["age"]`.

## ➕ Mở rộng: Một số lưu ý

- Sửa file module xong, cần **Restart Kernel** để notebook nạp lại bản mới (hoặc dùng `importlib.reload`).
- `from module import *` import mọi thứ — tiện nhưng dễ trùng tên, **không khuyến khích**.
- Khối `if __name__ == "__main__":` chỉ chạy khi file được chạy trực tiếp, không chạy khi bị import.

````python
import importlib
import mymodule

importlib.reload(mymodule)
print(mymodule.__name__)
````

> 🧪 **Try it Yourself** — Dùng module math có sẵn: import với alias m và tính căn bậc hai của 144.

````python
import math as m
print(m.sqrt(144))
````

---

## 🧠 Ghi nhớ nhanh

> - Module = file `.py` chứa code dùng lại được.
> - `import mymodule` → dùng `mymodule.ham()`; `import x as y` → đặt bí danh.
> - `from mymodule import ham` → dùng trực tiếp `ham()`.
> - `dir(module)` liệt kê nội dung; ⚠️ đừng đặt tên file trùng module có sẵn.
