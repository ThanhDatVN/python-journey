## Join Sets — Kết hợp các set

Có nhiều cách để kết hợp hai hoặc nhiều set, tương ứng với các **phép toán tập hợp** trong toán học:

| Phép toán | Phương thức | Toán tử | Kết quả |
|---|---|---|---|
| Hợp (union) | `union()`, `update()` | `\|` | Tất cả phần tử của cả hai set |
| Giao (intersection) | `intersection()`, `intersection_update()` | `&` | Chỉ phần tử có ở **cả hai** |
| Hiệu (difference) | `difference()`, `difference_update()` | `-` | Phần tử có ở set thứ nhất mà **không** có ở set thứ hai |
| Hiệu đối xứng | `symmetric_difference()`, `symmetric_difference_update()` | `^` | Phần tử **không có ở cả hai** |

## Union — Hợp

`union()` trả về set mới chứa tất cả phần tử của cả hai set:

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

set3 = set1.union(set2)
print(set3)
```

Dùng toán tử `|` cho kết quả tương tự:

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

set3 = set1 | set2
print(set3)
```

### Join Multiple Sets — Nối nhiều set

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"John", "Elena"}
set4 = {"apple", "bananas", "cherry"}

myset = set1.union(set2, set3, set4)
print(myset)

myset = set1 | set2 | set3 | set4
print(myset)
```

### Join a Set and a Tuple

`union()` nhận cả các kiểu khác như list, tuple; còn toán tử `|` **chỉ** dùng được giữa các set:

```python
x = {"a", "b", "c"}
y = (1, 2, 3)

z = x.union(y)
print(z)
```

### update()

`update()` thêm phần tử vào **set hiện tại** thay vì tạo set mới:

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

set1.update(set2)
print(set1)
```

> 📌 Cả `union()` và `update()` đều bỏ qua phần tử trùng lặp.

## Intersection — Giao

Giữ lại **chỉ các phần tử trùng nhau**:

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.intersection(set2)
print(set3)

print(set1 & set2)
```

`intersection_update()` cập nhật trực tiếp set gốc:

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set1.intersection_update(set2)
print(set1)
```

## Difference — Hiệu

Giữ lại phần tử có ở set thứ nhất mà **không** có ở set thứ hai:

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

print(set1.difference(set2))
print(set1 - set2)
```

## Symmetric Differences — Hiệu đối xứng

Giữ lại các phần tử **không** xuất hiện ở cả hai set:

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

print(set1.symmetric_difference(set2))
print(set1 ^ set2)
```

> 🧪 **Try it Yourself** — Tìm học sinh đăng ký CẢ HAI câu lạc bộ, và học sinh CHỈ đăng ký CLB Python.

```python
python_club = {"An", "Bình", "Chi", "Dũng"}
math_club = {"Chi", "Dũng", "Em"}

print("Cả hai:", python_club & math_club)
print("Chỉ Python:", python_club - math_club)
```

## 🧠 Ghi nhớ nhanh

> - Hợp `|` `union()` · Giao `&` `intersection()` · Hiệu `-` `difference()` · Hiệu đối xứng `^` `symmetric_difference()`.
> - Phương thức thường trả về **set mới**; bản `_update()` (và `update()`) sửa **set gốc**.
> - `union()` nhận cả list/tuple; toán tử `|` chỉ dùng giữa các set.
