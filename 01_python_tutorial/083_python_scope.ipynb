## Scope — Phạm vi

Một biến chỉ dùng được **bên trong vùng nơi nó được tạo ra**. Vùng đó gọi là **phạm vi (scope)**.

## Local Scope — Phạm vi cục bộ

Biến tạo **bên trong** một hàm thuộc phạm vi cục bộ của hàm đó, và chỉ dùng được trong hàm:

````python
def myfunc():
    x = 300
    print(x)

myfunc()
````

````python
def myfunc():
    x = 300

myfunc()
print(x)
````

### Function Inside Function — Hàm lồng trong hàm

Biến cục bộ của hàm ngoài **dùng được** trong hàm con bên trong nó:

````python
def myfunc():
    x = 300
    def myinnerfunc():
        print(x)
    myinnerfunc()

myfunc()
````

## Global Scope — Phạm vi toàn cục

Biến tạo trong phần chính của chương trình là **biến toàn cục**, dùng được ở mọi phạm vi:

````python
x = 300

def myfunc():
    print(x)

myfunc()
print(x)
````

### Naming Variables — Trùng tên biến

Nếu dùng cùng một tên biến ở trong và ngoài hàm, Python coi chúng là **hai biến riêng biệt**:

````python
x = 300

def myfunc():
    x = 200
    print(x)

myfunc()
print(x)
````

## Global Keyword — Từ khoá global

Muốn tạo hoặc **sửa** biến toàn cục từ bên trong hàm, dùng từ khoá `global`:

````python
def myfunc():
    global x
    x = 300

myfunc()
print(x)
````

````python
x = 300

def myfunc():
    global x
    x = 200

myfunc()
print(x)
````

## Nonlocal Keyword — Từ khoá nonlocal

`nonlocal` dùng trong **hàm lồng nhau**: nó cho phép hàm con sửa biến của **hàm bao ngoài** (không phải biến toàn cục):

````python
def myfunc1():
    x = "Jane"
    def myfunc2():
        nonlocal x
        x = "hello"
    myfunc2()
    return x

print(myfunc1())
````

## The LEGB Rule — Quy tắc LEGB

Khi gặp một tên biến, Python tìm theo thứ tự:

1. **L**ocal — bên trong hàm hiện tại
2. **E**nclosing — trong các hàm bao ngoài
3. **G**lobal — ở cấp module
4. **B**uilt-in — các tên có sẵn của Python (`print`, `len`…)

````python
x = "global"

def outer():
    x = "enclosing"
    def inner():
        x = "local"
        print("Inner:", x)
    inner()
    print("Outer:", x)

outer()
print("Global:", x)
````

> 🧪 **Try it Yourself** — Xoá dòng nonlocal count rồi chạy lại — chuyện gì xảy ra? Vì sao?

````python
def make_counter():
    count = 0
    def increase():
        nonlocal count
        count += 1
        return count
    return increase

counter = make_counter()
print(counter(), counter(), counter())
````

---

## 🧠 Ghi nhớ nhanh

> - Biến **trong hàm** = local (chỉ dùng trong hàm) · biến **ngoài hàm** = global.
> - Cùng tên trong/ngoài hàm → **hai biến khác nhau**.
> - `global x` → sửa biến toàn cục · `nonlocal x` → sửa biến của hàm **bao ngoài**.
> - Thứ tự tìm tên: **L → E → G → B** (Local, Enclosing, Global, Built-in).
