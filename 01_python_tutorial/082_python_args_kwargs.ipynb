## *args and **kwargs

Mặc định, hàm phải được gọi với đúng số đối số. Nhưng đôi khi bạn **không biết trước** sẽ có bao nhiêu đối số. Khi đó, dùng `*args` và `**kwargs`.

## Arbitrary Arguments - *args — Số lượng đối số tuỳ ý

Thêm dấu `*` trước tên tham số. Hàm sẽ nhận một **tuple** chứa tất cả đối số:

````python
def my_function(*kids):
    print("The youngest child is " + kids[2])

my_function("Emil", "Tobias", "Linus")
````

## What is *args? — *args là gì?

`*args` gom mọi đối số vị trí thành một **tuple**. Tên `args` chỉ là quy ước — dấu `*` mới là phần quan trọng:

````python
def my_function(*args):
    print("Type:", type(args))
    print("First argument:", args[0])
    print("Second argument:", args[1])
    print("All arguments:", args)

my_function("Emil", "Tobias", "Linus")
````

## Using *args with Regular Arguments — Kết hợp với tham số thường

Tham số thường phải đứng **trước** `*args`:

````python
def my_function(greeting, *names):
    for name in names:
        print(greeting, name)

my_function("Hello", "Emil", "Tobias", "Linus")
````

## Practical Example with *args — Ví dụ thực tế

Hàm tính tổng với số lượng đối số bất kỳ:

````python
def my_function(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total

print(my_function(1, 2, 3))
print(my_function(10, 20, 30, 40))
print(my_function(5))
````

Tìm giá trị lớn nhất:

````python
def my_function(*numbers):
    if len(numbers) == 0:
        return None
    max_num = numbers[0]
    for num in numbers:
        if num > max_num:
            max_num = num
    return max_num

print(my_function(3, 7, 2, 9, 1))
````

## Arbitrary Keyword Arguments - **kwargs — Đối số từ khoá tuỳ ý

Thêm hai dấu `**` trước tên tham số. Hàm sẽ nhận một **dictionary**:

````python
def my_function(**kid):
    print("His last name is " + kid["lname"])

my_function(fname="Tobias", lname="Refsnes")
````

## What is **kwargs? — **kwargs là gì?

`**kwargs` gom mọi đối số từ khoá thành một **dictionary**, với tên đối số là key:

````python
def my_function(**myvar):
    print("Type:", type(myvar))
    print("Name:", myvar["name"])
    print("Age:", myvar["age"])
    print("All data:", myvar)

my_function(name="Tobias", age=30, city="Bergen")
````

## Using **kwargs with Regular Arguments — Kết hợp với tham số thường

````python
def my_function(username, **details):
    print("Username:", username)
    print("Additional details:")
    for key, value in details.items():
        print(" ", key + ":", value)

my_function("emil123", age=25, city="Oslo", hobby="coding")
````

## Combining *args and **kwargs — Kết hợp cả hai

Thứ tự bắt buộc: **tham số thường → `*args` → `**kwargs`**:

````python
def my_function(title, *args, **kwargs):
    print("Title:", title)
    print("Positional arguments:", args)
    print("Keyword arguments:", kwargs)

my_function("User Info", "Emil", "Tobias", age=25, city="Oslo")
````

## Unpacking Arguments — Giải nén đối số

Dấu `*` và `**` cũng dùng được khi **gọi** hàm, để "trải" list/dict thành các đối số riêng lẻ.

Giải nén list với `*`:

````python
def my_function(a, b, c):
    return a + b + c

numbers = [1, 2, 3]
result = my_function(*numbers)   # tương đương my_function(1, 2, 3)
print(result)
````

Giải nén dictionary với `**`:

````python
def my_function(fname, lname):
    print("Hello", fname, lname)

person = {"fname": "Emil", "lname": "Refsnes"}
my_function(**person)            # tương đương my_function(fname="Emil", lname="Refsnes")
````

> 🧪 **Try it Yourself** — Viết hàm make_profile(name, **info) in ra tên và mọi thông tin kèm theo.

````python
def make_profile(name, **info):
    print("Hồ sơ:", name)
    for key, value in info.items():
        print(f"  - {key}: {value}")

make_profile("An", age=20, city="Huế", major="CNTT")
````

---

## 🧠 Ghi nhớ nhanh

> - `*args` → gom đối số **vị trí** thành **tuple**.
> - `**kwargs` → gom đối số **từ khoá** thành **dictionary**.
> - Thứ tự tham số: `thường → *args → **kwargs`.
> - Khi **gọi** hàm: `f(*list)` trải list, `f(**dict)` trải dict thành đối số.
