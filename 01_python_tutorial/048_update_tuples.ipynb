# Python - Update Tuples

> 📖 **Học song song trên W3Schools:** [python_tuples_update.asp](https://www.w3schools.com/python/python_tuples_update.asp)
>
> 🎯 **Trong bài này:** Change Tuple Values · Add Items · Remove Items

Tuple **không thay đổi được** — bạn không thể sửa, thêm hay xoá phần tử sau khi tạo.

Nhưng vẫn có vài "mẹo" để làm việc đó.

## Change Tuple Values — Thay đổi giá trị

Thử gán trực tiếp sẽ bị lỗi:

```python
x = ("apple", "banana", "cherry")
x[1] = "kiwi"
```

Mẹo: chuyển tuple thành **list**, sửa list, rồi chuyển ngược lại thành **tuple**:

```python
x = ("apple", "banana", "cherry")
y = list(x)
y[1] = "kiwi"
x = tuple(y)

print(x)
```

## Add Items — Thêm phần tử

Tuple không có phương thức `append()`, nhưng có hai cách khác:

**1. Chuyển sang list**, thêm phần tử, rồi chuyển lại:

```python
thistuple = ("apple", "banana", "cherry")
y = list(thistuple)
y.append("orange")
thistuple = tuple(y)
print(thistuple)
```

**2. Cộng tuple với tuple**: tạo tuple mới chứa phần tử cần thêm, rồi nối vào:

```python
thistuple = ("apple", "banana", "cherry")
y = ("orange",)
thistuple += y

print(thistuple)
```

> 📌 Khi tạo tuple có một phần tử, nhớ **dấu phẩy** sau phần tử đó.

## Remove Items — Xoá phần tử

Không thể xoá trực tiếp, nhưng có thể dùng mẹo chuyển sang list:

```python
thistuple = ("apple", "banana", "cherry")
y = list(thistuple)
y.remove("apple")
thistuple = tuple(y)
print(thistuple)
```

Hoặc xoá **toàn bộ** tuple bằng `del`:

```python
thistuple = ("apple", "banana", "cherry")
del thistuple
print(thistuple)  # lỗi: thistuple không còn tồn tại
```

> 💡 Thực chất, các "mẹo" trên không sửa tuple cũ mà **tạo tuple mới** rồi gán lại cho biến. Hàm `id()` sẽ cho thấy điều đó:

```python
t = (1, 2)
print(id(t))
t += (3,)
print(id(t))
```

> 🧪 **Try it Yourself** — Thêm "Python" vào cuối tuple languages bằng phép cộng tuple.

```python
languages = ("C", "Java")
languages += ("Python",)
print(languages)
```

---

## 🧠 Ghi nhớ nhanh

> - Tuple **bất biến**: `t[0] = x` → `TypeError`.
> - Mẹo sửa: `list(t)` → sửa → `tuple(...)`.
> - Thêm phần tử: `t += ("new",)`.
> - Thực chất các mẹo đều **tạo tuple mới**.
