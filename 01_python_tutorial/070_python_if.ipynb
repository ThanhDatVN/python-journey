## Python Conditions and If statements — Điều kiện và câu lệnh if

Python hỗ trợ các điều kiện logic quen thuộc từ toán học:

| Điều kiện | Cú pháp |
|---|---|
| Bằng | `a == b` |
| Khác | `a != b` |
| Nhỏ hơn | `a < b` |
| Nhỏ hơn hoặc bằng | `a <= b` |
| Lớn hơn | `a > b` |
| Lớn hơn hoặc bằng | `a >= b` |

Các điều kiện này thường dùng trong **câu lệnh `if`** và **vòng lặp**. Câu lệnh `if` được viết bằng từ khoá `if`:

````python
a = 33
b = 200
if b > a:
    print("b is greater than a")
````

Ở ví dụ trên, `b > a` là `True` (200 > 33), nên Python in ra màn hình.

## How If Statements Work — If hoạt động thế nào?

Câu lệnh `if` đánh giá một **biểu thức boolean**. Nếu kết quả là `True`, khối lệnh bên trong được thực thi; nếu `False`, khối lệnh bị bỏ qua:

````python
number = 15
if number > 0:
    print("The number is positive")
````

## Indentation — Thụt lề

Python dùng **thụt lề** để xác định phạm vi của khối lệnh. Ngôn ngữ khác thường dùng dấu ngoặc nhọn.

Câu lệnh `if` không có thụt lề sẽ báo lỗi:

````python
a = 33
b = 200
if b > a:
print("b is greater than a")
````

## Multiple Statements in If Block — Nhiều lệnh trong khối if

Khối `if` có thể chứa nhiều câu lệnh, miễn là chúng **thụt lề bằng nhau**:

````python
age = 20
if age >= 18:
    print("You are an adult")
    print("You can vote")
    print("You have full legal rights")
````

## Using Variables in Conditions — Dùng biến làm điều kiện

Biến boolean có thể dùng trực tiếp làm điều kiện:

````python
is_logged_in = True
if is_logged_in:
    print("Welcome back!")
````

Python còn tự đánh giá các giá trị khác: `0`, chuỗi rỗng `""`, `None`, và tập hợp rỗng là **False**; các giá trị còn lại là **True**:

````python
name = ""
cart = ["book"]

if name:
    print("Hello", name)
if cart:
    print("Cart has", len(cart), "item(s)")
````

> 🧪 **Try it Yourself** — Đổi giá trị temperature để thấy khối lệnh được/không được thực thi.

````python
temperature = 35
if temperature > 30:
    print("Trời nóng, nhớ uống đủ nước!")
````

---

## 🧠 Ghi nhớ nhanh

> - `if điều_kiện:` + khối lệnh **thụt lề** → chạy khi điều kiện `True`.
> - ⚠️ Quên dấu `:` hoặc quên thụt lề → lỗi cú pháp.
> - Giá trị rỗng/`0`/`None` được coi là `False` → viết gọn `if cart:` thay vì `if len(cart) > 0:`.
