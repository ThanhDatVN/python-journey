# Python - Change Dictionary Items

> 📖 **Học song song trên W3Schools:** [python_dictionaries_change.asp](https://www.w3schools.com/python/python_dictionaries_change.asp)
>
> 🎯 **Trong bài này:** Change Values · Update Dictionary · ➕ Mở rộng: Toán tử `|` và `|=` (Python 3.9+)

## Change Values — Thay đổi giá trị

Thay đổi giá trị của một phần tử bằng cách gán giá trị mới vào key của nó:

````python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict["year"] = 2018
print(thisdict)
````

## Update Dictionary — Cập nhật dictionary

Phương thức `update()` cập nhật dictionary bằng các phần tử từ tham số truyền vào. Tham số phải là một **dictionary** hoặc một **iterable chứa các cặp key-value**:

````python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict.update({"year": 2020})
print(thisdict)
````

`update()` có thể đổi nhiều giá trị cùng lúc:

````python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict.update({"model": "Ranger", "year": 2024})
print(thisdict)
````

## ➕ Mở rộng: Toán tử `|` và `|=` (Python 3.9+)

````python
car = {"brand": "Ford", "year": 1964}
changes = {"year": 2024, "color": "red"}

print(car | changes)   # tạo dict mới đã gộp
car |= changes         # cập nhật tại chỗ
print(car)
````

> 🧪 **Try it Yourself** — Tăng tuổi của người dùng thêm 1 và đổi thành phố bằng update().

````python
user = {"name": "An", "age": 20, "city": "Huế"}
user["age"] = user["age"] + 1
user.update({"city": "Hà Nội"})
print(user)
````

---

## 🧠 Ghi nhớ nhanh

> - Sửa giá trị: `d["key"] = giá_trị_mới`.
> - Sửa nhiều giá trị: `d.update({"k1": v1, "k2": v2})`.
> - Python 3.9+: `d1 | d2` gộp thành dict mới.
