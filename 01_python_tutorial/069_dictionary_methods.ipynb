## Dictionary Methods — Phương thức của dictionary

Python có các phương thức có sẵn sau để dùng với dictionary:

| Phương thức | Mô tả |
|---|---|
| `clear()` | Xoá tất cả phần tử |
| `copy()` | Trả về bản sao |
| `fromkeys()` | Tạo dictionary từ danh sách key và một giá trị chung |
| `get()` | Trả về value của key chỉ định |
| `items()` | Trả về view các tuple `(key, value)` |
| `keys()` | Trả về view các key |
| `pop()` | Xoá phần tử có key chỉ định |
| `popitem()` | Xoá cặp key-value được thêm vào cuối cùng |
| `setdefault()` | Trả về value của key; nếu key chưa có thì thêm key với value chỉ định |
| `update()` | Cập nhật dictionary với các cặp key-value chỉ định |
| `values()` | Trả về view các value |

Chi tiết: [Dictionary Methods Reference](../10_reference/dictionary_methods.md).

## ➕ Mở rộng: fromkeys()

```python
keys = ["math", "physics", "chemistry"]
scores = dict.fromkeys(keys, 0)
print(scores)
```

## ➕ Mở rộng: Ví dụ tổng hợp

```python
inventory = {"apple": 5, "banana": 0}

inventory.update({"cherry": 12})
inventory.setdefault("kiwi", 3)
sold_out = [k for k, v in inventory.items() if v == 0]
for k in sold_out:
    inventory.pop(k)

print(inventory)
print(sorted(inventory, key=inventory.get, reverse=True))   # sắp key theo value
```

> 🧪 **Try it Yourself** — Tạo dictionary chấm công: mọi nhân viên bắt đầu với 0 ngày (fromkeys), rồi cộng ngày cho An.

```python
staff = ["An", "Bình", "Chi"]
attendance = dict.fromkeys(staff, 0)
attendance["An"] += 5
print(attendance)
```
