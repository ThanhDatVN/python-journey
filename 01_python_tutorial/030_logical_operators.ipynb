## Logical Operators — Toán tử logic

Toán tử logic dùng để kết hợp các câu lệnh điều kiện:

| Toán tử | Mô tả | Ví dụ |
|---|---|---|
| `and` | `True` nếu **cả hai** vế đều đúng | `x < 5 and x < 10` |
| `or` | `True` nếu **ít nhất một** vế đúng | `x < 5 or x < 4` |
| `not` | Đảo ngược kết quả | `not(x < 5 and x < 10)` |

### The and Operator

````python
x = 5
print(x > 3 and x < 10)
````

### The or Operator

````python
x = 5
print(x < 5 or x > 3)
````

### The not Operator

````python
x = 5
print(not(x > 3 and x < 10))
````

## ➕ Mở rộng: Truth Table — Bảng chân trị

| A | B | `A and B` | `A or B` | `not A` |
|---|---|---|---|---|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

````python
for a in (True, False):
    for b in (True, False):
        print(a, b, "| and:", a and b, "| or:", a or b)
````

## ➕ Mở rộng: Short-circuit — Tính toán rút gọn

Python dừng đánh giá ngay khi đã biết kết quả:

- Với `and`: nếu vế trái là `False` thì không cần xét vế phải.
- Với `or`: nếu vế trái là `True` thì không cần xét vế phải.

Nhờ vậy, đoạn code sau không bị lỗi chia cho 0:

````python
x = 0
print(x != 0 and 10 / x > 1)
````

> 💡 `and` / `or` thực chất trả về **một trong hai giá trị** chứ không nhất thiết là `True`/`False`:

````python
print("" or "mặc định")
print("Python" and "Jupyter")
````

> 🧪 **Try it Yourself** — Kiểm tra điều kiện được vào cửa: tuổi từ 18 trở lên VÀ có vé.

````python
age = 20
has_ticket = True

print(age >= 18 and has_ticket)
````

## 🧠 Ghi nhớ nhanh

> - `and`: cả hai đúng · `or`: ít nhất một đúng · `not`: đảo ngược.
> - **Short-circuit**: `and` dừng khi gặp `False`, `or` dừng khi gặp `True`.
> - Mẹo nhớ: `and` = "**và** cả hai", `or` = "**hoặc** một trong hai".
