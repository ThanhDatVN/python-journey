## User Input — Nhận dữ liệu từ người dùng

Python cho phép nhận dữ liệu người dùng nhập vào bằng hàm **`input()`**.

Khi gặp `input()`, chương trình **dừng lại chờ** người dùng nhập và nhấn Enter.

> 📌 Trong Jupyter, một ô nhập liệu sẽ hiện ra ngay bên dưới cell. Kết quả lưu sẵn trong notebook này là ví dụ với dữ liệu mẫu.

```python
print("Enter your name:")
name = input()
print(f"Hello {name}")
```

## Using prompt — Dùng lời nhắc

Thay vì `print()` riêng, bạn có thể truyền **lời nhắc (prompt)** vào `input()`. Lời nhắc hiện trên cùng dòng với chỗ nhập:

```python
name = input("Enter your name:")
print(f"Hello {name}")
```

## Multiple Inputs — Nhập nhiều lần

Có thể gọi `input()` nhiều lần để thu thập nhiều thông tin:

```python
name = input("Enter your name: ")
print(f"Hello {name}")
fav1 = input("What is your favorite city: ")
fav2 = input("What is your favorite hobby: ")
print(f"So {name}, you like {fav1} and {fav2}.")
```

## Input Number — Nhập số

`input()` **luôn trả về chuỗi (str)**, kể cả khi bạn nhập số:

```python
x = input("Enter a number: ")
print(type(x))
```

Muốn tính toán, cần chuyển đổi kiểu bằng `int()` hoặc `float()`:

```python
x = input("Enter a number: ")
y = float(x)
print(f"The square root of {y} is {y ** 0.5}")
```

> ⚠️ Nếu người dùng nhập chữ, `float("abc")` sẽ báo lỗi `ValueError`.

## Validate Input — Kiểm tra dữ liệu nhập

Nên kiểm tra dữ liệu người dùng nhập. Dùng vòng lặp `while` kết hợp `try...except` để hỏi lại cho đến khi nhập đúng:

```python
while True:
    x = input("Enter a number: ")
    try:
        x = float(x)
        break
    except ValueError:
        print("Wrong input, please try again.")

print("Thank you!", x)
```

> 🧪 **Try it Yourself** — Chạy cell và nhập năm sinh của bạn; chương trình sẽ tính tuổi.

```python
birth_year = int(input("Năm sinh của bạn: "))
print("Năm 2026 bạn", 2026 - birth_year, "tuổi")
```

---

## 🧠 Ghi nhớ nhanh

> - `input("lời nhắc")` → dừng chờ người dùng nhập, **luôn trả về str**.
> - Cần số → `int(input(...))` hoặc `float(input(...))`.
> - ⚠️ Nhập sai kiểu → `ValueError`; kiểm tra bằng `while True` + `try/except` + `break`.
