## List

List dùng để lưu **nhiều phần tử** trong một biến duy nhất.

List là một trong 4 kiểu dữ liệu tập hợp có sẵn của Python. Ba kiểu còn lại là **Tuple**, **Set** và **Dictionary** — mỗi kiểu có đặc điểm và cách dùng riêng.

List được tạo bằng **dấu ngoặc vuông**:

```python
thislist = ["apple", "banana", "cherry"]
print(thislist)
```

## List Items — Phần tử của list

Phần tử trong list có ba đặc điểm: **có thứ tự (ordered)**, **thay đổi được (changeable)** và **cho phép trùng lặp (allow duplicates)**.

Phần tử được đánh chỉ số: phần tử đầu tiên có chỉ số `[0]`, phần tử thứ hai có chỉ số `[1]`…

### Ordered — Có thứ tự

Các phần tử có thứ tự xác định và thứ tự đó **không tự thay đổi**. Thêm phần tử mới thì nó sẽ nằm ở cuối list.

> Một số phương thức có thể thay đổi thứ tự, nhưng nhìn chung thứ tự là cố định.

### Changeable — Thay đổi được

Có thể sửa, thêm và xoá phần tử sau khi list đã được tạo.

### Allow Duplicates — Cho phép trùng lặp

Vì có chỉ số, list có thể chứa các phần tử giống nhau:

```python
thislist = ["apple", "banana", "cherry", "apple", "cherry"]
print(thislist)
```

## List Length — Độ dài list

Dùng `len()` để biết list có bao nhiêu phần tử:

```python
thislist = ["apple", "banana", "cherry"]
print(len(thislist))
```

## List Items - Data Types — Kiểu dữ liệu của phần tử

Phần tử có thể thuộc bất kỳ kiểu dữ liệu nào:

```python
list1 = ["apple", "banana", "cherry"]
list2 = [1, 5, 7, 9, 3]
list3 = [True, False, False]

print(list1, list2, list3)
```

Một list cũng có thể chứa **nhiều kiểu khác nhau**:

```python
list1 = ["abc", 34, True, 40, "male"]
print(list1)
```

## type()

Với Python, list là object có kiểu `'list'`:

```python
mylist = ["apple", "banana", "cherry"]
print(type(mylist))
```

## The list() Constructor — Hàm khởi tạo list()

Cũng có thể tạo list bằng hàm `list()`:

```python
thislist = list(("apple", "banana", "cherry"))  # chú ý: hai cặp ngoặc tròn
print(thislist)
```

## Python Collections (Arrays) — So sánh 4 kiểu tập hợp

| Kiểu | Thứ tự | Thay đổi được | Trùng lặp | Ví dụ |
|---|---|---|---|---|
| **List** | Có | Có | Cho phép | `["a", "b"]` |
| **Tuple** | Có | Không | Cho phép | `("a", "b")` |
| **Set** | Không | Không sửa được phần tử (*nhưng thêm/xoá được*) | Không | `{"a", "b"}` |
| **Dictionary** | Có (từ Python 3.7) | Có | Không trùng key | `{"k": "v"}` |

Chọn đúng kiểu tập hợp giúp code rõ ràng, an toàn và hiệu quả hơn.

> 🧪 **Try it Yourself** — Tạo list các môn học yêu thích của bạn rồi in ra số lượng môn.

```python
subjects = ["Toán", "Tin học", "Tiếng Anh"]
print(subjects)
print(len(subjects))
```

## 🧠 Ghi nhớ nhanh

> - List: `[ ]` — **có thứ tự · thay đổi được · cho phép trùng lặp**.
> - Chứa được mọi kiểu dữ liệu, kể cả lẫn lộn.
> - `len()` độ dài · `type()` → `list` · tạo bằng `list(iterable)`.
> - 4 kiểu tập hợp: **List** `[ ]`, **Tuple** `( )`, **Set** `{ }`, **Dict** `{k: v}`.
