## Many Values to Multiple Variables — Nhiều giá trị cho nhiều biến

Python cho phép gán giá trị cho nhiều biến **trên cùng một dòng**:

````python
x, y, z = "Orange", "Banana", "Cherry"

print(x)
print(y)
print(z)
````

> ⚠️ Số biến phải **bằng** số giá trị, nếu không Python sẽ báo lỗi:

````python
x, y = "Orange", "Banana", "Cherry"
````

## One Value to Multiple Variables — Một giá trị cho nhiều biến

Có thể gán **cùng một giá trị** cho nhiều biến trên một dòng:

````python
x = y = z = "Orange"

print(x)
print(y)
print(z)
````

## Unpack a Collection — Giải nén một tập hợp

Nếu có một tập hợp giá trị như list, tuple…, Python cho phép "rút" các giá trị ra các biến. Việc này gọi là **unpacking**:

````python
fruits = ["apple", "banana", "cherry"]
x, y, z = fruits

print(x)
print(y)
print(z)
````

Bạn sẽ học kỹ hơn trong bài *Unpack Tuples*.

## ➕ Mở rộng: Swap — Hoán đổi giá trị

Nhờ gán nhiều biến, việc hoán đổi giá trị hai biến chỉ cần một dòng:

````python
a = 1
b = 2
a, b = b, a

print(a, b)
````

> 🧪 **Try it Yourself** — Hãy gán tên, tuổi và thành phố của bạn cho ba biến trên cùng một dòng.

````python
name, age, city = "An", 20, "Hà Nội"
print(name, age, city)
````

---

## 🧠 Ghi nhớ nhanh

> - Nhiều giá trị → nhiều biến: `x, y, z = 1, 2, 3` (số lượng phải **khớp**).
> - Một giá trị → nhiều biến: `x = y = z = 0`.
> - Unpacking: `a, b, c = my_list`.
> - Mẹo đổi chỗ: `a, b = b, a`.
