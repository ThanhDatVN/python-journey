## Python Functions — Hàm

**Hàm (function)** là một khối code **chỉ chạy khi được gọi**.

Hàm có thể nhận dữ liệu (gọi là **tham số**) và có thể **trả về** dữ liệu.

Hàm giúp bạn **tái sử dụng code**: viết một lần, dùng nhiều lần.

## Creating a Function — Tạo hàm

Trong Python, hàm được định nghĩa bằng từ khoá `def`, theo sau là tên hàm và cặp ngoặc tròn, kết thúc bằng dấu `:`:

````python
def my_function():
    print("Hello from a function")
````

Thân hàm phải được **thụt lề**. Chạy cell trên **không in gì cả** — vì hàm mới chỉ được định nghĩa, chưa được gọi.

## Calling a Function — Gọi hàm

Gọi hàm bằng tên hàm kèm cặp ngoặc tròn:

````python
def my_function():
    print("Hello from a function")

my_function()
````

Có thể gọi một hàm **nhiều lần**:

````python
def my_function():
    print("Hello from a function")

my_function()
my_function()
my_function()
````

## Function Names — Đặt tên hàm

Tên hàm tuân theo quy tắc giống tên biến:

- Bắt đầu bằng chữ cái hoặc `_`, chỉ chứa chữ, số, `_`.
- Phân biệt hoa thường.
- Nên đặt tên **mô tả hành động**, viết kiểu `snake_case`: `calculate_sum()`, `print_report()`, `get_user_name()`.

## Why Use Functions? — Vì sao dùng hàm?

Giả sử cần đổi nhiều nhiệt độ từ °F sang °C. **Không dùng hàm**, công thức bị lặp lại:

````python
temp1 = 77
celsius1 = (temp1 - 32) * 5 / 9
print(celsius1)

temp2 = 95
celsius2 = (temp2 - 32) * 5 / 9
print(celsius2)
````

**Dùng hàm**, công thức chỉ viết một lần:

````python
def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5 / 9

print(fahrenheit_to_celsius(77))
print(fahrenheit_to_celsius(95))
print(fahrenheit_to_celsius(50))
````

Khi cần sửa công thức, bạn chỉ sửa **một chỗ**.

## Return Values — Giá trị trả về

Dùng câu lệnh `return` để hàm **gửi kết quả về** cho nơi gọi:

````python
def get_greeting():
    return "Hello from a function"

message = get_greeting()
print(message)
````

Có thể dùng trực tiếp giá trị trả về:

````python
def get_greeting():
    return "Hello from a function"

print(get_greeting())
````

> 📌 Hàm không có `return` sẽ trả về `None`. Khi gặp `return`, hàm **kết thúc ngay**.

## The pass Statement — Câu lệnh pass

Định nghĩa hàm không được để trống. Nếu chưa viết nội dung, dùng `pass`:

````python
def my_function():
    pass

print(my_function())
````

> 🧪 **Try it Yourself** — Viết hàm say_hello() in ra lời chào, rồi gọi nó 2 lần.

````python
def say_hello():
    print("Xin chào, mình đang học Python!")

say_hello()
say_hello()
````

## 🧠 Ghi nhớ nhanh

> - Định nghĩa: `def ten_ham():` + thân hàm **thụt lề**. Gọi: `ten_ham()`.
> - Định nghĩa thôi thì **chưa chạy** — phải gọi mới chạy.
> - `return` gửi kết quả về và **kết thúc hàm**; không có `return` → trả về `None`.
> - Hàm giúp **không lặp code**: sửa một chỗ, áp dụng mọi nơi.
