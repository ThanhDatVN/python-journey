## Python Dates — Ngày tháng

Python không có kiểu dữ liệu ngày riêng, nhưng có module **`datetime`** để làm việc với ngày giờ như những object.

Import module `datetime` và hiển thị ngày giờ hiện tại:

```python
import datetime

x = datetime.datetime.now()
print(x)
```

## Date Output — Hiển thị ngày

Kết quả ví dụ trên có dạng: năm, tháng, ngày, giờ, phút, giây, micro giây.

Module `datetime` có nhiều phương thức để lấy thông tin. Ví dụ năm và tên thứ trong tuần:

```python
import datetime

x = datetime.datetime.now()

print(x.year)
print(x.strftime("%A"))
```

## Creating Date Objects — Tạo đối tượng ngày

Dùng class `datetime()` của module `datetime`. Cần ba tham số: **năm, tháng, ngày**:

```python
import datetime

x = datetime.datetime(2020, 5, 17)
print(x)
```

Class `datetime()` còn nhận thêm giờ và múi giờ (giờ, phút, giây, micro giây, tzone) — tuỳ chọn, mặc định là `0` (`None` với múi giờ).

## The strftime() Method — Định dạng ngày

Object `datetime` có phương thức `strftime()` để **định dạng** ngày thành chuỗi dễ đọc. Tham số là một **mã định dạng**:

```python
import datetime

x = datetime.datetime(2018, 6, 1)
print(x.strftime("%B"))
```

### Bảng mã định dạng

| Mã | Ý nghĩa | Ví dụ |
|---|---|---|
| `%a` | Thứ, viết tắt | Wed |
| `%A` | Thứ, đầy đủ | Wednesday |
| `%w` | Thứ dạng số 0-6, 0 là Chủ nhật | 3 |
| `%d` | Ngày trong tháng 01-31 | 31 |
| `%b` | Tháng, viết tắt | Dec |
| `%B` | Tháng, đầy đủ | December |
| `%m` | Tháng dạng số 01-12 | 12 |
| `%y` | Năm, 2 chữ số | 18 |
| `%Y` | Năm, đầy đủ | 2018 |
| `%H` | Giờ 00-23 | 17 |
| `%I` | Giờ 00-12 | 05 |
| `%p` | AM/PM | PM |
| `%M` | Phút 00-59 | 41 |
| `%S` | Giây 00-59 | 08 |
| `%f` | Micro giây 000000-999999 | 548513 |
| `%z` | Chênh lệch UTC | +0100 |
| `%Z` | Múi giờ | CST |
| `%j` | Ngày trong năm 001-366 | 365 |
| `%U` | Tuần trong năm (CN đầu tuần) 00-53 | 52 |
| `%W` | Tuần trong năm (T2 đầu tuần) 00-53 | 52 |
| `%c` | Ngày giờ theo locale | Mon Dec 31 17:41:00 2018 |
| `%C` | Thế kỷ | 20 |
| `%x` | Ngày theo locale | 12/31/18 |
| `%X` | Giờ theo locale | 17:41:00 |
| `%%` | Ký tự `%` | % |
| `%G` | Năm theo ISO 8601 | 2018 |
| `%u` | Thứ theo ISO 8601 (1-7) | 1 |
| `%V` | Tuần theo ISO 8601 (01-53) | 01 |

```python
import datetime

x = datetime.datetime(2026, 10, 3, 14, 30)
print(x.strftime("%d/%m/%Y"))
print(x.strftime("%A, %d %B %Y - %H:%M"))
```

## ➕ Mở rộng: Tính khoảng cách ngày với timedelta

```python
from datetime import date, timedelta

today = date(2026, 10, 3)
print(today + timedelta(days=30))       # 30 ngày sau
new_year = date(2027, 1, 1)
print((new_year - today).days, "ngày nữa là Tết dương lịch")
```

Chuyển chuỗi thành ngày với `strptime()`:

```python
from datetime import datetime

d = datetime.strptime("25/12/2026", "%d/%m/%Y")
print(d, d.strftime("%A"))
```

> 🧪 **Try it Yourself** — Tính tuổi (theo năm) của bạn bằng cách trừ năm sinh khỏi năm hiện tại.

```python
import datetime

birth_year = 2005
now = datetime.datetime.now()
print("Tuổi:", now.year - birth_year)
```

---

## 🧠 Ghi nhớ nhanh

> - `import datetime` → `datetime.datetime.now()` lấy thời điểm hiện tại.
> - Tạo ngày: `datetime.datetime(năm, tháng, ngày)`.
> - Định dạng: `x.strftime("%d/%m/%Y")` — `%d` ngày, `%m` tháng, `%Y` năm, `%A` thứ, `%B` tên tháng, `%H:%M` giờ phút.
> - Cộng/trừ ngày: `timedelta(days=...)`; chuỗi → ngày: `strptime()`.
