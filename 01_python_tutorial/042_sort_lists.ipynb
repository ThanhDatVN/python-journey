## Sort List Alphanumerically — Sắp xếp theo chữ/số

Phương thức `sort()` sắp xếp list theo thứ tự chữ cái hoặc số, **tăng dần** theo mặc định:

```python
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort()
print(thislist)
```

```python
thislist = [100, 50, 65, 82, 23]
thislist.sort()
print(thislist)
```

## Sort Descending — Sắp xếp giảm dần

Dùng tham số `reverse=True`:

```python
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort(reverse=True)
print(thislist)
```

```python
thislist = [100, 50, 65, 82, 23]
thislist.sort(reverse=True)
print(thislist)
```

## Customize Sort Function — Tuỳ chỉnh cách sắp xếp

Tham số `key=function` cho phép tự định nghĩa tiêu chí sắp xếp. Hàm sẽ được gọi trên từng phần tử, và list được sắp xếp theo **giá trị hàm trả về**.

Sắp xếp theo khoảng cách tới số 50:

```python
def myfunc(n):
    return abs(n - 50)

thislist = [100, 50, 65, 82, 23]
thislist.sort(key=myfunc)
print(thislist)
```

## Case Insensitive Sort — Sắp xếp không phân biệt hoa thường

Mặc định `sort()` **phân biệt** hoa thường: mọi chữ hoa đứng trước chữ thường:

```python
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort()
print(thislist)
```

Dùng `str.lower` làm hàm key để bỏ qua hoa thường:

```python
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort(key=str.lower)
print(thislist)
```

## Reverse Order — Đảo ngược thứ tự

Phương thức `reverse()` đảo ngược thứ tự hiện tại, **không** quan tâm chữ cái:

```python
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.reverse()
print(thislist)
```

## ➕ Mở rộng: sort() hay sorted()?

- `list.sort()` sắp xếp **tại chỗ**, thay đổi list gốc, trả về `None`.
- `sorted(list)` **trả về list mới** đã sắp xếp, list gốc giữ nguyên.

```python
numbers = [3, 1, 2]
new_numbers = sorted(numbers)
print(numbers)
print(new_numbers)
```

> 🧪 **Try it Yourself** — Sắp xếp danh sách học sinh theo độ dài tên (gợi ý: key=len).

```python
students = ["Nguyên", "An", "Bình", "Khánh Linh"]
students.sort(key=len)
print(students)
```

---

## 🧠 Ghi nhớ nhanh

> - `a.sort()` tăng dần · `a.sort(reverse=True)` giảm dần.
> - `key=hàm` → sắp theo tiêu chí riêng; `key=str.lower` → không phân biệt hoa thường.
> - `a.reverse()` → chỉ đảo ngược thứ tự hiện tại.
> - `sort()` sửa list gốc (trả về `None`) · `sorted(a)` trả về **list mới**.
