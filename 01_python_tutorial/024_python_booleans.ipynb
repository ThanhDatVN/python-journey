# Python Booleans

> 📖 **Học song song trên W3Schools:** [python_booleans.asp](https://www.w3schools.com/python/python_booleans.asp)
>
> 🎯 **Trong bài này:** Boolean Values · Evaluate Values and Variables · Most Values are True · Some Values are False · Functions can Return a Boolean

Boolean biểu diễn một trong hai giá trị: `True` (đúng) hoặc `False` (sai).

## Boolean Values — Giá trị boolean

Trong lập trình, bạn thường cần biết một biểu thức là đúng hay sai. Python có thể đánh giá mọi biểu thức và trả về `True` hoặc `False`.

So sánh hai giá trị, kết quả là boolean:

```python
print(10 > 9)
print(10 == 9)
print(10 < 9)
```

Khi đặt điều kiện trong câu lệnh `if`, Python trả về `True` hoặc `False` để quyết định:

```python
a = 200
b = 33

if b > a:
    print("b is greater than a")
else:
    print("b is not greater than a")
```

## Evaluate Values and Variables — Đánh giá giá trị và biến

Hàm `bool()` cho phép đánh giá **bất kỳ** giá trị nào và trả về `True` hoặc `False`:

```python
print(bool("Hello"))
print(bool(15))
```

```python
x = "Hello"
y = 15

print(bool(x))
print(bool(y))
```

## Most Values are True — Hầu hết giá trị là True

- Mọi chuỗi đều là `True`, **trừ chuỗi rỗng** `""`.
- Mọi số đều là `True`, **trừ** `0`.
- Mọi list, tuple, set, dict đều là `True`, **trừ khi rỗng**.

```python
print(bool("abc"))
print(bool(123))
print(bool(["apple", "cherry", "banana"]))
```

## Some Values are False — Một số giá trị là False

Thực tế, không có nhiều giá trị cho ra `False`: các giá trị rỗng như `()`, `[]`, `{}`, `""`, số `0`, và `None`. Và tất nhiên là chính `False`:

```python
print(bool(False))
print(bool(None))
print(bool(0))
print(bool(""))
print(bool(()))
print(bool([]))
print(bool({}))
```

Một trường hợp nữa: object của class có phương thức `__len__` trả về `0`:

```python
class MyClass:
    def __len__(self):
        return 0

myobj = MyClass()
print(bool(myobj))
```

## Functions can Return a Boolean — Hàm trả về boolean

Bạn có thể viết hàm trả về giá trị boolean:

```python
def my_function():
    return True

print(my_function())
```

Và dùng kết quả đó để rẽ nhánh:

```python
def my_function():
    return True

if my_function():
    print("YES!")
else:
    print("NO!")
```

Python cũng có nhiều hàm có sẵn trả về boolean, ví dụ `isinstance()` kiểm tra một object có thuộc kiểu dữ liệu nào đó không:

```python
x = 200
print(isinstance(x, int))
```

> 🧪 **Try it Yourself** — Đoán trước kết quả của từng dòng, sau đó chạy để kiểm tra.

```python
print(bool(" "))
print(bool(0.0))
print(bool([0]))
print(bool("False"))
```

## 🧠 Ghi nhớ nhanh

> - Chỉ có hai giá trị: `True` và `False` (viết hoa chữ cái đầu).
> - Phép so sánh trả về boolean; `bool(x)` đánh giá mọi giá trị.
> - **Falsy**: `False`, `None`, `0`, `""`, `()`, `[]`, `{}` — còn lại hầu hết là **Truthy**.
> - `isinstance(x, int)` kiểm tra kiểu dữ liệu.
