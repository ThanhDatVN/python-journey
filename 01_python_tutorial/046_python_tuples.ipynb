## Tuple

Tuple dùng để lưu **nhiều phần tử** trong một biến.

Tuple là một trong 4 kiểu tập hợp có sẵn của Python (cùng với List, Set và Dictionary).

Tuple là tập hợp **có thứ tự** và **không thay đổi được (unchangeable)**. Tuple được viết bằng **dấu ngoặc tròn**:

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple)
```

## Tuple Items — Phần tử của tuple

Phần tử trong tuple: **có thứ tự**, **không thay đổi được**, và **cho phép trùng lặp**. Phần tử được đánh chỉ số từ `[0]`.

### Ordered — Có thứ tự

Các phần tử có thứ tự xác định và thứ tự đó không thay đổi.

### Unchangeable — Không thay đổi được

Sau khi tạo tuple, bạn **không thể** sửa, thêm hay xoá phần tử.

### Allow Duplicates — Cho phép trùng lặp

```python
thistuple = ("apple", "banana", "cherry", "apple", "cherry")
print(thistuple)
```

## Tuple Length — Độ dài tuple

```python
thistuple = ("apple", "banana", "cherry")
print(len(thistuple))
```

## Create Tuple With One Item — Tuple có một phần tử

Để tạo tuple chỉ có **một** phần tử, bạn phải thêm **dấu phẩy** sau phần tử đó; nếu không Python sẽ không hiểu đó là tuple:

```python
thistuple = ("apple",)
print(type(thistuple))

# KHÔNG phải tuple
thistuple = ("apple")
print(type(thistuple))
```

## Tuple Items - Data Types — Kiểu dữ liệu của phần tử

```python
tuple1 = ("apple", "banana", "cherry")
tuple2 = (1, 5, 7, 9, 3)
tuple3 = (True, False, False)
tuple4 = ("abc", 34, True, 40, "male")

print(tuple1, tuple2, tuple3, tuple4)
```

## type()

```python
mytuple = ("apple", "banana", "cherry")
print(type(mytuple))
```

## The tuple() Constructor — Hàm khởi tạo tuple()

```python
thistuple = tuple(("apple", "banana", "cherry"))  # chú ý hai cặp ngoặc
print(thistuple)
```

## ➕ Mở rộng: Khi nào dùng tuple?

- Dữ liệu **không nên thay đổi**: toạ độ `(x, y)`, ngày tháng `(2026, 10, 3)`, cấu hình…
- Tuple nhanh hơn và tốn ít bộ nhớ hơn list một chút.
- Tuple dùng được làm **key của dictionary** (list thì không).

```python
location = {(21.03, 105.85): "Hà Nội", (10.82, 106.63): "TP. Hồ Chí Minh"}
print(location[(21.03, 105.85)])
```

> 🧪 **Try it Yourself** — Tạo tuple lưu ngày sinh của bạn (năm, tháng, ngày) rồi in ra.

```python
birthday = (2005, 6, 15)
print(birthday)
print(type(birthday))
```
