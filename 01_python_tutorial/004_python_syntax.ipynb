## Python Indentation — Thụt lề

*Indentation* là các khoảng trắng ở đầu dòng code.

Ở nhiều ngôn ngữ khác, thụt lề chỉ để code dễ đọc. Trong Python, thụt lề **rất quan trọng** — Python dùng nó để biết dòng nào thuộc về khối lệnh nào.

### Example

```python
if 5 > 2:
    print("Five is greater than two!")
```

Nếu bỏ thụt lề, Python sẽ báo lỗi:

```python
if 5 > 2:
print("Five is greater than two!")
```

Số khoảng trắng do bạn chọn — phổ biến nhất là **4 khoảng trắng** — nhưng phải có **ít nhất 1**:

```python
if 5 > 2:
 print("Five is greater than two!")
if 5 > 2:
        print("Five is greater than two!")
```

Trong **cùng một khối lệnh**, các dòng phải thụt lề bằng nhau, nếu không sẽ lỗi:

```python
if 5 > 2:
    print("Five is greater than two!")
        print("Five is greater than two!")
```

## Python Variables — Biến

Trong Python, biến được tạo ra **ngay khi bạn gán giá trị** cho nó. Không cần lệnh khai báo riêng:

```python
x = 5
y = "Hello, World!"

print(x)
print(y)
```

Bạn sẽ học kỹ hơn trong chương *Python Variables*.

## Comments — Chú thích

Python cho phép viết chú thích (comment) để giải thích code. Comment bắt đầu bằng dấu `#`, Python sẽ bỏ qua phần còn lại của dòng đó:

```python
# Đây là một comment.
print("Hello, World!")
```

> 🧪 **Try it Yourself** — Thêm một dòng print thứ hai được thụt lề bên trong khối if, rồi chạy lại.

```python
if 10 > 3:
    print("Ten is greater than three!")
```

---

## 🧠 Ghi nhớ nhanh

> - **Thụt lề là cú pháp** trong Python, không chỉ để đẹp.
> - Chuẩn: **4 khoảng trắng**; các dòng trong cùng khối phải thụt lề **bằng nhau**.
> - ⚠️ Thiếu hoặc thừa thụt lề → `IndentationError`.
> - Biến tạo ra khi gán giá trị; comment bắt đầu bằng `#`.

## 📝 Exercise — Bài tập nhanh

**❓ Câu hỏi:** Đoạn code nào **đúng cú pháp** Python?

- **A.** `if (5 > 2) { print("Yes") }`
- **B.** `if 5 > 2: print("Yes") }`
- **C.** `if 5 > 2:` xuống dòng, rồi `    print("Yes")` (thụt lề 4 khoảng trắng)
- **D.** `if 5 > 2` xuống dòng, rồi `print("Yes")` (không thụt lề)

<details>
<summary>👉 Xem đáp án</summary>

✅ **Đáp án: C** — `if 5 > 2:` xuống dòng, rồi `    print("Yes")` (thụt lề 4 khoảng trắng)

Câu lệnh `if` kết thúc bằng dấu `:` và thân của nó phải được thụt lề.

</details>
