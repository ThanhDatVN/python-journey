# Python - Modify Strings

>
> 🎯 **Trong bài này:** Upper Case · Lower Case · Remove Whitespace · Replace String · Split String · ➕ Mở rộng: Chuỗi gốc không đổi

Python có sẵn rất nhiều **phương thức (method)** để xử lý chuỗi. Lưu ý: chuỗi là **bất biến (immutable)** — các phương thức này **trả về chuỗi mới**, chuỗi gốc không thay đổi.

## Upper Case — Chữ hoa

Phương thức `upper()` trả về chuỗi viết hoa toàn bộ:

````python
a = "Hello, World!"
print(a.upper())
````

## Lower Case — Chữ thường

Phương thức `lower()` trả về chuỗi viết thường toàn bộ:

````python
a = "Hello, World!"
print(a.lower())
````

## Remove Whitespace — Xoá khoảng trắng

*Whitespace* là khoảng trắng trước/sau nội dung chính — thứ bạn thường muốn loại bỏ.

Phương thức `strip()` xoá khoảng trắng ở **đầu và cuối** chuỗi:

````python
a = " Hello, World! "
print(a.strip())
````

Ngoài ra còn có `lstrip()` (chỉ xoá bên trái) và `rstrip()` (chỉ xoá bên phải):

````python
a = "   Hello   "
print("[" + a.lstrip() + "]")
print("[" + a.rstrip() + "]")
````

## Replace String — Thay thế chuỗi

Phương thức `replace()` thay thế một chuỗi con bằng chuỗi khác:

````python
a = "Hello, World!"
print(a.replace("H", "J"))
````

## Split String — Tách chuỗi

Phương thức `split()` tách chuỗi thành một **list** dựa trên ký tự phân tách:

````python
a = "Hello, World!"
print(a.split(","))
````

Không truyền tham số, `split()` sẽ tách theo khoảng trắng:

````python
sentence = "Python  is   easy to learn"
print(sentence.split())
````

## ➕ Mở rộng: Chuỗi gốc không đổi

````python
a = "Hello"
b = a.upper()

print(a)  # vẫn là Hello
print(b)
````

Xem toàn bộ phương thức chuỗi trong bài *String Methods* hoặc trang [String Methods Reference](../10_reference/string_methods.md).

> 🧪 **Try it Yourself** — Chuẩn hoá dữ liệu người dùng nhập: xoá khoảng trắng thừa và viết hoa toàn bộ.

````python
user_input = "   nguyen van an   "
print(user_input.strip().upper())
````

## 🧠 Ghi nhớ nhanh

> - `upper()` · `lower()` · `strip()` · `replace(a, b)` · `split(sep)`.
> - Chuỗi **bất biến**: phương thức trả về **chuỗi mới**, chuỗi gốc giữ nguyên.
> - ⚠️ Muốn giữ kết quả phải gán lại: `s = s.upper()`.
