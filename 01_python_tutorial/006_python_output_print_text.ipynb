## Print Text — In văn bản

Hàm `print()` dùng để hiển thị (xuất) dữ liệu ra màn hình. Bạn đã thấy nó nhiều lần:

````python
print("Hello World!")
````

### Print Many Lines — In nhiều dòng

Gọi `print()` nhiều lần thì mỗi lần sẽ in trên **một dòng mới**:

````python
print("Hello World!")
print("I am learning Python.")
print("It is awesome!")
````

## Double Quotes — Nháy kép hay nháy đơn?

Văn bản (chuỗi) phải được đặt trong dấu nháy. Dùng nháy kép `"` hay nháy đơn `'` đều được:

````python
print("This will work!")
print('This will also work!')
````

Nếu quên dấu nháy, Python sẽ hiểu chữ đó là tên biến hoặc cú pháp lạ và báo lỗi:

````python
print(This will cause an error)
````

## Print Without a New Line — In trên cùng một dòng

Mặc định, `print()` tự thêm ký tự xuống dòng ở cuối. Dùng tham số `end` để thay ký tự đó bằng thứ khác:

````python
print("Hello World!", end=" ")
print("I will print on the same line.")
````

Bạn có thể dùng bất kỳ chuỗi nào cho `end`:

````python
print("Loading", end="...")
print("Done!")
````

> 🧪 **Try it Yourself** — Đổi giá trị của end thành " | " hoặc " -> " rồi chạy lại.

````python
print("Python", end=" ")
print("Jupyter", end=" ")
print("W3Schools")
````

---

## 🧠 Ghi nhớ nhanh

> - `print()` hiển thị dữ liệu; văn bản phải nằm trong **dấu nháy** (`"..."` hoặc `'...'`).
> - Mỗi lần `print()` mặc định **xuống dòng** ở cuối.
> - `end=" "` → in tiếp trên cùng một dòng.
> - ⚠️ Quên dấu nháy → Python hiểu nhầm là tên biến/cú pháp → lỗi.
