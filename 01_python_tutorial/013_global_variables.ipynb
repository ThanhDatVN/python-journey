## Global Variables — Biến toàn cục

Biến được tạo **bên ngoài** mọi hàm gọi là **biến toàn cục (global variable)**. Biến toàn cục dùng được ở mọi nơi — cả bên trong lẫn bên ngoài hàm.

> Bạn sẽ học về hàm (`def`) trong chương *Python Functions*. Ở đây chỉ cần hiểu: `def myfunc():` định nghĩa một hàm, `myfunc()` gọi hàm đó.

### Example

````python
x = "awesome"

def myfunc():
    print("Python is " + x)

myfunc()
````

Nếu bạn tạo một biến **cùng tên bên trong hàm**, biến đó là **biến cục bộ (local)** và chỉ dùng được trong hàm. Biến toàn cục cùng tên vẫn giữ nguyên giá trị:

````python
x = "awesome"

def myfunc():
    x = "fantastic"
    print("Python is " + x)

myfunc()

print("Python is " + x)
````

## The global Keyword — Từ khoá global

Bình thường, biến tạo trong hàm là biến cục bộ. Muốn tạo biến **toàn cục** ngay bên trong hàm, dùng từ khoá `global`:

````python
def myfunc():
    global y
    y = "fantastic"

myfunc()

print("Python is " + y)
````

Dùng `global` cũng là cách để **thay đổi giá trị** của một biến toàn cục từ bên trong hàm:

````python
x = "awesome"

def myfunc():
    global x
    x = "fantastic"

myfunc()

print("Python is " + x)
````

> 🧪 **Try it Yourself** — Thử xoá dòng `global counter` rồi chạy lại — chuyện gì xảy ra?

````python
counter = 0

def increase():
    global counter
    counter = counter + 1

increase()
increase()
print(counter)
````

---

## 🧠 Ghi nhớ nhanh

> - Biến tạo **ngoài hàm** = biến toàn cục (global), dùng được ở mọi nơi.
> - Biến tạo **trong hàm** = biến cục bộ (local), chỉ sống trong hàm.
> - Muốn **tạo/sửa** biến toàn cục từ trong hàm → khai báo `global x`.
