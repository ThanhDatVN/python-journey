## Tuple Methods — Phương thức của tuple

Vì không thay đổi được, tuple chỉ có **hai** phương thức:

| Phương thức | Mô tả |
|---|---|
| `count()` | Trả về số lần một giá trị xuất hiện trong tuple |
| `index()` | Tìm giá trị trong tuple và trả về vị trí đầu tiên của nó |

Chi tiết: [Tuple Methods Reference](../10_reference/tuple_methods.md).

### count()

```python
thistuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)
x = thistuple.count(5)
print(x)
```

### index()

```python
thistuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)
x = thistuple.index(8)
print(x)
```

`index()` báo lỗi nếu giá trị không tồn tại:

```python
thistuple = (1, 3, 7)
thistuple.index(99)
```

## ➕ Mở rộng: Hàm built-in dùng với tuple

```python
numbers = (5, 2, 9, 1)

print(len(numbers))
print(min(numbers), max(numbers), sum(numbers))
print(sorted(numbers))   # trả về list
```

> 🧪 **Try it Yourself** — Đếm xem điểm 10 xuất hiện bao nhiêu lần trong tuple điểm.

```python
scores = (8, 10, 7, 10, 9, 10)
print(scores.count(10))
```

## 🧠 Ghi nhớ nhanh

> - Tuple chỉ có **2** phương thức: `count(x)` và `index(x)`.
> - ⚠️ `index()` báo lỗi nếu không tìm thấy.
