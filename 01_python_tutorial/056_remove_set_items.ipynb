## Remove Item — Xoá phần tử

Có hai phương thức để xoá phần tử: `remove()` và `discard()`.

```python
thisset = {"apple", "banana", "cherry"}
thisset.remove("banana")
print(thisset)
```

> ⚠️ Nếu phần tử **không tồn tại**, `remove()` sẽ **báo lỗi**:

```python
thisset = {"apple", "banana", "cherry"}
thisset.remove("mango")
```

```python
thisset = {"apple", "banana", "cherry"}
thisset.discard("banana")
print(thisset)
```

> ✅ Nếu phần tử không tồn tại, `discard()` **không** báo lỗi:

```python
thisset = {"apple", "banana", "cherry"}
thisset.discard("mango")
print(thisset)
```

## pop()

`pop()` cũng xoá phần tử, nhưng vì set không có thứ tự nên bạn **không biết trước** phần tử nào bị xoá. Giá trị trả về là phần tử đã bị xoá:

```python
thisset = {"apple", "banana", "cherry"}
x = thisset.pop()

print(x)
print(thisset)
```

## clear() và del

`clear()` làm rỗng set:

```python
thisset = {"apple", "banana", "cherry"}
thisset.clear()
print(thisset)
```

`del` xoá set hoàn toàn:

```python
thisset = {"apple", "banana", "cherry"}
del thisset
print(thisset)
```

| Cách | Khi phần tử không tồn tại |
|---|---|
| `remove(x)` | Báo lỗi `KeyError` |
| `discard(x)` | Không làm gì |
| `pop()` | Xoá phần tử bất kỳ (lỗi nếu set rỗng) |

> 🧪 **Try it Yourself** — Đổi remove thành discard và xoá một phần tử không có trong set.

```python
online_users = {"an", "binh", "chi"}
online_users.discard("binh")
print(online_users)
```

## 🧠 Ghi nhớ nhanh

> - `remove(x)` → lỗi nếu không có · `discard(x)` → **không** lỗi.
> - `pop()` → xoá phần tử **ngẫu nhiên**, trả về phần tử đó.
> - `clear()` làm rỗng · `del s` xoá hẳn biến.
