## Loop Items — Duyệt set

Dùng vòng lặp `for` để duyệt qua các phần tử của set:

```python
thisset = {"apple", "banana", "cherry"}

for x in thisset:
    print(x)
```

> 📌 Thứ tự duyệt **không được đảm bảo**. Nếu cần thứ tự cố định, hãy sắp xếp trước bằng `sorted()`:

```python
thisset = {"cherry", "apple", "banana"}

for x in sorted(thisset):
    print(x)
```

> 🧪 **Try it Yourself** — Duyệt set các số và chỉ in ra số chẵn.

```python
numbers = {1, 2, 3, 4, 5, 6, 7, 8}
for n in sorted(numbers):
    if n % 2 == 0:
        print(n)
```

---

## 🧠 Ghi nhớ nhanh

> - `for x in s:` — thứ tự **không được đảm bảo**.
> - Cần thứ tự cố định → `for x in sorted(s):`.
