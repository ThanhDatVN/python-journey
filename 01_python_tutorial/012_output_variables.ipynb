## Output Variables — In giá trị của biến

Hàm `print()` thường được dùng để in giá trị của biến:

````python
x = "Python is awesome"
print(x)
````

Trong `print()`, bạn có thể in **nhiều biến** bằng cách ngăn cách chúng bởi dấu phẩy:

````python
x = "Python"
y = "is"
z = "awesome"
print(x, y, z)
````

Cũng có thể dùng toán tử `+` để **nối chuỗi**:

````python
x = "Python "
y = "is "
z = "awesome"
print(x + y + z)
````

> 💡 Để ý các khoảng trắng sau `"Python "` và `"is "` — thiếu chúng, kết quả sẽ thành `Pythonisawesome`.

Với số, dấu `+` là phép **cộng**:

````python
x = 5
y = 10
print(x + y)
````

Nhưng khi kết hợp chuỗi và số bằng `+`, Python báo lỗi:

````python
x = 5
y = "John"
print(x + y)
````

Cách tốt nhất để in nhiều biến khác kiểu là dùng **dấu phẩy** — nó hoạt động với mọi kiểu dữ liệu:

````python
x = 5
y = "John"
print(x, y)
````

> 🧪 **Try it Yourself** — Tạo hai biến first_name và last_name rồi in họ tên đầy đủ bằng hai cách: dấu phẩy và dấu +.

````python
first_name = "Nguyễn"
last_name = "An"

print(first_name, last_name)
print(first_name + " " + last_name)
````

---

## 🧠 Ghi nhớ nhanh

> - `print(x, y, z)` — dấu phẩy, chạy được với **mọi kiểu**, tự thêm khoảng trắng.
> - `+` với chuỗi = **nối**; `+` với số = **cộng**.
> - ⚠️ `5 + "John"` → `TypeError`. Dùng dấu phẩy, `str()` hoặc f-string.
