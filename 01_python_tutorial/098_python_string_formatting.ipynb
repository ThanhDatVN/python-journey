# Python String Formatting

> 📖 **Học song song trên W3Schools:** [python_string_formatting.asp](https://www.w3schools.com/python/python_string_formatting.asp)
>
> 🎯 **Trong bài này:** F-Strings · Placeholders and Modifiers · Perform Operations in F-Strings · Execute Functions in F-Strings · More Modifiers · String format() · Multiple Values · Index Numbers · Named Indexes

F-string có từ Python 3.6 và hiện là cách **được ưa chuộng** để định dạng chuỗi.

Trước Python 3.6, ta dùng phương thức `format()`.

## F-Strings

F-string cho phép định dạng các phần được chọn trong chuỗi.

Để tạo f-string, thêm chữ `f` trước chuỗi:

````python
txt = f"The price is 49 dollars"
print(txt)
````

## Placeholders and Modifiers — Placeholder và bộ định dạng

Để định dạng giá trị trong f-string, đặt **placeholder** `{}` — có thể chứa biến, phép tính, hàm, modifier:

````python
price = 59
txt = f"The price is {price} dollars"
print(txt)
````

Modifier đặt sau dấu `:`. Định dạng giá trị số thực có 2 chữ số thập phân:

````python
price = 59
txt = f"The price is {price:.2f} dollars"
print(txt)
````

Cũng có thể định dạng trực tiếp một giá trị:

````python
txt = f"The price is {95:.2f} dollars"
print(txt)
````

## Perform Operations in F-Strings — Tính toán trong f-string

Thực hiện phép toán trong placeholder:

````python
txt = f"The price is {20 * 59} dollars"
print(txt)
````

Với biến:

````python
price = 59
tax = 0.25
txt = f"The price is {price + (price * tax)} dollars"
print(txt)
````

Dùng câu lệnh `if...else` (dạng ternary) trong placeholder:

````python
price = 49
txt = f"It is very {'Expensive' if price > 50 else 'Cheap'}"
print(txt)
````

## Execute Functions in F-Strings — Gọi hàm trong f-string

Gọi phương thức có sẵn:

````python
fruit = "apples"
txt = f"I love {fruit.upper()}"
print(txt)
````

Gọi hàm tự viết:

````python
def myconverter(x):
    return x * 0.3048

txt = f"The plane is flying at a {myconverter(30000)} meter altitude"
print(txt)
````

## More Modifiers — Thêm bộ định dạng

Dùng dấu phẩy làm phân cách hàng nghìn:

````python
price = 59000
txt = f"The price is {price:,} dollars"
print(txt)
````

Danh sách các kiểu định dạng:

| Modifier | Ý nghĩa |
|---|---|
| `:<` | Căn trái (trong độ rộng chỉ định) |
| `:>` | Căn phải |
| `:^` | Căn giữa |
| `:=` | Đặt dấu ở vị trí ngoài cùng bên trái |
| `:+` | Luôn hiện dấu `+` / `-` |
| `:-` | Chỉ hiện dấu `-` cho số âm |
| `: ` | Thêm khoảng trắng trước số dương |
| `:,` | Dấu phẩy phân cách hàng nghìn |
| `:_` | Dấu gạch dưới phân cách hàng nghìn |
| `:b` | Nhị phân |
| `:c` | Ký tự Unicode tương ứng |
| `:d` | Số thập phân (nguyên) |
| `:e` / `:E` | Dạng khoa học, `e` thường / `E` hoa |
| `:f` / `:F` | Số thực có số chữ số cố định |
| `:g` / `:G` | Dạng tổng quát |
| `:o` | Bát phân |
| `:x` / `:X` | Thập lục phân, thường / hoa |
| `:n` | Định dạng số theo locale |
| `:%` | Phần trăm |

````python
n = 42
print(f"[{n:<8}] [{n:>8}] [{n:^8}]")
print(f"{n:+} {n:b} {n:o} {n:x} {n:X} {n:e}")
print(f"{0.856:%} {0.856:.1%}")
print(f"{1234567:_}")
````

## String format()

Phương thức `format()` vẫn dùng được, nhưng f-string nhanh và gọn hơn.

`format()` thay các placeholder `{}` bằng giá trị truyền vào:

````python
price = 49
txt = "The price is {} dollars"
print(txt.format(price))
````

Thêm modifier vào trong ngoặc nhọn:

````python
price = 49
txt = "The price is {:.2f} dollars"
print(txt.format(price))
````

## Multiple Values — Nhiều giá trị

Truyền thêm giá trị vào `format()` và thêm placeholder tương ứng:

````python
quantity = 3
itemno = 567
price = 49
myorder = "I want {} pieces of item number {} for {:.2f} dollars."
print(myorder.format(quantity, itemno, price))
````

## Index Numbers — Chỉ số

Dùng **chỉ số** `{0}` để chắc chắn giá trị được đặt đúng chỗ:

````python
quantity = 3
itemno = 567
price = 49
myorder = "I want {0} pieces of item number {1} for {2:.2f} dollars."
print(myorder.format(quantity, itemno, price))
````

Dùng chỉ số khi muốn **lặp lại** một giá trị nhiều lần:

````python
age = 36
name = "John"
txt = "His name is {1}. {1} is {0} years old."
print(txt.format(age, name))
````

## Named Indexes — Chỉ số có tên

Đặt **tên** trong ngoặc nhọn, rồi truyền giá trị theo tên:

````python
myorder = "I have a {carname}, it is a {model}."
print(myorder.format(carname="Ford", model="Mustang"))
````

> 🧪 **Try it Yourself** — In hoá đơn căn lề đẹp: tên căn trái 10 ký tự, giá căn phải 10 ký tự có dấu phẩy.

````python
items = [("Phở", 45000), ("Trà đá", 5000), ("Bánh mì", 25000)]
for name, price in items:
    print(f"{name:<10}{price:>10,}")
````

---

## 🧠 Ghi nhớ nhanh

> - F-string: `f"... {biểu_thức:modifier} ..."` — cách **ưu tiên** dùng.
> - Modifier hay dùng: `.2f` (2 số lẻ) · `,` (phân cách nghìn) · `%` · `<` `>` `^` (căn lề) · `b` `x` (nhị phân, hex).
> - Trong `{}` đặt được phép tính, ternary, lời gọi hàm.
> - Cách cũ: `"{} {}".format(a, b)` · `{0}` theo chỉ số · `{name}` theo tên.
