# Python - Add Set Items

> 📖 **Học song song trên W3Schools:** [python_sets_add.asp](https://www.w3schools.com/python/python_sets_add.asp)
>
> 🎯 **Trong bài này:** Add Items · Add Sets · Add Any Iterable

## Add Items — Thêm phần tử

Sau khi tạo set, bạn không sửa được phần tử nhưng có thể **thêm** phần tử mới bằng `add()`:

````python
thisset = {"apple", "banana", "cherry"}
thisset.add("orange")
print(thisset)
````

Thêm một giá trị đã có thì không có gì thay đổi (và không báo lỗi):

````python
thisset = {"apple", "banana", "cherry"}
thisset.add("apple")
print(thisset)
````

## Add Sets — Thêm phần tử từ set khác

Dùng phương thức `update()` để thêm các phần tử từ một set khác:

````python
thisset = {"apple", "banana", "cherry"}
tropical = {"pineapple", "mango", "papaya"}

thisset.update(tropical)
print(thisset)
````

## Add Any Iterable — Thêm từ bất kỳ iterable nào

`update()` nhận **mọi iterable**: tuple, list, dictionary…

````python
thisset = {"apple", "banana", "cherry"}
mylist = ["kiwi", "orange"]

thisset.update(mylist)
print(thisset)
````

> 🧪 **Try it Yourself** — Thu thập các thẻ (tag) của nhiều bài viết vào một set chung, không trùng lặp.

````python
all_tags = set()
all_tags.update(["python", "jupyter"])
all_tags.update(["python", "data"])
all_tags.add("ai")
print(all_tags)
````

---

## 🧠 Ghi nhớ nhanh

> - `add(x)` → thêm **một** phần tử (đã có thì bỏ qua).
> - `update(iterable)` → thêm nhiều phần tử từ set/list/tuple/dict.
