# Python - Loop Tuples

> 📖 **Học song song trên W3Schools:** [python_tuples_loop.asp](https://www.w3schools.com/python/python_tuples_loop.asp)
>
> 🎯 **Trong bài này:** Loop Through a Tuple · Loop Through the Index Numbers · Using a While Loop

## Loop Through a Tuple — Duyệt tuple bằng for

````python
thistuple = ("apple", "banana", "cherry")
for x in thistuple:
    print(x)
````

## Loop Through the Index Numbers — Duyệt theo chỉ số

Dùng `range()` và `len()`:

````python
thistuple = ("apple", "banana", "cherry")
for i in range(len(thistuple)):
    print(thistuple[i])
````

Hoặc `enumerate()`:

````python
thistuple = ("apple", "banana", "cherry")
for i, fruit in enumerate(thistuple, start=1):
    print(i, fruit)
````

## Using a While Loop — Duyệt bằng while

````python
thistuple = ("apple", "banana", "cherry")
i = 0
while i < len(thistuple):
    print(thistuple[i])
    i = i + 1
````

> 🧪 **Try it Yourself** — In ra các tháng trong quý 1, kèm số thứ tự bắt đầu từ 1.

````python
quarter1 = ("Tháng 1", "Tháng 2", "Tháng 3")
for i, month in enumerate(quarter1, start=1):
    print(i, month)
````

---

## 🧠 Ghi nhớ nhanh

> - `for x in t:` · `for i in range(len(t)):` · `for i, x in enumerate(t):` · `while i < len(t):`.
> - Giống hệt cách duyệt list.
