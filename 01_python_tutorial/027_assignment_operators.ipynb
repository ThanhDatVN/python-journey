## Assignment Operators — Toán tử gán

Toán tử gán dùng để gán giá trị cho biến:

| Toán tử | Ví dụ | Tương đương |
|---|---|---|
| `=` | `x = 5` | `x = 5` |
| `+=` | `x += 3` | `x = x + 3` |
| `-=` | `x -= 3` | `x = x - 3` |
| `*=` | `x *= 3` | `x = x * 3` |
| `/=` | `x /= 3` | `x = x / 3` |
| `%=` | `x %= 3` | `x = x % 3` |
| `//=` | `x //= 3` | `x = x // 3` |
| `**=` | `x **= 3` | `x = x ** 3` |
| `&=` | `x &= 3` | `x = x & 3` |
| `\|=` | `x \|= 3` | `x = x \| 3` |
| `^=` | `x ^= 3` | `x = x ^ 3` |
| `>>=` | `x >>= 3` | `x = x >> 3` |
| `<<=` | `x <<= 3` | `x = x << 3` |
| `:=` | `print(x := 3)` | `x = 3` rồi `print(x)` |

### Examples

````python
x = 5
x += 3
print(x)

x -= 2
print(x)

x *= 4
print(x)

x /= 3
print(x)
````

Toán tử gán kết hợp cũng dùng được với chuỗi:

````python
greeting = "Hello"
greeting += ", World!"
print(greeting)
````

## The Walrus Operator — Toán tử "con hải mã" `:=`

Python 3.8 giới thiệu toán tử `:=`. Nó **gán giá trị cho biến và đồng thời trả về giá trị đó** — vì vậy có thể dùng ngay bên trong một biểu thức:

````python
numbers = [1, 2, 3, 4, 5]

if (count := len(numbers)) > 3:
    print(f"List has {count} elements")
````

So sánh với cách viết thông thường:

````python
numbers = [1, 2, 3, 4, 5]
count = len(numbers)
if count > 3:
    print(f"List has {count} elements")
````

> 🧪 **Try it Yourself** — Mô phỏng tài khoản ngân hàng: thêm vài lần nạp (+=) và rút (-=) tiền.

````python
balance = 1000
balance += 500   # nạp tiền
balance -= 200   # rút tiền
print("Số dư:", balance)
````

## 🧠 Ghi nhớ nhanh

> - `x += 3` ⇔ `x = x + 3` (tương tự `-= *= /= %= //= **=`…).
> - Walrus `:=` gán **và** trả về giá trị ngay trong biểu thức: `if (n := len(a)) > 3:`.
