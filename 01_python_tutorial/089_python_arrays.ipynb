> ⚠️ **Lưu ý:** Python **không có** kiểu mảng (array) có sẵn như C/Java, nhưng **List** có thể dùng thay thế.

## Arrays — Mảng

Mảng dùng để lưu **nhiều giá trị** trong một biến:

```python
cars = ["Ford", "Volvo", "BMW"]
print(cars)
```

## What is an Array? — Mảng là gì?

Mảng là biến đặc biệt có thể chứa nhiều giá trị cùng lúc.

Nếu có danh sách các hãng xe, lưu mỗi hãng vào một biến riêng sẽ như sau:

```python
car1 = "Ford"
car2 = "Volvo"
car3 = "BMW"
```

Nhưng nếu muốn duyệt qua tất cả các xe để tìm một xe cụ thể? Hoặc có tới 300 xe thay vì 3?

Giải pháp là mảng: lưu nhiều giá trị dưới **một tên**, truy cập từng giá trị bằng **chỉ số**.

## Access the Elements of an Array — Truy cập phần tử

```python
cars = ["Ford", "Volvo", "BMW"]
x = cars[0]
print(x)
```

Sửa giá trị phần tử đầu tiên:

```python
cars = ["Ford", "Volvo", "BMW"]
cars[0] = "Toyota"
print(cars)
```

## The Length of an Array — Độ dài mảng

```python
cars = ["Ford", "Volvo", "BMW"]
x = len(cars)
print(x)
```

> 📌 Độ dài mảng luôn lớn hơn chỉ số lớn nhất **1 đơn vị**.

## Looping Array Elements — Duyệt mảng

```python
cars = ["Ford", "Volvo", "BMW"]
for x in cars:
    print(x)
```

## Adding Array Elements — Thêm phần tử

```python
cars = ["Ford", "Volvo", "BMW"]
cars.append("Honda")
print(cars)
```

## Removing Array Elements — Xoá phần tử

Xoá theo chỉ số bằng `pop()`:

```python
cars = ["Ford", "Volvo", "BMW"]
cars.pop(1)
print(cars)
```

Xoá theo giá trị bằng `remove()`:

```python
cars = ["Ford", "Volvo", "BMW"]
cars.remove("Volvo")
print(cars)
```

> 📌 `remove()` chỉ xoá lần xuất hiện **đầu tiên** của giá trị.

## ➕ Mở rộng: module `array` và NumPy

Khi cần mảng **cùng kiểu số** hiệu quả về bộ nhớ, Python có module `array`. Với tính toán khoa học, thư viện **NumPy** là lựa chọn chuẩn (xem `extras/numpy`):

```python
import array

nums = array.array("i", [1, 2, 3])   # "i" = số nguyên
nums.append(4)
print(nums)
print(nums[0], len(nums))
```

> 🧪 **Try it Yourself** — Tạo mảng nhiệt độ 7 ngày, in nhiệt độ cao nhất và trung bình.

```python
temps = [31, 33, 29, 35, 32, 30, 34]
print("Cao nhất:", max(temps))
print("Trung bình:", round(sum(temps) / len(temps), 1))
```

## 🧠 Ghi nhớ nhanh

> - Python **không có array có sẵn** → dùng **list** làm mảng.
> - Truy cập `a[0]`, sửa `a[0] = x`, độ dài `len(a)`, duyệt `for x in a`.
> - Thêm `append()`, xoá `pop(i)` / `remove(x)`.
> - Mảng số hiệu quả: module `array` hoặc **NumPy**.
