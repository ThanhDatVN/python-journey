## Built-in Data Types — Các kiểu dữ liệu có sẵn

Trong lập trình, **kiểu dữ liệu** là khái niệm quan trọng: mỗi kiểu cho phép làm những việc khác nhau.

Python có sẵn các kiểu dữ liệu sau, chia theo nhóm:

| Nhóm | Kiểu |
|---|---|
| Text (văn bản) | `str` |
| Numeric (số) | `int`, `float`, `complex` |
| Sequence (dãy) | `list`, `tuple`, `range` |
| Mapping (ánh xạ) | `dict` |
| Set (tập hợp) | `set`, `frozenset` |
| Boolean (logic) | `bool` |
| Binary (nhị phân) | `bytes`, `bytearray`, `memoryview` |
| None | `NoneType` |

## Getting the Data Type — Lấy kiểu dữ liệu

Dùng hàm `type()` để biết kiểu dữ liệu của một đối tượng bất kỳ:

```python
x = 5
print(type(x))
```

## Setting the Data Type — Kiểu dữ liệu được xác định khi gán

Trong Python, kiểu dữ liệu được xác định **tự động** theo giá trị bạn gán:

| Ví dụ | Kiểu dữ liệu |
|---|---|
| `x = "Hello World"` | `str` |
| `x = 20` | `int` |
| `x = 20.5` | `float` |
| `x = 1j` | `complex` |
| `x = ["apple", "banana", "cherry"]` | `list` |
| `x = ("apple", "banana", "cherry")` | `tuple` |
| `x = range(6)` | `range` |
| `x = {"name": "John", "age": 36}` | `dict` |
| `x = {"apple", "banana", "cherry"}` | `set` |
| `x = frozenset({"apple", "banana", "cherry"})` | `frozenset` |
| `x = True` | `bool` |
| `x = b"Hello"` | `bytes` |
| `x = bytearray(5)` | `bytearray` |
| `x = memoryview(bytes(5))` | `memoryview` |
| `x = None` | `NoneType` |

Chạy cell sau để kiểm chứng:

```python
values = ["Hello World", 20, 20.5, 1j, ["apple", "banana"], ("apple", "banana"),
          range(6), {"name": "John", "age": 36}, {"apple", "banana"},
          frozenset({"apple", "banana"}), True, b"Hello", bytearray(5),
          memoryview(bytes(5)), None]

for x in values:
    print(type(x))
```

## Setting the Specific Data Type — Chỉ định kiểu cụ thể

Muốn chỉ định rõ kiểu dữ liệu, dùng các **hàm khởi tạo (constructor)** tương ứng:

| Ví dụ | Kiểu dữ liệu |
|---|---|
| `x = str("Hello World")` | `str` |
| `x = int(20)` | `int` |
| `x = float(20.5)` | `float` |
| `x = complex(1j)` | `complex` |
| `x = list(("apple", "banana", "cherry"))` | `list` |
| `x = tuple(("apple", "banana", "cherry"))` | `tuple` |
| `x = range(6)` | `range` |
| `x = dict(name="John", age=36)` | `dict` |
| `x = set(("apple", "banana", "cherry"))` | `set` |
| `x = frozenset(("apple", "banana", "cherry"))` | `frozenset` |
| `x = bool(5)` | `bool` |
| `x = bytes(5)` | `bytes` |
| `x = bytearray(5)` | `bytearray` |
| `x = memoryview(bytes(5))` | `memoryview` |

```python
x = dict(name="John", age=36)
print(x)
print(type(x))
```

> 🧪 **Try it Yourself** — Đoán kiểu dữ liệu của từng giá trị trước, rồi chạy để kiểm tra.

```python
print(type(3.0))
print(type("3"))
print(type([3]))
print(type(3 > 2))
```

## 🧠 Ghi nhớ nhanh

> - 8 nhóm: Text `str` · Numeric `int float complex` · Sequence `list tuple range` · Mapping `dict` · Set `set frozenset` · Boolean `bool` · Binary `bytes bytearray memoryview` · `NoneType`.
> - Kiểu được xác định **tự động** khi gán giá trị; xem bằng `type()`.
> - Ngoặc: `[ ]` → list · `( )` → tuple · `{ }` → dict/set.
