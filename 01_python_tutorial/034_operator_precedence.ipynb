## Operator Precedence — Độ ưu tiên toán tử

Độ ưu tiên quyết định **phép nào được tính trước** khi có nhiều toán tử trong cùng một biểu thức.

Dấu ngoặc có độ ưu tiên cao nhất, nên biểu thức trong ngoặc được tính trước:

```python
print((6 + 3) - (6 + 3))
```

Phép nhân `*` có độ ưu tiên cao hơn phép cộng `+`, nên được tính trước:

```python
print(100 + 5 * 3)
```

## Precedence Order — Bảng độ ưu tiên

Từ **cao nhất** (trên cùng) đến **thấp nhất** (dưới cùng):

| Toán tử | Mô tả |
|---|---|
| `()` | Dấu ngoặc |
| `**` | Luỹ thừa |
| `+x`, `-x`, `~x` | Dấu dương, dấu âm, bitwise NOT |
| `*`, `/`, `//`, `%` | Nhân, chia, chia lấy nguyên, chia lấy dư |
| `+`, `-` | Cộng, trừ |
| `<<`, `>>` | Dịch bit |
| `&` | Bitwise AND |
| `^` | Bitwise XOR |
| `\|` | Bitwise OR |
| `==`, `!=`, `>`, `>=`, `<`, `<=`, `is`, `is not`, `in`, `not in` | So sánh, định danh, thành viên |
| `not` | Logic NOT |
| `and` | Logic AND |
| `or` | Logic OR |

## Left-to-Right Evaluation — Tính từ trái sang phải

Nếu hai toán tử có **cùng** độ ưu tiên, biểu thức được tính **từ trái sang phải**:

```python
print(5 + 4 - 7 + 3)   # ((5 + 4) - 7) + 3
print(100 / 10 * 2)    # (100 / 10) * 2
```

Ngoại lệ: luỹ thừa `**` tính **từ phải sang trái**:

```python
print(2 ** 3 ** 2)     # 2 ** (3 ** 2) = 2 ** 9
print((2 ** 3) ** 2)
```

## ➕ Mở rộng: Một vài "cái bẫy"

```python
print(-3 ** 2)        # -(3 ** 2) = -9, vì ** ưu tiên hơn dấu âm
print((-3) ** 2)      # 9

print(not True == False)   # not (True == False) -> True
print(True or False and False)   # True or (False and False)
```

> 💡 Khi không chắc chắn, hãy **dùng dấu ngoặc** — code vừa đúng vừa dễ đọc hơn.

> 🧪 **Try it Yourself** — Đoán kết quả trước, sau đó thêm dấu ngoặc để biểu thức cho ra 21.

```python
print(1 + 2 * 3 + 4)
```

## 🧠 Ghi nhớ nhanh

> - Thứ tự (cao → thấp): `()` → `**` → `+x -x ~x` → `* / // %` → `+ -` → `<< >>` → `&` → `^` → `|` → so sánh/`is`/`in` → `not` → `and` → `or`.
> - Cùng mức → tính **trái sang phải**; riêng `**` tính **phải sang trái**.
> - ⚠️ `-3 ** 2` → `-9`. Không chắc? → **dùng ngoặc**.
