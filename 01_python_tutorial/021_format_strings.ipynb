## String Format — Định dạng chuỗi

Như bài trước, ta không thể nối chuỗi và số bằng `+`:

````python
age = 36
txt = "My name is John, I am " + age
print(txt)
````

Giải pháp hiện đại nhất là dùng **f-string** hoặc phương thức `format()`.

## F-Strings

F-string (có từ Python 3.6) là cách được ưa chuộng nhất để định dạng chuỗi.

Thêm chữ `f` **trước** dấu nháy mở, rồi đặt biến hoặc biểu thức trong cặp ngoặc nhọn `{}` (gọi là *placeholder*):

````python
age = 36
txt = f"My name is John, I am {age}"
print(txt)
````

## Placeholders and Modifiers — Placeholder và bộ định dạng

Placeholder có thể chứa biến, phép tính, lời gọi hàm và **bộ định dạng (modifier)**.

Thêm placeholder cho biến `price`:

````python
price = 59
txt = f"The price is {price} dollars"
print(txt)
````

Modifier được viết sau dấu `:`. Ví dụ `.2f` nghĩa là *số thực với 2 chữ số sau dấu thập phân*:

````python
price = 59
txt = f"The price is {price:.2f} dollars"
print(txt)
````

Một số modifier hữu ích khác:

````python
big = 1234567.891
print(f"{big:,}")      # phân cách hàng nghìn
print(f"{big:,.2f}")   # kết hợp
print(f"{0.25:%}")     # phần trăm
print(f"{0.25:.0%}")
print(f"[{'Hi':>10}]") # căn phải trong 10 ký tự
````

## Perform Operations in F-Strings — Tính toán trong f-string

Bạn có thể đặt **phép tính** ngay trong placeholder:

````python
txt = f"The price is {20 * 59} dollars"
print(txt)
````

````python
price = 59
tax = 0.25
txt = f"The price is {price + (price * tax)} dollars"
print(txt)
````

Hoặc gọi hàm/phương thức:

````python
fruit = "apples"
txt = f"I love {fruit.upper()}"
print(txt)
````

Bạn sẽ gặp lại f-string và phương thức `format()` chi tiết hơn trong bài *Python String Formatting*.

> 🧪 **Try it Yourself** — Tạo hoá đơn: in ra tên sản phẩm, số lượng và thành tiền với 2 chữ số thập phân.

````python
product = "Cà phê"
quantity = 3
unit_price = 25.5

print(f"{product}: {quantity} x {unit_price} = {quantity * unit_price:.2f}")
````

---

## 🧠 Ghi nhớ nhanh

> - F-string: `f"Tôi {age} tuổi"` — chữ `f` trước dấu nháy, biểu thức trong `{ }`.
> - Modifier sau dấu `:` → `{price:.2f}` (2 chữ số thập phân), `{n:,}` (phân cách nghìn).
> - Đặt được phép tính và lời gọi hàm trong `{ }`: `f"{20 * 59}"`, `f"{name.upper()}"`.
