## Nested If Statements — If lồng nhau

Bạn có thể đặt câu lệnh `if` **bên trong** một câu lệnh `if` khác — gọi là **if lồng nhau**:

````python
x = 41

if x > 10:
    print("Above ten,")
    if x > 20:
        print("and also above 20!")
    else:
        print("but not above 20.")
````

## How Nested If Works — If lồng nhau hoạt động thế nào?

Điều kiện **bên ngoài** được kiểm tra trước. Chỉ khi nó đúng, điều kiện **bên trong** mới được kiểm tra:

````python
age = 25
has_license = True

if age >= 18:
    if has_license:
        print("You can drive")
    else:
        print("You need a license")
else:
    print("You are too young to drive")
````

## Multiple Levels of Nesting — Lồng nhiều tầng

````python
score = 85
attendance = 90
submitted = True

if score >= 60:
    if attendance >= 80:
        if submitted:
            print("You passed with good standing")
        else:
            print("Submit your assignment")
    else:
        print("Attendance too low")
else:
    print("Score too low")
````

> ⚠️ Lồng quá nhiều tầng làm code khó đọc. Thông thường, không nên lồng quá **2–3 tầng**.

## Nested If vs Logical Operators — If lồng nhau hay toán tử logic?

Nhiều trường hợp, `if` lồng nhau có thể thay bằng `and`:

````python
temperature = 25
is_sunny = True

# Cách 1: if lồng nhau
if temperature > 20:
    if is_sunny:
        print("Perfect beach weather!")

# Cách 2: dùng and — gọn hơn
if temperature > 20 and is_sunny:
    print("Perfect beach weather!")
````

Dùng `if` lồng nhau khi mỗi tầng cần **xử lý riêng** (ví dụ thông báo lỗi khác nhau); dùng `and` khi chỉ cần kiểm tra tất cả điều kiện cùng lúc.

## More Examples — Thêm ví dụ

````python
username = "Emil"
password = "python123"
is_active = True

if username:
    if password:
        if is_active:
            print("Login successful")
        else:
            print("Account is not active")
    else:
        print("Password required")
else:
    print("Username required")
````

> 🧪 **Try it Yourself** — Đổi giá trị các biến để thử lần lượt từng thông báo.

````python
balance = 500
amount = 200
card_locked = False

if not card_locked:
    if amount <= balance:
        print("Rút tiền thành công")
    else:
        print("Số dư không đủ")
else:
    print("Thẻ đã bị khoá")
````

## 🧠 Ghi nhớ nhanh

> - If lồng nhau: tầng trong chỉ được kiểm tra khi tầng ngoài **đúng**.
> - Mỗi tầng thụt lề thêm một mức.
> - Không cần xử lý riêng từng tầng → gộp bằng `and` cho gọn.
