## String Methods — Các phương thức của chuỗi

Python có sẵn nhiều phương thức dùng với chuỗi.

> 📌 **Lưu ý:** mọi phương thức chuỗi đều **trả về giá trị mới**, không thay đổi chuỗi gốc.

| Phương thức | Mô tả |
|---|---|
| `capitalize()` | Viết hoa ký tự đầu tiên |
| `casefold()` | Chuyển thành chữ thường (mạnh hơn `lower()`) |
| `center()` | Trả về chuỗi được căn giữa |
| `count()` | Đếm số lần một giá trị xuất hiện |
| `encode()` | Trả về phiên bản đã mã hoá (bytes) của chuỗi |
| `endswith()` | `True` nếu chuỗi kết thúc bằng giá trị chỉ định |
| `expandtabs()` | Đặt độ rộng tab |
| `find()` | Tìm vị trí của giá trị; trả về `-1` nếu không thấy |
| `format()` | Định dạng các giá trị trong chuỗi |
| `format_map()` | Định dạng các giá trị trong chuỗi (từ dict) |
| `index()` | Giống `find()` nhưng báo lỗi nếu không thấy |
| `isalnum()` | `True` nếu mọi ký tự là chữ hoặc số |
| `isalpha()` | `True` nếu mọi ký tự là chữ cái |
| `isascii()` | `True` nếu mọi ký tự là ký tự ASCII |
| `isdecimal()` | `True` nếu mọi ký tự là chữ số thập phân |
| `isdigit()` | `True` nếu mọi ký tự là chữ số |
| `isidentifier()` | `True` nếu chuỗi là một tên định danh hợp lệ |
| `islower()` | `True` nếu mọi chữ cái đều viết thường |
| `isnumeric()` | `True` nếu mọi ký tự là ký tự số |
| `isprintable()` | `True` nếu mọi ký tự in được |
| `isspace()` | `True` nếu mọi ký tự là khoảng trắng |
| `istitle()` | `True` nếu chuỗi theo quy tắc tiêu đề |
| `isupper()` | `True` nếu mọi chữ cái đều viết hoa |
| `join()` | Nối các phần tử của iterable bằng chuỗi này |
| `ljust()` | Căn trái chuỗi |
| `lower()` | Chuyển thành chữ thường |
| `lstrip()` | Xoá khoảng trắng bên trái |
| `maketrans()` | Tạo bảng chuyển đổi dùng cho `translate()` |
| `partition()` | Tách chuỗi thành tuple 3 phần |
| `removeprefix()` | Xoá tiền tố (nếu có) |
| `removesuffix()` | Xoá hậu tố (nếu có) |
| `replace()` | Thay thế một giá trị bằng giá trị khác |
| `rfind()` | Tìm vị trí xuất hiện **cuối cùng**; `-1` nếu không thấy |
| `rindex()` | Giống `rfind()` nhưng báo lỗi nếu không thấy |
| `rjust()` | Căn phải chuỗi |
| `rpartition()` | Tách thành tuple 3 phần, tính từ lần xuất hiện cuối |
| `rsplit()` | Tách chuỗi thành list, tính từ bên phải |
| `rstrip()` | Xoá khoảng trắng bên phải |
| `split()` | Tách chuỗi thành list |
| `splitlines()` | Tách chuỗi theo dòng thành list |
| `startswith()` | `True` nếu chuỗi bắt đầu bằng giá trị chỉ định |
| `strip()` | Xoá khoảng trắng hai đầu |
| `swapcase()` | Đảo hoa thành thường và ngược lại |
| `title()` | Viết hoa chữ cái đầu mỗi từ |
| `translate()` | Trả về chuỗi đã chuyển đổi theo bảng |
| `upper()` | Chuyển thành chữ hoa |
| `zfill()` | Thêm số `0` vào đầu cho đủ độ dài |

Chi tiết từng phương thức (tham số, ví dụ): [String Methods Reference](../10_reference/string_methods.md).

## ➕ Mở rộng: Ví dụ các phương thức hay dùng

### Viết hoa / viết thường

```python
txt = "hello, and welcome to my world."
print(txt.capitalize())
print(txt.title())
print(txt.upper())
print("Hello World".swapcase())
```

### Tìm kiếm và đếm

```python
txt = "I love apples, apple are my favorite fruit"
print(txt.count("apple"))
print(txt.find("apple"))
print(txt.rfind("apple"))
print(txt.find("banana"))
```

### Kiểm tra nội dung

```python
print("Company12".isalnum())
print("CompanyX".isalpha())
print("50800".isdigit())
print("   ".isspace())
print("Hello World".startswith("Hello"))
print("photo.jpg".endswith(".jpg"))
```

### Căn lề và đệm

```python
txt = "banana"
print("[" + txt.center(20) + "]")
print("[" + txt.ljust(20, "*") + "]")
print("[" + txt.rjust(20, "*") + "]")
print("42".zfill(5))
```

### Tách và ghép

```python
txt = "apple#banana#cherry#orange"
print(txt.split("#"))
print(txt.split("#", 1))
print(txt.partition("#"))
print("Line 1\nLine 2\nLine 3".splitlines())
print(", ".join(["a", "b", "c"]))
```

### Xoá tiền tố / hậu tố (Python 3.9+)

```python
print("report_2026.csv".removesuffix(".csv"))
print("https://example.com".removeprefix("https://"))
```

> 🧪 **Try it Yourself** — Kiểm tra một tên file có phải ảnh không (đuôi .png hoặc .jpg), không phân biệt hoa thường.

```python
filename = "Holiday.JPG"
print(filename.lower().endswith((".png", ".jpg")))
```

## 🧠 Ghi nhớ nhanh

> - Nhóm **đổi hoa/thường**: `upper lower capitalize title swapcase casefold`.
> - Nhóm **tìm kiếm**: `find` (không thấy → `-1`) · `index` (không thấy → lỗi) · `count` · `startswith` · `endswith`.
> - Nhóm **kiểm tra** (`is...`): `isdigit isalpha isalnum isspace isupper islower`.
> - Nhóm **tách/ghép/căn lề**: `split join partition strip center ljust rjust zfill`.
