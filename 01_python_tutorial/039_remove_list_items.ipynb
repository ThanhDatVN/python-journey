## Remove Specified Item — Xoá theo giá trị

Phương thức `remove()` xoá phần tử có giá trị chỉ định:

````python
thislist = ["apple", "banana", "cherry"]
thislist.remove("banana")
print(thislist)
````

Nếu có **nhiều** phần tử trùng giá trị, `remove()` chỉ xoá **lần xuất hiện đầu tiên**:

````python
thislist = ["apple", "banana", "cherry", "banana", "kiwi"]
thislist.remove("banana")
print(thislist)
````

Xoá giá trị không tồn tại sẽ gây lỗi:

````python
thislist = ["apple", "banana", "cherry"]
thislist.remove("mango")
````

## Remove Specified Index — Xoá theo chỉ số

Phương thức `pop()` xoá phần tử tại chỉ số chỉ định **và trả về phần tử đó**:

````python
thislist = ["apple", "banana", "cherry"]
removed = thislist.pop(1)
print(thislist)
print("Đã xoá:", removed)
````

Không truyền chỉ số, `pop()` xoá phần tử **cuối cùng**:

````python
thislist = ["apple", "banana", "cherry"]
thislist.pop()
print(thislist)
````

Từ khoá `del` cũng xoá được phần tử theo chỉ số:

````python
thislist = ["apple", "banana", "cherry"]
del thislist[0]
print(thislist)
````

`del` còn có thể xoá **toàn bộ list** — sau đó biến không còn tồn tại:

````python
thislist = ["apple", "banana", "cherry"]
del thislist
print(thislist)  # lỗi vì thislist không còn tồn tại
````

## Clear the List — Làm rỗng list

Phương thức `clear()` xoá hết phần tử nhưng **list vẫn còn**, chỉ là rỗng:

````python
thislist = ["apple", "banana", "cherry"]
thislist.clear()
print(thislist)
````

| Cách | Xoá theo | Ghi chú |
|---|---|---|
| `remove(x)` | Giá trị | Lần xuất hiện đầu tiên |
| `pop(i)` | Chỉ số | Trả về phần tử bị xoá; mặc định xoá phần tử cuối |
| `del list[i]` | Chỉ số | Cũng xoá được cả list hoặc một khoảng `del a[1:3]` |
| `clear()` | — | List còn lại nhưng rỗng |

> 🧪 **Try it Yourself** — Lấy ra (pop) khách hàng đầu tiên trong hàng đợi, rồi in danh sách còn lại.

````python
queue = ["An", "Bình", "Chi", "Dũng"]
first = queue.pop(0)
print("Phục vụ:", first)
print("Còn lại:", queue)
````

---

## 🧠 Ghi nhớ nhanh

> - `remove(x)` → xoá theo **giá trị** (lần xuất hiện đầu; không có → lỗi).
> - `pop(i)` → xoá theo **chỉ số** và **trả về** phần tử; `pop()` xoá phần tử cuối.
> - `del a[i]` xoá theo chỉ số; `del a` xoá cả biến.
> - `clear()` → list **rỗng** nhưng vẫn tồn tại.
