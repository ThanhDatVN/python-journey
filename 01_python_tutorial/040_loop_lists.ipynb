## Loop Through a List — Duyệt list bằng for

Dùng vòng lặp `for` để duyệt qua từng phần tử:

````python
thislist = ["apple", "banana", "cherry"]
for x in thislist:
    print(x)
````

Tìm hiểu kỹ hơn về `for` trong bài *Python For Loops*.

## Loop Through the Index Numbers — Duyệt theo chỉ số

Có thể duyệt theo chỉ số bằng cách kết hợp `range()` và `len()`:

````python
thislist = ["apple", "banana", "cherry"]
for i in range(len(thislist)):
    print(i, thislist[i])
````

`range(len(thislist))` tạo ra dãy `0, 1, 2`.

Cách "chuẩn Python" hơn để có cả chỉ số lẫn giá trị là dùng `enumerate()`:

````python
thislist = ["apple", "banana", "cherry"]
for i, fruit in enumerate(thislist):
    print(i, fruit)
````

## Using a While Loop — Duyệt bằng while

Dùng `while` với biến đếm, nhớ **tăng biến đếm** sau mỗi vòng:

````python
thislist = ["apple", "banana", "cherry"]
i = 0
while i < len(thislist):
    print(thislist[i])
    i = i + 1
````

Tìm hiểu thêm trong bài *Python While Loops*.

## Looping Using List Comprehension — Duyệt bằng list comprehension

List comprehension là cú pháp ngắn gọn nhất để duyệt list:

````python
thislist = ["apple", "banana", "cherry"]
[print(x) for x in thislist]
````

> 💡 Cách trên chỉ để minh hoạ: list comprehension thực chất dùng để **tạo list mới** (bài tiếp theo), còn khi chỉ cần in thì nên dùng `for` bình thường.

> 🧪 **Try it Yourself** — Tính tổng điểm và điểm trung bình bằng vòng lặp for.

````python
scores = [8, 7.5, 9, 6, 10]
total = 0
for s in scores:
    total += s
print("Tổng:", total)
print("Trung bình:", total / len(scores))
````

---

## 🧠 Ghi nhớ nhanh

> - `for x in a:` → duyệt từng phần tử.
> - `for i in range(len(a)):` → duyệt theo chỉ số.
> - `for i, x in enumerate(a):` → có **cả chỉ số lẫn giá trị** (cách Pythonic).
> - `while` → nhớ **tăng biến đếm**, nếu không sẽ lặp vô hạn.
