## Python Iterators

**Iterator** là một đối tượng chứa một số lượng giá trị **duyệt được** — nghĩa là có thể đi qua lần lượt từng giá trị.

Về kỹ thuật, iterator là đối tượng cài đặt **giao thức iterator**, gồm hai phương thức `__iter__()` và `__next__()`.

## Iterator vs Iterable

List, tuple, dictionary, set và chuỗi đều là **iterable** — những "thùng chứa" mà bạn có thể lấy ra iterator từ chúng.

Tất cả đều có phương thức `iter()` để lấy iterator:

```python
mytuple = ("apple", "banana", "cherry")
myit = iter(mytuple)

print(next(myit))
print(next(myit))
print(next(myit))
```

Chuỗi cũng là iterable, trả về iterator duyệt từng ký tự:

```python
mystr = "banana"
myit = iter(mystr)

print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
print(next(myit))
```

## Looping Through an Iterator — Duyệt iterator bằng for

```python
mytuple = ("apple", "banana", "cherry")
for x in mytuple:
    print(x)
```

```python
mystr = "banana"
for x in mystr:
    print(x)
```

> 📌 Vòng `for` thực chất **tự tạo một iterator** và gọi `next()` ở mỗi vòng lặp.

## Create an Iterator — Tự tạo iterator

Để biến một object/class thành iterator, cần cài đặt hai phương thức `__iter__()` và `__next__()`.

> Bạn sẽ học về class trong chương *Python Classes*. Tạm hiểu: `__init__()` chạy khi tạo object.

- `__iter__()` trả về chính đối tượng iterator (có thể khởi tạo giá trị ở đây).
- `__next__()` trả về **phần tử tiếp theo**.

Iterator trả về các số, bắt đầu từ 1, mỗi lần tăng 1 (1, 2, 3, 4, 5…):

```python
class MyNumbers:
    def __iter__(self):
        self.a = 1
        return self

    def __next__(self):
        x = self.a
        self.a += 1
        return x

myclass = MyNumbers()
myiter = iter(myclass)

print(next(myiter))
print(next(myiter))
print(next(myiter))
print(next(myiter))
print(next(myiter))
```

## StopIteration — Dừng lặp

Ví dụ trên có thể chạy mãi nếu ta dùng `for`. Để **dừng**, `__next__()` cần phát ra `StopIteration` khi đủ số lần:

```python
class MyNumbers:
    def __iter__(self):
        self.a = 1
        return self

    def __next__(self):
        if self.a <= 20:
            x = self.a
            self.a += 1
            return x
        else:
            raise StopIteration

myclass = MyNumbers()
myiter = iter(myclass)

for x in myiter:
    print(x, end=" ")
```

> 💡 Cách nhanh hơn để tạo iterator là dùng **generator** (bài [Python Generators](087_python_generators.ipynb)).

> 🧪 **Try it Yourself** — Tạo iterator từ list và lấy từng phần tử bằng next().

```python
colors = ["đỏ", "xanh", "vàng"]
it = iter(colors)
print(next(it))
print(next(it))
```

---

## 🧠 Ghi nhớ nhanh

> - **Iterable**: list, tuple, dict, set, str… → lấy iterator bằng `iter(obj)`.
> - **Iterator**: đối tượng có `__iter__()` và `__next__()`; lấy phần tử kế tiếp bằng `next(it)`.
> - Hết phần tử → `StopIteration`; vòng `for` tự xử lý việc này.
> - Tự tạo iterator: viết class có `__iter__` + `__next__` (hoặc dùng generator cho gọn).
