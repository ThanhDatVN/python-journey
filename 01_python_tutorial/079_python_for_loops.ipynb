## Python For Loops — Vòng lặp for

Vòng lặp `for` dùng để **duyệt qua một dãy** (list, tuple, dictionary, set, hoặc chuỗi).

Nó khác từ khoá `for` ở nhiều ngôn ngữ khác, và giống với phương thức *iterator* trong các ngôn ngữ hướng đối tượng hơn.

Với `for`, ta thực thi một khối lệnh **cho mỗi phần tử** trong list, tuple, set…:

````python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print(x)
````

`for` **không cần** biến đếm được khởi tạo từ trước.

## Looping Through a String — Duyệt qua chuỗi

Chuỗi cũng là đối tượng duyệt được, nó chứa một dãy ký tự:

````python
for x in "banana":
    print(x)
````

## The break Statement — Câu lệnh break

`break` dừng vòng lặp trước khi duyệt hết các phần tử.

Thoát khi `x` là "banana":

````python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    print(x)
    if x == "banana":
        break
````

Thoát khi `x` là "banana", nhưng lần này `break` đứng **trước** `print`:

````python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    if x == "banana":
        break
    print(x)
````

## The continue Statement — Câu lệnh continue

`continue` bỏ qua vòng hiện tại và tiếp tục với phần tử kế tiếp.

Không in "banana":

````python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
    if x == "banana":
        continue
    print(x)
````

## The range() Function — Hàm range()

Để lặp một số lần nhất định, dùng hàm `range()`.

`range()` trả về một dãy số, mặc định **bắt đầu từ 0**, tăng 1, và **dừng trước** số chỉ định:

````python
for x in range(6):
    print(x)
````

> 📌 `range(6)` là các số từ 0 đến 5, **không phải** 0 đến 6.

Chỉ định giá trị bắt đầu bằng tham số thứ nhất — `range(2, 6)` nghĩa là từ 2 đến 6 (không gồm 6):

````python
for x in range(2, 6):
    print(x)
````

Tham số thứ ba là **bước nhảy** — `range(2, 30, 3)`:

````python
for x in range(2, 30, 3):
    print(x)
````

## Else in For Loop — else trong vòng for

Khối `else` chạy **khi vòng lặp kết thúc bình thường**:

````python
for x in range(6):
    print(x)
else:
    print("Finally finished!")
````

> 📌 Khối `else` **không** chạy nếu vòng lặp bị dừng bằng `break`:

````python
for x in range(6):
    if x == 3:
        break
    print(x)
else:
    print("Finally finished!")
````

## Nested Loops — Vòng lặp lồng nhau

Vòng lặp **trong** chạy hết một lượt cho **mỗi** vòng của vòng lặp **ngoài**:

````python
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
    for y in fruits:
        print(x, y)
````

## The pass Statement — Câu lệnh pass

Vòng lặp `for` không được để trống. Nếu cần, đặt `pass` để tránh lỗi:

````python
for x in [0, 1, 2]:
    pass
````

## ➕ Mở rộng: Bảng cửu chương

````python
for i in range(1, 4):
    for j in range(1, 10):
        print(f"{i} x {j} = {i * j:2}", end="   ")
    print()
````

> 🧪 **Try it Yourself** — Tính tổng các số từ 1 đến 100 bằng vòng lặp for.

````python
total = 0
for n in range(1, 101):
    total += n
print(total)
````

## 🧠 Ghi nhớ nhanh

> - `for x in dãy:` → duyệt từng phần tử của list/tuple/set/dict/chuỗi.
> - `range(stop)` · `range(start, stop)` · `range(start, stop, step)` — **không gồm** `stop`.
> - `break` dừng · `continue` bỏ qua · `else` chạy khi lặp xong (không `break`).
> - Vòng lặp lồng nhau: vòng trong chạy hết cho **mỗi** vòng ngoài.
