## Nested Dictionaries — Dictionary lồng nhau

Một dictionary có thể chứa các dictionary khác — gọi là **dictionary lồng nhau (nested)**.

Tạo dictionary chứa ba dictionary:

```python
myfamily = {
    "child1": {"name": "Emil", "year": 2004},
    "child2": {"name": "Tobias", "year": 2007},
    "child3": {"name": "Linus", "year": 2011}
}
print(myfamily)
```

Hoặc tạo ba dictionary riêng, rồi gộp vào một dictionary mới:

```python
child1 = {"name": "Emil", "year": 2004}
child2 = {"name": "Tobias", "year": 2007}
child3 = {"name": "Linus", "year": 2011}

myfamily = {
    "child1": child1,
    "child2": child2,
    "child3": child3
}
print(myfamily)
```

## Access Items in Nested Dictionaries — Truy cập phần tử lồng nhau

Dùng tên key của từng tầng, từ ngoài vào trong:

```python
myfamily = {
    "child1": {"name": "Emil", "year": 2004},
    "child2": {"name": "Tobias", "year": 2007},
    "child3": {"name": "Linus", "year": 2011}
}
print(myfamily["child2"]["name"])
```

## Loop Through Nested Dictionaries — Duyệt dictionary lồng nhau

Dùng `items()` ở tầng ngoài, rồi lồng thêm một vòng lặp cho tầng trong:

```python
myfamily = {
    "child1": {"name": "Emil", "year": 2004},
    "child2": {"name": "Tobias", "year": 2007},
    "child3": {"name": "Linus", "year": 2011}
}

for x, obj in myfamily.items():
    print(x)
    for y in obj:
        print(y + ":", obj[y])
```

## ➕ Mở rộng: Dữ liệu thực tế thường lồng nhau

Dữ liệu từ API/JSON thường kết hợp dict và list:

```python
student = {
    "name": "An",
    "scores": {"math": 9, "physics": 7.5},
    "clubs": ["python", "chess"]
}
print(student["scores"]["math"])
print(student["clubs"][0])
student["scores"]["chemistry"] = 8
print(student)
```

> 🧪 **Try it Yourself** — In năm sinh của Linus, rồi thêm child4 vào gia đình.

```python
myfamily = {
    "child1": {"name": "Emil", "year": 2004},
    "child3": {"name": "Linus", "year": 2011}
}
print(myfamily["child3"]["year"])
myfamily["child4"] = {"name": "Anna", "year": 2015}
print(myfamily)
```

## 🧠 Ghi nhớ nhanh

> - Dictionary có thể chứa dictionary: `d["outer"]["inner"]`.
> - Duyệt lồng nhau: `for k, sub in d.items():` rồi `for kk in sub:`.
> - Dữ liệu JSON/API thường là dict + list lồng nhau.
