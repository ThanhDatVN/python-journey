# Python JSON

**JSON** là cú pháp để lưu trữ và trao đổi dữ liệu. JSON là **văn bản**, viết theo ký pháp object của JavaScript.

## JSON in Python

Python có module có sẵn tên `json` để làm việc với dữ liệu JSON:

```python
import json
```

## Parse JSON - Convert from JSON to Python — Đọc JSON

Có chuỗi JSON? Dùng `json.loads()` để chuyển thành **dictionary** Python:

```python
import json

# chuỗi JSON
x = '{ "name":"John", "age":30, "city":"New York"}'

# đọc JSON
y = json.loads(x)

# kết quả là một dictionary Python
print(y["age"])
```

## Convert from Python to JSON — Chuyển Python thành JSON

Có object Python? Dùng `json.dumps()` để chuyển thành chuỗi JSON:

```python
import json

# dictionary Python
x = {
    "name": "John",
    "age": 30,
    "city": "New York"
}

# chuyển thành JSON
y = json.dumps(x)

# kết quả là một chuỗi JSON
print(y)
```

Các kiểu Python sau chuyển được sang JSON: `dict`, `list`, `tuple`, `str`, `int`, `float`, `True`, `False`, `None`:

```python
import json

print(json.dumps({"name": "John", "age": 30}))
print(json.dumps(["apple", "bananas"]))
print(json.dumps(("apple", "bananas")))
print(json.dumps("hello"))
print(json.dumps(42))
print(json.dumps(31.76))
print(json.dumps(True))
print(json.dumps(False))
print(json.dumps(None))
```

Bảng chuyển đổi Python → JSON:

| Python | JSON |
|---|---|
| `dict` | Object |
| `list` | Array |
| `tuple` | Array |
| `str` | String |
| `int` | Number |
| `float` | Number |
| `True` | `true` |
| `False` | `false` |
| `None` | `null` |

Chuyển object Python chứa đủ các kiểu dữ liệu:

```python
import json

x = {
    "name": "John",
    "age": 30,
    "married": True,
    "divorced": False,
    "children": ("Ann", "Billy"),
    "pets": None,
    "cars": [
        {"model": "BMW 230", "mpg": 27.5},
        {"model": "Ford Edge", "mpg": 24.1}
    ]
}

print(json.dumps(x))
```

## Format the Result — Định dạng kết quả

Chuỗi JSON ở trên khó đọc vì không có thụt lề. Tham số `indent` giúp dễ đọc hơn:

```python
import json

x = {"name": "John", "age": 30, "cars": [{"model": "BMW 230", "mpg": 27.5}]}
print(json.dumps(x, indent=4))
```

Có thể đổi ký tự phân cách bằng tham số `separators` (mặc định là `", "` và `": "`):

```python
import json

x = {"name": "John", "age": 30}
print(json.dumps(x, indent=4, separators=(". ", " = ")))
```

## Order the Result — Sắp xếp kết quả

Tham số `sort_keys=True` sắp xếp các key:

```python
import json

x = {"name": "John", "age": 30, "city": "New York"}
print(json.dumps(x, indent=4, sort_keys=True))
```

## ➕ Mở rộng: Tiếng Việt trong JSON

Mặc định, ký tự không phải ASCII bị chuyển thành mã `\uXXXX`. Dùng `ensure_ascii=False` để giữ nguyên tiếng Việt:

```python
import json

data = {"tên": "Nguyễn An", "thành_phố": "Hà Nội"}
print(json.dumps(data))
print(json.dumps(data, ensure_ascii=False))
```

> 🧪 **Try it Yourself** — Đọc chuỗi JSON danh sách sản phẩm và in tên sản phẩm đầu tiên.

```python
import json

text = '[{"name": "Laptop", "price": 1500}, {"name": "Mouse", "price": 25}]'
products = json.loads(text)
print(products[0]["name"])
print(type(products))
```

---

## 🧠 Ghi nhớ nhanh

> - `json.loads(chuỗi)` → **đọc** JSON thành dict/list Python (*s* = string).
> - `json.dumps(obj)` → **ghi** Python thành chuỗi JSON.
> - `indent=4` cho dễ đọc · `sort_keys=True` sắp key · `ensure_ascii=False` giữ tiếng Việt.
> - `True/False/None` ↔ `true/false/null`.
