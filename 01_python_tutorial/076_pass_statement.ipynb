## The pass Statement — Câu lệnh pass

Câu lệnh `if` **không được để trống**. Nếu vì lý do nào đó bạn cần một khối `if` chưa có nội dung, hãy đặt câu lệnh `pass` vào để tránh lỗi:

```python
a = 33
b = 200

if b > a:
    pass
```

`pass` nghĩa là **"không làm gì cả"**.

## Why Use pass? — Vì sao cần pass?

- Khi phác thảo cấu trúc chương trình, chưa viết chi tiết.
- Khi cú pháp **bắt buộc** phải có câu lệnh nhưng bạn không muốn làm gì.
- Đánh dấu chỗ sẽ viết code sau.

## pass in Development — pass khi đang phát triển

```python
age = 16

if age < 18:
    pass  # TODO: thêm logic cho trẻ vị thành niên
else:
    print("Access granted")
```

## pass vs Comments — pass khác comment thế nào?

Comment bị Python **bỏ qua hoàn toàn**, nên một khối chỉ có comment vẫn được coi là **trống** và gây lỗi:

```python
if 5 > 2:
    # sẽ làm sau
```

`pass` là một câu lệnh thật nên khối lệnh hợp lệ:

```python
if 5 > 2:
    # sẽ làm sau
    pass
```

## pass with Multiple Conditions — pass trong nhiều nhánh

```python
value = 50

if value < 0:
    print("Negative value")
elif value == 0:
    pass  # chưa cần xử lý số 0
else:
    print("Positive value")
```

## 🧠 Ghi nhớ nhanh

> - Khối lệnh (`if`, `for`, `def`, `class`…) **không được trống** → dùng `pass` làm chỗ giữ chỗ.
