# Python Shorthand If

> 📖 **Học song song trên W3Schools:** [python_if_shorthand.asp](https://www.w3schools.com/python/python_if_shorthand.asp)
>
> 🎯 **Trong bài này:** Short Hand If · Short Hand If ... Else · Assign a Value With If ... Else · Multiple Conditions on One Line · Practical Examples · When to Use Shorthand If

## Short Hand If — If viết gọn

Nếu chỉ có **một** câu lệnh cần chạy, bạn có thể đặt nó trên cùng dòng với `if`:

```python
a = 200
b = 33
if a > b: print("a is greater than b")
```

## Short Hand If ... Else — If...else viết gọn

Nếu chỉ có một lệnh cho `if` và một lệnh cho `else`, có thể viết tất cả trên một dòng:

```python
a = 2
b = 330
print("A") if a > b else print("B")
```

Cách viết này gọi là **biểu thức điều kiện** hay **toán tử ternary** (xem lại bài [Ternary Operator](028_ternary_operator.ipynb)).

## Assign a Value With If ... Else — Gán giá trị bằng if...else

Biểu thức điều kiện rất hay dùng để **chọn giá trị** gán cho biến:

```python
a = 10
b = 20
bigger = a if a > b else b
print("Bigger is", bigger)
```

Công thức: `giá_trị_nếu_đúng if điều_kiện else giá_trị_nếu_sai`.

## Multiple Conditions on One Line — Nhiều điều kiện trên một dòng

Có thể nối nhiều biểu thức điều kiện:

```python
a = 330
b = 330
print("A") if a > b else print("=") if a == b else print("B")
```

## Practical Examples — Ví dụ thực tế

Tìm giá trị lớn nhất:

```python
x = 15
y = 20
max_value = x if x > y else y
print("Maximum value:", max_value)
```

Gán giá trị mặc định khi dữ liệu trống:

```python
username = ""
display_name = username if username else "Guest"
print("Welcome,", display_name)
```

## When to Use Shorthand If — Khi nào nên viết gọn?

- ✅ Dùng khi điều kiện **đơn giản**, hai nhánh ngắn gọn.
- ❌ Tránh khi logic phức tạp hoặc nhiều nhánh — code khó đọc, khó gỡ lỗi.

> 🧪 **Try it Yourself** — Viết một dòng gán biến status là "Đậu" nếu điểm >= 5, ngược lại "Trượt".

```python
score = 6.5
status = "Đậu" if score >= 5 else "Trượt"
print(status)
```

## 🧠 Ghi nhớ nhanh

> - Một lệnh: `if a > b: print("...")` trên cùng một dòng.
> - If...else một dòng: `x = A if điều_kiện else B`.
> - Chỉ viết gọn khi logic **đơn giản** — ưu tiên dễ đọc.
