# Python Try Except

> 📖 **Học song song trên W3Schools:** [python_try_except.asp](https://www.w3schools.com/python/python_try_except.asp)
>
> 🎯 **Trong bài này:** Exception Handling · Many Exceptions · Else · Finally · Raise an exception · ➕ Mở rộng: Lấy thông tin lỗi với `as`

- Khối `try` cho phép **thử** chạy một đoạn code có thể lỗi.
- Khối `except` cho phép **xử lý** lỗi.
- Khối `else` chạy khi **không có lỗi**.
- Khối `finally` **luôn chạy**, bất kể có lỗi hay không.

## Exception Handling — Xử lý ngoại lệ

Khi có lỗi (gọi là **exception** — ngoại lệ), Python thường **dừng chương trình** và hiện thông báo lỗi.

Các exception này có thể được xử lý bằng câu lệnh `try`.

Khối `try` gây lỗi vì `x` chưa được định nghĩa:

````python
try:
    print(x_undefined)
except:
    print("An exception occurred")
````

Vì có lỗi, khối `except` được chạy. **Không có** khối `try`, chương trình sẽ dừng và báo lỗi:

````python
print(x_undefined)
````

## Many Exceptions — Nhiều loại ngoại lệ

Có thể viết nhiều khối `except` để xử lý **từng loại lỗi** khác nhau:

````python
try:
    print(x_undefined)
except NameError:
    print("Variable x is not defined")
except:
    print("Something else went wrong")
````

Xem danh sách loại lỗi: [Built-in Exceptions](../10_reference/exceptions.md).

## Else — Khối else

Khối `else` chạy khi **không có lỗi** xảy ra trong `try`:

````python
try:
    print("Hello")
except:
    print("Something went wrong")
else:
    print("Nothing went wrong")
````

## Finally — Khối finally

Khối `finally` **luôn được chạy**, dù `try` có lỗi hay không:

````python
try:
    print(x_undefined)
except:
    print("Something went wrong")
finally:
    print("The 'try except' is finished")
````

`finally` hữu ích để **đóng/giải phóng tài nguyên**, ví dụ đóng file:

````python
try:
    f = open("demofile.txt", "w", encoding="utf-8")
    try:
        f.write("Lorum Ipsum")
    except:
        print("Something went wrong when writing to the file")
    finally:
        f.close()
except:
    print("Something went wrong when opening the file")
print("File closed:", f.closed)
````

Chương trình tiếp tục chạy mà không để file mở.

## Raise an exception — Chủ động phát sinh lỗi

Bạn có thể **chủ động** phát sinh lỗi khi điều kiện không thoả, bằng từ khoá `raise`:

````python
x = -1

if x < 0:
    raise Exception("Sorry, no numbers below zero")
````

Có thể chỉ định **loại** lỗi và thông báo:

````python
x = "hello"

if not type(x) is int:
    raise TypeError("Only integers are allowed")
````

## ➕ Mở rộng: Lấy thông tin lỗi với `as`

````python
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("Lỗi:", e)
    print("Loại lỗi:", type(e).__name__)
````

Bắt nhiều loại lỗi trong một `except`:

````python
def to_number(text):
    try:
        return int(text)
    except (ValueError, TypeError):
        return None

print(to_number("42"), to_number("abc"), to_number(None))
````

> 🧪 **Try it Yourself** — Viết hàm chia an toàn: chia cho 0 thì trả về None thay vì báo lỗi.

````python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Không thể chia cho 0!")
        return None

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

## 🧠 Ghi nhớ nhanh

> - `try` thử · `except` xử lý lỗi · `else` khi **không** lỗi · `finally` **luôn chạy**.
> - Bắt **đúng loại lỗi**: `except ValueError:` tốt hơn `except:` chung chung.
> - `except Exception as e:` → lấy thông tin lỗi.
> - `raise TypeError("...")` → chủ động phát sinh lỗi.
