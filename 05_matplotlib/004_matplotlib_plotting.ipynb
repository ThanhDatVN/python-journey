## Plotting x and y points — Vẽ các điểm x, y

Hàm `plot()` dùng để vẽ các điểm (marker) trên đồ thị.

Mặc định, `plot()` vẽ **đường nối** từ điểm này sang điểm khác.

- Tham số 1: mảng chứa các toạ độ trên **trục x**.
- Tham số 2: mảng chứa các toạ độ trên **trục y**.

Muốn vẽ đường từ (1, 3) đến (8, 10), truyền hai mảng `[1, 8]` và `[3, 10]`:

```python
import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

plt.plot(xpoints, ypoints)
plt.show()
```

> 📌 Trục **x** là trục ngang, trục **y** là trục dọc.

## Plotting Without Line — Chỉ vẽ điểm

Để chỉ vẽ điểm (không nối), dùng tham số rút gọn `'o'` (nghĩa là "rings" — vòng tròn):

```python
import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

plt.plot(xpoints, ypoints, 'o')
plt.show()
```

## Multiple Points — Nhiều điểm

Vẽ bao nhiêu điểm cũng được, chỉ cần hai mảng có **cùng số phần tử**.

Vẽ đường đi qua (1, 3), (2, 8), (6, 1), (8, 10):

```python
import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([1, 2, 6, 8])
ypoints = np.array([3, 8, 1, 10])

plt.plot(xpoints, ypoints)
plt.show()
```

## Default X-Points — Toạ độ x mặc định

Nếu **không** truyền toạ độ x, Matplotlib tự dùng `0, 1, 2, 3…` (tuỳ theo số điểm y):

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10, 5, 7])

plt.plot(ypoints)
plt.show()
```

> Toạ độ x trong ví dụ trên là `[0, 1, 2, 3, 4, 5]`.

> 🧪 **Try it Yourself** — Vẽ doanh thu 6 tháng (đơn vị: triệu đồng) chỉ với mảng y.

```python
import matplotlib.pyplot as plt
import numpy as np

revenue = np.array([120, 135, 128, 150, 170, 165])
plt.plot(revenue)
plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - `plt.plot(x, y)` → nối các điểm bằng đường; `plt.plot(x, y, 'o')` → chỉ vẽ điểm.
> - Mảng x và y phải **cùng độ dài**.
> - Bỏ x → Matplotlib dùng `0, 1, 2…` làm toạ độ x.
