## Creating Scatter Plots — Biểu đồ phân tán

Hàm `scatter()` vẽ **một chấm cho mỗi quan sát**. Cần hai mảng cùng độ dài: một cho trục x, một cho trục y.

Ví dụ: dữ liệu về 13 chiếc xe — trục x là **tuổi xe**, trục y là **tốc độ**:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])

plt.scatter(x, y)
plt.show()
```

Có vẻ như **xe càng mới thì chạy càng nhanh** — nhưng đó có thể chỉ là tình cờ, vì ta mới có 13 xe.

## Compare Plots — So sánh hai nhóm dữ liệu

Vẽ hai nhóm dữ liệu trên cùng một hình:

```python
import matplotlib.pyplot as plt
import numpy as np

# ngày 1:
x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])
plt.scatter(x, y)

# ngày 2:
x = np.array([2, 2, 8, 1, 15, 8, 12, 9, 7, 3, 11, 4, 7, 14, 12])
y = np.array([100, 105, 84, 105, 90, 99, 90, 95, 94, 100, 79, 112, 91, 80, 85])
plt.scatter(x, y)

plt.show()
```

## Colors — Màu

Đặt màu cho từng nhóm bằng `color` hoặc `c`:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])
plt.scatter(x, y, color='hotpink')

x = np.array([2, 2, 8, 1, 15, 8, 12, 9, 7, 3, 11, 4, 7, 14, 12])
y = np.array([100, 105, 84, 105, 90, 99, 90, 95, 94, 100, 79, 112, 91, 80, 85])
plt.scatter(x, y, color='#88c999')

plt.show()
```

## Color Each Dot — Màu cho từng chấm

Truyền **mảng màu** cho tham số `c`:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])
colors = np.array(["red", "green", "blue", "yellow", "pink", "black", "orange",
                   "purple", "beige", "brown", "gray", "cyan", "magenta"])

plt.scatter(x, y, c=colors)
plt.show()
```

## ColorMap — Bảng màu

**Colormap** là một dải màu, mỗi màu ứng với một giá trị từ 0 đến 100. Ví dụ colormap `'viridis'` đi từ **tím** (0) qua xanh tới **vàng** (100).

### How to Use the ColorMap — Cách dùng

Đặt `cmap='viridis'`, rồi truyền mảng giá trị (0-100) cho `c`:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])
colors = np.array([0, 10, 20, 30, 40, 45, 50, 55, 60, 70, 80, 90, 100])

plt.scatter(x, y, c=colors, cmap='viridis')
plt.show()
```

### Include the Actual ColorMap — Hiển thị thanh màu

Thêm `plt.colorbar()`:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])
colors = np.array([0, 10, 20, 30, 40, 45, 50, 55, 60, 70, 80, 90, 100])

plt.scatter(x, y, c=colors, cmap='viridis')
plt.colorbar()
plt.show()
```

### Available ColorMaps — Các colormap có sẵn

Một số colormap phổ biến: `'viridis'`, `'plasma'`, `'inferno'`, `'magma'`, `'cividis'`, `'Blues'`, `'Greens'`, `'Reds'`, `'coolwarm'`, `'rainbow'`, `'jet'`… Thêm hậu tố `_r` để **đảo ngược** dải màu (ví dụ `'viridis_r'`).

```python
import matplotlib.pyplot as plt

print(len(plt.colormaps()), "colormaps, ví dụ:", list(plt.colormaps())[:8])
```

## Size — Kích thước chấm

Tham số `s` đặt kích thước; truyền mảng để mỗi chấm có kích thước riêng:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])
sizes = np.array([20, 50, 100, 200, 500, 1000, 60, 90, 10, 300, 600, 800, 75])

plt.scatter(x, y, s=sizes)
plt.show()
```

## Alpha — Độ trong suốt

Tham số `alpha` (0 → 1) điều chỉnh độ trong suốt:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6])
y = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])
sizes = np.array([20, 50, 100, 200, 500, 1000, 60, 90, 10, 300, 600, 800, 75])

plt.scatter(x, y, s=sizes, alpha=0.5)
plt.show()
```

## Combine Color Size and Alpha — Kết hợp màu, kích thước, độ trong suốt

```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(seed=1)
x = rng.integers(100, size=100)
y = rng.integers(100, size=100)
colors = rng.integers(100, size=100)
sizes = 10 * rng.integers(100, size=100)

plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='nipy_spectral')
plt.colorbar()
plt.show()
```

> 🧪 **Try it Yourself** — Vẽ scatter chiều cao - cân nặng của 6 bạn, kích thước chấm theo tuổi.

```python
import matplotlib.pyplot as plt
import numpy as np

height = np.array([160, 172, 155, 180, 168, 175])
weight = np.array([52, 68, 48, 77, 60, 70])
age = np.array([18, 22, 17, 25, 20, 23])
plt.scatter(height, weight, s=age * 15, c=age, cmap="plasma", alpha=0.7)
plt.colorbar()
plt.show()
```
