# Matplotlib Pie Charts

> 📖 **Học song song trên W3Schools:** [matplotlib_pie_charts.asp](https://www.w3schools.com/python/matplotlib_pie_charts.asp)
>
> 🎯 **Trong bài này:** Creating Pie Charts · Labels · Start Angle · Explode · Shadow · Colors · Legend · ➕ Mở rộng: Hiển thị phần trăm

## Creating Pie Charts — Biểu đồ tròn

Dùng hàm `pie()`:

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])

plt.pie(y)
plt.show()
```

Mỗi giá trị trong mảng là một **miếng (wedge)**. Mặc định, miếng đầu tiên bắt đầu từ trục x và đi **ngược chiều kim đồng hồ**.

> 📌 Kích thước mỗi miếng = giá trị đó chia cho tổng các giá trị: `x / sum(x)`.

## Labels — Nhãn

Dùng tham số `labels` (một mảng nhãn cho mỗi miếng):

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.pie(y, labels=mylabels)
plt.show()
```

## Start Angle — Góc bắt đầu

Mặc định góc bắt đầu là trục x (0 độ). Đổi bằng `startangle` (đơn vị độ):

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.pie(y, labels=mylabels, startangle=90)
plt.show()
```

## Explode — Tách miếng

Tham số `explode` (mảng, mỗi phần tử ứng với một miếng) tách miếng ra khỏi tâm:

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
myexplode = [0.2, 0, 0, 0]

plt.pie(y, labels=mylabels, explode=myexplode)
plt.show()
```

## Shadow — Bóng đổ

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
myexplode = [0.2, 0, 0, 0]

plt.pie(y, labels=mylabels, explode=myexplode, shadow=True)
plt.show()
```

## Colors — Màu

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
mycolors = ["black", "hotpink", "b", "#4CAF50"]

plt.pie(y, labels=mylabels, colors=mycolors)
plt.show()
```

## Legend — Chú giải

Thêm danh sách giải thích cho từng miếng bằng `legend()`:

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.pie(y, labels=mylabels)
plt.legend()
plt.show()
```

### Legend With Header — Chú giải có tiêu đề

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]

plt.pie(y, labels=mylabels)
plt.legend(title="Four Fruits:")
plt.show()
```

## ➕ Mở rộng: Hiển thị phần trăm

```python
import matplotlib.pyplot as plt

expenses = [4500, 2000, 1500, 1000, 1000]
labels = ["Nhà ở", "Ăn uống", "Đi lại", "Giải trí", "Tiết kiệm"]
plt.pie(expenses, labels=labels, autopct="%1.1f%%", startangle=90)
plt.title("Chi tiêu tháng (nghìn đồng)")
plt.show()
```

> 🧪 **Try it Yourself** — Vẽ biểu đồ tròn thời gian trong ngày của bạn, tách miếng "Học Python".

```python
import matplotlib.pyplot as plt

hours = [8, 8, 3, 2, 3]
labels = ["Ngủ", "Làm việc", "Học Python", "Ăn uống", "Khác"]
plt.pie(hours, labels=labels, explode=[0, 0, 0.15, 0, 0], autopct="%1.0f%%")
plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - `plt.pie(values)` — mỗi giá trị là một miếng, tỉ lệ `x / sum(x)`.
> - `labels` nhãn · `startangle` góc bắt đầu · `explode` tách miếng · `shadow` bóng · `colors` màu.
> - `plt.legend(title=...)` chú giải · `autopct="%1.1f%%"` hiện phần trăm.
