## Histogram — Biểu đồ tần suất

**Histogram** là biểu đồ thể hiện **phân bố tần suất**: số lượng quan sát rơi vào mỗi **khoảng** giá trị cho trước.

Ví dụ: khảo sát chiều cao của 250 người, histogram có thể cho thấy:

- 2 người cao từ 140 đến 145 cm
- 5 người cao từ 145 đến 150 cm
- 15 người cao từ 151 đến 156 cm
- …

## Create Histogram — Vẽ histogram

Dùng hàm `hist()`. Hàm nhận một mảng số và tự chia khoảng, đếm tần suất.

Để đơn giản, ta dùng NumPy tạo ngẫu nhiên 250 giá trị, tập trung quanh **170**, độ lệch chuẩn **10** (xem thêm [Normal Data Distribution](../06_machine_learning/006_ml_normal_data_distribution.ipynb)):

```python
import numpy as np

np.random.seed(0)
x = np.random.normal(170, 10, 250)

print(x[:10])
```

`hist()` đọc mảng và vẽ histogram:

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)
x = np.random.normal(170, 10, 250)

plt.hist(x)
plt.show()
```

## ➕ Mở rộng: Số cột (bins) và tuỳ chỉnh

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)
x = np.random.normal(170, 10, 250)

plt.hist(x, bins=20, color="orange", edgecolor="black")
plt.title("Chiều cao của 250 người")
plt.xlabel("cm")
plt.ylabel("Số người")
plt.show()
```

> 🧪 **Try it Yourself** — Vẽ histogram điểm thi của 500 học sinh: trung bình 6.5, độ lệch chuẩn 1.5, 15 bins.

```python
import matplotlib.pyplot as plt
import numpy as np

scores = np.random.normal(6.5, 1.5, 500)
plt.hist(scores, bins=15, edgecolor="white")
plt.show()
```
