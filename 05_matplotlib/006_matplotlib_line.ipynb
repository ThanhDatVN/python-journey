## Linestyle — Kiểu đường

Dùng tham số `linestyle` (viết tắt `ls`) để đổi kiểu đường:

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, linestyle='dotted')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, linestyle='dashed')
plt.show()
```

## Shorter Syntax — Cú pháp ngắn

- `linestyle` viết tắt là `ls`.
- `dotted` viết tắt là `:`; `dashed` viết tắt là `--`.

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, ls=':')
plt.show()
```

## Line Styles — Bảng kiểu đường

| Kiểu | Viết tắt |
|---|---|
| `'solid'` (mặc định) | `'-'` |
| `'dotted'` | `':'` |
| `'dashed'` | `'--'` |
| `'dashdot'` | `'-.'` |
| `'None'` | `''` hoặc `' '` |

## Line Color — Màu đường

Dùng `color` hoặc viết tắt `c`:

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, color='r')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, c='#4CAF50')
plt.show()
```

## Line Width — Độ dày đường

Dùng `linewidth` hoặc viết tắt `lw`. Giá trị là số thực, đơn vị điểm (points):

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, linewidth='20.5')
plt.show()
```

## Multiple Lines — Nhiều đường

Vẽ nhiều đường bằng cách gọi `plt.plot()` nhiều lần:

```python
import matplotlib.pyplot as plt
import numpy as np

y1 = np.array([3, 8, 1, 10])
y2 = np.array([6, 2, 7, 11])

plt.plot(y1)
plt.plot(y2)
plt.show()
```

Hoặc truyền nhiều cặp x, y trong **một** lần gọi `plt.plot()`:

```python
import matplotlib.pyplot as plt
import numpy as np

x1 = np.array([0, 1, 2, 3])
y1 = np.array([3, 8, 1, 10])
x2 = np.array([0, 1, 2, 3])
y2 = np.array([6, 2, 7, 11])

plt.plot(x1, y1, x2, y2)
plt.show()
```

> 🧪 **Try it Yourself** — So sánh nhiệt độ 2 thành phố: một đường nét liền đỏ, một đường nét đứt xanh, dày 3.

```python
import matplotlib.pyplot as plt
import numpy as np

hanoi = np.array([17, 19, 22, 26, 29, 31])
hcm = np.array([27, 28, 29, 30, 29, 28])
plt.plot(hanoi, c='r', lw=3)
plt.plot(hcm, c='b', ls='--', lw=3)
plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - `ls` kiểu đường: `'-'` liền · `':'` chấm · `'--'` đứt · `'-.'` gạch-chấm.
> - `c` màu · `lw` độ dày.
> - Nhiều đường: gọi `plt.plot()` nhiều lần, hoặc `plt.plot(x1, y1, x2, y2)`.
