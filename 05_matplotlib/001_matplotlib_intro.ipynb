## What is Matplotlib? — Matplotlib là gì?

**Matplotlib** là thư viện vẽ đồ thị **cấp thấp** của Python, dùng để **trực quan hoá dữ liệu**.

- Được tạo bởi **John D. Hunter**.
- Là phần mềm **mã nguồn mở**, dùng miễn phí.
- Phần lớn viết bằng Python; một số phần viết bằng C, Objective-C và JavaScript để tương thích đa nền tảng.

Với Matplotlib bạn vẽ được: đường (line), điểm (scatter), cột (bar), histogram, biểu đồ tròn (pie)… — và chương này sẽ đi qua lần lượt từng loại.

## Where is the Matplotlib Codebase? — Mã nguồn ở đâu?

Mã nguồn Matplotlib nằm trên GitHub: [github.com/matplotlib/matplotlib](https://github.com/matplotlib/matplotlib)

Xem trước sức mạnh của Matplotlib chỉ với vài dòng code:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
plt.legend()
plt.title("Matplotlib preview")
plt.show()
```

> 💡 Chương này dùng thêm thư viện **NumPy** (`np`) để tạo dữ liệu số. Bạn có thể học NumPy tại [04_python_modules/numpy](../04_python_modules/numpy/README.md).
