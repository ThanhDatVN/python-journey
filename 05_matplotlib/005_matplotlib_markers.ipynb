## Markers — Điểm đánh dấu

Dùng tham số `marker` để đánh dấu mỗi điểm bằng một ký hiệu.

Đánh dấu mỗi điểm bằng hình tròn:

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, marker='o')
plt.show()
```

Đánh dấu bằng hình sao:

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, marker='*')
plt.show()
```

## Marker Reference — Bảng ký hiệu marker

| Marker | Mô tả | Marker | Mô tả |
|---|---|---|---|
| `'o'` | Tròn | `'*'` | Sao |
| `'.'` | Chấm | `','` | Pixel |
| `'x'` | Chữ X | `'X'` | X (tô đặc) |
| `'+'` | Dấu cộng | `'P'` | Dấu cộng (tô đặc) |
| `'s'` | Vuông | `'D'` | Kim cương |
| `'d'` | Kim cương (mảnh) | `'p'` | Ngũ giác |
| `'H'` / `'h'` | Lục giác | `'v'` | Tam giác xuống |
| `'^'` | Tam giác lên | `'<'` / `'>'` | Tam giác trái / phải |
| `'1'` `'2'` `'3'` `'4'` | Tri (chạc ba) các hướng | `'\|'` / `'_'` | Vạch đứng / ngang |

## Format Strings fmt — Chuỗi định dạng

Có thể dùng **chuỗi định dạng rút gọn** `fmt` theo cú pháp `marker|line|color`:

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, 'o:r')
plt.show()
```

`'o:r'` = marker tròn, nét chấm `:`, màu đỏ `r`.

### Line Reference — Kiểu đường

| Cú pháp | Mô tả |
|---|---|
| `'-'` | Nét liền |
| `':'` | Nét chấm |
| `'--'` | Nét đứt |
| `'-.'` | Nét gạch-chấm |

> 📌 Nếu bỏ phần *line* trong `fmt`, sẽ chỉ vẽ marker, không có đường nối.

### Color Reference — Màu

| Cú pháp | Màu | Cú pháp | Màu |
|---|---|---|---|
| `'r'` | Đỏ | `'g'` | Xanh lá |
| `'b'` | Xanh dương | `'c'` | Lục lam |
| `'m'` | Tím hồng | `'y'` | Vàng |
| `'k'` | Đen | `'w'` | Trắng |

## Marker Size — Kích thước marker

Dùng `markersize` hoặc viết tắt `ms`:

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, marker='o', ms=20)
plt.show()
```

## Marker Color — Màu marker

- `markeredgecolor` / `mec`: màu **viền** marker.
- `markerfacecolor` / `mfc`: màu **bên trong** marker.

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, marker='o', ms=20, mec='r')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, marker='o', ms=20, mec='r', mfc='r')
plt.show()
```

Có thể dùng mã màu **hex** hoặc **tên màu**:

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, marker='o', ms=20, mec='#4CAF50', mfc='hotpink')
plt.show()
```

> 🧪 **Try it Yourself** — Thử các fmt khác nhau: 's--g', '^-.b', 'D:k'.

```python
import matplotlib.pyplot as plt
import numpy as np

ypoints = np.array([3, 8, 1, 10])
plt.plot(ypoints, 's--g')
plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - `marker='o'` (tròn), `'*'` (sao), `'s'` (vuông), `'^'` (tam giác)…
> - fmt rút gọn: `'marker line color'` → `'o:r'` = tròn + chấm + đỏ.
> - `ms` kích thước · `mec` màu viền · `mfc` màu trong; màu dùng được tên, `'r'`, hoặc hex `'#4CAF50'`.
