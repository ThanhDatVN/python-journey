## Installation of Matplotlib — Cài đặt

Nếu đã có Python và PIP, cài Matplotlib bằng lệnh:

```bash
pip install matplotlib
```

Trong notebook:

````python
%pip install matplotlib
````

> 📌 Nếu bạn đã chạy `pip install -r requirements.txt` của repo này thì Matplotlib đã được cài sẵn.

## Import Matplotlib

Sau khi cài, import vào chương trình bằng từ khoá `import`:

````python
import matplotlib
````

Giờ Matplotlib đã sẵn sàng để dùng.

## Checking Matplotlib Version — Kiểm tra phiên bản

Chuỗi phiên bản được lưu trong thuộc tính `__version__`:

````python
import matplotlib

print(matplotlib.__version__)
````

> 📌 Chú ý: hai dấu gạch dưới ở mỗi bên của `version`.
