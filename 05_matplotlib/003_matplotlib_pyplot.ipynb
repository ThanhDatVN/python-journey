## Pyplot

Hầu hết tiện ích của Matplotlib nằm trong module con **`pyplot`**, thường được import với bí danh **`plt`**:

```python
import matplotlib.pyplot as plt
```

Giờ có thể gọi các hàm vẽ qua tên `plt`.

### Example — Vẽ đường thẳng

Vẽ một đường thẳng từ điểm (0, 0) đến điểm (6, 250):

```python
import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([0, 6])
ypoints = np.array([0, 250])

plt.plot(xpoints, ypoints)
plt.show()
```

`plt.show()` hiển thị đồ thị. Trong Jupyter, đồ thị hiện ngay bên dưới cell.

Bạn sẽ tìm hiểu kỹ về cách vẽ ở các bài tiếp theo.

> 🧪 **Try it Yourself** — Đổi điểm cuối thành (10, 100) và quan sát độ dốc của đường thay đổi.

```python
import matplotlib.pyplot as plt
import numpy as np

plt.plot(np.array([0, 10]), np.array([0, 100]))
plt.show()
```
