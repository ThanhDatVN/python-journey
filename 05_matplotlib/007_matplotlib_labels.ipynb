# Matplotlib Labels and Title

> 📖 **Học song song trên W3Schools:** [matplotlib_labels.asp](https://www.w3schools.com/python/matplotlib_labels.asp)
>
> 🎯 **Trong bài này:** Create Labels for a Plot · Create a Title for a Plot · Set Font Properties for Title and Labels · Position the Title · ➕ Mở rộng: Tiếng Việt trong đồ thị

## Create Labels for a Plot — Nhãn trục

Dùng `xlabel()` và `ylabel()` để đặt nhãn cho trục x và trục y:

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.plot(x, y)

plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.show()
```

## Create a Title for a Plot — Tiêu đề

Dùng `title()`:

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.plot(x, y)

plt.title("Sports Watch Data")
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.show()
```

## Set Font Properties for Title and Labels — Font chữ

Dùng tham số `fontdict` để đặt font cho tiêu đề và nhãn:

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

font1 = {'family': 'serif', 'color': 'blue', 'size': 20}
font2 = {'family': 'serif', 'color': 'darkred', 'size': 15}

plt.title("Sports Watch Data", fontdict=font1)
plt.xlabel("Average Pulse", fontdict=font2)
plt.ylabel("Calorie Burnage", fontdict=font2)

plt.plot(x, y)
plt.show()
```

## Position the Title — Vị trí tiêu đề

Tham số `loc` của `title()` nhận `'left'`, `'right'` hoặc `'center'` (mặc định):

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.title("Sports Watch Data", loc='left')
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.plot(x, y)
plt.show()
```

## ➕ Mở rộng: Tiếng Việt trong đồ thị

Font mặc định của Matplotlib (DejaVu Sans) hiển thị tốt tiếng Việt có dấu:

```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3], [2, 4, 3], marker="o")
plt.title("Điểm trung bình theo học kỳ")
plt.xlabel("Học kỳ")
plt.ylabel("Điểm")
plt.show()
```

> 🧪 **Try it Yourself** — Thêm tiêu đề căn phải và nhãn trục cho đồ thị số bước chân 7 ngày.

```python
import matplotlib.pyplot as plt

steps = [6500, 7200, 8100, 5400, 9000, 12000, 10500]
plt.plot(steps, marker="o")
plt.title("Số bước chân trong tuần", loc="right")
plt.xlabel("Ngày")
plt.ylabel("Số bước")
plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - `plt.xlabel()` · `plt.ylabel()` · `plt.title()`.
> - Font: `fontdict={'family': ..., 'color': ..., 'size': ...}`.
> - Vị trí tiêu đề: `plt.title(..., loc='left' | 'center' | 'right')`.
