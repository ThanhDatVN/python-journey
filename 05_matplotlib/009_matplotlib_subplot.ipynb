## Display Multiple Plots — Nhiều đồ thị trong một hình

Hàm `subplot()` cho phép vẽ **nhiều đồ thị** trong cùng một hình:

````python
import matplotlib.pyplot as plt
import numpy as np

# plot 1:
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(1, 2, 1)
plt.plot(x, y)

# plot 2:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(1, 2, 2)
plt.plot(x, y)

plt.show()
````

## The subplot() Function — Hàm subplot()

`subplot()` nhận **ba tham số** mô tả bố cục: **số hàng**, **số cột**, và **vị trí** của đồ thị hiện tại.

```py
plt.subplot(1, 2, 1)
# hình có 1 hàng, 2 cột, đây là đồ thị THỨ NHẤT

plt.subplot(1, 2, 2)
# hình có 1 hàng, 2 cột, đây là đồ thị THỨ HAI
```

Muốn xếp các đồ thị **chồng lên nhau** theo chiều dọc, dùng 2 hàng 1 cột:

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(2, 1, 1)
plt.plot(x, y)

x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(2, 1, 2)
plt.plot(x, y)

plt.show()
````

Vẽ 6 đồ thị (2 hàng × 3 cột):

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

for i in range(1, 7):
    plt.subplot(2, 3, i)
    plt.plot(x, y * i)

plt.show()
````

## Title — Tiêu đề từng đồ thị

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title("SALES")

x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(1, 2, 2)
plt.plot(x, y)
plt.title("INCOME")

plt.show()
````

## Super Title — Tiêu đề chung

Dùng `suptitle()` để đặt tiêu đề cho **cả hình**:

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title("SALES")

x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(1, 2, 2)
plt.plot(x, y)
plt.title("INCOME")

plt.suptitle("MY SHOP")
plt.show()
````

## ➕ Mở rộng: plt.subplots() — cách hướng đối tượng

````python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(1, 2, figsize=(8, 3))
axes[0].plot([3, 8, 1, 10])
axes[0].set_title("SALES")
axes[1].bar(["A", "B", "C"], [5, 7, 3])
axes[1].set_title("PRODUCTS")
fig.suptitle("MY SHOP")
plt.show()
````

---

## 🧠 Ghi nhớ nhanh

> - `plt.subplot(hàng, cột, vị_trí)` — vị trí đếm từ **1**, trái → phải, trên → dưới.
> - `plt.title()` cho từng đồ thị · `plt.suptitle()` cho cả hình.
> - Hiện đại: `fig, axes = plt.subplots(hàng, cột)`.
