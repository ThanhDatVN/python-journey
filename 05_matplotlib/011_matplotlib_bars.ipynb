# Matplotlib Bars

> 📖 **Học song song trên W3Schools:** [matplotlib_bars.asp](https://www.w3schools.com/python/matplotlib_bars.asp)
>
> 🎯 **Trong bài này:** Creating Bars · Horizontal Bars · Bar Color · Bar Width · Bar Height · ➕ Mở rộng: Hiển thị giá trị trên cột

## Creating Bars — Biểu đồ cột

Dùng hàm `bar()` để vẽ biểu đồ cột:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.bar(x, y)
plt.show()
```

`bar()` nhận các tham số mô tả bố cục: danh mục (trục x) và giá trị (chiều cao cột). Có thể dùng list Python bình thường:

```python
import matplotlib.pyplot as plt

x = ["APPLES", "BANANAS"]
y = [400, 350]
plt.bar(x, y)
plt.show()
```

## Horizontal Bars — Cột nằm ngang

Dùng `barh()`:

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.barh(x, y)
plt.show()
```

## Bar Color — Màu cột

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.bar(x, y, color="red")
plt.show()
```

### Color Names — Tên màu

Có thể dùng một trong [140 tên màu được hỗ trợ](https://www.w3schools.com/colors/colors_names.asp):

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.bar(x, y, color="hotpink")
plt.show()
```

### Color Hex — Mã màu hex

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.bar(x, y, color="#4CAF50")
plt.show()
```

## Bar Width — Độ rộng cột

Tham số `width` (mặc định `0.8`):

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.bar(x, y, width=0.1)
plt.show()
```

> 📌 Với cột nằm ngang, dùng `height` thay vì `width`.

## Bar Height — Độ cao cột ngang

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.barh(x, y, height=0.1)
plt.show()
```

## ➕ Mở rộng: Hiển thị giá trị trên cột

```python
import matplotlib.pyplot as plt

fruits = ["Táo", "Chuối", "Cam", "Xoài"]
sold = [35, 50, 28, 42]
bars = plt.bar(fruits, sold, color="#4CAF50")
plt.bar_label(bars)
plt.title("Số kg trái cây bán được")
plt.show()
```

> 🧪 **Try it Yourself** — Vẽ biểu đồ cột ngang số giờ học Python mỗi ngày trong tuần, màu "skyblue".

```python
import matplotlib.pyplot as plt

days = ["T2", "T3", "T4", "T5", "T6", "T7", "CN"]
hours = [1, 1.5, 0.5, 2, 1, 3, 2.5]
plt.barh(days, hours, color="skyblue")
plt.xlabel("Giờ học")
plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - `plt.bar(x, y)` cột đứng · `plt.barh(x, y)` cột ngang.
> - `color` (tên màu hoặc hex) · `width` (cột đứng) · `height` (cột ngang).
> - Hiện số trên cột: `plt.bar_label(bars)`.
