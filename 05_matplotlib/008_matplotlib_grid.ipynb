# Matplotlib Adding Grid Lines

> 📖 **Học song song trên W3Schools:** [matplotlib_grid.asp](https://www.w3schools.com/python/matplotlib_grid.asp)
>
> 🎯 **Trong bài này:** Add Grid Lines to a Plot · Specify Which Grid Lines to Display · Set Line Properties for the Grid

## Add Grid Lines to a Plot — Thêm lưới

Dùng hàm `grid()` để thêm đường lưới:

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.title("Sports Watch Data")
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.plot(x, y)

plt.grid()

plt.show()
```

## Specify Which Grid Lines to Display — Chọn đường lưới

Tham số `axis` nhận `'x'`, `'y'` hoặc `'both'` (mặc định):

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.title("Sports Watch Data")
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.plot(x, y)

plt.grid(axis='x')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.plot(x, y)
plt.grid(axis='y')
plt.show()
```

## Set Line Properties for the Grid — Kiểu đường lưới

Dùng `grid(color='color', linestyle='linestyle', linewidth=number)`:

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.title("Sports Watch Data")
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.plot(x, y)

plt.grid(color='green', linestyle='--', linewidth=0.5)

plt.show()
```

---

## 🧠 Ghi nhớ nhanh

> - `plt.grid()` thêm lưới; `axis='x' | 'y' | 'both'`.
> - Tuỳ chỉnh: `plt.grid(color=..., linestyle=..., linewidth=...)`.
